# CONTROL CON RECHAZO ACTIVO DE PERTURBACIONES PARA LA REGULACIÓN DE OXÍGENO DISUELTO EN LA ETAPA DE LODOS ACTIVADOS DE UNA PTAR

## 1.Model

### 1.1 Mass balance 
$$
\text{Acumulación}=\text{Entrada}-\text{Salida}+\text{Generación}-\text{Consumo}
$$

$$
\frac{dS}{dt}=\frac{1}{V}\sum_{i=1}^3 Q_{in,i}S_{in,i}-\frac{Q_{out}}{V}S+K_{L_a}\left(S^{*}-S\right)-r_z(S)
$$

$$
\underbrace{\frac{dS}{dt}}_{\frac{\mathrm{mg}}{\mathrm{L}\cdot\mathrm{h}}}
=
\underbrace{\frac{1}{V}\sum_{i=1}^3 Q_{in,i}S_{in,i}}_{\frac{1}{\mathrm{L}}
\left(\frac{\mathrm{L}}{\mathrm{h}}\right)
\left(\frac{\mathrm{mg}}{\mathrm{L}}\right)
=
\frac{\mathrm{mg}}{\mathrm{L}\cdot\mathrm{h}}}
-
\underbrace{\frac{Q_{out}}{V}S}_{\left(\frac{\mathrm{L}/\mathrm{h}}{\mathrm{L}}\right)
\left(\frac{\mathrm{mg}}{\mathrm{L}}\right)
=
\frac{\mathrm{mg}}{\mathrm{L}\cdot\mathrm{h}}}
+
\underbrace{K_{La}(S^{*}-S)}_{\left(\frac{1}{\mathrm{h}}\right)
\left(\frac{\mathrm{mg}}{\mathrm{L}}\right)
=
\frac{\mathrm{mg}}{\mathrm{L}\cdot\mathrm{h}}}
-
\underbrace{r_z(S)}_{\frac{\mathrm{mg}}{\mathrm{L}\cdot\mathrm{h}}}
$$
With $r_Z:$
\begin{equation}
    r_{Z} = -\frac{1 - Y_H}{Y_H}\,\rho_1 - \frac{4.57 - Y_A}{Y_A}\,\rho_3
\end{equation}

\begin{equation}
    \rho_1 = \mu_H \left(\frac{S_S}{K_S + S_S}\right)\left(\frac{S_O}{K_{O,H} + S_O}\right) X_{B,H}
\end{equation}

\begin{equation}
    \rho_3 = \mu_A \left(\frac{S_{NH}}{K_{NH} + S_{NH}}\right)\left(\frac{S_O}{K_{O,A} + S_O}\right) X_{B,A}
\end{equation}

If the oxygen transfer coefficient is considered proportional to the control input,

\begin{equation}
K_{La} \propto u
\end{equation}

then,

\begin{equation}
K_{La} = \alpha u
\end{equation}

where $\alpha$ is a proportionality constant.

Substituting into the dissolved oxygen mass balance yields

\begin{equation}
\dot S
=
\underbrace{
\frac{1}{V}\sum_{i=1}^3 Q_{in,i}S_{in,i}
-\frac{Q_{out}}{V}S
-r_z
}_{f(S,t)}
+
u\,
\underbrace{
\alpha\left(S^{*}-S\right)
}_{g(S,t)}.
\end{equation}

\begin{equation}
\dot S = f(S,t) + g(S,t)\,u
\end{equation}

where

\begin{equation}
g(S,t)=\alpha\left(S^{*}-S\right)
\end{equation}

and

\begin{equation}
f(S,t)
=
\frac{1}{V}\sum_{i=1}^3 Q_{in,i}S_{in,i}
-\frac{Q_{out}}{V}S
-r_z(S)
\end{equation}


All dynamics that differ from 
a pure integrator are grouped as a total disturbance $f$, 


\begin{equation}
    \dot{y} = f + b_0 \cdot K_{L_a}
\end{equation}

where the total disturbance $f$ is defined as:

\begin{equation}
f(S,t)
=
\frac{1}{V}\sum_{i=1}^3 Q_{in,i}S_{in,i}
-\frac{Q_{out}}{V}S
-r_z(S)+K_{L_a}\left(S^{*}-S\right)-b_0 \cdot K_{L_a}
\end{equation}

relationship between Oxygen transfer coefficient $K_{La}$ and $u$. Oxygen transfer coefficient is 
proportional to the control signal $u$ applied to the 
blowers:

\begin{equation}
    K_{L_a} \propto u
\end{equation}

\begin{equation}
     K_{L_a} = \alpha \cdot u
\end{equation}

with $k = bo \cdot \alpha$ 

$$\dot{y} = f + k \cdot u$$


The oxygen mass balance is affine in the transfer coefficient:

\begin{equation}
\dot S_O
=
\underbrace{
\frac{1}{V}\sum_{i=1}^{3} Q_{in,i}S_{O,in,i}
-\frac{Q_{out}}{V}S_O
- r_{O}(\mathbf{x})
}_{f(\mathbf{x},t)}
\;+\;
\underbrace{\left(S_O^{*}-S_O\right)}_{g(S_O)}\,K_{L_a}
\end{equation}

where $f$ collects convective transport and biological uptake, and
$g$ is the driving force for oxygen transfer, in $\mathrm{mg/L}$.

Over the operating range of the blowers,

\begin{equation}
K_{L_a} = k_a\,u,
\qquad k_a = \frac{K_{L_a}^{\max}}{100},
\qquad u \in [0,100]\,\%
\end{equation}

Substituting into the dissolved oxygen mass balance yields

\begin{equation}
\dot S_O
=
\underbrace{
\frac{1}{V}\sum_{i=1}^{3} Q_{in,i}S_{O,in,i}
-\frac{Q_{out}}{V}S_O
- r_{O}(\mathbf{x})
}_{f(\mathbf{x},t)}
\;+\;
\underbrace{k_a\left(S_O^{*}-S_O\right)}_{g(S_O)}\,u
\end{equation}

### 1.2 Simplified model
The ASM Transfer function is given by:
First order system
$$
G(s)=\frac{Y(s)}{U(s)}
=\frac{k_{dc}}{\tau s+1}
$$
In chain of integrators:
$$
s\cdot Y(s)=-\frac{Y(s)}{\tau}+\frac{U(s)\cdot k_{d_c}}{\tau}
$$

$$
\dot{y}(t)=-\frac{y(t)}{\tau}+\frac{u(t)\cdot k_{d_c}}{\tau}
$$

In disturbance form, grouping the plant's own pole into the total disturbance:
$$
\dot{y}(t)=\xi(t) + \frac{k_{d_c}}{\tau}\cdot u(t),
\qquad \xi(t):=-\frac{y(t)}{\tau}
$$

with
$$
b_0:=\frac{k_{d_c}}{\tau}
$$

$$
\dot{y}(t)=\xi(t) + b_0\cdot u(t)
$$

```python
# ============================================================
# Other functions
# ============================================================

def tf2sym(tf,symb=None):
    s = symbols(symb) if symb is not None else symbols('s')
    numTF = tf.num[0][0]   
    denTF = tf.den[0][0]
    num_symb = sum(numTF[i]*s**(len(numTF)-1-i)
                   for i in range(len(numTF)))
    den_symb = sum(denTF[i]*s**(len(denTF)-1-i)
                   for i in range(len(denTF)))
    return (num_symb / den_symb).simplify()

def sym2tf(expr,symb=None,Ts=None):

    s = symbols(symb) if symb is not None else symbols('s')
    num, den = fraction(expr)
    num_poly = Poly(num, s)
    den_poly = Poly(den, s)

    num_coeff = [float(i) for i  in num_poly.as_poly().all_coeffs()]
    den_coeff = [float(i) for i in den_poly.as_poly().all_coeffs()]
    if Ts!=None:
        return tf(num_coeff,den_coeff,Ts)
    else:
        return tf(num_coeff,den_coeff)

```

### 1.4 Modelo interno

La perturbación $\xi$ se describe con un modelo interno de orden $m$,
$p_\xi(s) = s^m + \gamma_{m-1}s^{m-1} + \cdots + \gamma_1 s + \gamma_0$, con el estado
$\mathbf{x}_\xi = [\,\xi\ \ \dot\xi\ \ \cdots\ \ \xi^{(m-1)}\,]^\top$:

$$\dot{\mathbf{x}}_\xi(t) = \mathbf{A}_\xi\,\mathbf{x}_\xi(t) + \mathbf{B}_\xi\,p_\xi(\xi,t),
\qquad \xi(t) = \mathbf{C}_\xi\,\mathbf{x}_\xi(t),$$

donde $p_\xi(\xi,t)$ es el residuo, nulo cuando $\xi$ pertenece exactamente a la familia
que genera el modelo. Con $m = 1$ y $\gamma_0 = 0$ (perturbación tipo escalón):
$\mathbf{A}_\xi = 0$, $\mathbf{B}_\xi = 1$, $\mathbf{C}_\xi = 1$.

```python
# ============================================================
#   State Space disturbance system
# ============================================================

# One Integral
Ad = np.block([[0]])
Bd = np.block([[1]])
Cd = np.block([[1]])
Dd = np.block([[0]])

#Two Integrals
# Ad = np.block([[0,1],[0,0]])
# Bd = np.block([[1],[0]])
# Cd = np.block([[1,0]])
# Dd = np.block([[0]])

sys_d = ss(Ad,Bd,Cd,Dd)
display(Latex(r"State Space of Disturbance System:"))
display(sys_d)
```

```python
L1_t,L2_t,kp_t = symbols(['L1','L2','K1'])
A_t = np.block([[0,1],[0,0]])
B_t =  np.block([[1],[0]])
L_t = np.block([[L1_t],[L2_t]])
C_t = np.block([1,0]).reshape(1,2)
K_t = np.block([kp_t,0]).reshape(1,2)
print(L_t.shape)
display(A_t-L_t@C_t-B_t@K_t)
```

## 2. Control

### 2.1 Control Law with observer

#### 2.1.1 Continuos Time
**Now, we have to design to the ADRC**

We rewrite the second-order system as a second-order error system. Let $e_y = y - r$:

$$\dot{e}_y = k\cdot u + (\bar{\xi} -u^{*})$$

Defining $\xi = \bar{\xi} - u^{*}$:

$$\dot{e}_y = k\cdot u + \xi$$

Applying the same procedure as before, the estimation error dynamics become: We have to select to the states of the system.
$$\dot{x_1}=x_2+k\cdot u$$
$$\dot{x_2}=PMI(\xi,t)$$

**Extended Dynamics System**
$$\dot{X}_{au} = A_{au} \cdot X_{au} + B_{au} \cdot u + B_{\xi_{au}} \cdot \xi^{(m)}$$

with the state vectors defined as:

$$X_p = \begin{bmatrix} e_y  \end{bmatrix}, \qquad X_{\xi} = \begin{bmatrix} \xi \end{bmatrix}$$

and the augmented state:

$$X_{au} = \begin{bmatrix} X_p \\ X_{\xi}\end{bmatrix}$$
**Observer System**
$$\dot{\widehat{X}}_{au} = (A_{au} - LC_{au})\widehat{X}_{au} + B_{au} u + Le_y$$

with $e_y=C_{au}\cdot X_{au}$

**Estimation Error Dynamics**
$$(\dot{X}_{au} - \dot{\widehat{X}}_{au}) = (A_{au} - LC_{au})(X_{au} - \widehat{X}_{au}) + B_{{\xi}_{au}} {\xi}^{(m)}$$

$$\dot{\bar{e}} = (A_{au} - LC_{au})\bar{e} + B_{\xi_{au}} \xi^{(m)}$$





The control law is given by the following expression, where the observer provides the estimated plant and disturbance states:

\begin{equation}
    e^{(1)}=v 
\end{equation}
Ideal model:
\begin{equation}
   f_\delta^{(n)}=\beta \cdot u+\phi(f,\dot{f}_\delta,...,f^{(n-1)}_\delta) 
\end{equation}

Donde $\beta$, $f_\delta^{(n)}$ y $\phi$ corresponde a la dinámica no deseada.
Ahora bien de aquí se puede establecer la señal de control.

\begin{equation}
    u=\frac{1}{\beta}[-\phi-v(ef^{(n-1)},..,ef,f^{(n)*})]
\end{equation}



Now:


\begin{equation}
    u=\frac{1}{b_0}[-\hat{f}-v(ey^{(0)})]
\end{equation}



with $v$:
$$v = k_1 \cdot X_{au_1} + k_2 \cdot \widehat{X}_{au_2}$$


$$v=K\cdot \widehat{X}_{au}$$

#### 2.1.2 Sistema aumentado

Con la planta de diseño (cadena de $n$ integradores: $\mathbf{A}_p$, $\mathbf{B}_p$,
$\mathbf{C}_p$) y el modelo interno ($\mathbf{A}_\xi$, $\mathbf{B}_\xi$, $\mathbf{C}_\xi$),
el sistema aumentado de orden $n+m$ es

$$\dot{\mathbf{X}}(t) = \mathbf{A}\,\mathbf{X}(t) + \mathbf{B}\,u_0(t) + \mathbf{E}\,p_\xi(\xi,t),
\qquad e_y(t) = \mathbf{C}\,\mathbf{X}(t),$$

$$\mathbf{A} = \begin{bmatrix}\mathbf{A}_p & \mathbf{B}_p\,\mathbf{C}_\xi\\ \mathbf{0}_{m\times n} & \mathbf{A}_\xi\end{bmatrix},\qquad
\mathbf{B} = \begin{bmatrix}\mathbf{B}_p\\ \mathbf{0}_{m\times 1}\end{bmatrix},\qquad
\mathbf{C} = \begin{bmatrix}\mathbf{C}_p & \mathbf{0}_{1\times m}\end{bmatrix},\qquad
\mathbf{E} = \begin{bmatrix}\mathbf{0}_{n\times 1}\\ \mathbf{B}_\xi\end{bmatrix}.$$

Para $n = 1$, $m = 1$:
$\mathbf{A} = \begin{bmatrix}0&1\\0&0\end{bmatrix}$,
$\mathbf{B} = \begin{bmatrix}1\\0\end{bmatrix}$,
$\mathbf{C} = \begin{bmatrix}1&0\end{bmatrix}$,
$\mathbf{E} = \begin{bmatrix}0\\1\end{bmatrix}$. En el código: `Aau`, `Bau`, `Cau`.

```python
n=Ap.shape[0]
m=Ad.shape[0]

Aau = np.block([[Ap,Bp@Cd],
                [np.zeros((m,n)),Ad]])
Bau = np.block([[Bp],
                [np.zeros((m,1))]])
Cau = np.block([Cp,np.zeros((1,m))])
Dau = np.block([np.zeros((1,1))])   
sys_au = ss(Aau,Bau,Cau,Dau)

display(Latex(r"State Space of Augmented System:"))
display(sys_au)
```

#### 2.1.3 Diseño de las ganancias

Por asignación de polos (Ackermann), con los $n$ polos del controlador en $-\omega_c$ y
los $n+m$ del observador en $-\omega_o$ (sintonía por ancho de banda):

$$\mathbf{K} = \begin{bmatrix}\mathbf{K}_p & \mathbf{C}_\xi\end{bmatrix},\qquad
\Delta_c(s) = (s + \omega_c)^n,\qquad \Delta_o(s) = (s + \omega_o)^{n+m}.$$

Para $n = m = 1$: $k_p = \omega_c$ y $\mathbf{L} = \begin{bmatrix}2\,\omega_o\\ \omega_o^2\end{bmatrix}$,
con $\omega_c = 4/\tau = 3{,}33\times10^{-3}$ rad/s y $\omega_o = 5\,\omega_c = 0{,}0167$ rad/s:
$\mathbf{K} = [\,3{,}33\times10^{-3}\ \ 1\,]$ y $\mathbf{L} = [\,0{,}0333\ \ 2{,}78\times10^{-4}\,]^\top$.

```python
# ============================================================
# Control-Loop Design
# ============================================================


pols = np.array([
    -4/Tau,              # 4 veces mas rapido que el lazo abierto: tau_c = Tau/4 = 300 s
])
K_1= acker(Ap,Bp,pols).reshape(1,n)
K = np.block([K_1,1]).reshape(1,n+m)
display(Latex(r"State Feedback Gain K:"))
display(K_1)
display(K)

# ============================================================
# Observer Design
# ============================================================
pols_L = np.array([
-0.0045,-0.0035, -0.0025
])
pols_L = np.array([
-5*4/Tau, -5*4/Tau     # 5 veces mas rapido que el controlador: 20/Tau = 0,0167 rad/s (60 1/h)
])
#   72 y 86 1/h.  Antes estaba en -0.05,-0.06 (180 1/h): con esa el ESO
#   metia 16,4 puntos de soplador de RUIDO DE SONDA puro (ver el barrido
#   de la seccion 4).  A 72 1/h bajan a 7,1, y el IAE solo empeora 0,020
#   mg/L, que esta por debajo del ruido del propio sensor.
#   Coincide con el optimo 60-100 1/h del analisis de dinamica, 14.3.
L = acker(Aau.T,Cau.T, pols_L.T).reshape(n+m,1)
display(Latex(r"Observer Gain L:"))
display(L)  

# ============================================================
# Validation of the design
# ============================================================

# Closed-loop poles
eig_cl = np.linalg.eigvals(Ap - Bp @ K_1)

# Observer poles
eig_obs = np.linalg.eigvals(Aau - L @ Cau)
print("Desired controller poles :", np.sort_complex(pols))

print("Obtained controller poles:", np.sort_complex(eig_cl))

print("Desired observer poles :", np.sort_complex(pols_L))
print("Obtained observer poles:", np.sort_complex(eig_obs))
```

**Pole dynamics**

```python

# ============================================================
# Time vectors
# ============================================================

# Continuous time
t = np.linspace(0, 3000.0, 2000)


# ============================================================
# CONTROLLER DYNAMICS
# ============================================================

# Continuous dynamics
responses_controller = np.array([
    np.exp(p * t)
    for p in pols
])



# ============================================================
# Plot Controller Dynamics
# ============================================================

fig, ax = plt.subplots(
    figsize=(8, 6),
    layout='constrained'
)

# Continuous poles
for i, p in enumerate(pols):

    ax.plot(
        t,
        responses_controller[i],
        linewidth=2,
        linestyle='-',
        label=fr'Continuous: $s={p}$'
    )

ax.set_title(
    'Controller Pole Dynamics'
)

ax.set_xlabel(
    'Time [s]'
)

ax.set_ylabel(
    r'Modal response'
)

ax.set_xlim(
    0,
    3000
)



ax.minorticks_on()

ax.grid(
    visible=True,
    which='both'
)

ax.legend()

plt.show()


# ============================================================
# OBSERVER DYNAMICS
# ============================================================

# Continuous dynamics
responses_observer = np.array([
    np.exp(p * t)
    for p in pols_L
])



# ============================================================
# Plot Observer Dynamics
# ============================================================

fig, ax = plt.subplots(
    figsize=(8, 6),
    layout='constrained'
)

# Continuous poles
for i, p in enumerate(pols_L):

    ax.plot(
        t,
        responses_observer[i],
        linewidth=2,
        linestyle='-',
        label=fr'Continuous: $s={p}$'
    )

ax.set_title(
    'Observer Pole Dynamics'
)

ax.set_xlabel(
    'Time [s]'
)

ax.set_ylabel(
    r'Modal response'
)

ax.set_xlim(
    0,
    120
)


ax.minorticks_on()

ax.grid(
    visible=True,
    which='both'
)

ax.legend()

plt.show()
```

**Observador de estado extendido (ESO)**

$$\dot{\widehat{\mathbf{X}}}(t) = \bigl(\mathbf{A} - \mathbf{L}\,\mathbf{C}\bigr)\,\widehat{\mathbf{X}}(t)
+ \begin{bmatrix}\mathbf{L} & \mathbf{B}\end{bmatrix}\begin{bmatrix}e_y(t)\\ u_0(t)\end{bmatrix}
= \mathbf{A}_o\,\widehat{\mathbf{X}}(t) + \mathbf{B}_o\,\mathbf{u}_t(t),
\qquad \mathbf{C}_o = \mathbf{I},\quad \mathbf{D}_o = \mathbf{0}.$$

Error de estimación, $\bar{\mathbf{e}} = \mathbf{X} - \widehat{\mathbf{X}}$:
$\dot{\bar{\mathbf{e}}}(t) = \bigl(\mathbf{A} - \mathbf{L}\,\mathbf{C}\bigr)\,\bar{\mathbf{e}}(t) + \mathbf{E}\,p_\xi(\xi,t)$.
En el código: `Ao`, `Bo`.

```python
#####################################
#Observer
#####################################
Ao = (Aau - L@Cau)
print(Ao)
Bo = np.block([L,Bau])
Co = np.eye(n+m)
Do = np.zeros((n+m,2))
sys_o = ss(Ao,Bo,Co,Do)

print(sys_o)
display(Latex(r"Observer Gain L:"))
print(L)
```

**Ley de control y lazo cerrado**

$$u(t) = \frac{u_0(t)}{b_0} = -\frac{1}{b_0}\bigl(\mathbf{K}_p\,\hat{\mathbf{x}}_p(t) + \hat{\xi}(t)\bigr)
= -\frac{1}{b_0}\,\mathbf{K}\,\widehat{\mathbf{X}}(t),\qquad
\mathbf{K} = \begin{bmatrix}\mathbf{K}_p & \mathbf{C}_\xi\end{bmatrix}.$$

Por el principio de separación,
$\operatorname{eig}(\text{lazo cerrado}) = \operatorname{eig}(\mathbf{A} - \mathbf{B}\mathbf{K}) \cup \operatorname{eig}(\mathbf{A} - \mathbf{L}\,\mathbf{C})$.

Con la planta real $\dot{\mathbf{x}}_g = \mathbf{A}_g\mathbf{x}_g + \mathbf{B}_g\bar{u}$,
$\bar{u} = u + \varepsilon$ y $e_y = \mathbf{C}_g\mathbf{x}_g - r$, sin saturación:

$$\begin{bmatrix}\dot{\mathbf{x}}_g(t)\\ \dot{\widehat{\mathbf{X}}}(t)\end{bmatrix}
= \begin{bmatrix}\mathbf{A}_g & -\frac{1}{b_0}\,\mathbf{B}_g\mathbf{K}\\ \mathbf{L}\,\mathbf{C}_g & \mathbf{A} - \mathbf{L}\,\mathbf{C} - \mathbf{B}\mathbf{K}\end{bmatrix}
\begin{bmatrix}\mathbf{x}_g(t)\\ \widehat{\mathbf{X}}(t)\end{bmatrix}
+ \begin{bmatrix}\mathbf{0} & \mathbf{B}_g\\ -\mathbf{L} & \mathbf{0}\end{bmatrix}
\begin{bmatrix}r(t)\\ \varepsilon(t)\end{bmatrix}.$$

$b_0$ solo aparece en la fila de la planta: el observador recibe $u_0$, no $u$.

```python
## Lazo cerrado: ADRC basado en el error sobre integrador ideal
#
# El observador estima e_y = y - r.  Esa variable lleva la consigna dentro de su
# definicion, asi que el sistema del error tiene TRES entradas, no dos:
#
#       e_y' = e_y'(planta)  +  u0            <- conocida, se la pasas
#                            +  xi            <- desconocida, la estima el ESO
#                            -  r'            <- CONOCIDA, y aqui NO se le pasa
#
# Esta celda deja el lazo "desnudo", sin tratar la consigna: el observador no se
# entera de que r salta, deduce una xi falsa y la manda al actuador.  Es la
# propiedad que Herbst & Madonski documentan en la Fig. 6.13 ("giant control
# efforts") y que se mide abajo: u pico 419 %, t63 de 93 s con tau_c = 1063 s.
#
# NO es un error: el ADRC basado en el error es el de salida al que se le ha
# quitado el prefiltro (seccion 8.4, p. 141).  Las dos secciones siguientes se
# lo devuelven, cada una a su manera, SIN tocar ninguna de estas matrices.
#
# bo aparece en UN solo sitio: u = u0/bo, al cruzar al hardware.



Aau_cl = np.block([[Ap_r,      -Bp_r @ K / bo],
                  [L @ Cp_r,   Aau - L @ Cau - Bau @ K]])

Bau_cl = np.block([[np.zeros((n,1)),Bp_r],
                  [-L,np.zeros((n+m,1))]])

Cau_cl = np.block([Cp_r, np.zeros((1, n + m))])
Dau_cl = np.zeros((1, 2))

sys_au_cl = ss(Aau_cl, Bau_cl, Cau_cl, Dau_cl)

print('max Re(eig) del lazo cerrado : %.3e' % np.real(np.linalg.eigvals(Aau_cl)).max())
print('L = %s   (la misma de la celda 37)' % np.array2string(L.ravel(), precision=6))
display(Latex(r"State Space of Extended System:"))
display(sys_au_cl)
```

```python
# ============================================================
# Simulacion del lazo cerrado
#   U = [r, xi, w]    r  -> consigna [mg/L]
#                     xi -> perturbacion de entrada [% de u]
#                     w  -> -dr/dt, la entrada conocida del sistema del error
# ============================================================
t  = np.arange(0, 10000, T_s)
r  = np.where(t >= 4000, 2.2, 2.0)          # consigna: arranca en 2.0, sube a 3.0 en t = 4000 s
xi = np.where(t >= 7000, -15.0, 0.0)        # perturbacion [% de u] en t = 7000 s
w  = np.r_[-np.diff(r), 0.0] / T_s          # -dr/dt: un pulso de una muestra en cada cambio
u  = np.column_stack([r, xi])

xp0    = np.zeros(n)                        # planta en reposo, y(0) = 0
xp0_es = np.array([0.0])             # e_hat_y(0) = y(0) - r(0): el error inicial MEDIDO
xd0_es = np.zeros(m)                        # de la perturbacion no sabemos nada
x0 = np.block([xp0, xp0_es, xd0_es])

[Y, T, X] = lsim(sys_au_cl, U=u, T=t, X0=x0)

# ------------------------------------------------------------
#   X[:, 0] = x_p       estado de la PLANTA
#   X[:, 1] = e_hat_y   estimacion del ERROR      [mg/L]   (Cp_z = 1)
#   X[:, 2] = xi_hat    estimacion de la PERTURBACION  [mg/L/s]
# ------------------------------------------------------------
y_out   = (Cp_r @ X[:, :n].T).ravel()
e_est   = X[:, n]                           # e_y ESTIMADO por el observador
xi_hat  = X[:, n + 1]           # xi1_hat: el indice SIEMPRE es n+1, no n+m (ver seccion 4)

e_real  = y_out - r                         # el error DE VERDAD

u0      = -K_1[0, 0] * e_est - xi_hat       # = -K x_hat   [mg/L/s]
u_ctrl  = u0 / bo                           # control fisico al soplador [%]

# xi_hat lleva dentro la autorregulacion de la planta (-y/Tau), que el
# integrador ideal mando a la perturbacion total. Lo que queda, en % de u:
xi_fisica = (xi_hat + y_out / Tau) / bo

# trayectoria que impone el polo del controlador, vectorizada:
#   en cada tramo de consigna,  y(t) = r + (y(t0) - r) e^{-omega_c (t - t0)}
w_c    = -pols[0]
k_pert = int(np.flatnonzero(xi != 0)[0])

nuevo = np.r_[True, np.diff(r) != 0]          # True en cada cambio de consigna
tramo = np.cumsum(nuevo) - 1                  # numero de tramo de cada muestra
k0    = np.flatnonzero(nuevo)[tramo]          # muestra en que arranco su tramo

y_teo = r + (y_out[k0] - r) * np.exp(-w_c * (t - t[k0]))
y_teo[k_pert:] = np.nan                       # a partir de xi manda el observador, no el polo

print('omega_c = %.6f rad/s  ->  tau_c = %.1f s      bo = %.4e\n' % (w_c, 1 / w_c, bo))
print('SEGUIMIENTO DE CONSIGNA  (%d escalones de consigna)' % nuevo.sum())
print('   max|y - trayectoria|   = %.2e mg/L' % np.nanmax(np.abs(y_out - y_teo)))
print('   t63 del primer escalon = %.1f s   (tau_c = %.1f s)'
      % (t[np.argmax(y_out >= 0.632 * r[0])], 1 / w_c))
print('\nestimacion del error : max|e_real - e_est| = %.2e mg/L' % np.abs(e_real - e_est).max())
print('perturbacion fisica  : estimada %.2f %%   ·   aplicada %.1f %%' % (xi_fisica[-1], xi[-1]))
print('control              : u pico %.1f %%, minimo %.1f %%' % (u_ctrl.max(), u_ctrl.min()))
```

```python
fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.5), layout='constrained')

# a) salida contra consigna y contra el polo diseñado
ax = axs[0, 0]
ax.plot(T, y_out, color='C0', linewidth=2.4, label='$y$  (OD medido)')
ax.plot(T, r, color='k', linewidth=1.4, linestyle='--', label='$r$  (consigna)')
ax.axvline(T[k_pert], color='C3', linewidth=1, alpha=.5)
ax.annotate('entra $\\varepsilon$', xy=(T[k_pert], 0.15), xytext=(6, 0),
            textcoords='offset points', fontsize=8.5, color='C3')
ax.set_title('a · Salida contra consigna')
ax.set_ylabel('OD [mg/L]'); 

# b) el error: real contra estimado
ax = axs[0, 1]
ax.plot(T, e_real, color='C0', linewidth=3, label='$e_y$ REAL $= y - r$')
ax.plot(T, e_est, color='C1', linewidth=1.6, linestyle='--',
        label='$\\hat e_y$  ESTIMADO (estado del observador)')
ax.set_title('b · El error: real contra estimado')
ax.set_ylabel('error [mg/L]')
ax.text(.97, .06, 'máx. error de estimación: %.1e mg/L' % np.abs(e_real - e_est).max(),
        transform=ax.transAxes, ha='right', fontsize=9, color='#555555')

# c) la perturbación física: estimada contra aplicada
ax = axs[1, 0]
ax.plot(T, xi_fisica, color='C2', linewidth=2,
        label='$\\hat\\xi$  perturbación total')
ax.plot(T, xi, color='k', linewidth=1.5, linestyle='--', label='$\\varepsilon$  perturbación exógena')
ax.set_title('c · Perturbación exógena y perturbación total')
ax.set_ylabel('[% de $u$]')

# d) señal de control
ax = axs[1, 1]
ax.plot(T, u_ctrl, color='C4', linewidth=2, label='$u = u_0/b_0$  (al soplador)')
ax.axhline(100, color='C3', linestyle=':', linewidth=1.2, label='límite del soplador')
ax.axhline(0, color='C3', linestyle=':', linewidth=1.2)
ax.set_title('d · Señal de control')
ax.set_ylabel('$u$ [%]'); ax.set_ylim(-8, 110)

for ax in axs.flat:
    ax.minorticks_on(); ax.legend(fontsize=8.5, frameon=False, loc='best')
    ax.grid(visible=True, which='major', alpha=0.35)
    ax.set_xlabel('tiempo [s]'); eje_minutos(ax)

fig.suptitle('Lazo cerrado',
             fontsize=13, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/adrc_lazo_cerrado.pdf', bbox_inches='tight')
plt.show()
```

```python
# ============================================================
# Margenes de estabilidad y anchos de banda
#   C(s): el ADRC visto de -e_y a u (observador + ley de control, con 1/b0)
#   L(s) = C(s) G(s), con la planta real G(s) = K_dc/(Tau s + 1)
# ============================================================
import control as ct

C_adrc = ct.tf(minreal(ss(Aau - L @ Cau - Bau @ K, L, K / bo, 0), verbose=False))
den_c = C_adrc.den[0][0]                              # minreal deja el integrador en ~1e-19, no en 0: sin esto margin()
den_c = np.where(np.abs(den_c) < 1e-12 * np.abs(den_c).max(), 0.0, den_c)   # ve un falso cruce de -180 en w ~ 0
C_adrc = ct.tf(C_adrc.num[0][0], den_c)
L_lazo = C_adrc * Gs_real
gm, pm, w_gm, w_pm = margin(L_lazo)
w_o = -pols_L[0]

print('C(s) =', C_adrc)
print('margen de fase     = %.1f grados en w = %.4f rad/s (%.1f rad/h)' % (pm, w_pm, w_pm * 3600))
print('margen de ganancia = %s' % ('infinito (la fase no cruza -180 grados)' if not np.isfinite(gm)
                                    else '%.1f dB en %.4f rad/s' % (20 * np.log10(gm), w_gm)))

w_f  = np.logspace(-5, 0, 3000)
Lw = ct.frequency_response(L_lazo, w_f).fresp[0, 0]
mag_dB = 20 * np.log10(np.abs(Lw))
fase   = np.degrees(np.unwrap(np.angle(Lw)))

T_lazo = feedback(L_lazo, 1)                          # y/r con la planta real
T_c    = tf([w_c], [1, w_c])                          # dinamica impuesta al error
T_o    = tf([w_o**2], [1, 2 * w_o, w_o**2])           # xi_hat/xi del observador
mag = lambda sis: 20 * np.log10(np.abs(ct.frequency_response(sis, w_f).fresp[0, 0]))

fig, axs = plt.subplots(3, 1, figsize=(10, 9), sharex=True, layout='constrained')

ax = axs[0]
ax.semilogx(w_f * 3600, mag_dB, color='C0', linewidth=2, label='$|G_L(j\\omega)| = |C(j\\omega)\\,G(j\\omega)|$')
ax.axhline(0, color='k', linewidth=0.8)
ax.axvline(w_pm * 3600, color='gray', linestyle=':', linewidth=1)             # cruce de ganancia
mg_txt = '$\\infty$ (la fase no cruza $-180°$)' if not np.isfinite(gm) else '%s dB' % ('%.1f' % (20 * np.log10(gm))).replace('.', ',')
ax.set_ylabel('magnitud [dB]'); ax.set_title('a · Magnitud:  MG = %s' % mg_txt)

ax = axs[1]
ax.semilogx(w_f * 3600, fase, color='C0', linewidth=2, label='$\\angle G_L(j\\omega)$')
ax.axhline(-180, color='k', linewidth=0.8)
ax.axvline(w_pm * 3600, color='gray', linestyle=':', linewidth=1)             # cruce de ganancia
f_pm = np.interp(np.log10(w_pm), np.log10(w_f), fase)
ax.plot([w_pm * 3600] * 2, [-180, f_pm], color='C3', linewidth=2.2)            # MF: de -180 a la curva (como margin de Matlab)
ax.plot(w_pm * 3600, f_pm, 'o', color='C3', ms=4)
ax.set_ylabel('fase [°]'); ax.set_ylim(-200, -60)
ax.set_title('b · Fase:  MF = %s°  (en %s rad/h)' % (('%.1f' % pm).replace('.', ','), ('%.1f' % (w_pm * 3600)).replace('.', ',')))

ax = axs[2]
ax.semilogx(w_f * 3600, mag(T_c), color='C1', linewidth=2, label='$\\omega_c/(s+\\omega_c)$  (dinámica impuesta al error)')
ax.semilogx(w_f * 3600, mag(T_o), color='C2', linewidth=2, label='$\\omega_o^2/(s+\\omega_o)^2$  (observador, $\\hat\\xi/\\xi$)')
ax.semilogx(w_f * 3600, mag(T_lazo), color='C0', linewidth=1.4, linestyle='-.', label='$G_o = y/r$  lazo cerrado con la planta')
ax.axhline(-3, color='gray', linestyle=':', linewidth=1)
for x, txt in ((1 / Tau, '$1/\\tau$'), (w_c, '$\\omega_c$'), (w_o, '$\\omega_o$')):
    ax.axvline(x * 3600, color='gray', linestyle='--', linewidth=1)
    ax.text(x * 3600 * 1.08, -38, txt + ' = %.1f rad/h' % (x * 3600), fontsize=9, color='#555555', rotation=90, va='bottom')
m_bw = -20 * np.log10(np.sqrt(2))                                             # -3 dB: |T| = 1/sqrt(2), como en la seccion 7
m_lazo = mag(T_lazo); i_bw = np.flatnonzero(m_lazo < m_bw)[0]                  # ancho de banda del lazo cerrado (-3 dB)
w_bw = 10 ** np.interp(m_bw, [m_lazo[i_bw], m_lazo[i_bw - 1]], [np.log10(w_f[i_bw]), np.log10(w_f[i_bw - 1])])
ax.axvline(w_bw * 3600, color='C0', linestyle='--', linewidth=1)
ax.text(w_bw * 3600 * 0.93, -38, '$\\omega_{BW}$ = %.1f rad/h' % (w_bw * 3600), fontsize=9, color='C0', rotation=90, va='bottom', ha='right')
print('ancho de banda del lazo cerrado (-3 dB) = %.4f rad/s (%.1f rad/h)' % (w_bw, w_bw * 3600))
ax.set_ylim(-40, 6); ax.set_ylabel('magnitud [dB]'); ax.set_xlabel('frecuencia [rad/h]')
ax.set_title('c · Anchos de banda: controlador, observador y lazo cerrado')

for ax in axs:
    ax.grid(visible=True, which='both', alpha=0.3); ax.legend(fontsize=9, frameon=False, loc='lower left')
axs[1].legend(fontsize=9, frameon=False, loc='upper right')
fig.suptitle('Márgenes de estabilidad y anchos de banda del ADRC', fontsize=13, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/adrc_frecuencia.pdf', bbox_inches='tight')
plt.show()
```

**Márgenes con retardo.** El mismo lazo abierto con el retardo de la planta, $L(s) = C(s)\,G(s)\,e^{-\theta s}$, con $\theta$ = 50 s (retardo exacto en la respuesta en frecuencia). El retardo no cambia la magnitud, solo quita fase, $\theta\omega$: el margen de fase baja y aparece un cruce de $-180°$, así que el margen de ganancia deja de ser infinito.

```python
# ============================================================
# Margenes de estabilidad con retardo:  L(s) = C(s) G(s) e^{-theta s}
#   C(s): el ADRC de -e_y a u (la misma C_adrc de la celda anterior)
#   G(s) = K_dc/(Tau s + 1);  el retardo se evalua exacto, e^{-j w theta}
#   Margenes al estilo de margin() de Matlab: segmento en w_180 (MG) y en w_gc (MF)
# ============================================================
THETA_B = 50.0                                                # retardo [s]
w_b  = np.logspace(-5, 0, 20000)
L_w0 = ct.frequency_response(L_lazo, w_b).frdata[0, 0]        # sin retardo

def margenes(Lw):
    mag, fase = np.abs(Lw), np.degrees(np.unwrap(np.angle(Lw)))
    ig = np.flatnonzero(np.diff(np.sign(mag - 1)))[0]
    ip = np.flatnonzero(np.diff(np.sign(fase + 180)))
    wp, mg = (w_b[ip[0]], 1 / mag[ip[0]]) if len(ip) else (np.nan, np.inf)
    return mag, fase, w_b[ig], 180 + fase[ig], wp, mg

coma = lambda x, f='%.1f': (f % x).replace('.', ',')
fig, axs = plt.subplots(2, 1, figsize=(10, 7.5), sharex=True, layout='constrained')
for th, sty, lab in ((0.0, dict(color='0.55', lw=1.4, ls='--'), 'sin retardo'),
                     (THETA_B, dict(color='C0', lw=2), r'$\theta$ = %d s' % THETA_B)):
    mag, fase, wg, mf, wp, mg = margenes(L_w0 * np.exp(-1j * w_b * th))
    axs[0].semilogx(w_b * 3600, 20 * np.log10(mag), label=lab, **sty)
    axs[1].semilogx(w_b * 3600, fase, label=lab, **sty)
    print('theta = %2.0f s:  MF = %.1f grados en %.1f rad/h   MG = %s'
          % (th, mf, wg * 3600, ('%.2f (%.1f dB) en %.1f rad/h' % (mg, 20 * np.log10(mg), wp * 3600)) if np.isfinite(mg) else 'infinito'))
    if th > 0:
        mf_b, mg_b, wg_b, wp_b = mf, mg, wg, wp
        for ax in axs:
            ax.axvline(wg * 3600, color='gray', ls=':', lw=1); ax.axvline(wp * 3600, color='gray', ls=':', lw=1)
        axs[0].plot([wp * 3600] * 2, [-20 * np.log10(mg), 0], color='C3', lw=2.2)      # MG: de la curva a 0 dB
        axs[0].plot(wp * 3600, -20 * np.log10(mg), 'o', color='C3', ms=4)
        axs[1].plot([wg * 3600] * 2, [-180, mf - 180], color='C3', lw=2.2)              # MF: de -180 a la curva
        axs[1].plot(wg * 3600, mf - 180, 'o', color='C3', ms=4)
axs[0].axhline(0, color='k', lw=0.8); axs[1].axhline(-180, color='k', lw=0.8)
axs[0].set_ylim(-60, 40); axs[0].set_ylabel('magnitud [dB]')
axs[0].set_title('a · Magnitud:  MG = %s dB  (en %s rad/h)' % (coma(20 * np.log10(mg_b)), coma(wp_b * 3600)))
axs[1].set_ylim(-360, -60); axs[1].set_ylabel('fase [°]')
axs[1].set_title('b · Fase:  MF = %s°  (en %s rad/h)' % (coma(mf_b), coma(wg_b * 3600)))
axs[1].set_xlabel('frecuencia [rad/h]'); axs[1].set_xlim(1, 1000)
for ax in axs:
    ax.grid(visible=True, which='both', alpha=0.3); ax.legend(fontsize=9, frameon=False, loc='lower left')
fig.suptitle('Márgenes de estabilidad del ADRC con retardo, $L = C\\,G\\,e^{-\\theta s}$', fontsize=13, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/adrc_margenes_retardo.pdf', bbox_inches='tight')
plt.show()
```

#### Lazo cerrado con saturación del actuador

El soplador solo entrega entre 0 y 100 %. Al observador se le realimenta la orden **aplicada**, $b_0\,u_{sat}$, que está en las unidades de $u_0$: es el anti-windup del ADRC (Herbst & Madoński, sec. 9.1.1). Mismo escenario de la simulación anterior.

```python
# ============================================================
# Lazo cerrado CON saturacion del soplador (0-100 %)
#   Mismo escenario de la celda de simulacion (r, xi, t).  La saturacion es no
#   lineal, asi que lsim no sirve: se recorre muestra a muestra, con la planta y
#   el observador CONTINUOS discretizados exactos (ZOH) a T_s.
#   Al observador entra la orden APLICADA, b0*u_sat (anti-windup, Herbst 9.1.1).
# ============================================================
U_MIN, U_MAX = 0.0, 100.0

P_d = c2d(ss(Ap_r, Bp_r, Cp_r, 0), T_s)                                   # planta real
O_d = c2d(ss(Aau - L @ Cau, np.block([L, Bau]), np.eye(n + m), np.zeros((n + m, 2))), T_s)   # observador: entradas [e_y, u0]
Ap_d, Bp_d = np.asarray(P_d.A), np.asarray(P_d.B)
Ao_d, Bo_d = np.asarray(O_d.A), np.asarray(O_d.B)

N_s   = len(t)
xp    = np.zeros((n, 1))                     # planta en reposo
xh    = np.zeros((n + m, 1))                 # observador en cero
y_s   = np.empty(N_s); u_s = np.empty(N_s); u_pedida = np.empty(N_s)
eh_s  = np.empty(N_s); xih_s = np.empty(N_s)
for k in range(N_s):
    y_k   = (Cp_r @ xp).item()
    e_k   = y_k - r[k]
    u_k   = -(K @ xh).item() / bo            # lo que pide el controlador [%]
    us_k  = min(max(u_k, U_MIN), U_MAX)      # lo que entrega el soplador
    y_s[k], u_s[k], u_pedida[k], eh_s[k], xih_s[k] = y_k, us_k, u_k, xh[0, 0], xh[n, 0]
    xp = Ap_d @ xp + Bp_d * (us_k + xi[k])
    xh = Ao_d @ xh + Bo_d @ np.array([[e_k], [bo * us_k]])   # <- b0*u_sat, no u0

e_real_s  = y_s - r
xi_fis_s  = (xih_s + y_s / Tau) / bo         # sin la fuga de la planta, en % de u
k_pert    = int(np.flatnonzero(xi != 0)[0])
sat_frac  = 100 * np.mean((u_pedida < U_MIN) | (u_pedida > U_MAX))

print('CON saturacion 0-100 %:')
print('   t63 del primer escalon = %.1f s' % t[np.argmax(y_s >= 0.632 * r[0])])
print('   OD maximo = %.3f mg/L   ·   error final |y - r| = %.2e mg/L' % (y_s.max(), abs(e_real_s[-1])))
print('   u pedida entre %.0f y %.0f %%  ·  saturada el %.2f %% del tiempo' % (u_pedida.min(), u_pedida.max(), sat_frac))
print('   u en regimen: %.1f %% (t=3900 s), %.1f %% (t=6900 s), %.1f %% (t=9900 s)'
      % (u_s[int(3900 / T_s)], u_s[int(6900 / T_s)], u_s[int(9900 / T_s)]))
print('   perturbacion: estimada %.2f %%   ·   aplicada %.1f %%' % (xi_fis_s[-1], xi[-1]))

fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.5), layout='constrained')

ax = axs[0, 0]
ax.plot(t, y_s, color='C0', linewidth=2.4, label='$y$  (OD medido)')
ax.plot(t, r, color='k', linewidth=1.4, linestyle='--', label='$r$  (consigna)')
ax.axvline(t[k_pert], color='C3', linewidth=1, alpha=.5)
ax.annotate('entra $\\varepsilon$', xy=(t[k_pert], 0.15), xytext=(6, 0),
            textcoords='offset points', fontsize=8.5, color='C3')
ax.set_title('a · Salida contra consigna'); ax.set_ylabel('OD [mg/L]')

ax = axs[0, 1]
ax.plot(t, e_real_s, color='C0', linewidth=3, label='$e_y$ REAL $= y - r$')
ax.plot(t, eh_s, color='C1', linewidth=1.6, linestyle='--', label='$\\hat e_y$  ESTIMADO (estado del observador)')
ax.set_title('b · El error: real contra estimado'); ax.set_ylabel('error [mg/L]')

ax = axs[1, 0]
ax.plot(t, xi_fis_s, color='C2', linewidth=2, label='$\\hat\\xi$  perturbación total')
ax.plot(t, xi, color='k', linewidth=1.5, linestyle='--', label='$\\varepsilon$  perturbación exógena')
ax.set_title('c · Perturbación exógena y perturbación total'); ax.set_ylabel('[% de $u$]')

ax = axs[1, 1]
ax.plot(t, u_pedida, color='C7', linewidth=1.1, linestyle=':', label='$u = u_0/b_0$  (pedida)')
ax.plot(t, u_s, color='C4', linewidth=2, label='$u_{sat}$  (al soplador)')
ax.axhline(100, color='C3', linestyle=':', linewidth=1.2, label='límite del soplador')
ax.axhline(0, color='C3', linestyle=':', linewidth=1.2)
ax.set_title('d · Señal de control'); ax.set_ylabel('$u$ [%]'); ax.set_ylim(-8, 110)

for ax in axs.flat:
    ax.minorticks_on(); ax.legend(fontsize=8.5, frameon=False, loc='best')
    ax.grid(visible=True, which='major', alpha=0.35); ax.set_xlabel('tiempo [s]'); eje_minutos(ax)
fig.suptitle('Lazo cerrado con saturación del actuador', fontsize=13, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/adrc_lazo_cerrado_saturacion.pdf', bbox_inches='tight')
plt.show()
```

#### Variación de los parámetros de la planta

Solo el escalón de consigna de 2,0 a 2,2 mg/L. El controlador queda fijo (diseño nominal, esquema original sin saturación: al observador entra $u_0 = b_0 u$) y se varía un parámetro de la planta a la vez: $K_{dc}$ y $\tau$ en los rangos intercuartílicos de los escalones de M2 y M3 (capítulo 2), y el retardo $\theta$ de 0 a 60 s. La nominal ideal ($K_{dc}$ = 0,05, $\tau$ = 20 min, $\theta$ = 0) va siempre de referencia.

```python
# ============================================================
# Variacion de los parametros de la planta en el escalon 2,0 -> 2,2 mg/L
#   Controlador FIJO: el del diseno nominal (K, L, bo), esquema original sin
#   saturacion (al observador entra u0 = b0*u).  Cambia solo la planta real:
#   K_dc y tau (y con ellos su b0 real = K_dc/tau) y el retardo theta.
#   Sin saturacion, u no se limita: el 100 % del soplador queda solo de referencia.
# ============================================================
Ts_v = 0.5                                   # paso de simulacion [s] (polos lentos)
t_e  = 20000.0                               # el escalon, ya en regimen en 2,0 mg/L
t_v  = np.arange(0, t_e + 3000, Ts_v)
r_v  = np.where(t_v >= t_e, 2.2, 2.0)

O_v = c2d(ss(Aau - L @ Cau, np.block([L, Bau]), np.eye(n + m), np.zeros((n + m, 2))), Ts_v)
Ao_v, Bo_v = np.asarray(O_v.A), np.asarray(O_v.B)

def sim_variacion(Kdc=K_dc, tau=Tau, theta=0.0):
    P = c2d(ss([[-1 / tau]], [[Kdc / tau]], [[1.0]], [[0.0]]), Ts_v)
    a, b = P.A.item(), P.B.item()
    d = int(round(theta / Ts_v)); buf = np.zeros(d + 1)        # retardo en la entrada de la planta
    y_k = 0.0; xh = np.zeros((n + m, 1))
    Y = np.empty(len(t_v)); U = np.empty(len(t_v))
    for k in range(len(t_v)):
        e_k = y_k - r_v[k]
        u_k = -(K @ xh).item() / bo
        buf = np.roll(buf, 1); buf[0] = u_k
        Y[k], U[k] = y_k, u_k
        y_k = a * y_k + b * buf[d]
        xh  = Ao_v @ xh + Bo_v @ np.array([[e_k], [bo * u_k]])
    return Y, U

# desajuste de b0: planta NOMINAL ideal y b0 del CONTROLADOR por un factor
P_n = c2d(ss([[-1 / Tau]], [[K_dc / Tau]], [[1.0]], [[0.0]]), Ts_v)
a_n, b_n = P_n.A.item(), P_n.B.item()

def sim_b0(fac):
    b0_c = fac * bo                          # b0 que usa el controlador
    y_k = 0.0; xh = np.zeros((n + m, 1))
    Y = np.empty(len(t_v)); U = np.empty(len(t_v))
    for k in range(len(t_v)):
        e_k = y_k - r_v[k]
        u_k = -(K @ xh).item() / b0_c
        Y[k], U[k] = y_k, u_k
        y_k = a_n * y_k + b_n * u_k
        xh  = Ao_v @ xh + Bo_v @ np.array([[e_k], [b0_c * u_k]])
    return Y, U

# K_dc: la nominal dividida y multiplicada por 2 y 4 (comprobacion adicional); la variacion de la ganancia se
# discute con el barrido de b0 del panel d, que abarca el rango intercuartilico de b0 de M2 y M3 (Tabla 2.3).
# tau: Q1 y Q3 de M2 y de M3 (11-26 y 15-40 min).
casos_v = [(r'K_{dc}', [(0.0125, '0,0125 (1/4 × nominal)'), (0.025, '0,025 (1/2 × nominal)'), (0.1, '0,1 (2 × nominal)'), (0.2, '0,2 (4 × nominal)')], 'Kdc'),
           (r'\tau',  [(11 * 60, '11 min'), (15 * 60, '15 min'), (26 * 60, '26 min'), (40 * 60, '40 min')], 'tau'),
           (r'\theta', [(20, '20 s'), (40, '40 s'), (50, '50 s'), (60, '60 s')], 'theta')]

fig, axs = plt.subplots(4, 2, figsize=(12.5, 11), sharex=True, layout='constrained')
ven = t_v >= t_e - 300; tt_v = (t_v[ven] - t_e) / 60
Y_nom, U_nom = sim_variacion()
colores = plt.cm.viridis(np.linspace(0.05, 0.85, 4))
for i, (sym, vals, arg) in enumerate(casos_v):
    axy, axu = axs[i]
    axy.plot(tt_v, Y_nom[ven], 'k', lw=2.2, label=r'nominal ideal ($K_{dc}$ = 0,05, $\tau$ = 20 min, $\theta$ = 0)')
    axu.plot(tt_v, U_nom[ven], 'k', lw=2.2)
    for (v, txt), c in zip(vals, colores):
        Y, U = sim_variacion(**{arg: v})
        lab = '$%s$ = %s' % (sym, txt)
        if arg in ('Kdc', 'tau'):
            lab += r'  ($b_{0,\mathrm{planta}}/b_0$ = %s)' % ('%.2f' % ((v / Tau if arg == 'Kdc' else K_dc / v) / bo)).replace('.', ',')
        axy.plot(tt_v, Y[ven], color=c, lw=1.4, ls='--', alpha=0.6, label=lab)
        axu.plot(tt_v, U[ven], color=c, lw=1.3, ls='--', alpha=0.6)
        print('%-6s = %-24s  y final %.3f   y max %.3f   u max %.1f %%   u final %.1f %%' % (arg, txt, Y[-1], Y[ven].max(), U[ven].max(), U[-1]))
    axy.plot(tt_v, r_v[ven], 'k:', lw=1); axy.set_ylabel('OD [mg/L]'); axy.set_ylim(1.95, 2.35)
    axu.axhline(100, color='C3', ls=':', lw=1); axu.set_ylabel('$u$ [%]'); axu.set_ylim(bottom=0)
    axy.set_title('%s · variación de $%s$: salida' % ('abc'[i], sym))
    axu.set_title('%s · variación de $%s$: señal de control' % ('abc'[i], sym))
    axy.legend(fontsize=7.8, frameon=False, loc='lower right')
# d · desajuste de b0 del controlador
axy, axu = axs[3]
axy.plot(tt_v, Y_nom[ven], 'k', lw=2.2, label=r'$b_0$ nominal ($k_{dc}/\tau$)')
axu.plot(tt_v, U_nom[ven], 'k', lw=2.2)
for f, txt, c in zip([1 / 6, 1 / 4, 1 / 2, 2, 4, 6], ['1/6', '1/4', '1/2', '2', '4', '6'],
                     plt.cm.viridis(np.linspace(0.05, 0.9, 6))):
    Y, U = sim_b0(f)
    axy.plot(tt_v, Y[ven], color=c, lw=1.4, ls='--', alpha=0.6, label=r'$b_0$ del controlador = %s $\times$ nominal' % txt)
    axu.plot(tt_v, U[ven], color=c, lw=1.3, ls='--', alpha=0.6)
    print('b0 x %-5s  y max %.3f   u max %.1f %%   u min %.1f %%' % (txt, Y[ven].max(), U[ven].max(), U[ven].min()))
axy.plot(tt_v, r_v[ven], 'k:', lw=1); axy.set_ylabel('OD [mg/L]'); axy.set_ylim(1.95, 2.35)
axu.axhline(100, color='C3', ls=':', lw=1); axu.set_ylabel('$u$ [%]'); axu.set_ylim(bottom=0)
axy.set_title('d · desajuste de $b_0$ del controlador: salida')
axu.set_title('d · desajuste de $b_0$ del controlador: señal de control')
axy.legend(fontsize=7.8, frameon=False, loc='lower right')

# sobreimpulso de la nominal, en % del escalon de 0,2 mg/L
k_pk = np.argmax(Y_nom[ven]); Mp_nom = 100 * (Y_nom[ven][k_pk] - 2.2) / 0.2
print('sobreimpulso de la nominal = %.1f %%' % Mp_nom)
# tiempo de establecimiento de diseno: t_s = 4 tau_c = 4/w_c (98,2 % del escalon)
fuera = np.flatnonzero(np.abs(Y_nom[ven] - 2.2) > 0.02 * 0.2)  # t_s: ultima salida de la banda de 2 % del escalon
ts_nom = tt_v[fuera[-1]]                        # [min]
y_ts = np.interp(ts_nom, tt_v, Y_nom[ven])
print('t_s (banda de 2 %%) = %.1f min   (4/w_c = %.1f min)' % (ts_nom, 4 / w_c / 60))
ax_a = axs[0, 0]
ax_a.plot([ts_nom, ts_nom], [1.95, y_ts], color='0.75', ls='--', lw=1)          # vertical hasta la curva
ax_a.plot([tt_v[0], ts_nom], [y_ts, y_ts], color='0.75', ls='--', lw=1)         # horizontal hasta la curva
ax_a.plot(ts_nom, y_ts, 'o', ms=3.5, color='k', zorder=5)
ax_a.annotate('$t_s$: %s min' % ('%.1f' % ts_nom).replace('.', ','), xy=(ts_nom, y_ts), xytext=(0, 6),
            textcoords='offset points', ha='center', va='bottom', fontsize=8.5)

for axy in axs[:1, 0]:                         # solo en el primer panel
    axy.plot(tt_v[k_pk], Y_nom[ven][k_pk], 'o', ms=3.5, color='k', zorder=5)
    axy.annotate('$M_p$: %s %%' % ('%.1f' % Mp_nom).replace('.', ','), xy=(tt_v[k_pk], Y_nom[ven][k_pk]), xytext=(0, 6),
                 textcoords='offset points', ha='center', fontsize=8.5)

for ax in axs.flat:
    ax.grid(alpha=.3); ax.minorticks_on()
for ax in axs[3]:
    ax.set_xlabel('tiempo desde el escalón [min]')
fig.suptitle('Escalón de 2,0 a 2,2 mg/L con variación de la planta y de $b_0$ (sin saturación)',
             fontsize=12.5, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/adrc_variacion_parametros.pdf', bbox_inches='tight')
plt.show()
```

**Lugar de las raíces con desajuste de $b_0$.** Con el $b_0$ del controlador distinto del de la planta, $b_{0,\mathrm{planta}} = K_{dc}/\tau$, el lazo cerrado es

$$s\,(s+k_2)\,(s+1/\tau) + g\,(k_1 s + k_0) = 0,\qquad g = \frac{b_{0,\mathrm{planta}}}{b_0},$$

un lugar de las raíces en $g$ ($g > 1$: $b_0$ subestimado). Sin retardo y con $\theta$ = 50 s (Padé de orden 6).

```python
# ============================================================
# Lugar de las raices al variar g = b0_planta/b0, sin y con retardo
#   L(s) = g (k1 s + k0)/(s (s + k2)(s + 1/Tau))   [ADRC con ganancias GPI equivalentes]
#   El eje en rad/h: s se escala por 3600.  Retardo: Padé de orden 6.
# ============================================================
kp_r, l1_r, l2_r = K[0, 0], L[0, 0], L[1, 0]
k2_r, k1_r, k0_r = kp_r + l1_r, kp_r * l1_r + l2_r, kp_r * l2_r
Hs = 3600.0
L_rl = ct.tf([k1_r * Hs**2, k0_r * Hs**3], np.polymul([1, k2_r * Hs, 0], [1, Hs / Tau]))
g_rl = np.r_[0, np.geomspace(1e-3, 1e3, 4000)]

fig, axs = plt.subplots(1, 2, figsize=(12.5, 5.2), layout='constrained')
for ax, th, tit in ((axs[0], 0.0, 'a · sin retardo'), (axs[1], 50.0, r'b · con retardo $\theta$ = 50 s')):
    Ls = L_rl if th == 0 else L_rl * ct.tf(*ct.pade(th / Hs, 6))
    ct.root_locus_plot(Ls, gains=g_rl, ax=ax)                   # presentacion de python-control, sin cuadricula
    if th > 0:
        mg_rl = ct.stability_margins(Ls); g_lim, w_lim = mg_rl[0], mg_rl[3]
        ax.annotate(r'inestable si $b_0 < b_{0,\mathrm{planta}}/%s$' % ('%.2f' % g_lim).replace('.', ','), xy=(0, w_lim),
                    xytext=(-150, 10), textcoords='offset points', fontsize=9, color='C3')
        print('theta = %.0f s: inestable para g > %.2f (cruce en %.1f rad/h)' % (th, g_lim, w_lim))
    else:
        print('theta = 0 s: estable para todo g > 0')
    ax.axvline(0, color='k', lw=0.8, zorder=0); ax.axhline(0, color='k', lw=0.8, zorder=0)
    ax.set_xlim(-170, 40); ax.set_ylim(-110, 110)
    ax.set_title(tit); ax.set_xlabel('Re [rad/h]'); ax.set_ylabel('Im [rad/h]')
fig.suptitle(r'Lugar de las raíces del ADRC al variar $b_{0,\mathrm{planta}}/b_0$', fontsize=12.5, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/adrc_rlocus_b0.pdf', bbox_inches='tight')
plt.show()
```

#### Escalón al variar los anchos de banda

Escalón de 2,0 a 2,2 mg/L variando, $\times$ 1/4 a 4 sobre el diseño ($\omega_c = 4/\tau$, $\omega_o = 5\,\omega_c$), $\omega_c$ con $\omega_o$ fijo, $\omega_o$ con $\omega_c$ fijo y los dos con $\omega_o/\omega_c = 5$, con la planta real sin retardo. Las funciones de abajo (MF, $M_s$, IAE...) las usa también la sección del GPI. Para cada caso, $k_p=\omega_c$, $l_1=2\omega_o$, $l_2=\omega_o^2$ y el ADRC visto de $-e_y$ a $u$ es

$$C(s)=\frac{1}{b_0}\,\frac{(k_pl_1+l_2)\,s+k_pl_2}{s\,(s+k_p+l_1)},\qquad L(s)=C(s)\,G(s),\qquad G(s)=\frac{K_{dc}}{\tau s+1}.$$

- Margen de fase: $\mathrm{MF}=180°+\angle L(j\omega_{gc})$, con $|L(j\omega_{gc})|=1$ (el menor si hay varios cruces).
- Retardo máximo tolerable: $\theta_{\max}=\dfrac{\pi}{180}\,\dfrac{\mathrm{MF}}{\omega_{gc}}$ [s].
- Sensibilidad máxima: $M_s=\max_\omega |S(j\omega)|$, con $S=1/(1+L)$.
- Escalón de 2,0 a 2,2 mg/L con $T=L/(1+L)$: $e(t)=\Delta r\,(1-y_{e}(t))$, $\Delta r=0{,}2$ mg/L, hasta $t_f=20\,000$ s con paso de 0,5 s.
  $\mathrm{IAE}=\int_0^{t_f}|e|\,dt$, $\mathrm{ITAE}=\int_0^{t_f}t\,|e|\,dt$, $M_p=100\max(y_{e}-1,0)$ en % del escalón.

Lineal: sin saturación, retardo ni ruido.

```python
# ============================================================
# Robustez y desempeno variando un parametro a la vez (planta sin retardo)
#   C(s) = (1/b0) (k1 s + k0)/(s (s + k2)),  k2 = kp + l1, k1 = kp l1 + l2, k0 = kp l2
#   L = C G,   S = 1/(1 + L),   T = L/(1 + L)
#   MF: 180 + fase de L en |L| = 1 (el menor si hay varios cruces);  theta_max = MF[rad]/w_gc
#   Ms = max |S|;   escalon de 0,2 mg/L con T: IAE, ITAE y sobreimpulso
# ============================================================
from scipy import signal

w_m = np.logspace(-6, 0.5, 5000); s_m = 1j * w_m
G_m = K_dc / (Tau * s_m + 1)
t_m = np.arange(0, 20000, 0.5); dr = 0.2
coma = lambda x, f='%.1f': (f % x).replace('.', ',')

def robustez(k2, k1, k0):
    L_ = (k1 * s_m + k0) / (bo * s_m * (s_m + k2)) * G_m
    lm, fase = np.log10(np.abs(L_)), np.degrees(np.unwrap(np.angle(L_)))
    mf, th = [], []
    for i in np.flatnonzero(np.diff(np.sign(lm))):                   # cruces de |L| = 1
        a  = -lm[i] / (lm[i + 1] - lm[i])                            # interpolacion en log w
        pm = 180 + fase[i] + a * (fase[i + 1] - fase[i])
        wg = 10 ** (np.log10(w_m[i]) + a * np.log10(w_m[i + 1] / w_m[i]))
        mf.append(pm); th.append(np.radians(pm) / wg)
    return (min(mf), min(th)) if mf else (np.nan, np.nan), np.abs(1 / (1 + L_)).max()

def desempeno(k2, k1, k0):
    num = K_dc * np.array([k1, k0])
    den = np.polyadd(bo * np.polymul([1, k2, 0], [Tau, 1]), num)   # b0 s (s + k2)(Tau s + 1) + K_dc (k1 s + k0)
    if np.any(np.real(np.roots(den)) >= 0):
        return np.nan, np.nan, np.nan
    _, y = signal.step(signal.lti(num, den), T=t_m)
    e = dr * (1 - y)
    return np.trapezoid(np.abs(e), t_m), np.trapezoid(t_m * np.abs(e), t_m), 100 * max(y.max() - 1, 0)

def indicadores(k):
    (mf, th), ms = robustez(*k)
    return np.array([mf, ms, th, *desempeno(*k)])                   # MF, Ms, theta_max, IAE, ITAE, Mp

def ganancias(wc, wo):
    kp, l1, l2 = wc, 2 * wo, wo**2
    return kp + l1, kp * l1 + l2, kp * l2

f_v = np.geomspace(1 / 4, 4, 41)                                    # factor sobre el diseno (f_v[20] = 1)
paneles = [('a · Margen de fase', 'MF [°]', '%.1f°'), ('b · Sensibilidad máxima', '$M_s$', '%.3f'),
           ('c · Retardo máximo tolerable', r'$\theta_{\max}$ [s]', '%.0f s'), ('d · IAE', 'IAE [mg·s/L]', '%.1f'),
           ('e · ITAE', 'ITAE [10³ mg·s²/L]', '%.2f'), ('f · Sobreimpulso', '$M_p$ [% del escalón]', '%.1f %%')]

def figura_variacion(curvas, archivo, titulo):
    """curvas: [(etiqueta, [ganancias (k2, k1, k0) para cada factor de f_v])]"""
    fig, axs = plt.subplots(2, 3, figsize=(13.5, 7.5), sharex=True, layout='constrained')
    for (lab, ks), c in zip(curvas, ['C0', 'C1', 'C2']):
        R = np.array([indicadores(k) for k in ks]); R[:, 4] /= 1e3
        for j, ax in enumerate(axs.flat):
            ax.plot(f_v, R[:, j], color=c, lw=2, label=lab)
    R0 = indicadores(curvas[0][1][20]); R0[4] /= 1e3
    for j, (ax, (tit, unid, fmt)) in enumerate(zip(axs.flat, paneles)):
        ax.axvline(1, color='gray', ls='--', lw=1)
        ax.plot(1, R0[j], 'o', ms=8, mfc='white', mec='k', mew=1.6, zorder=5)
        ax.annotate('diseño: ' + coma(R0[j], fmt), xy=(1, R0[j]), xytext=(8, 8), textcoords='offset points', fontsize=8.5)
        ax.set_xscale('log', base=2)
        ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: {0.25: '1/4', 0.5: '1/2'}.get(v, '%g' % v)))
        ax.set_title(tit); ax.set_ylabel(unid); ax.grid(visible=True, which='both', alpha=.3)
    axs[0, 2].axhline(50, color='C3', ls=':', lw=1.2, label='retardo de 50 s'); axs[0, 2].legend(fontsize=8.5, frameon=False)
    axs[0, 0].legend(fontsize=9, frameon=False, loc='best')
    for ax in axs[1]:
        ax.set_xlabel('factor sobre el diseño')
    fig.suptitle(titulo, fontsize=12.5, fontweight='bold')
    fig.savefig('Latex/Tesis/figuras/%s_variacion_robustez.pdf' % archivo, bbox_inches='tight'); plt.show()
    print('diseno: MF %.1f°  Ms %.3f  theta_max %.0f s  IAE %.2f mg·s/L  ITAE %.0f mg·s²/L  Mp %.1f %%'
          % (R0[0], R0[1], R0[2], R0[3], R0[4] * 1e3, R0[5]))

wc_d, wo_d = 4 / Tau, 20 / Tau

# --- Escalon de 2,0 a 2,2 mg/L variando w_c, w_o y los dos (planta sin retardo ni saturacion) ---
#   y = 2,0 + 0,2 * escalon de T,   u = 40 + 0,2 * escalon de C/(1 + CG)
#   T = K_dc (k1 s + k0)/den,  C/(1 + CG) = (k1 s + k0)(Tau s + 1)/den,  den = b0 s (s + k2)(Tau s + 1) + K_dc (k1 s + k0)
def _pasos_anchos():
    t_p = np.arange(0, 3000, 0.5)
    fac = [1 / 4, 1 / 2, 1, 2, 4]; ftxt = ['1/4', '1/2', '1', '2', '4']
    casos = [('$\\omega_c$', '$\\omega_o$ fijo', lambda f: ganancias(f * wc_d, wo_d)),
             ('$\\omega_o$', '$\\omega_c$ fijo', lambda f: ganancias(wc_d, f * wo_d)),
             ('$\\omega_c$ y $\\omega_o$', '$\\omega_o/\\omega_c$ = 5', lambda f: ganancias(f * wc_d, f * wo_d))]
    cols = plt.cm.coolwarm(np.linspace(0, 1, 5))
    fig, axs = plt.subplots(3, 2, figsize=(12.5, 9), sharex=True, layout='constrained')
    print('Escalon de 2,0 a 2,2 mg/L:  IAE [mg·s/L]   Mp [% del escalon]   ts (banda 2 %) [min]   u max [%]')
    for i, (sym, fijo, gan) in enumerate(casos):
        axy, axu = axs[i]; fila = []
        for f, ft in zip(fac, ftxt):
            k2, k1, k0 = gan(f)
            den = np.polyadd(bo * np.polymul([1, k2, 0], [Tau, 1]), K_dc * np.array([k1, k0]))
            _, yT = signal.step(signal.lti(K_dc * np.array([k1, k0]), den), T=t_p)
            _, yU = signal.step(signal.lti(np.polymul([k1, k0], [Tau, 1]), den), T=t_p)
            y, u = 2.0 + 0.2 * yT, 2.0 / K_dc + 0.2 * yU
            iae, _, mp = desempeno(k2, k1, k0)
            _, yL = signal.step(signal.lti(K_dc * np.array([k1, k0]), den), T=t_m)
            ts = t_m[np.flatnonzero(np.abs(yL - 1) > 0.02)[-1]] / 60
            fila.append('x %-4s IAE %6.1f   Mp %5.1f   ts %5.1f   umax %5.0f' % (ft, iae, mp, ts, u.max()))
            if f == 1:
                st = dict(color='k', lw=2.4, zorder=5); lab = 'diseño'
            else:
                st = dict(color=cols[fac.index(f)], lw=1.4, ls='--', alpha=0.7); lab = '%s $\\times$ %s' % (sym, ft)
            tm = np.r_[-5, t_p / 60]
            axy.plot(tm, np.r_[2.0, y], label=lab, **st); axu.plot(tm, np.r_[2.0 / K_dc, u], **st)
        print(sym.replace('$', '').replace('\\omega', 'w') + ' (' + fijo.replace('$', '').replace('\\omega', 'w') + '):')
        print('\n'.join('    ' + c for c in fila))
        axy.plot([-5, 0, 0, 50], [2.0, 2.0, 2.2, 2.2], 'k:', lw=1)
        axy.set_ylim(1.95, 2.35); axy.set_ylabel('OD [mg/L]'); axy.legend(fontsize=8, frameon=False, loc='lower right')
        axu.axhline(100, color='C3', ls=':', lw=1); axu.set_ylabel('$u$ [%]'); axu.set_ylim(0, 300)
        axy.set_title('%s · variación de %s (%s): salida' % ('abc'[i], sym, fijo))
        axu.set_title('%s · variación de %s (%s): señal de control' % ('abc'[i], sym, fijo))
    for ax in axs.flat:
        ax.grid(alpha=.3); ax.minorticks_on(); ax.set_xlim(-5, 50)
    for ax in axs[2]:
        ax.set_xlabel('tiempo desde el escalón [min]')
    fig.suptitle('ADRC: escalón de 2,0 a 2,2 mg/L variando los anchos de banda (planta sin retardo ni saturación)',
                 fontsize=12.5, fontweight='bold')
    fig.savefig('Latex/Tesis/figuras/adrc_pasos_anchos.pdf', bbox_inches='tight'); plt.show()

_pasos_anchos()
```

#### Funciones de transferencia del lazo al variar $\omega_o/\omega_c$ (Herbst & Madoński, Fig. 5.8 y 5.9)

Lazo $u = C\,(r - y)$, $y = G\,(u + \varepsilon) + n$:

| entrada / salida | $-e_y = r - y$ | $u$ | $y$ |
|---|---|---|---|
| $r$ | $1/(1+CG)$ | $C/(1+CG)$ | $CG/(1+CG)$ |
| $\varepsilon$ | $-G/(1+CG)$ | $-CG/(1+CG)$ | $G/(1+CG)$ |
| $n$ | $-1/(1+CG)$ | $-C/(1+CG)$ | $1/(1+CG)$ |

Se grafican las nueve funciones; aparte, el lazo abierto $CG$ con su margen de fase y el controlador $C$. $\omega_c$ fijo, planta real sin retardo, $\omega_o/\omega_c$ de 1,25 a 20 (5 en negro).

```python
# ============================================================
# Funciones de transferencia del lazo al variar k_ESO = w_o / w_c
#   Herbst & Madonski, Fig. 5.8 y 5.9.  w_c = 4/Tau fijo, planta real sin retardo.
#   u = C (r - y),  y = G (u + eps) + n
#   Figura 1: las nueve de r, eps y n hacia -e_y, u, y.   Figura 2: lazo abierto CG y margen de fase,
#   a la izquierda con w_c fijo (varia w_o) y a la derecha con w_o = 5 w_c fijo (varia w_c).   Figura 3: C.
#   Al final, las mismas medidas variando w_c (w_o fijo) y los dos con w_o/w_c = 5, para la tabla.
# ============================================================
def _gang_kESO():
    wc_g = 4 / Tau
    k_eso = np.geomspace(1.25, 20, 9)                      # 5 en el centro
    w = np.logspace(-2, 4, 6000) / 3600                    # [rad/s]
    s = 1j * w; G = K_dc / (Tau * s + 1)
    cols = plt.cm.coolwarm(np.linspace(0, 1, len(k_eso)))
    fig, axs = plt.subplots(3, 3, figsize=(14, 9), sharex=True, layout='constrained')
    figL, axL = plt.subplots(2, 2, figsize=(14, 6.5), sharex=True, sharey='row', layout='constrained')
    fig2, axc = plt.subplots(figsize=(8, 4), layout='constrained')
    def lazo(wc, wo):
        k2, k1, k0 = wc + 2 * wo, 2 * wc * wo + wo**2, wc * wo**2
        C = (k1 * s + k0) / (bo * s * (s + k2))
        L = C * G
        # margen de fase: cruce de |L| = 1, interpolado en log w
        lm = np.log10(np.abs(L)); fase = np.degrees(np.unwrap(np.angle(L)))
        i = np.flatnonzero(np.diff(np.sign(lm)))[0]; a = -lm[i] / (lm[i + 1] - lm[i])
        wg = 10 ** (np.log10(w[i]) + a * np.log10(w[i + 1] / w[i])); fg = fase[i] + a * (fase[i + 1] - fase[i])
        return C, L, fase, wg, fg, 180 + fg

    fmt = '%8.2f %11.1f %8.1f %15.1f %14.0f %14.3f %17.1f %20.1f'
    def fila(wc, wo):
        C, L, fase, wg, fg, mf = lazo(wc, wo); D = 1 + L
        print(fmt % (wo / wc, wo * 3600, mf, wg * 3600, np.radians(mf) / wg,
              np.abs(1 / D).max(), 20 * np.log10(np.abs(G / D).max()), 20 * np.log10(np.abs(C / D)[-1])))

    cab = '%8s %11s %8s %15s %14s %14s %17s %20s' % ('wo/wc', 'wo [rad/h]', 'MF [°]', 'w_gc [rad/h]', 'theta_max [s]', 'max|1/(1+CG)|', 'max|G/(1+CG)| [dB]', '|C/(1+CG)| 1e4 [dB]')
    print(cab)
    for k, c in zip(k_eso, cols):
        wo = k * wc_g
        C, L, fase, wg, fg, mf = lazo(wc_g, wo); D = 1 + L
        fts = (1 / D, C / D, L / D, -G / D, -L / D, G / D, -1 / D, -C / D, 1 / D)
        nom = np.isclose(k, 5)
        st = dict(color='k', lw=2.4, zorder=5) if nom else dict(color=c, lw=1.3, ls='--', alpha=0.55)
        ktxt = ('%.2f' % k).rstrip('0').rstrip('.').replace('.', ',')
        lab = '$\\omega_o/\\omega_c$ = %s' % ktxt
        for ax, H in zip(axs.flat, fts):
            ax.semilogx(w * 3600, 20 * np.log10(np.abs(H)), label=lab, **st)
        # lazo abierto: izquierda w_c fijo (varia w_o); derecha w_o = 5 w_c fijo (varia w_c)
        for j, (Lj, fasej, wgj, fgj, mfj) in enumerate(((L, fase, wg, fg, mf), lazo(5 * wc_g / k, 5 * wc_g)[1:])):
            labL = '%s:  MF = %s°' % (lab, ('%.1f' % mfj).replace('.', ','))
            axL[0, j].semilogx(w * 3600, 20 * np.log10(np.abs(Lj)), label=labL, **st)
            axL[1, j].semilogx(w * 3600, fasej, **st)
            axL[1, j].plot([wgj * 3600] * 2, [-180, fgj], color='k' if nom else c, lw=2.2 if nom else 1.2, alpha=1 if nom else 0.7)
            axL[1, j].plot(wgj * 3600, fgj, 'o', ms=4, color='k' if nom else c)
        axc.semilogx(w * 3600, 20 * np.log10(np.abs(C)), label=lab, **st)
        print(fmt % (k, wo * 3600, mf, wg * 3600, np.radians(mf) / wg,
              np.abs(1 / D).max(), 20 * np.log10(np.abs(G / D).max()), 20 * np.log10(np.abs(C / D)[-1])))
    print('\nVariando w_c con w_o = 5 w_c(diseño) fijo, factor 1/4 a 4 sobre w_c\n' + cab)
    for f in (1 / 4, 1 / 2, 1, 2, 4):
        fila(f * wc_g, 5 * wc_g)
    print('\nVariando los dos con w_o/w_c = 5, factor 1/4 a 4\n' + cab)
    for f in (1 / 4, 1 / 2, 1, 2, 4):
        fila(f * wc_g, 5 * f * wc_g)
    tit = ('a · $r \\to -e_y$:  $\\dfrac{1}{1 + CG}$', 'b · $r \\to u$:  $\\dfrac{C}{1 + CG}$  [dB re 1 % por mg/L]',
           'c · $r \\to y$:  $\\dfrac{CG}{1 + CG}$',
           'd · $\\varepsilon \\to -e_y$:  $\\dfrac{-G}{1 + CG}$  [dB re 1 mg/L por %]', 'e · $\\varepsilon \\to u$:  $\\dfrac{-CG}{1 + CG}$',
           'f · $\\varepsilon \\to y$:  $\\dfrac{G}{1 + CG}$  [dB re 1 mg/L por %]',
           'g · $\\eta \\to -e_y$:  $\\dfrac{-1}{1 + CG}$', 'h · $\\eta \\to u$:  $\\dfrac{-C}{1 + CG}$  [dB re 1 % por mg/L]',
           'i · $\\eta \\to y$:  $\\dfrac{1}{1 + CG}$')
    lims = ((-60, 10), (0, 100), (-60, 10), (-80, 0), (-60, 10), (-80, 0), (-60, 10), (0, 100), (-60, 10))
    for ax, t_, lim in zip(axs.flat, tit, lims):
        ax.set_title(t_, fontsize=10); ax.set_ylim(*lim); ax.grid(which='both', alpha=.3)
        ax.axvline(wc_g * 3600, color='0.5', ls=':', lw=1)
    for ax in axs[:, 0]:
        ax.set_ylabel('magnitud [dB]')
    for ax in axs[2]:
        ax.set_xlabel('frecuencia [rad/h]')
    axs[0, 0].legend(fontsize=7.5, frameon=False, loc='lower right')
    fig.suptitle('ADRC: funciones de transferencia del lazo variando $\\omega_o/\\omega_c$ (planta sin retardo)',
                 fontsize=12.5, fontweight='bold')
    fig.savefig('Latex/Tesis/figuras/adrc_gang_kESO.pdf', bbox_inches='tight')
    for j, txt in enumerate(('$\\omega_c$ fijo, varía $\\omega_o$', '$\\omega_o$ fijo, varía $\\omega_c$')):
        axL[0, j].axhline(0, color='k', lw=0.8); axL[1, j].axhline(-180, color='k', lw=0.8)
        axL[0, j].set_title('%s · Magnitud de $C(s)\\,G(s)$, %s' % ('ab'[j], txt))
        axL[1, j].set_title('%s · Fase de $C(s)\\,G(s)$, %s' % ('cd'[j], txt))
        axL[1, j].set_xlabel('frecuencia [rad/h]')
        axL[0, j].legend(fontsize=7.5, frameon=False, loc='upper right')
    axL[0, 0].set_ylim(-60, 80); axL[0, 0].set_ylabel('magnitud [dB]')
    axL[1, 0].set_ylim(-190, -80); axL[1, 0].set_ylabel('fase [°]')
    for ax in axL.flat:
        ax.grid(which='both', alpha=.3); ax.axvline(wc_g * 3600, color='0.5', ls=':', lw=1)
    figL.suptitle('ADRC: lazo abierto y margen de fase variando $\\omega_o/\\omega_c$ (planta sin retardo)',
                  fontsize=12.5, fontweight='bold')
    figL.savefig('Latex/Tesis/figuras/adrc_margen_fase_kESO.pdf', bbox_inches='tight')
    axc.set_title('Controlador $C(s)$  [dB re 1 % por mg/L]'); axc.set_ylabel('magnitud [dB]'); axc.set_xlabel('frecuencia [rad/h]')
    axc.set_ylim(-20, 130); axc.grid(which='both', alpha=.3); axc.axvline(wc_g * 3600, color='0.5', ls=':', lw=1)
    axc.legend(fontsize=7.5, frameon=False, loc='lower left', ncol=2)
    fig2.savefig('Latex/Tesis/figuras/adrc_C_kESO.pdf', bbox_inches='tight')
    plt.show()

_gang_kESO()
```

#### Ruido: controlador estrictamente propio contra bipropio

El GPI es un filtro por un PI, $C(s) = \dfrac{1}{b_0}\,\dfrac{1}{s + k_2}\cdot\dfrac{k_1 s + k_0}{s}$ (estrictamente propio). Sin el filtro, con su ganancia estática $1/k_2$ para no cambiar el controlador a baja frecuencia, queda el PI $C_{PI}(s) = \dfrac{1}{b_0 k_2}\,\dfrac{k_1 s + k_0}{s}$ (bipropio, $C_{PI}(\infty) = k_1/(b_0 k_2)$). Se comparan $n \to -e_y$, $-1/(1+CG)$, y $n \to u$, $-C/(1+CG)$, con la planta real, y la $\sigma_u$ simulada con el ruido blanco del sensor ($\sigma = 0{,}0011$ mg/L a 50 ms), en régimen en 2,0 mg/L, sin saturación.

```python
# ============================================================
# Ruido: GPI (estrictamente propio) contra el mismo PI sin el filtro 1/(s + k2) (bipropio)
#   C_GPI = (1/b0) (k1 s + k0)/(s (s + k2)),   C_PI = (1/(b0 k2)) (k1 s + k0)/s   (misma ganancia a baja frecuencia)
#   n -> -e_y = -1/(1 + CG),   n -> u = -C/(1 + CG)
#   sigma_u: simulacion a Ts = 50 ms con el ruido blanco del sensor, en regimen en 2,0 mg/L (u = 40 %)
# ============================================================
def _ruido_propio():
    wc_r, wo_r = 4 / Tau, 20 / Tau
    k2, k1, k0 = wc_r + 2 * wo_r, 2 * wc_r * wo_r + wo_r**2, wc_r * wo_r**2
    Ts_r, sig = 0.05, 0.0011
    w = np.logspace(-2, np.log10(np.pi / Ts_r * 3600), 6000) / 3600; s = 1j * w
    G = K_dc / (Tau * s + 1)
    Kp_r, Ki_r = k1 / (bo * k2), k0 / (bo * k2)
    ctrl = {'GPI (estrictamente propio)': (k1 * s + k0) / (bo * s * (s + k2)),
            'PI sin filtro (bipropio)': (k1 * s + k0) / (bo * k2 * s)}
    # --- simulacion muestra a muestra ---
    n_r = int(3000 / Ts_r); ruido = sig * np.random.default_rng(1).standard_normal(n_r)
    ag, bg = np.exp(-Ts_r / Tau), K_dc * (1 - np.exp(-Ts_r / Tau))
    Dg = c2d(ss(tf([k1 / bo, k0 / bo], [1, k2, 0])), Ts_r)               # GPI discretizado ZOH (sin termino directo)
    Ad, Bd, Cd = np.asarray(Dg.A), np.asarray(Dg.B).ravel(), np.asarray(Dg.C).ravel()
    def sigma_u(tipo):
        y = 2.0; U = np.empty(n_r)
        if tipo == 'GPI':                                                # estado de regimen: (A - I) x = 0, C x = 40
            x = np.linalg.lstsq(np.vstack([Ad - np.eye(2), Cd]), np.r_[0, 0, 2.0 / K_dc], rcond=None)[0]
        else:
            I = 2.0 / K_dc
        for k in range(n_r):
            e = 2.0 - (y + ruido[k])
            if tipo == 'GPI':
                u = Cd @ x; x = Ad @ x + Bd * e
            else:
                u = Kp_r * e + I; I += Ki_r * Ts_r * e
            U[k] = u; y = ag * y + bg * u
        return U[int(1000 / Ts_r):].std()
    sim = {'GPI (estrictamente propio)': sigma_u('GPI'), 'PI sin filtro (bipropio)': sigma_u('PI')}
    fig, axs = plt.subplots(1, 2, figsize=(12.5, 4.3), layout='constrained')
    print('C_PI(inf) = k1/(b0 k2) = %.1f %%/(mg/L) = %.1f dB' % (Kp_r, 20 * np.log10(Kp_r)))
    for (nom, C), st in zip(ctrl.items(), (dict(color='k', lw=2.2), dict(color='C3', lw=1.6, ls='--'))):
        L = C * G; D = 1 + L
        lm = np.log10(np.abs(L)); fase = np.degrees(np.unwrap(np.angle(L))); i = np.flatnonzero(np.diff(np.sign(lm)))[0]
        axs[0].semilogx(w * 3600, 20 * np.log10(np.abs(-1 / D)), label=nom, **st)
        axs[1].semilogx(w * 3600, 20 * np.log10(np.abs(-C / D)), label='%s:  $\\sigma_u$ = %s %%' % (nom, ('%.4f' % sim[nom]).replace('.', ',')), **st)
        print('%-28s MF %.1f°   |C/(1+CG)| en 1e4 rad/h %.1f dB, en Nyquist %.1f dB   sigma_u simulada %.4f %%' % (
            nom, 180 + fase[i], 20 * np.log10(np.abs(C / D)[np.argmin(abs(w * 3600 - 1e4))]), 20 * np.log10(np.abs(C / D)[-1]), sim[nom]))
    axs[0].set_title('a · $\\eta \\to -e_y$:  $\\dfrac{-1}{1 + CG}$'); axs[0].set_ylim(-60, 10)
    axs[1].set_title('b · $\\eta \\to u$:  $\\dfrac{-C}{1 + CG}$  [dB re 1 % por mg/L]'); axs[1].set_ylim(-20, 80)
    for ax in axs:
        ax.set_xlabel('frecuencia [rad/h]'); ax.set_ylabel('magnitud [dB]'); ax.grid(which='both', alpha=.3)
        ax.axvline(wc_r * 3600, color='0.5', ls=':', lw=1); ax.legend(fontsize=8.5, frameon=False, loc='lower left')
    fig.suptitle('Ruido con el GPI y con el mismo PI sin el filtro (planta, $\\omega_o = 5\\,\\omega_c$, hasta Nyquist a 50 ms)',
                 fontsize=12, fontweight='bold')
    fig.savefig('Latex/Tesis/figuras/ruido_gpi_vs_pi.pdf', bbox_inches='tight'); plt.show()


_ruido_propio()
```

#### 2.1.5 Relacion GPI with ESO:
$$\dot{\widehat{X}}_{au} = (A_{au} - LC_{au})\widehat{X}_{au} + B_{au} u + Le_y$$
The observer only sees the (1/s) term because the augmented system contains only the (1/s) dynamics. Therefore, (b_0) is not used by the observer.
$$u=-(K\cdot\widehat{X}_{au})$$


$$\dot{\widehat{X}}_{au} = (A_{au} - LC_{au}-B_{au}K)\widehat{X}_{au}  + Le_y$$
$$A_c = A_{au} - LC_{au}-B_{au}K$$
$$B_c = L_{(n+m,1)}$$
$$C_c = I_{(n+m,n+m)}$$
$$D_c = 0_{(n+m,1)}$$

Transfer function:

$$\frac{X_{au}(s)}{E_y(s)}=G_{au}(s)=C_c[sI-A_c]^{-1}B_c+D_c$$

$$G_{au}(s)=\begin{bmatrix} G_{au_1}(s) \\ G_{au_2}(s)\\ \vdots \\ G_{au_{(n+m)}}(s)\end{bmatrix}$$

$$U(s) = -\frac{1}{b_0} \cdot K\cdot G_{au}(s) \cdot E_y$$
$$U(s) = -\frac{1}{b_0} \cdot K_{(1,n+m)}\cdot G_{au}(s)_{(n+m,1)} \cdot E_y$$
$$U(s) = -\frac{1}{b_0} \cdot K_{(1,n+m)}\cdot G_{au}(s)_{(n+m,1)} \cdot E_y$$



---


#### 2.1.5.1 Building the controller from integrators (GPI / flat filter)

Error dynamics of the design model, with $u_0=b_0u$ and $\xi$ the total disturbance:

$$
\dot e_y=u_0+\xi
$$

Only $e_y$, $u_0$ and their integrals may be used: no derivatives, no observer. The gains are numbered so that the subscript matches the power of $s$ it ends up multiplying in the characteristic polynomial.

**A. Proportional only.**

$$
u_0=-k_0e_y
\qquad\Longrightarrow\qquad
e_y\,(s+k_0)=\xi
\qquad\Longrightarrow\qquad
e_y=\frac{\xi}{s+k_0}
$$

For a constant $\xi$ of value $\xi(\infty)$ the final value is $\frac{\xi(\infty)}{k_0}\neq0$: offset. The internal model of $\xi$ is missing.

**B. Proportional plus integral (minimum GPI).**

$$
u_0=-k_1e_y-k_0\!\int_0^t\! e_y(\tau)\,d\tau
\qquad\Longrightarrow\qquad
s\,e_y+k_1e_y+\frac{k_0}{s}e_y=\xi
$$

$$
e_y\big[s^{2}+k_1s+k_0\big]=s\,\xi
\qquad\Longrightarrow\qquad
\lim_{t\to\infty}e_y=0
$$

The $s$ on the right-hand side is the integrator of the controller: it is what kills a constant $\xi$. This already works, and it is the pure GPI, $C(s)=(k_1s+k_0)/s$.

**C. Replacing the measurement by a reconstruction.** Integrating the plant once,

$$
\int_0^t\!\dot e_y\,d\tau=\int_0^t\! u_0(\tau)\,d\tau+\int_0^t\!\xi(\tau)\,d\tau
$$
$$
\qquad\Longrightarrow\qquad
(e_y(t)- e_y(0))=\int_0^t\! u_0\,d\tau+\int_0^t\!\xi\,d\tau
\qquad\Longrightarrow\qquad
e_y(t)=\int_0^t\! u_0\,d\tau+\int_0^t\!\xi\,d\tau+ e_y(0)
$$

Dropping what is unknown, $e_y(0)$ and $\xi$, leaves the integral reconstructor of the output:

$$
\widehat e_y(t)=\int_0^t\! u_0(\tau)\,d\tau
$$

Feeding back that reconstruction instead of the measured error:

$$
u_0=-k_2\!\int_0^t\! u_0\,d\tau-k_1\!\int_0^t\! e_y\,d\tau
\qquad\Longrightarrow\qquad
u_0\Big(1+\frac{k_2}{s}\Big)=-\frac{k_1}{s}e_y
\qquad\Longrightarrow\qquad
u_0=-\frac{k_1}{s+k_2}\,e_y
$$

$$
e_y\big[s^{2}+k_2s+k_1\big]=(s+k_2)\,\xi
\qquad\Longrightarrow\qquad
\lim_{t\to\infty}e_y=\frac{k_2\,\xi(\infty)}{k_1}\neq0
$$

It fails. The $\int u_0$ moved the $1/s$ into the denominator, so $C(s)=k_1/(s+k_2)$ has no pole at the origin: the integrator that rejected $\xi$ in **B** was spent on the reconstructor.

**D. One more integral of the error: the flat filter.**

$$
u_0=-k_2\!\int_0^t\! u_0(\tau)\,d\tau-k_1\!\int_0^t\! e_y(\tau)\,d\tau-k_0\!\iint e_y(\tau)\,d\tau
$$

$$
u_0(s)=-\frac{k_2}{s}u_0(s)-\frac{k_1}{s}e_y(s)-\frac{k_0}{s^{2}}e_y(s)
\qquad\Longrightarrow\qquad
u_0\Big(1+\frac{k_2}{s}\Big)=-\frac{k_1s+k_0}{s^{2}}\,e_y
$$

$$
\boxed{\;u_0(s)=-\frac{k_1s+k_0}{s\,(s+k_2)}\,e_y(s)\;}
$$

Closing with $\dot e_y=u_0+\xi$:

$$
s\,e_y=-\frac{k_1s+k_0}{s\,(s+k_2)}e_y+\xi
\qquad\Longrightarrow\qquad
s\,e_y\,s(s+k_2)+(k_1s+k_0)\,e_y=s\,(s+k_2)\,\xi
$$

$$
e_y\big[s^{3}+k_2s^{2}+k_1s+k_0\big]=s\,(s+k_2)\,\xi
\qquad\Longrightarrow\qquad
\lim_{t\to\infty}e_y=0
$$

**Bookkeeping.** What has to be counted is not how many integrals are written, but how many survive in $C(s)$:

$$
\text{net integrators of }C(s)=\#\!\int\! e_y-\#\!\int\! u_0=2-1=1=m
$$

one per degree of the polynomial model of $\xi$. A ramp disturbance would need three integrals of the error and still one of $u_0$.

**Implementation, integrators only.** With $x=e_y/[s(s+k_2)]$:

$$
\dot x_1=x_2,\qquad \dot x_2=-k_2x_2+e_y,\qquad u_0=-k_0x_1-k_1x_2,\qquad u=\frac{u_0}{b_0}
$$

All the internal signals are $x$ and its derivatives: this is what makes it a *flat* filter. Design is pole placement on $s^{3}+k_2s^{2}+k_1s+k_0$, three coefficients for three roots.

#### 2.1.5.2 Error-based observer $\rightarrow$ $C(s)$, step by step ($n=1$, $m=1$, continuous)

Closing the observer with the control law gives a state-space system whose input is $e_y$ and whose output is $\widehat{x}_{au}$:

$$
C_c\big[sI-A_c\big]^{-1}B_c+D_c,
\qquad
A_c=A_{au}-LC_{au}-B_{au}K,\quad
B_c=L_{(n+m,1)},\quad
C_c=I_{(n+m,n+m)},\quad
D_c=0_{(n+m,1)}
$$

Design model (integrator chain + constant disturbance) and gains, with $k_p$ the control gain:

$$
A_{au}=\begin{bmatrix}0&1\\0&0\end{bmatrix},\quad
B_{au}=\begin{bmatrix}1\\0\end{bmatrix},\quad
C_{au}=\begin{bmatrix}1&0\end{bmatrix},\quad
L=\begin{bmatrix}l_1\\l_2\end{bmatrix},\quad
K=\begin{bmatrix}k_p&1\end{bmatrix}
$$

The second entry of $K$ is $1$: it is the term that cancels $\widehat{x}_2=\widehat f$ in $u_0=-K\widehat{x}_{au}$.

**Closed-loop observer matrix.**

$$
A_c=\underbrace{\begin{bmatrix}0&1\\0&0\end{bmatrix}}_{A_{au}}
-\underbrace{\begin{bmatrix}l_1\\l_2\end{bmatrix}\begin{bmatrix}1&0\end{bmatrix}}_{LC_{au}}
-\underbrace{\begin{bmatrix}1\\0\end{bmatrix}\begin{bmatrix}k_p&1\end{bmatrix}}_{B_{au}K}
=\begin{bmatrix}0&1\\0&0\end{bmatrix}
-\begin{bmatrix}l_1&0\\l_2&0\end{bmatrix}
-\begin{bmatrix}k_p&1\\0&0\end{bmatrix}
$$

$$
A_c=\begin{bmatrix}-(k_p+l_1)&0\\-l_2&0\end{bmatrix}
$$

The $1$ of $A_{au}$ and the $1$ of $B_{au}K$ cancel, so the $(1,2)$ entry is zero.

**Resolvent.** Subtracting $A_c$ changes the sign of every entry:

$$
sI-A_c=\begin{bmatrix}s+k_p+l_1&0\\l_2&s\end{bmatrix},
\qquad
\det(sI-A_c)=s\,(s+k_p+l_1)
$$

$$
\big[sI-A_c\big]^{-1}=\frac{1}{s\,(s+k_p+l_1)}
\begin{bmatrix}s&0\\-l_2&s+k_p+l_1\end{bmatrix}
$$

**States as a function of the error.** The input matrix is $B_c=L$, so the resolvent multiplies the gain vector row by row:

$$
\frac{\widehat{x}_{au}(s)}{e_y(s)}=\big[sI-A_c\big]^{-1}L
=\frac{1}{s\,(s+k_p+l_1)}
\begin{bmatrix}s&0\\-l_2&s+k_p+l_1\end{bmatrix}
\begin{bmatrix}l_1\\l_2\end{bmatrix}
$$

$$
=\frac{1}{s\,(s+k_p+l_1)}
\begin{bmatrix}s\cdot l_1+0\cdot l_2\\-l_2\cdot l_1+(s+k_p+l_1)\,l_2\end{bmatrix}
=\frac{1}{s\,(s+k_p+l_1)}
\begin{bmatrix}l_1s\\l_2(s+k_p)\end{bmatrix}
$$

In the second row $-l_1l_2$ cancels with the $+l_1l_2$ of the product, which is what leaves $l_2(s+k_p)$.

**Control law.** With $u=-\dfrac{1}{b_0}K\widehat{x}_{au}=-\dfrac{1}{b_0}\big(k_p\widehat{x}_1+\widehat{x}_2\big)$, the transfer function from $-e_y$ to $u$ is

$$
b_0\,C(s)=k_p\frac{l_1s}{s(s+k_p+l_1)}+\frac{l_2(s+k_p)}{s(s+k_p+l_1)}
=\frac{(k_pl_1+l_2)\,s+k_pl_2}{s\,(s+k_p+l_1)}
$$


**Same thing as a GPI (flat filter).** Writing the controller as $b_0C(s)=\dfrac{k_1s+k_0}{s\,(s+k_2)}$, the GPI gains are read directly from the observer gains:

$$
\boxed{\;k_2=k_p+l_1,\qquad k_1=k_pl_1+l_2,\qquad k_0=k_pl_2\;}
$$

$$
e_y\Big[s^{3}+k_2s^{2}+k_1s+k_0\Big]=s\,(s+k_2)\,\xi
$$

Both the characteristic polynomial and the zeros of $e_y/\xi$ agree, so this is the same controller written as a compensation network. Note the $s$ in the denominator: without it the controller has no integrator and a constant $\xi$ is not rejected. Going back, $k_p$ is any real root of the cubic and then $l_1=k_2-k_p$, $l_2=k_0/k_p$.

```python
#####################################
#transfer function: observer + control law  (-e_y -> u)
#####################################

A_cl_n = Aau - L@Cau - Bau@K
B_cl_n = L.reshape(n+m,1)
C_cl_n = np.eye(n+m)
D_cl_n = np.zeros((n+m,1))
sys_cl_n  = ss(A_cl_n, B_cl_n, C_cl_n, D_cl_n)
sys_cl_tf = ss2tf(sys_cl_n)
display(Latex(r"Transfer functions from $e_y$ to $\hat{x}_{au}$:"))
display(sys_cl_tf)

u_tf = (K[0,0] * sys_cl_tf[0,0] + K[0,1] * sys_cl_tf[1,0]) / bo
u_tf = minreal(u_tf , tol=1e-6)
display(Latex(r"$C(s)$ from $-e_y$ to $u$:"))
display(u_tf)
pzmap(u_tf)

u_symb_s = u_tf                 # ya es continuo: no hay que pasar de z a s
print(u_symb_s.zeros())
print(u_symb_s.poles())

# C(s) = PI(s) * F(s)
s_z = np.real(u_symb_s.zeros())[0]
print(s_z)
s_p = np.real(u_symb_s.poles()).min()
print(s_p)
k_s = u_symb_s.num[0][0][0] / u_symb_s.den[0][0][0]
wf  = -s_p
K1  = (k_s / wf)*bo
K2  = (-K1 * s_z)

GPI_s = tf([K1, K2], [1, 0])*1/bo
F_s  = tf([wf], [1, wf])

display(Latex(r"$F(s)$, observer filter:"))
display(tf2sym(F_s))
display(Latex(r"GPI form $\dfrac{1}{b_0}\dfrac{k_1 s + k_0}{s}$:  $k_1 = %.4e$,  $k_0 = %.4e$" % (K1, K2)))
print('K1 = %.4f   K2 = %.6f   Tf = %.1f s   fc=%.4f ' % (K1, K2, 1/wf,wf))
```

#### 2.1.6 Anti-Integral windup


**Controller implementation with saturation feedback** (*The Control Handbook*, 2nd ed., §19.1.3)

Any linear controller can be written as a ratio of polynomials, with $L(s)$ monic of degree $n$: For that, we use the faster poles.

$$
U(s) = C(s)\,E(s) = \frac{P(s)}{L(s)}\,E(s), \qquad L(s) = s^{n} + l_{n-1}s^{n-1} + \cdots + l_0
$$

It is implemented with saturation feedback, using a stable monic polynomial $E_1(s)$ of the same degree $n$:

$$
V(s) = \frac{P(s)}{E_1(s)}\,E(s) - \frac{L(s)-E_1(s)}{E_1(s)}\,U(s), \qquad U = \mathrm{sat}(V)
$$

**When the actuator does not saturate ($V = U$)** the original controller is recovered:

$$
\begin{aligned}
U &= \frac{P}{E_1}\,E - \frac{L-E_1}{E_1}\,U \\[4pt]
U\left[1 + \frac{L-E_1}{E_1}\right] &= \frac{P}{E_1}\,E \\[4pt]
U\left[\frac{L}{E_1}\right] &= \frac{P}{E_1}\,E \\[4pt]
U &= \frac{P}{L}\,E
\end{aligned}
$$

**Example** (book): plant $\dfrac{1}{s}$ and controller

$$
C(s) = \frac{6s+5}{s} \qquad\Longrightarrow\qquad P(s) = 6s+5, \quad L(s) = s, \quad n = 1
$$

Closed loop:

$$
G_{cl}(s) = \frac{\dfrac{6s+5}{s}\cdot\dfrac{1}{s}}{1+\dfrac{6s+5}{s}\cdot\dfrac{1}{s}} = \frac{6s+5}{s^{2}+6s+5}
$$

Poles: $(s+1)(s+5)$. $E_1$ takes the $n = 1$ fastest one:

$$
E_1(s) = s+5
$$

Then:

$$
\begin{aligned}
V &= \frac{6s+5}{s+5}\,E(s) - \frac{s-(s+5)}{s+5}\,U(s) \\[4pt]
V &= \frac{6s+5}{s+5}\,E(s) - \frac{-5}{s+5}\,U(s)
\end{aligned}
$$


```python
def antiWindup(C, Gp, symb=None):
    """C = P/L -> [P/E1, (L-E1)/E1].  Handbook 19.1.3.  symb: 's', 'z' or None (auto)."""
    if symb is None:
        symb = 'z' if getattr(C, 'dt', 0) else 's'
    P = np.asarray(C.num[0][0], float) / C.den[0][0][0]
    L = np.asarray(C.den[0][0], float) / C.den[0][0][0]     # monic
    n = len(L) - 1
    p = feedback(C * Gp, 1).poles()
    
    # E1: the n fastest closed-loop poles
    fast = p[np.argsort(np.abs(p))][:n] if symb == 'z' else p[np.argsort(p.real)][:n]
    E1 = np.poly(fast)
    arg = (T_s,) if symb == 'z' else ()
    return [tf(P, E1, *arg), tf(L - E1, E1, *arg)]

for nom, C, Gp in (('s', Ctrl_tf, Gs_real),):
    C1, C2 = antiWindup(C, Gp, nom)
    display(Latex(r"$E_1(%s) = %s$" % (nom, C1.den)))
    display(Latex(r"$P/E_1$  (on the error):")); display(C1)
    display(Latex(r"$(L-E_1)/E_1$  (on the saturated $u$):")); display(C2)
```

**Anti-windup escrito con las ganancias del GPI** (sin calcular polos del lazo cerrado)

Con la planta de diseño $b_0/s$, la orden se calcula en unidades de $u_0$ y $u = u_0/b_0$.
En unidades de $u_0$:

$$C_0(s) = b_0\,C(s) = \frac{k_1 s + k_0}{s\,(s + k_2)} = \frac{P(s)}{L(s)},\qquad
P(s) = k_1 s + k_0,\qquad L(s) = s^2 + k_2 s\quad (n = 2),$$

y el polinomio característico del lazo es
$1 + C_0(s)\,\frac{1}{s} = 0 \;\Rightarrow\; \Delta(s) = s^3 + k_2 s^2 + k_1 s + k_0$.
Se toma $E_1$ como $\Delta$ sin su término independiente, dividido por $s$: mónico, de
grado 2, construido solo con las ganancias, sin factorizar,
$E_1(s) = s^2 + k_2 s + k_1$. Con $C = C_0/b_0$, el anti-windup deja $1/b_0$ justo después
del bloque de $P/E_1$, y a la rama de realimentación llega $u_{\mathrm{sat}}$:

$$v_1 = \frac{k_1 s + k_0}{s^2 + k_2 s + k_1}\,(-e_y),\qquad
v_2 = \frac{\bigl[s^2 + k_2 s\bigr] - \bigl[s^2 + k_2 s + k_1\bigr]}{s^2 + k_2 s + k_1}\,u_{\mathrm{sat}}
= \frac{-k_1}{s^2 + k_2 s + k_1}\,u_{\mathrm{sat}},$$

$$v = \frac{v_1}{b_0} - v_2,\qquad u_{\mathrm{sat}} = \mathrm{sat}(v).$$

Dos propiedades salen sin costo: $E_1$ es estable siempre que $k_2, k_1 > 0$, y la ganancia
estática de la rama de realimentación es exactamente $-k_1/k_1 = -1$, así que la acción
integral se recupera sin restar números casi iguales. Si el actuador no se satura,
$u_{\mathrm{sat}} = v$ y el lazo vuelve a $u = \frac{1}{b_0}(P/L)(-e_y) = C(s)(-e_y)$.
$E_1$ es de libre elección: cualquier $s^2 + \alpha_1 s + \alpha_0$ mónico y estable sirve,
y solo cambia $v_2 = \bigl[(k_2 - \alpha_1)s - \alpha_0\bigr]/E_1\cdot u_{\mathrm{sat}}$.

### 2.2 Control Law with GPI

### 2.2.1 Control law






whit n=1 and m=1 
$$
    u(s) = -\frac{1}{b_0} 
    \left[ 
    \frac{k_{1}s +   k_0}
    {s\cdot(s+k_{2})} 
    \right] e_y(s)
$$




**error dynamics**
\begin{align*}
y &= \frac{N_G}{D_G}\Big[u+\bar{\xi}\Big] \\[4pt]
y-y^{*} &= \frac{N_G}{D_G}\Big[u+\bar{\xi}\Big]-y^{*} \\[4pt]
e_y &= \frac{N_G}{D_G}\left[u+\bar{\xi}-\frac{D_G}{N_G}\,y^{*}\right] \\[4pt]
e_y &= \frac{N_G}{D_G}\Big[u+\xi\Big] &&(y^{*}=0) \\[4pt]
e_yD_G &= N_G\Big[u+\xi\Big] \\[4pt]
e_yD_G &= N_G\,u(s) + N_G\,\xi \\[4pt]
\end{align*}
**Law control**

$$
u = \left[\frac{N_C}{s^{m}D_C}\right](-e_y)
$$

**"Replace": closing the loop**
\begin{align*}
e_yD_G &= N_G\left[\frac{N_C}{s^{m}D_C}\right](-e_y) + N_G\,\xi \\[4pt]
e_yD_G &= \frac{-N_GN_C\,e_y}{s^{m}D_C} + N_G\,\xi
   &&\big(D_A := s^{m}\big) \\[4pt]
e_yD_G &= \frac{-N_GN_C\,e_y + D_AD_CN_G\,\xi}{D_AD_C} \\[4pt]
e_yD_GD_AD_C + N_CN_G\,e_y &= D_AD_CN_G\,\xi
\end{align*}
$$
\big[D_GD_CD_A+N_CN_G\big]\,e_y = D_AD_CN_G\,\xi \;=: P_{e_y}$$

**With $b_0,\ n=1,\ N_G=1,\ m=1$:**

\begin{align*}
\big[D_GD_CD_A+N_CN_G\big]\,e_y &= D_AD_CN_G\,\xi \\[4pt]
\big[D_GD_CD_A+N_C\big]\,e_y &= D_AD_C\,\xi
   &&\big(N_G=1,\ G(s)=\tfrac1s\big) \\[4pt]
\big[D_GD_Cs+N_C\big]\,e_y &= D_Cs\,\xi
   &&(D_A=s)
\end{align*}

**with $G(s)=\dfrac{1}{s}$, $u(s) = -\left[\frac{k_{1}s + k_0}{s\cdot(s+k_{2})}\right] e_y(s)$, $m=1$ and $(D_C=(s+k_2))$**


\begin{align*}
\big[sD_Cs+N_C\big]\,e_y &= sD_C\,\xi \\[4pt]
\big[s^{2}(s+k_2)+N_C\big]\,e_y &= s\,(s+k_2)\,\xi  \\[4pt]
\big[s^{3}+k_2s^{2}+k_1s+k_0\big]\,e_y &= s\,(s+k_2)\,\xi
\end{align*}



```python
# ============================================================
# GPI  n=1, m=2   ·   diseno por colocacion de polos
#
#   De la celda 2.2.1:   [s^3 + k2 s^2 + k1 s + k0] e_y = s^2 xi
#   -> los k son, uno a uno, los coeficientes del polinomio deseado.
#
#   TRES cosas que hay que respetar en el codigo, y que eran las que
#   inestabilizaban:
#
#   1. EL SIGNO.  La ley es  u = -(1/b0)*N_C/s^m * e_y  con e_y = y - r.
#      Pero feedback() cierra con  e = r - y = -e_y, asi que el menos YA
#      esta en el lazo.  Si ademas se deja el menos en C, la realimentacion
#      queda POSITIVA y aparece un polo en +3.96e-3.
#
#   2. feedback(), no L/(1+L).  La forma algebraica no cancela polos y ceros
#      y devuelve 6 polos en vez de 3, con dos espurios en s = 0.
#
#   3. CONTRA QUE PLANTA se cierra.  Los k se calcularon para la planta de
#      DISENO b0/s.  Contra ella los polos caen exactos donde se pidieron.
#      Contra la planta REAL K_dc/(Tau s+1) se corren, porque su polo propio
#      (-1/Tau = -8.3e-4) cae justo en medio del ancho de banda de diseno.
# ============================================================

s, kc_0, kc_1, kc_2, kc_sys = symbols(['s', 'K_0', 'K_1', 'K_2', 'b0'])

# --- dinamica deseada en lazo cerrado ---
dynamic = (s + 0.0014) * (s + 0.0012) * (s + 0.001)
#dynamic = (s + 0.0014) * (s + 0.0012)
display(Latex(r"Desired closed-loop dynamics:"))
display(dynamic)

polyDinamic = simplify(dynamic).as_poly()
coeff = polyDinamic.all_coeffs()          # [s^3, s^2, s^1, s^0]
display(Latex(r"Coefficients of the desired polynomial:"))
display(coeff)

# --- controlador GPI para n=1, m=2 ---
display(Latex(r"Controller  n=1, m=2:"))
#Ctrl_s = (1 / kc_sys) * (kc_2 * s**2 + kc_1 * s + kc_0) / s**2
Ctrl_s = (1 / kc_sys) * (kc_1 * s + kc_0) / ((s + kc_2)*s)
display(Ctrl_s)

print(K)
print(L)
#Ctrl_s = Ctrl_s.subs({kc_0: coeff[3], kc_1: coeff[2], kc_2: coeff[1], kc_sys: bo})

  
   

Ctrl_s = Ctrl_s.subs({
    kc_0: 0.000941* 0.000900000000000,
    kc_1: 0.000941* 0.060000000000000+0.000900000000000,
    kc_2: 0.000941+ 0.060000000000000,
    kc_sys: bo
})
#Ctrl_s = Ctrl_s.subs({ kc_0: coeff[2], kc_1: coeff[1], kc_sys: bo})
Ctrl_tf = sym2tf(Ctrl_s)
display(Latex(r"Controller transfer function $C(s)$:"))
print(Ctrl_tf)

# ============================================================
# Lazo cerrado
# ============================================================
Gs_dis = tf([bo], [1, 0])                 # planta de DISENO: b0/s

T_dis  = feedback(Ctrl_tf * Gs_dis,  1)   # contra la de diseno
T_real = feedback(Ctrl_tf * Gs_real, 1)   # contra la real

print('\npolos pedidos                      :', np.array2string(
      np.sort_complex(np.roots([float(c) for c in coeff])), precision=6))
print('polos contra la planta de diseno   :', np.array2string(
      np.sort_complex(pole(T_dis)), precision=6), ' <- exactos')
print('polos contra la planta REAL        :', np.array2string(
      np.sort_complex(pole(T_real)), precision=6))
print('estable contra la real?             ', 'SI' if (np.real(pole(T_real)) < 0).all() else 'NO')

# ============================================================
# Respuesta al escalon del lazo cerrado (continuo)
#   los polos son lentos (tau ~ 1700 s): hace falta un vector de tiempo
#   explicito, con el automatico se corta antes de llegar al regimen
# ============================================================
t_cl = np.arange(0, 8000, T_s)
y_cl, _ = step(T_real,   t_cl)

print('\nescalon: y_final = %.4f   ·   sobreimpulso %.1f %%'
      % (y_cl[-1], 100 * (y_cl.max() - y_cl[-1]) / y_cl[-1]))

fig, ax = plt.subplots(1, 1, figsize=(8, 5), layout='constrained')
ax.plot(t_cl, y_cl, color='C0', linewidth=2, label='continuo  $T(s)$')
ax.axhline(1.0, color='k', linewidth=1, linestyle=':')
ax.set_title('GPI n=1, m=1 · escalon en lazo cerrado contra la planta real', fontsize=12)
ax.set_xlabel('Time [s]'); ax.set_ylabel('y(t)')
ax.minorticks_on(); ax.grid(visible=True, which='both', alpha=.35); ax.legend()
plt.show()
```

### 2.2.3 Closed-loop simulation

Escalón de consigna de 2,0 a 2,5 mg/L y escalón de perturbación, contra la planta
real $K_{dc}/(\tau s+1)$.

**Sobre la "perturbación estimada":** esta forma del GPI no tiene observador, así
que no hay ningún estado $\hat\xi$ que leer. La estimación está **implícita en la
acción integral**. Con la ley de 2.2.1, el controlador es estrictamente propio y no
tiene parte proporcional: se parte en su rama integradora y su rama de reconstrucción,

$$b_0\,C(s) = \underbrace{\frac{k_0/k_2}{s}}_{\text{modelo interno de }f}
+\underbrace{\frac{(k_1k_2-k_0)/k_2}{s+k_2}}_{\text{reconstructor}}$$

Comparando con la forma del ADRC, $b_0u=u_0=-k_p\widehat{x}_1-\widehat f$, donde
$\widehat{x}_1$ es el error reconstruido (la rama de arriba, el error filtrado por
$1/(s+k_2)$), la estimación se lee **directamente de la señal de control**, sin
simular ningún integrador suelto:

$$\hat f = -b_0\,u-k_p\,\widehat{x}_1
\qquad\xrightarrow{\;e_y\to0\;}\qquad
\hat f = -b_0\,u \qquad [\mathrm{mg/(L\cdot s)}]$$

El término que falta es proporcional al error reconstruido, así que desaparece en
estacionario y en los tramos sin cambio de consigna; es lo que se grafica abajo. (El
integrador suelto en lazo abierto no sirve aquí: con 1,8 millones de muestras
acumula su propio error numérico y la estimación se va.)

Lo acumulado por la integral contiene en estacionario dos cosas: el sesgo que sostiene la
consigna ($r/K_{dc}$) y la compensación de la perturbación. Despejando la segunda:

$$\hat\xi = \frac{r}{K_{dc}} + \frac{\hat f}{b_0} \qquad [\%\ \text{de }u]$$

Ésa es la lectura comparable con la $\hat\xi$ del ESO de la sección 2.1, y es lo que
se grafica abajo.

```python
# ============================================================
# Simulacion en lazo cerrado del GPI  (n=1, m=2)  ·  POR BLOQUES, EN CONTINUO
#
#                            xi (perturbacion, % de u)
#                                       |
#    r --->(+)---- e ---->[ C(s) ]--u-->(+)--->[ G(s) ]----+---> y
#          -^                                              |
#           +----------------------------------------------+
#
#   C(s)                                                       -> Ctrl_tf, celda 2.2.2
#   G(s) = K_dc/(Tau s+1)                                      -> Gs_real, la planta REAL
#
#   Dos entradas (r y xi) en un lazo lineal = SUPERPOSICION: cada una tiene su
#   propia funcion de transferencia y las respuestas se suman.  Las cuatro salen
#   de feedback(), sin escribir un solo cociente a mano:
#
#       y = G_yr r + G_yd xi              u = G_ur r + G_ud xi
#
#       G_yr = CG/(1+CG)                  G_ur =  C/(1+CG)
#       G_yd =  G/(1+CG)                  G_ud = -CG/(1+CG) = -G_yr
#
#   (G_ud = -G_yr no es casualidad: lo que el control tiene que quitar es
#    exactamente lo que la perturbacion mete por el mismo punto de suma.)
#
#   Y la perturbacion estimada sale de la identidad del ADRC, leyendo la propia
#   senal de control  --  sin simular el integrador suelto, que en lazo abierto
#   y con millones de muestras acumula su error numerico y se va:
#
#       b0 u = u0 - f_est,   u0 = k2 e      ->      f_est = k2 e - b0 u
# ============================================================

# lsim() siempre devuelve tambien el estado interno; con una funcion de
# transferencia eso lanza un aviso que aqui no aplica (solo se usa la salida)
import warnings; warnings.filterwarnings('ignore', message='return_x specified')

#k2_, k1_, k0_ = float(coeff[1]), float(coeff[2]), float(coeff[3])
k1_, k0_ = float(coeff[1]), float(coeff[2])
# --- los bloques ---
G_tf = Gs_real                             # la planta real, continua, como TF

G_yr = feedback(Ctrl_tf * G_tf, 1)         # r  -> y
G_yd = feedback(G_tf, Ctrl_tf)             # xi -> y
G_ur = feedback(Ctrl_tf, G_tf)             # r  -> u
G_ud = -G_yr                               # xi -> u

print('polos del lazo cerrado en s :', np.array2string(
      np.sort_complex(pole(G_yr)), precision=6))

# --- simulacion ---
t_g  = np.arange(0, 20000, T_s)
r_g  = np.where(t_g >= 4000,  2.5,   2.0)      # consigna [mg/L]
xi_g = np.where(t_g >= 12000, -15.0, 0.0)      # perturbacion [% de u]

y_g = lsim(G_yr, r_g, t_g)[0] + lsim(G_yd, xi_g, t_g)[0]     # salida
u_g = lsim(G_ur, r_g, t_g)[0] + lsim(G_ud, xi_g, t_g)[0]     # senal de control [%]

e_g      = r_g - y_g                                          # error
#f_est_g  = k2_ * e_g - bo * u_g                               # f estimada [mg/(L*s)]
f_est_g  = - bo * u_g      
xi_est_g = r_g / K_dc + f_est_g / bo                          # la misma, en % de u

k_ref = int(11000 / T_s); k_fin = -1
print('\nantes de la perturbacion (t=11000 s): y=%.4f  u=%.2f %%  xi_est=%+.2f %%  (aplicada %+.1f)'
      % (y_g[k_ref], u_g[k_ref], xi_est_g[k_ref], xi_g[k_ref]))
print('al final                 (t=20000 s): y=%.4f  u=%.2f %%  xi_est=%+.2f %%  (aplicada %+.1f)'
      % (y_g[k_fin], u_g[k_fin], xi_est_g[k_fin], xi_g[k_fin]))
print('error final |y - r| = %.2e mg/L' % abs(y_g[k_fin] - r_g[k_fin]))

# ============================================================
# Figuras
# ============================================================
fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.5), layout='constrained')
ax = axs[0, 0]
ax.plot(t_g, y_g, color='C0', linewidth=2, label='$y$  (OD)')
ax.plot(t_g, r_g, color='k', linewidth=1.4, linestyle='--', label='$r$  (consigna)')
ax.set_title('a · Salida contra consigna'); ax.set_ylabel('OD [mg/L]')

ax = axs[0, 1]
ax.plot(t_g, e_g, color='C0', linewidth=2, label='$e = r - y$')
ax.axhline(0, color='k', linewidth=1, linestyle=':')
ax.set_title('b · El error'); ax.set_ylabel('error [mg/L]')
ax.text(.97, .9, 'error final: %.1e mg/L' % abs(e_g[-1]), transform=ax.transAxes,
        ha='right', fontsize=9, color='#555555')

ax = axs[1, 0]
ax.plot(t_g, xi_est_g, color='C2', linewidth=2,
        label=r'$\hat\xi = r/K_{dc} + \hat f/b_0$  (leida de $u$)')
ax.plot(t_g, xi_g, color='k', linewidth=1.5, linestyle='--', label='$\\xi$ APLICADA')
ax.set_title('c · La perturbación: estimada contra aplicada'); ax.set_ylabel('[% de $u$]')

ax = axs[1, 1]
ax.plot(t_g, u_g, color='C4', linewidth=2, label='$u$  (al soplador)')
ax.axhline(100, color='C3', linestyle=':', linewidth=1.2, label='límite del soplador')
ax.axhline(0, color='C3', linestyle=':', linewidth=1.2)
ax.set_title('d · Señal de control'); ax.set_ylabel('$u$ [%]')

for ax in axs.flat:
    ax.minorticks_on(); ax.legend(fontsize=8.5, frameon=False, loc='best')
    ax.grid(visible=True, which='major', alpha=.35); ax.set_xlabel('tiempo [s]')

fig.suptitle('GPI $n=1$, $m=2$ · lazo cerrado CONTINUO contra la planta real',
             fontsize=13, fontweight='bold')
plt.show()
```

### 2.2.4 Caso de estudio: variación de $k_0$, $k_1$ y $k_2$

Planta real de primer orden sin retardo, $G(s)=K_{dc}/(\tau s+1)$, y el GPI

$$u(s) = -\frac{1}{b_0}\,\frac{k_1 s + k_0}{s\,(s+k_2)}\,e_y(s).$$

Las ganancias nominales salen del ESO de la sección 2.1 ($k_2 = k_{p_1}+l_1$, $k_1 = k_{p_1}l_1+l_2$, $k_0 = k_{p_1}l_2$). Se cambia una ganancia a la vez por $\times$ 1/4, 1/2, 2 y 4, con el escalón de consigna de 2,0 a 2,2 mg/L desde el régimen (soplador en $2{,}0/K_{dc} = 40$ %), sin saturación.

Métricas del transitorio: $M_p$ en % del escalón, $t_s$ con banda de 2 % del escalón, IAE $=\int|y-r|\,dt$ desde el escalón y $u_{\max}$.

Sensibilidad respecto a la perturbación total: $e_y = S(s)\,\xi$, con $S(s) = \dfrac{s\,(s+k_2)}{s^3+k_2s^2+k_1s+k_0}$.

Al final, el GPI nominal contra el ADRC con observador de la sección 2.1 en el mismo escalón.

```python
# ============================================================
# GPI n=1, m=1 · variacion de k0, k1, k2 en el escalon 2,0 -> 2,2 mg/L
#   Ganancias nominales desde el ESO (ec. eso2gpi):
#       k2 = kp + l1,   k1 = kp l1 + l2,   k0 = kp l2
#   Lazo lineal contra la planta REAL, sin saturacion: se ve la u que se pide.
# ============================================================
kp_, l1_, l2_ = K[0, 0], L[0, 0], L[1, 0]
k_nom = {'k2': kp_ + l1_, 'k1': kp_ * l1_ + l2_, 'k0': kp_ * l2_}
wc_g, wo_g = -pols[0], -pols_L[0]

def C_gpi(k2, k1, k0):
    """C(s) de -e_y a u:  (1/b0) (k1 s + k0) / (s (s + k2))"""
    return tf([k1 / bo, k0 / bo], [1, k2, 0])

r0, r1 = 2.0, 2.2                         # escalon de consigna [mg/L]
u_ss   = r0 / K_dc                        # soplador en regimen a 2,0 mg/L [%]
t_cs   = np.arange(0, 3600, 0.5)          # tiempo desde el escalon [s]
factores = [(1 / 4, '1/4'), (1 / 2, '1/2'), (2, '2'), (4, '4')]
param    = [('k0', 'k_0'), ('k1', 'k_1'), ('k2', 'k_2')]
coma     = lambda x, f='%.1f': (f % x).replace('.', ',')

def respuesta(k):
    C = C_gpi(**k)
    y = r0 + (r1 - r0) * step(feedback(C * Gs_real, 1), t_cs)[0]     # y/r
    u = u_ss + (r1 - r0) * step(feedback(C, Gs_real), t_cs)[0]       # u/r
    return y, u

def metricas(y, u, banda=0.02):
    d     = r1 - r0
    Mp    = max(0.0, 100 * (y.max() - r1) / d)                       # % del escalon
    fuera = np.flatnonzero(np.abs(y - r1) > banda * d)
    ts    = t_cs[fuera[-1] + 1] / 60 if len(fuera) else 0.0          # [min]
    iae   = np.trapezoid(np.abs(y - r1), t_cs)                       # [mg*s/L]
    return round(float(Mp), 1), round(float(ts), 1), round(float(iae), 1), round(float(u.max()), 1)

colores = plt.cm.viridis(np.linspace(0.05, 0.85, 4))
pre = np.arange(-300, 0, 0.5)                                        # 5 min antes, en regimen
tt  = np.r_[pre, t_cs] / 60
pad = lambda x, x0: np.r_[np.full(len(pre), x0), x]

Y_nom, U_nom = respuesta(k_nom)
filas = [('nominal', '1', *metricas(Y_nom, U_nom))]

fig, axs = plt.subplots(3, 2, figsize=(12.5, 8.6), sharex=True, layout='constrained')
lab_nom = r'nominal ($k_2$ = %s, $k_1$ = %s$\times10^{-4}$, $k_0$ = %s$\times10^{-6}$)' % (
    coma(k_nom['k2'], '%.4f'), coma(k_nom['k1'] * 1e4, '%.2f'), coma(k_nom['k0'] * 1e6, '%.2f'))
for i, (clave, sym) in enumerate(param):
    axy, axu = axs[i]
    axy.plot(tt, pad(Y_nom, r0), 'k', lw=2.2, label=lab_nom)
    axu.plot(tt, pad(U_nom, u_ss), 'k', lw=2.2)
    for (f, txt), c in zip(factores, colores):
        k = dict(k_nom); k[clave] = f * k_nom[clave]
        Y, U = respuesta(k)
        filas.append(('$%s$' % sym, txt, *metricas(Y, U)))
        axy.plot(tt, pad(Y, r0), color=c, lw=1.4, ls='--', alpha=0.7, label=r'$%s$ = %s $\times$ nominal' % (sym, txt))
        axu.plot(tt, pad(U, u_ss), color=c, lw=1.3, ls='--', alpha=0.7)
    axy.plot(tt, pad(np.full_like(t_cs, r1), r0), 'k:', lw=1)
    axy.set_ylabel('OD [mg/L]'); axy.set_ylim(1.95, 2.35)
    axu.axhline(100, color='C3', ls=':', lw=1.2, label='límite del soplador'); axu.axhline(0, color='C3', ls=':', lw=1.2)
    axu.set_ylabel('$u$ [%]'); axu.set_ylim(-10, 175)
    axy.set_title('%s · variación de $%s$: salida' % ('abc'[i], sym))
    axu.set_title('%s · variación de $%s$: señal de control' % ('abc'[i], sym))
    axy.legend(fontsize=7.8, frameon=False, loc='lower right')
    axu.legend(fontsize=7.8, frameon=False, loc='upper right')

# M_p y t_s de la nominal, en el primer panel
Mp_n, ts_n = filas[0][2], filas[0][3]
k_pk = np.argmax(Y_nom); y_ts = np.interp(ts_n, t_cs / 60, Y_nom)
ax = axs[0, 0]
ax.plot(t_cs[k_pk] / 60, Y_nom[k_pk], 'o', ms=3.5, color='k', zorder=5)
ax.annotate('$M_p$: %s %%' % coma(Mp_n), xy=(t_cs[k_pk] / 60, Y_nom[k_pk]), xytext=(4, 8),
            textcoords='offset points', ha='left', fontsize=8.5)
ax.plot([ts_n, ts_n], [1.95, y_ts], color='0.75', ls='--', lw=1)
ax.plot(ts_n, y_ts, 'o', ms=3.5, color='k', zorder=5)
ax.annotate('$t_s$: %s min' % coma(ts_n), xy=(ts_n, y_ts), xytext=(0, -14),
            textcoords='offset points', ha='center', fontsize=8.5)

for ax in axs.flat:
    ax.grid(alpha=.3); ax.minorticks_on()
for ax in axs[2]:
    ax.set_xlabel('tiempo desde el escalón [min]')
fig.suptitle('GPI: escalón de 2,0 a 2,2 mg/L variando $k_0$, $k_1$ y $k_2$ (planta sin saturación)',
             fontsize=12.5, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/gpi_variacion_ganancias.pdf', bbox_inches='tight')
plt.show()

tabla_gpi = pd.DataFrame(filas, columns=['ganancia', 'factor', 'Mp [%]', 'ts 2% [min]', 'IAE [mg·s/L]', 'u max [%]'])
display(tabla_gpi)
```

```python
# ============================================================
# Sensibilidad en lazo cerrado respecto a la perturbacion total xi
#   e_y' = u0 + xi   ->   e_y = S(s) xi,   S(s) = s (s + k2) / (s^3 + k2 s^2 + k1 s + k0)
#   xi ya contiene el polo -y/tau de la planta real, asi que S no depende de ella.
#   Unidades: xi en mg/(L s), e_y en mg/L  ->  S en s.   Sin control: e_y = xi/s.
# ============================================================
w_s = np.logspace(-5, 0, 3000)
H_w = lambda sis: ct.frequency_response(sis, w_s).frdata[0, 0]
S_gpi = lambda k2, k1, k0: tf([1, k2, 0], [1, k2, k1, k0])

def bode_S(k):
    H = H_w(S_gpi(**k))
    return 20 * np.log10(np.abs(H)), np.degrees(np.unwrap(np.angle(H)))

print('S(s) nominal =', S_gpi(**k_nom))
I_w = H_w(tf([1], [1, 0]))                                          # sin control: 1/s
I_db, I_fase = 20 * np.log10(np.abs(I_w)), np.degrees(np.angle(I_w))
S_nom_db, S_nom_fase = bode_S(k_nom)
marcas = ((1 / Tau, '$1/\\tau$'), (wc_g, '$\\omega_c$'), (wo_g, '$\\omega_o$'))

fig, axs = plt.subplots(3, 2, figsize=(12.5, 8.6), sharex=True, layout='constrained')
picos = []
for i, (clave, sym) in enumerate(param):
    axm, axf = axs[i]
    axm.semilogx(w_s * 3600, I_db, color='0.6', lw=1.4, ls=':', label='$1/s$  sin control')
    axf.semilogx(w_s * 3600, I_fase, color='0.6', lw=1.4, ls=':')
    axm.semilogx(w_s * 3600, S_nom_db, 'k', lw=2.2, label='nominal')
    axf.semilogx(w_s * 3600, S_nom_fase, 'k', lw=2.2)
    for (f, txt), c in zip(factores, colores):
        k = dict(k_nom); k[clave] = f * k_nom[clave]
        m_db, fase_ = bode_S(k)
        picos.append(('$%s$' % sym, txt, m_db.max(), w_s[m_db.argmax()] * 3600))
        axm.semilogx(w_s * 3600, m_db, color=c, lw=1.4, ls='--', alpha=0.8, label=r'$%s$ = %s $\times$ nominal' % (sym, txt))
        axf.semilogx(w_s * 3600, fase_, color=c, lw=1.3, ls='--', alpha=0.8)
    for x, txt in marcas:
        for ax in (axm, axf):
            ax.axvline(x * 3600, color='gray', ls='--', lw=1)
        axm.text(x * 3600 * 1.08, -38, txt, fontsize=9, color='#555555', va='bottom')
    axm.set_ylim(-40, 60); axm.set_ylabel('$|S_\\xi(j\\omega)|$ [dB]')
    axf.set_ylim(-100, 100); axf.set_ylabel('fase [°]')
    axm.set_title('%s · variación de $%s$: magnitud de $S_\\xi = e_y/\\xi$' % ('abc'[i], sym))
    axf.set_title('%s · variación de $%s$: fase de $S_\\xi = e_y/\\xi$' % ('abc'[i], sym))
    axm.legend(fontsize=7.8, frameon=False, loc='lower right')
for ax in axs.flat:
    ax.grid(visible=True, which='both', alpha=0.3)
for ax in axs[2]:
    ax.set_xlabel('frecuencia [rad/h]')
fig.suptitle('Sensibilidad en lazo cerrado del GPI respecto a $\\xi$, $S_\\xi(s) = s\\,(s+k_2)/(s^3+k_2s^2+k_1s+k_0)$',
             fontsize=12.5, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/gpi_sensibilidad.pdf', bbox_inches='tight')
plt.show()

print('nominal: pico %.1f dB en %.1f rad/h' % (S_nom_db.max(), w_s[S_nom_db.argmax()] * 3600))
display(pd.DataFrame(picos, columns=['ganancia', 'factor', 'pico |S_xi| [dB]', 'en w [rad/h]']).round(1))
```

```python
# ============================================================
# GPI contra ADRC con observador (seccion 2.1), escalon 2,0 -> 2,2 mg/L
#   ADRC: el espacio de estados de la celda del lazo cerrado (Aau_cl, Bau_cl),
#         planta real + ESO + K, con salidas y y u = -K x_hat / b0.
#   GPI : C(s) con las ganancias nominales (k_nom), por feedback().
#   Los dos parten del regimen en 2,0 mg/L (u = 40 %): se simula el incremento.
# ============================================================
C_y_u = np.block([[Cau_cl], [np.zeros((1, n)), -K / bo]])     # salidas [y, u]
adrc_r = ss(Aau_cl, Bau_cl[:, :1], C_y_u, np.zeros((2, 1)))   # entrada r
adrc_e = ss(Aau_cl, Bau_cl[:, 1:], C_y_u, np.zeros((2, 1)))   # entrada eps

# escalon en t = 0 y perturbacion de entrada eps = -15 % a los 40 min, por superposicion
t_ep, eps_p = 2400.0, -15.0
t_cp = np.arange(0, 4800, 0.5)                                # 80 min desde el escalon
k_ep = int(round(t_ep / 0.5))
tarde = lambda h: np.r_[np.zeros(k_ep), step(h, t_cp[:len(t_cp) - k_ep])[0]]
C_n = C_gpi(**k_nom)

Y_adrc = r0 + (r1 - r0) * step(adrc_r[0, 0], t_cp)[0] + eps_p * tarde(adrc_e[0, 0])
U_adrc = u_ss + (r1 - r0) * step(adrc_r[1, 0], t_cp)[0] + eps_p * tarde(adrc_e[1, 0])
Y_gpi = r0 + (r1 - r0) * step(feedback(C_n * Gs_real, 1), t_cp)[0] + eps_p * tarde(feedback(Gs_real, C_n))
U_gpi = u_ss + (r1 - r0) * step(feedback(C_n, Gs_real), t_cp)[0] + eps_p * tarde(-feedback(C_n * Gs_real, 1))

print('C(s) GPI  =', C_n)
print('max |y_ADRC - y_GPI| = %.1e mg/L   ·   max |u_ADRC - u_GPI| = %.1e %%'
      % (np.abs(Y_adrc - Y_gpi).max(), np.abs(U_adrc - U_gpi).max()))
antes = t_cp < t_ep
print('escalon: Mp = %.1f %%, u max = %.1f %%   ·   perturbacion: OD min = %.3f mg/L, u max = %.1f %%, u final = %.1f %%'
      % (100 * (Y_gpi[antes].max() - r1) / (r1 - r0), U_gpi[antes].max(), Y_gpi[~antes].min(), U_gpi[~antes].max(), U_gpi[-1]))

pre_p = np.arange(-300, 0, 0.5)
tt_p  = np.r_[pre_p, t_cp] / 60
pad_p = lambda x, x0: np.r_[np.full(len(pre_p), x0), x]

fig, axs = plt.subplots(1, 2, figsize=(12.5, 4.2), sharex=True, layout='constrained')
ax = axs[0]
ax.plot(tt_p, pad_p(Y_adrc, r0), color='C0', lw=3.5, alpha=0.5, label='ADRC con observador')
ax.plot(tt_p, pad_p(Y_gpi, r0), color='k', lw=1.4, ls='--', label='GPI')
ax.plot(tt_p, pad_p(np.full_like(t_cp, r1), r0), 'k:', lw=1, label='$r$  (consigna)')
ax.set_ylabel('OD [mg/L]'); ax.set_ylim(1.95, 2.3); ax.set_title('a · Salida contra consigna')
ax = axs[1]
ax.plot(tt_p, pad_p(U_adrc, u_ss), color='C4', lw=3.5, alpha=0.5, label='ADRC con observador')
ax.plot(tt_p, pad_p(U_gpi, u_ss), color='k', lw=1.4, ls='--', label='GPI')
ax.axhline(100, color='C3', ls=':', lw=1.2, label='límite del soplador'); ax.axhline(0, color='C3', ls=':', lw=1.2)
ax.set_ylabel('$u$ [%]'); ax.set_ylim(-8, 110); ax.set_title('b · Señal de control')
for ax in axs:
    ax.grid(alpha=.3); ax.minorticks_on(); ax.legend(fontsize=8.5, frameon=False, loc='lower right' if ax is axs[0] else 'upper right')
    ax.set_xlabel('tiempo desde el escalón [min]')
    ax.axvline(t_ep / 60, color='0.5', ls='--', lw=1)
    ax.text(t_ep / 60 + 0.8, ax.get_ylim()[1], r'$\varepsilon = -15$ %', fontsize=9, color='0.35', va='top')
fig.suptitle(r'Escalón de 2,0 a 2,2 mg/L y $\varepsilon = -15$ % a los 40 min: GPI y ADRC con observador (sin saturación)',
             fontsize=12.5, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/gpi_vs_adrc.pdf', bbox_inches='tight')
plt.show()
```

```python
# ============================================================
# GPI: resumen en frecuencia variando k0, k1 y k2 uno a la vez (los otros nominales)
#   Robustez con la planta real sin retardo (mismas funciones que en el ADRC, seccion 2.1):
#   MF, theta_max y Ms de L = C G.   Sensibilidad a xi: S(s) = s (s + k2)/(s^3 + k2 s^2 + k1 s + k0),
#   su pico y la frecuencia del pico, y la asintota S ~ (k2/k0) s a baja frecuencia.
# ============================================================
def _gpi_frecuencia():
    k_d = {'k2': k_nom['k2'], 'k1': k_nom['k1'], 'k0': k_nom['k0']}
    w = np.logspace(-6, 0, 20000); s = 1j * w
    print('%-8s %-6s %8s %10s %8s %16s %18s %16s' % ('ganancia', 'factor', 'MF [°]', 'theta_max', 'Ms', 'pico |S| [dB]', 'en w [rad/h]', 'k2/k0 [h]'))
    filas = [('nominal', '1', dict(k_d))]
    for clave in ('k0', 'k1', 'k2'):
        for f, txt in zip([1 / 4, 1 / 2, 2, 4], ['1/4', '1/2', '2', '4']):
            k = dict(k_d); k[clave] = f * k_d[clave]; filas.append((clave, txt, k))
    for nom, txt, k in filas:
        (mf, th), ms = robustez(k['k2'], k['k1'], k['k0'])
        S = s * (s + k['k2']) / (s**3 + k['k2'] * s**2 + k['k1'] * s + k['k0'])
        i = np.argmax(np.abs(S))
        print('%-8s %-6s %8.1f %10.0f %8.3f %16.1f %18.1f %16.1f' % (nom, txt, mf, th, ms, 20 * np.log10(np.abs(S[i])), w[i] * 3600,
                                                                    k['k2'] / k['k0'] / 3600))

_gpi_frecuencia()

# Lazo abierto C G con la planta nominal y margen de fase, variando k0, k1 y k2 (como la figura del ADRC)
def _gpi_margen():
    w = np.logspace(-2, 4, 6000) / 3600; s = 1j * w
    G = K_dc / (Tau * s + 1)
    def lazo(k2, k1, k0):
        L = (k1 * s + k0) / (bo * s * (s + k2)) * G
        lm, fase = np.log10(np.abs(L)), np.degrees(np.unwrap(np.angle(L)))
        cruces = []
        for i in np.flatnonzero(np.diff(np.sign(lm))):                 # todos los cruces de |L| = 1
            a = -lm[i] / (lm[i + 1] - lm[i])
            wg = 10 ** (np.log10(w[i]) + a * np.log10(w[i + 1] / w[i])); fg = fase[i] + a * (fase[i + 1] - fase[i])
            cruces.append((180 + fg, wg, fg))
        mf, wg, fg = min(cruces)                                       # el menor margen, como en la tabla
        return L, fase, wg, fg, mf
    fig, axs = plt.subplots(2, 3, figsize=(15, 6.8), sharex=True, sharey='row', layout='constrained')
    for j, (clave, sym) in enumerate(param):
        casos = [('nominal', None, dict(k_nom))] + [(txt, c, {**k_nom, clave: f * k_nom[clave]}) for (f, txt), c in zip(factores, colores)]
        for txt, c, k in casos:
            L, fase, wg, fg, mf = lazo(**k)
            nom = c is None
            st = dict(color='k', lw=2.2, zorder=5) if nom else dict(color=c, lw=1.3, ls='--', alpha=0.8)
            lab = ('nominal' if nom else r'$%s$ = %s $\times$ nominal' % (sym, txt)) + ':  MF = %s°' % coma(mf)
            axs[0, j].semilogx(w * 3600, 20 * np.log10(np.abs(L)), label=lab, **st)
            axs[1, j].semilogx(w * 3600, fase, **st)
            axs[1, j].plot([wg * 3600] * 2, [-180, fg], color='k' if nom else c, lw=2.0 if nom else 1.2)
            axs[1, j].plot(wg * 3600, fg, 'o', ms=4, color='k' if nom else c)
        axs[0, j].set_title('%s · Magnitud de $C(s)\\,G(s)$, variación de $%s$' % ('abc'[j], sym))
        axs[1, j].set_title('%s · Fase de $C(s)\\,G(s)$, variación de $%s$' % ('def'[j], sym))
        axs[0, j].legend(fontsize=7.5, frameon=False, loc='upper right')
        axs[0, j].axhline(0, color='k', lw=0.8); axs[1, j].axhline(-180, color='k', lw=0.8)
        axs[1, j].set_xlabel('frecuencia [rad/h]')
    axs[0, 0].set_ylim(-60, 80); axs[0, 0].set_ylabel('magnitud [dB]')
    axs[1, 0].set_ylim(-190, -80); axs[1, 0].set_ylabel('fase [°]')
    for ax in axs.flat:
        ax.grid(which='both', alpha=.3); ax.axvline(wc_g * 3600, color='0.5', ls=':', lw=1)
    fig.suptitle('GPI: lazo abierto y margen de fase variando $k_0$, $k_1$ y $k_2$ (planta sin retardo)',
                 fontsize=12.5, fontweight='bold')
    fig.savefig('Latex/Tesis/figuras/gpi_margen_fase.pdf', bbox_inches='tight')
    plt.show()

_gpi_margen()
```

### 2.2.5 Saturación: ADRC con observador y GPI ante variaciones de la planta

Los dos controladores con el soplador limitado entre 0 y 100 %, cada uno con su anti-windup:

- **ADRC con observador** (2.1.3): al observador entra la orden aplicada, $b_0\,u_{sat}$.
- **GPI** (2.1.6): $v = \dfrac{P}{E_1}(-e_y) - \dfrac{L-E_1}{E_1}\,u_{sat}$, $u_{sat}=\mathrm{sat}(v)$, con $P=(k_1s+k_0)/b_0$, $L=s^2+k_2s$ y $E_1=s^2+k_2s+k_1$, es decir $v = \dfrac{P}{E_1}(-e_y)+\dfrac{k_1}{E_1}\,u_{sat}$.

La primera figura es la misma de la variación de la planta de 2.1.3 (mismas variaciones y colores), ahora con saturación: el ADRC en discontinua y el GPI en discontinua delgada del mismo color (en el panel d, con seis casos muy juntos, el ADRC va como banda continua y el GPI como continua delgada). La segunda es el arranque desde 0 mg/L con la planta nominal, donde el soplador se queda en 100 %; ahí se agrega el GPI **sin** anti-windup ($u_{sat}=\mathrm{sat}(C\,(-e_y))$) para ver el *windup*.

```python
# ============================================================
# Saturacion 0-100 %: GPI con anti-windup ante variaciones de la planta
#   (el ADRC con observador da lo mismo: se simula para la tabla impresa, pero no se grafica)
#   ADRC: observador con entrada [e_y, b0 u_sat]  (anti-windup de Herbst 9.1.1)
#   GPI : v = P/E1 (-e_y) + k1/E1 u_sat,  u_sat = sat(v)   (Handbook 19.1.3)
#         P = (k1 s + k0)/b0,  L = s^2 + k2 s,  E1 = s^2 + k2 s + k1
#   GPI sin anti-windup (solo en el arranque): u_sat = sat(C(s) (-e_y))
#   Realizaciones en forma observable, discretizadas ZOH a Ts_v, como la planta.
# ============================================================
U_MIN, U_MAX = 0.0, 100.0
k2_, k1_, k0_ = k_nom['k2'], k_nom['k1'], k_nom['k0']

def gpi_aw(b0_c, e1=None, e0=None):
    """estado x de E1 = s^2 + e1 s + e0 en forma observable; entradas [-e_y, u_sat], salida v.
    Por defecto E1 = s^2 + k2 s + k1; con E1 = s^2 + l1 s + l2 (observador) es identico al ADRC."""
    e1 = k2_ if e1 is None else e1; e0 = k1_ if e0 is None else e0
    A = np.array([[-e1, 1.0], [-e0, 0.0]])
    B = np.array([[k1_ / b0_c, -(k2_ - e1)], [k0_ / b0_c, e0]])   # rama de u_sat: -(L - E1)/E1
    D = c2d(ss(A, B, [[1.0, 0.0]], [[0.0, 0.0]]), Ts_v)
    return np.asarray(D.A), np.asarray(D.B)

def gpi_sin_aw(b0_c):
    """C(s) = (1/b0)(k1 s + k0)/(s (s + k2)) en forma observable; entrada -e_y"""
    A = np.array([[-k2_, 1.0], [0.0, 0.0]])
    B = np.array([[k1_ / b0_c], [k0_ / b0_c]])
    D = c2d(ss(A, B, [[1.0, 0.0]], [[0.0]]), Ts_v)
    return np.asarray(D.A), np.asarray(D.B)

def sim_sat(ctrl, Kdc=K_dc, tau=Tau, theta=0.0, b0_c=bo, t=t_v, r=r_v):
    P = c2d(ss([[-1 / tau]], [[Kdc / tau]], [[1.0]], [[0.0]]), Ts_v)
    a, b = P.A.item(), P.B.item()
    d = int(round(theta / Ts_v)); buf = np.zeros(d + 1)
    if ctrl == 'gpi': Ac, Bc = gpi_aw(b0_c)
    if ctrl == 'gpi_obs': Ac, Bc = gpi_aw(b0_c, l1_, l2_)
    if ctrl == 'gpi_sin': Ac, Bc = gpi_sin_aw(b0_c)
    y_k = 0.0; x = np.zeros((2, 1))
    Y = np.empty(len(t)); U = np.empty(len(t))
    for k in range(len(t)):
        e_k = y_k - r[k]
        v_k = -(K @ x).item() / b0_c if ctrl == 'adrc' else x[0, 0]
        u_k = min(max(v_k, U_MIN), U_MAX)
        buf = np.roll(buf, 1); buf[0] = u_k
        Y[k], U[k] = y_k, u_k
        y_k = a * y_k + b * buf[d]
        if ctrl == 'adrc':
            x = Ao_v @ x + Bo_v @ np.array([[e_k], [b0_c * u_k]])
        elif ctrl in ('gpi', 'gpi_obs'):
            x = Ac @ x + Bc @ np.array([[-e_k], [u_k]])
        else:
            x = Ac @ x + Bc * (-e_k)
    return Y, U

from matplotlib.lines import Line2D

# --- misma figura que la variacion de la planta de 2.1.3, ahora con saturacion ---
fig, axs = plt.subplots(4, 2, figsize=(12.5, 11), sharex=True, layout='constrained')
ven = t_v >= t_e - 300; tt_v = (t_v[ven] - t_e) / 60
Ya0, Ua0 = sim_sat('adrc'); Yg0, Ug0 = sim_sat('gpi')
estilo_a = dict(lw=1.4, ls='--', alpha=0.6)                 # ADRC: como en 2.1.3
estilo_g = dict(lw=0.8, ls='--')                            # GPI: discontinua delgada, mismo color
filas_d = [(sym, [(dict([(arg, v)]), v, txt) for v, txt in vals], arg) for sym, vals, arg in casos_v]
filas_d.append((r'b_0', [(dict(b0_c=f * bo), f, txt) for f, txt in zip([1 / 6, 1 / 4, 1 / 2, 2, 4, 6],
                                                                       ['1/6', '1/4', '1/2', '2', '4', '6'])], 'b0'))
for sym, vals, arg in filas_d:                              # caso extremo de retardo: el soplador se satura
    if arg == 'theta': vals.append((dict(theta=100), 100, '100 s (extremo)'))
print('nominal: max|y_ADRC - y_GPI| = %.1e mg/L' % np.abs(Ya0 - Yg0)[ven].max())
print('%-6s %-22s %9s %9s %9s %9s %10s' % ('', 'caso', 'ymax ADRC', 'ymax GPI', 'umax ADRC', 'umax GPI', 'max|dy|'))
for i, (sym, vals, arg) in enumerate(filas_d):
    axy, axu = axs[i]
    colores = plt.cm.viridis(np.linspace(0.05, 0.85, len(vals))) if arg != 'b0' else plt.cm.viridis(np.linspace(0.05, 0.9, 6))
    lab_nom = (r'nominal ideal ($K_{dc}$ = 0,05, $\tau$ = 20 min, $\theta$ = 0)' if arg != 'b0'
               else r'$b_0$ nominal ($k_{dc}/\tau$)')
    axy.plot(tt_v, Yg0[ven], 'k', lw=2.6, zorder=5, label=lab_nom); axu.plot(tt_v, Ug0[ven], 'k', lw=2.6, zorder=5)
    est_g = dict(lw=1.4, ls='--', alpha=0.85)                 # solo el GPI con anti-windup
    ymin_k = 2.0
    for (kw, v, txt), c in zip(vals, colores):
        Ya, Ua = sim_sat('adrc', **kw); Yg, Ug = sim_sat('gpi', **kw)
        ymin_k = min(ymin_k, Yg[ven].min())
        if arg == 'b0':
            lab = r'$b_0$ del controlador = %s $\times$ nominal' % txt
        else:
            lab = '$%s$ = %s' % (sym, txt)
            if arg in ('Kdc', 'tau'):
                lab += r'  ($b_{0,\mathrm{planta}}/b_0$ = %s)' % ('%.2f' % ((v / Tau if arg == 'Kdc' else K_dc / v) / bo)).replace('.', ',')
        axy.plot(tt_v, Yg[ven], color=c, label=lab, **est_g); axu.plot(tt_v, Ug[ven], color=c, **est_g)
        print('%-6s %-22s %9.3f %9.3f %9.1f %9.1f %10.1e' % (arg, txt, Ya[ven].max(), Yg[ven].max(),
                                                            Ua[ven].max(), Ug[ven].max(), np.abs(Ya - Yg)[ven].max()))
    axy.plot(tt_v, r_v[ven], 'k:', lw=1); axy.set_ylabel('OD [mg/L]')
    axy.set_ylim(1.95, 2.45 if arg == 'theta' else 2.35)
    if ymin_k < 1.95:                                        # K_dc/4: el soplador al 100 % no sostiene el OD
        axy.annotate('$K_{dc}$/4: el OD cae a %s mg/L, fuera del eje' % ('%.2f' % ymin_k).replace('.', ','),
                     xy=(0.02, 0.95), xycoords='axes fraction', ha='left', va='top', fontsize=8, color='0.35')
    axu.axhline(100, color='C3', ls=':', lw=1); axu.set_ylabel('$u_{sat}$ [%]'); axu.set_ylim(0, 105)
    nom = 'desajuste de $b_0$ del controlador' if arg == 'b0' else 'variación de $%s$' % sym
    axy.set_title('%s · %s: salida' % ('abcd'[i], nom)); axu.set_title('%s · %s: señal de control' % ('abcd'[i], nom))
    axy.legend(fontsize=7.8, frameon=False, loc='lower right')
axs[1, 1].legend(handles=[Line2D([], [], color='C3', ls=':', lw=1)], labels=['límite del soplador'],
                 fontsize=8.5, frameon=False, loc='upper right')

# M_p y t_s de la nominal, como en 2.1.3
k_pk = np.argmax(Yg0[ven]); Mp_nom = 100 * (Yg0[ven][k_pk] - 2.2) / 0.2
fuera = np.flatnonzero(np.abs(Yg0[ven] - 2.2) > 0.02 * 0.2)  # t_s: banda de 2 % del escalon
ts_nom = tt_v[fuera[-1]]; y_ts = np.interp(ts_nom, tt_v, Yg0[ven])
ax_a = axs[0, 0]
ax_a.plot([ts_nom, ts_nom], [1.95, y_ts], color='0.75', ls='--', lw=1)
ax_a.plot([tt_v[0], ts_nom], [y_ts, y_ts], color='0.75', ls='--', lw=1)
ax_a.plot(ts_nom, y_ts, 'o', ms=3.5, color='k', zorder=5)
ax_a.annotate('$t_s$: %s min' % ('%.1f' % ts_nom).replace('.', ','), xy=(ts_nom, y_ts), xytext=(0, 6),
              textcoords='offset points', ha='center', va='bottom', fontsize=8.5)
ax_a.plot(tt_v[k_pk], Yg0[ven][k_pk], 'o', ms=3.5, color='k', zorder=5)
ax_a.annotate('$M_p$: %s %%' % ('%.1f' % Mp_nom).replace('.', ','), xy=(tt_v[k_pk], Yg0[ven][k_pk]), xytext=(0, 6),
              textcoords='offset points', ha='center', fontsize=8.5)

for ax in axs.flat:
    ax.grid(alpha=.3); ax.minorticks_on()
for ax in axs[3]:
    ax.set_xlabel('tiempo desde el escalón [min]')
fig.suptitle('GPI con anti-windup: escalón de 2,0 a 2,2 mg/L con variación de la planta y de $b_0$ (saturación 0-100 %)',
             fontsize=12.5, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/saturacion_adrc_gpi_variacion.pdf', bbox_inches='tight')
plt.show()
```

```python
# ============================================================
# Arranque desde 0 mg/L con la planta nominal y saturacion 0-100 %
#   ADRC con observador, GPI con anti-windup (E1 = s^2 + k2 s + k1 y E1 del
#   observador, s^2 + l1 s + l2) y GPI sin anti-windup
#   (mismas funciones que la celda anterior)
# ============================================================
t_a = np.arange(0, 3600, Ts_v); r_a = np.full_like(t_a, 2.0)
fig, axs = plt.subplots(1, 2, figsize=(12.5, 4.2), sharex=True, layout='constrained')
for ctrl, lab, sty in (('adrc', 'ADRC con observador', dict(color='C0', lw=3.5, alpha=0.5)),
                       ('gpi', 'GPI con anti-windup, $E_1 = s^2 + k_2 s + k_1$', dict(color='k', lw=1.4, ls='--')),
                       ('gpi_obs', 'GPI con anti-windup, $E_1 = s^2 + l_1 s + l_2$', dict(color='C2', lw=1.8, ls=':')),
                       ('gpi_sin', 'GPI sin anti-windup', dict(color='C3', lw=1.6))):
    Y, U = sim_sat(ctrl, t=t_a, r=r_a)
    if ctrl == 'adrc': Y_ad, U_ad = Y, U
    if ctrl == 'gpi_obs': print('E1 del observador: max|y_ADRC - y_GPI| = %.1e mg/L' % np.abs(Y - Y_ad).max())
    axs[0].plot(t_a / 60, Y, label=lab, **sty); axs[1].plot(t_a / 60, U, label=lab, **sty)
    fuera = np.flatnonzero(np.abs(Y - 2.0) > 0.02 * 2.0)
    print('arranque %-46s ymax %.3f mg/L   t_s(2 %%) %.1f min   tiempo en 100 %%: %.0f s'
          % (lab, Y.max(), t_a[fuera[-1] + 1] / 60, Ts_v * np.sum(U >= U_MAX)))
axs[0].plot(t_a / 60, r_a, 'k:', lw=1, label='$r$  (consigna)')
axs[0].set_ylabel('OD [mg/L]'); axs[0].set_title('a · Salida contra consigna')
axs[1].axhline(100, color='C3', ls=':', lw=1.2, label='límite del soplador'); axs[1].axhline(0, color='C3', ls=':', lw=1.2)
axs[1].set_ylabel('$u_{sat}$ [%]'); axs[1].set_ylim(-5, 105); axs[1].set_title('b · Señal de control')
for ax in axs:
    ax.grid(alpha=.3); ax.minorticks_on(); ax.set_xlabel('tiempo [min]')
axs[0].legend(fontsize=8.5, frameon=False, loc='lower right'); axs[1].legend(fontsize=8.5, frameon=False, loc='upper right', bbox_to_anchor=(1, 0.9))
fig.suptitle('Arranque desde 0 mg/L con la planta nominal (saturación 0-100 %)', fontsize=12.5, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/saturacion_arranque.pdf', bbox_inches='tight')
plt.show()
```

**Rechazo de la perturbación con retardo.** Con el sistema en estacionario en 2,0 mg/L entra $\varepsilon = -15$ % de $u$, con la planta nominal y retardo $\theta$ = 0, 20, 40, 50 y 60 s en la entrada, saturación y anti-windup. ADRC en discontinua y GPI en discontinua delgada; la perturbación estimada es la del observador del ADRC, $(\hat\xi + y/\tau)/b_0$.

```python
# ============================================================
# Rechazo de la perturbacion con retardo con el GPI (en la tesis, capitulo del GPI)
#   Se simulan el ADRC y el GPI (con anti-windup); el soplador no llega al limite, asi que el
#   GPI se comporta como el lineal. Se grafica solo el GPI; la tabla impresa compara los dos.
#   OD en regimen en 2,0 mg/L; en t_e entra eps = -15 % de u (como en 2.1.3).
#   Saturacion 0-100 %, planta nominal con retardo theta en la entrada (u + eps).
#   La perturbacion estimada solo existe en el ADRC: (xi_hat + y/tau)/b0 [% de u].
# ============================================================
EPS = -15.0
t_p = np.arange(0, t_e + 3000, Ts_v)
eps_p = np.where(t_p >= t_e, EPS, 0.0)

def sim_pert(ctrl, theta):
    P = c2d(ss([[-1 / Tau]], [[K_dc / Tau]], [[1.0]], [[0.0]]), Ts_v); a, b = P.A.item(), P.B.item()
    d = int(round(theta / Ts_v)); buf = np.zeros(d + 1)
    if ctrl == 'gpi': Ac, Bc = gpi_aw(bo)
    y_k = 0.0; x = np.zeros((2, 1))
    Y = np.empty(len(t_p)); U = np.empty(len(t_p)); XI = np.empty(len(t_p))
    for k in range(len(t_p)):
        e_k = y_k - 2.0
        v_k = -(K @ x).item() / bo if ctrl == 'adrc' else x[0, 0]
        u_k = min(max(v_k, U_MIN), U_MAX)
        Y[k], U[k] = y_k, u_k
        XI[k] = (x[1, 0] + y_k / Tau) / bo if ctrl == 'adrc' else np.nan
        buf = np.roll(buf, 1); buf[0] = u_k + eps_p[k]          # retardo en la entrada de la planta
        y_k = a * y_k + b * buf[d]
        if ctrl == 'adrc':
            x = Ao_v @ x + Bo_v @ np.array([[e_k], [bo * u_k]])
        else:
            x = Ac @ x + Bc @ np.array([[-e_k], [u_k]])
    return Y, U, XI

thetas_p = [(20, '20 s'), (40, '40 s'), (50, '50 s'), (60, '60 s')]
colores = plt.cm.viridis(np.linspace(0.05, 0.85, 4))
ven = t_p >= t_e - 300; tt_p = (t_p[ven] - t_e) / 60
fig, (axy, axu) = plt.subplots(1, 2, figsize=(12.5, 4.4), layout='constrained')

Ya0, Ua0, Xa0 = sim_pert('adrc', 0.0); Yg0, Ug0, _ = sim_pert('gpi', 0.0)
axy.plot(tt_p, Yg0[ven], 'k', lw=2.4, zorder=5, label=r'sin retardo ($\theta$ = 0)')
axu.plot(tt_p, Ug0[ven], 'k', lw=2.4, zorder=5)
print('%-6s %10s %10s %10s %10s %14s' % ('theta', 'OD min A', 'OD min G', 'u max A', 'u max G', 'max|dy| A-G'))
print('%-6s %10.3f %10.3f %10.1f %10.1f %14.1e' % ('0 s', Ya0[ven].min(), Yg0[ven].min(), Ua0[ven].max(), Ug0[ven].max(), np.abs(Ya0 - Yg0)[ven].max()))
for (th, txt), c in zip(thetas_p, colores):
    Ya, Ua, Xa = sim_pert('adrc', th); Yg, Ug, _ = sim_pert('gpi', th)
    axy.plot(tt_p, Yg[ven], color=c, lw=1.5, ls='--', label=r'$\theta$ = %s' % txt)
    axu.plot(tt_p, Ug[ven], color=c, lw=1.5, ls='--')
    print('%-6s %10.3f %10.3f %10.1f %10.1f %14.1e' % (txt, Ya[ven].min(), Yg[ven].min(), Ua[ven].max(), Ug[ven].max(), np.abs(Ya - Yg)[ven].max()))
axy.axhline(2.0, color='k', ls=':', lw=1); axy.set_ylabel('OD [mg/L]'); axy.set_title('a · Salida')
axy.legend(fontsize=8, frameon=False, loc='lower right')
axu.axhline(100, color='C3', ls=':', lw=1, label='límite del soplador'); axu.set_ylabel('$u$ [%]'); axu.set_ylim(0, 105)
axu.set_title('b · Señal de control'); axu.legend(fontsize=8, frameon=False, loc='lower right')
for ax in (axy, axu):
    ax.grid(alpha=.3); ax.minorticks_on(); ax.set_xlabel('tiempo desde la entrada de la perturbación [min]')
fig.suptitle(r'GPI: perturbación $\varepsilon$ = $-15$ % con retardo (planta nominal)', fontsize=12.5, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/gpi_perturbacion_retardo.pdf', bbox_inches='tight')
plt.show()
```

## 3.With Prefiltro LD:

### 3.1 Prefiltro de consigna

El ADRC basado en el error es el de salida **al que se le ha quitado el prefiltro**
(Herbst & Madoński, §8.4, p. 141). Por eso la consigna entra cruda al observador y
su derivada acaba dentro de la perturbación total: con un escalón, `ṙ` es un impulso
y el ESO responde con lo que el libro llama *"giant control efforts"* (Fig. 6.13).

El prefiltro exacto del libro, $C_{PF}(z)$ (Tablas A.6.1 y A.6.2), es discreto: sus
celdas quedan abajo pero no se usan en la versión continua. Aquí se usa el prefiltro
simple $F(s)$ de la sección 3.2.

```python
# ============================================================
# NO SE USA EN LA VERSION CONTINUA.
#   C_PF(z) (Herbst & Madonski, Tablas A.6.1 y A.6.2) es el prefiltro del ADRC
#   DISCRETO: sus coeficientes mezclan T_s con las ganancias discretas. Su
#   equivalente continuo, C_PF(s) = (1 + g1 s + g2 s^2)/(1 + b1 s) (Tabla A.2.1),
#   no es realizable. Aqui solo se usa el prefiltro simple F(s) de la seccion 3.2.
# ============================================================
# # ============================================================
# # Prefiltro de consigna C_PF(z)   ·   Herbst & Madonski, Tablas A.6.1 y A.6.2
# #   Las matrices del lazo (Aau_e, Bau_e, Cau_e) NO se tocan.
# # ============================================================

# k1 = K_1[0, 0]                 # ganancia del controlador, de la celda de diseno
# l1 = L[0, 0]                   # ganancias del observador
# l2 = L[1, 0]

# # Tabla A.6.2, columna "General terms".  b0 se cancela en C_PF, asi que se omite.
# beta_0  = k1 * l1 + l2
# beta_1  = T_s * k1 * l2 - k1 * l1 - l2
# gamma_0 = k1
# gamma_1 = k1 * (T_s * l2 + l1 - 2)
# gamma_2 = k1 * (1 - l1)

# print('k1 = %.6e     l1 = %.6e     l2 = %.6e' % (k1, l1, l2))
# print('beta_0  = %+.6e   beta_1  = %+.6e' % (beta_0, beta_1))
# print('gamma_0 = %+.6e   gamma_1 = %+.6e   gamma_2 = %+.6e' % (gamma_0, gamma_1, gamma_2))

# # Tabla A.6.1, primer orden
# C_PF = tf([gamma_0 / beta_0, gamma_1 / beta_0, gamma_2 / beta_0],
#           [1.0,              beta_1  / beta_0, 0.0], T_s)

# display(Latex(r"Reference signal prefilter $C_{PF}(z)$:"))
# display(C_PF)
# print('ganancia DC = %.6f   (tiene que ser 1: el prefiltro no cambia el regimen)'
#       % float(dcgain(C_PF)))
```

### 3.2 Closed-Loop System for simulation — con prefiltro simple de elección propia

El libro ofrece en la Fig. 6.16 un prefiltro **de elección propia** en el camino de
la consigna.

> *"a custom prefilter `F(s)` could be optionally attached in the reference signal
> path… to help with the transient tracking performance"* (§6.4, receta)

> *"Using customized prefilters for error-based ADRC enables **true 2DOF design
> possibilities**, whereas the output-based ADRC design is fixed at a critically
> damped response"* (Fig. 6.17)

Para `N = 1` el más simple es de primer orden, con **un solo parámetro**:

$$F(s)=\frac{1}{T_{ref}\,s+1}$$

Igual que antes, **las matrices del lazo no cambian**: solo se filtra `r` antes de
entrar.

### Cómo se elige `T_ref`

`T_ref` es literalmente *"cuánto quiero que tarde la consigna en llegar"*. Y se
traduce directo a esfuerzo de control, porque lo que la consigna le pide al
actuador es proporcional a $\dot r = \Delta r/T_{ref}$. La celda de abajo lo barre.

El punto interesante está en `T_ref = 600 s`: da **el mismo pico de control que el
`C_PF` exacto** (75,1 %) siendo casi el doble de rápido, y sin el pico de 72 mg/L.
Pasado ese punto, subir `T_ref` ya no baja `u` — porque ese 75,1 % lo fija el
escalón de perturbación de `t` = 7000 s, no la consigna.


```python
# ============================================================
# Prefiltro simple de eleccion propia   ·   Herbst & Madonski, Fig. 6.16
#   F(s) = 1/(T_ref s + 1).  Un parametro.
#   Las matrices del lazo (Aau_e, Bau_e, Cau_e) siguen sin tocarse.
# ============================================================

T_ref = 1065.0                               # [s] cuanto quiero que tarde la consigna

F_ref = tf([1.0], [T_ref, 1.0])
display(Latex(r"Prefiltro de eleccion propia $F(s)$:"))
display(F_ref)
print('ganancia DC = %.6f' % float(dcgain(F_ref)))

t  = np.arange(0, 10000, T_s)
r  = np.where(t >= 4000, 3.0, 2.0)          # consigna que teclea el operario [mg/L]
xi = np.where(t >= 7000, -15.0, 0.0)        # perturbacion [% de u] en t = 7000 s

# --- barrido de T_ref, para elegir con numeros ---
def _corre(rr):
    _, _, Xq = lsim(sys_au_cl, U=np.column_stack([rr, xi]), T=t, X0=np.zeros(n + n + m))
    yq = (Cp_r @ Xq[:, :n].T).ravel()
    uq = -(K @ Xq[:, n:].T).ravel() / bo
    return yq, uq

print('\n  filtro                    u pico     t63 de y    pico de r filtrada')
yq, uq = _corre(r)
print('  sin filtro              %8.1f %% %10.1f s %14.2f mg/L'
      % (uq.max(), t[np.argmax(yq >= 0.632 * r[0])], r.max()))
for Tq in [150.0, 300.0, 600.0, 1063.0, 2000.0]:
    rq = np.asarray(lsim(tf([1.0], [Tq, 1.0]), U=r, T=t)[0]).ravel()
    yq, uq = _corre(rq)
    marca = '   <- T_ref elegido' if Tq == T_ref else ''
    print('  F(s)=1/(%6.0f s+1)     %8.1f %% %10.1f s %14.2f mg/L%s'
          % (Tq, uq.max(), t[np.argmax(yq >= 0.632 * r[0])], rq.max(), marca))

# ============================================================
# Simulacion con el T_ref elegido
# ============================================================
r_f  = np.asarray(lsim(F_ref, U=r, T=t)[0]).ravel()      # <-- el prefiltro
x0   = np.zeros(n + n + m)

[Y, T, X] = lsim(sys_au_cl, U=np.column_stack([r_f, xi]), T=t, X0=x0)

y_out   = (Cp_r @ X[:, :n].T).ravel()
xhat_au = X[:, n:].T
e_est   = xhat_au[0, :]
xi_hat  = xhat_au[1, :]
e_real  = y_out - r                         # error contra la consigna DEL OPERARIO
e_obs   = y_out - r_f                       # error que el observador SI ve
u_ctrl  = -(K @ xhat_au).ravel() / bo

xi_fisica = (xi_hat + y_out / Tau) / bo     # -> % de u, comparable con la xi aplicada

w_c    = -pols[0]
k_pert = int(np.flatnonzero(xi != 0)[0])
nuevo  = np.r_[True, np.diff(r) != 0]
tramo  = np.cumsum(nuevo) - 1
k0     = np.flatnonzero(nuevo)[tramo]
y_teo  = r + (y_out[k0] - r) * np.exp(-w_c * (t - t[k0]))
y_teo[k_pert:] = np.nan

print('\nCON F(s) = 1/(%.0f s + 1):' % T_ref)
print('   t63 del primer escalon = %.1f s   (tau_c = %.1f s)'
      % (t[np.argmax(y_out >= 0.632 * r[0])], 1 / w_c))
print('   u pico = %.1f %%   minimo = %.1f %%' % (u_ctrl.max(), u_ctrl.min()))
print('   max|e_obs - e_est| = %.2e mg/L   (sin picos: F no amplifica en alta frecuencia)'
      % np.abs(e_obs - e_est)[20:].max())
print('   perturbacion: estimada %.2f %%   ·   aplicada %.1f %%' % (xi_fisica[-1], xi[-1]))
```

```python
# para comparar en la figura: las dos versiones
r_sin  = r
u_sin  = _corre(r_sin)[1]

fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.5), layout='constrained')

# a) salida contra consigna cruda y filtrada
ax = axs[0, 0]
ax.plot(T, y_out, color='C0', linewidth=2.4, label='$y$  (OD medido)')
ax.plot(T, r, color='k', linewidth=1.4, linestyle='--', label='$r$  (consigna del operario)')
ax.plot(T, r_f, color='C2', linewidth=1.6, label='$F(s)\\,r$  (lo que ve el lazo)')
ax.axvline(T[k_pert], color='C3', linewidth=1, alpha=.5)
ax.set_title('a · Salida, consigna cruda y consigna filtrada')
ax.set_ylabel('OD [mg/L]'); ax.set_ylim(0, 3.4)

# b) el error: el que ve el observador contra el que estima
ax = axs[0, 1]
ax.plot(T, e_obs, color='C0', linewidth=3, label='$e_y$ que ve el observador $= y - F r$')
ax.plot(T, e_est, color='C1', linewidth=1.6, linestyle='--', label='$\\hat e_y$  ESTIMADO')
ax.plot(T, e_real, color='C7', linewidth=1.1, alpha=.8, label='$y - r$  (contra la cruda)')
ax.set_title('b · El error: el que ve el observador contra el que estima')
ax.set_ylabel('error [mg/L]')

# c) la perturbación física
ax = axs[1, 0]
ax.plot(T, xi_fisica, color='C2', linewidth=2,
        label='$(\\hat\\xi + y/\\tau)/b_0$  estimada')
ax.plot(T, xi, color='k', linewidth=1.6, linestyle='--', label='$\\xi$ APLICADA')
ax.set_title('c · La perturbación física: estimada contra aplicada')
ax.set_ylabel('[% de $u$]')

# d) las tres señales de control
ax = axs[1, 1]
ax.plot(T, u_sin, color='C3', linewidth=1.1, alpha=.6, label='sin filtro (%.0f %%)' % u_sin.max())
ax.plot(T, u_ctrl, color='C4', linewidth=2.2, label='$F(s)=1/(%.0f s+1)$  (%.0f %%)' % (T_ref, u_ctrl.max()))
ax.axhline(100, color='C3', linestyle=':', linewidth=1.2, label='límite del soplador')
ax.axhline(0, color='C3', linestyle=':', linewidth=1.2)
ax.set_title('d · Señal de control: los dos casos')
ax.set_ylabel('$u$ [%]'); ax.set_ylim(-20, 200)

for ax in axs.flat:
    ax.minorticks_on(); ax.legend(fontsize=8.5, frameon=False, loc='best')
    ax.grid(visible=True, which='major', alpha=0.35)
    ax.set_xlabel('tiempo [s]')

fig.suptitle('Lazo cerrado con prefiltro simple $F(s)=1/(T_{ref}s+1)$  ·  Fig. 6.16',
             fontsize=13, fontweight='bold')
plt.show()
```

### 3.3 Referencia con polinomio de Bézier cúbico

El mismo lazo de 2.1.3 con saturación 0-100 % y anti-windup ($b_0\,u_{sat}$ al observador), y el mismo escenario (arranque, cambio de consigna de 2,0 a 2,2 mg/L en $t$ = 4000 s y $\varepsilon = -15$ % en $t$ = 7000 s), pero la consigna pasa por el polinomio cúbico de Bézier (sección 5.2 de la tesis):

$$y^*(t) = y_0 + \Delta\,\tau^2(3-2\tau),\qquad \dot y^*(t) = \frac{\Delta}{t_f}\,6\,\tau(1-\tau),\qquad \tau = \frac{t-t_0}{t_f}\in[0,1].$$

Duración: $t_f = \max(20\ \text{min},\ 24\ \text{min por mg/L}\cdot|\Delta|)$, para que la pendiente máxima $1{,}5\,\Delta/t_f$ no pase de 3,75 mg/(L·h), lo que le sobra al soplador después de compensar la perturbación de diseño. El arranque parte del OD medido, $y_0 = 0$.

Como el ADRC está basado en el error, $\dot y^*$ entra en la perturbación total: $\hat\xi = \bar\xi - \dot y^*$. Para comparar con la $\varepsilon$ aplicada se grafica $(\hat\xi + y/\tau + \dot y^*)/b_0$.

```python
# ============================================================
# ADRC con saturacion 0-100 % y referencia de Bezier cubica
#   Mismo lazo que 2.1.3 (planta real, observador con b0*u_sat), mismo escenario;
#   la consigna r(t) en escalones pasa por y*(t) = y0 + D tau^2 (3 - 2 tau).
#   t_f = max(20 min, 24 min por mg/L): pendiente maxima 1,5 D/t_f <= 3,75 mg/(L h)
# ============================================================
U_MIN, U_MAX = 0.0, 100.0
t_b  = np.arange(0, 10000, T_s)
xi_b = np.where(t_b >= 7000, -15.0, 0.0)                 # perturbacion [% de u] en t = 7000 s
MAN_B = [(0.0, 0.0, 2.0), (4000.0, 2.0, 2.2)]            # (inicio [s], y0, yf): arranque y cambio de consigna
tf_b  = lambda D: 60.0 * max(20.0, 24.0 * abs(D))        # duracion de la maniobra [s]

def bezier3(t, maniobras):
    """y*, dy*/dt y la consigna en escalones; las maniobras no se solapan (sin replanificacion)."""
    ys = np.full(len(t), maniobras[0][1]); vs = np.zeros(len(t)); r_esc = ys.copy()
    for t0, y0, yf in maniobras:
        tf = tf_b(yf - y0); tau = np.clip((t - t0) / tf, 0, 1); en = t >= t0
        ys[en] = (y0 + (yf - y0) * tau**2 * (3 - 2 * tau))[en]
        vs[en] = ((yf - y0) / tf * 6 * tau * (1 - tau))[en]
        r_esc[en] = yf
    return ys, vs, r_esc

y_ref, v_ref, r_esc = bezier3(t_b, MAN_B)
print('maniobras: ' + '   '.join('%.1f -> %.1f mg/L en t_f = %.0f min' % (y0, yf, tf_b(yf - y0) / 60) for _, y0, yf in MAN_B))

P_b = c2d(ss(Ap_r, Bp_r, Cp_r, 0), T_s)
O_b = c2d(ss(Aau - L @ Cau, np.block([L, Bau]), np.eye(n + m), np.zeros((n + m, 2))), T_s)
Ap_b, Bp_b = np.asarray(P_b.A), np.asarray(P_b.B)
Ao_b, Bo_b = np.asarray(O_b.A), np.asarray(O_b.B)

N_b  = len(t_b)
xp   = np.zeros((n, 1)); xh = np.zeros((n + m, 1))
y_b  = np.empty(N_b); u_b = np.empty(N_b); up_b = np.empty(N_b); eh_b = np.empty(N_b); xih_b = np.empty(N_b)
for k in range(N_b):
    y_k  = (Cp_r @ xp).item()
    e_k  = y_k - y_ref[k]                                 # el error contra la referencia de Bezier
    u_k  = -(K @ xh).item() / bo
    us_k = min(max(u_k, U_MIN), U_MAX)
    y_b[k], u_b[k], up_b[k], eh_b[k], xih_b[k] = y_k, us_k, u_k, xh[0, 0], xh[n, 0]
    xp = Ap_b @ xp + Bp_b * (us_k + xi_b[k])
    xh = Ao_b @ xh + Bo_b @ np.array([[e_k], [bo * us_k]])

e_b      = y_b - y_ref
xi_fis_b = (xih_b + y_b / Tau + v_ref) / bo              # sin la fuga de la planta ni dy*/dt, en % de u
k_pert   = int(np.flatnonzero(xi_b != 0)[0])
print('max |e_y| = %.3f mg/L   ·   OD maximo = %.3f mg/L   ·   error final = %.1e mg/L'
      % (np.abs(e_b).max(), y_b.max(), abs(e_b[-1])))
print('u pedida entre %.0f y %.0f %%  ·  saturada el %.2f %% del tiempo' % (up_b.min(), up_b.max(),
      100 * np.mean((up_b < U_MIN) | (up_b > U_MAX))))
print('perturbacion: estimada %.2f %%   ·   aplicada %.1f %%' % (xi_fis_b[-1], xi_b[-1]))

fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.5), layout='constrained')
ax = axs[0, 0]
ax.plot(t_b, y_b, color='C0', linewidth=2.4, label='$y$  (OD medido)')
ax.plot(t_b, y_ref, color='k', linewidth=1.4, linestyle='--', label='$y^*$  (Bézier)')
ax.plot(t_b, r_esc, color='0.6', linewidth=1.0, linestyle=':', label='$r$  (consigna)')
ax.axvline(t_b[k_pert], color='C3', linewidth=1, alpha=.5)
ax.annotate('entra $\\varepsilon$', xy=(t_b[k_pert], 0.15), xytext=(6, 0),
            textcoords='offset points', fontsize=8.5, color='C3')
ax.set_title('a · Salida contra consigna'); ax.set_ylabel('OD [mg/L]')

ax = axs[0, 1]
ax.plot(t_b, e_b, color='C0', linewidth=3, label='$e_y$ REAL $= y - y^*$')
ax.plot(t_b, eh_b, color='C1', linewidth=1.6, linestyle='--', label='$\\hat e_y$  ESTIMADO (estado del observador)')
ax.set_title('b · El error: real contra estimado'); ax.set_ylabel('error [mg/L]')

ax = axs[1, 0]
ax.plot(t_b, xi_fis_b, color='C2', linewidth=2, label='$\\hat\\xi$  perturbación total')
ax.plot(t_b, xi_b, color='k', linewidth=1.5, linestyle='--', label='$\\varepsilon$  perturbación exógena')
ax.set_title('c · Perturbación exógena y perturbación total'); ax.set_ylabel('[% de $u$]')

ax = axs[1, 1]
ax.plot(t_b, up_b, color='C7', linewidth=1.1, linestyle=':', label='$u = u_0/b_0$  (pedida)')
ax.plot(t_b, u_b, color='C4', linewidth=2, label='$u_{sat}$  (al soplador)')
ax.axhline(100, color='C3', linestyle=':', linewidth=1.2, label='límite del soplador')
ax.axhline(0, color='C3', linestyle=':', linewidth=1.2)
ax.set_title('d · Señal de control'); ax.set_ylabel('$u$ [%]'); ax.set_ylim(-8, 110)

for ax in axs.flat:
    ax.minorticks_on(); ax.legend(fontsize=8.5, frameon=False, loc='best')
    ax.grid(visible=True, which='major', alpha=0.35); ax.set_xlabel('tiempo [s]'); eje_minutos(ax)
fig.suptitle('Lazo cerrado con saturación del actuador y referencia de Bézier', fontsize=13, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/adrc_lazo_cerrado_bezier.pdf', bbox_inches='tight')
plt.show()
```

**Escalón contra Bézier con retardo.** El cambio de 2,0 a 2,2 mg/L con la planta nominal y retardo $\theta$ = 20, 40, 50 y 60 s en la entrada (las mismas variaciones de 2.1.3), con saturación y anti-windup: (a) con la consigna en escalón y (b) con la referencia de Bézier ($t_f$ = 20 min).

```python
# ============================================================
# Escalon contra Bezier con retardo: cambio de 2,0 a 2,2 mg/L
#   ADRC con saturacion 0-100 % y anti-windup (b0 u_sat al observador), planta
#   nominal con retardo theta en la entrada.  Mismas variaciones de theta que en
#   2.1.3 y misma referencia de Bezier que la celda anterior (t_f = 20 min).
# ============================================================
t_r  = np.arange(0, t_e + 3000, Ts_v)                       # escalon en t_e = 20000 s, ya en regimen
tf_r = tf_b(0.2)
tau_r = np.clip((t_r - t_e) / tf_r, 0, 1)
refs = {'escalón': np.where(t_r >= t_e, 2.2, 2.0),
        'Bézier':  np.where(t_r >= t_e, 2.0 + 0.2 * tau_r**2 * (3 - 2 * tau_r), 2.0)}

def sim_ret(r, theta):
    P = c2d(ss([[-1 / Tau]], [[K_dc / Tau]], [[1.0]], [[0.0]]), Ts_v); a, b = P.A.item(), P.B.item()
    d = int(round(theta / Ts_v)); buf = np.zeros(d + 1)
    y_k = 0.0; xh = np.zeros((n + m, 1)); Y = np.empty(len(t_r)); U = np.empty(len(t_r))
    for k in range(len(t_r)):
        e_k = y_k - r[k]
        u_k = min(max(-(K @ xh).item() / bo, U_MIN), U_MAX)
        buf = np.roll(buf, 1); buf[0] = u_k
        Y[k], U[k] = y_k, u_k
        y_k = a * y_k + b * buf[d]
        xh  = Ao_v @ xh + Bo_v @ np.array([[e_k], [bo * u_k]])
    return Y, U

thetas = [(20, '20 s'), (40, '40 s'), (50, '50 s'), (60, '60 s')]
colores = plt.cm.viridis(np.linspace(0.05, 0.85, 4))
ven = t_r >= t_e - 300; tt_r = (t_r[ven] - t_e) / 60
fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.5), sharex=True, layout='constrained')
print('%-8s %-6s %10s %12s %9s' % ('', 'theta', 'OD max', 'sobrepaso', 'u max'))
for i, (nom, r) in enumerate(refs.items()):
    axy, axu = axs[i]
    Y0, U0 = sim_ret(r, 0.0)
    z0, zc, al = (5, 2, 0.6) if nom == 'escalón' else (1, 3, 0.9)   # en (b) las curvas coinciden: la negra va debajo
    axy.plot(tt_r, Y0[ven], 'k', lw=2.6, zorder=z0, label=r'sin retardo ($\theta$ = 0)'); axu.plot(tt_r, U0[ven], 'k', lw=2.6, zorder=z0)
    for (th, txt), c in zip(thetas, colores):
        Y, U = sim_ret(r, th)
        axy.plot(tt_r, Y[ven], color=c, lw=1.4, ls='--', alpha=al, zorder=zc, label=r'$\theta$ = %s' % txt)
        axu.plot(tt_r, U[ven], color=c, lw=1.3, ls='--', alpha=al, zorder=zc)
        print('%-8s %-6s %10.3f %12.3f %9.1f' % (nom, txt, Y[ven].max(), max(0, Y[ven].max() - 2.2), U[ven].max()))
    axy.plot(tt_r, r[ven], 'k:', lw=1); axy.set_ylabel('OD [mg/L]'); axy.set_ylim(1.95, 2.35)
    axu.axhline(100, color='C3', ls=':', lw=1); axu.set_ylabel('$u_{sat}$ [%]'); axu.set_ylim(0, 105)
    axy.set_title('%s · referencia en %s: salida' % ('ab'[i], nom)); axu.set_title('%s · referencia en %s: señal de control' % ('ab'[i], nom))
    axy.legend(fontsize=7.8, frameon=False, loc='lower right')
for ax in axs.flat:
    ax.grid(alpha=.3); ax.minorticks_on()
for ax in axs[1]:
    ax.set_xlabel('tiempo desde el cambio de consigna [min]')
fig.suptitle('Cambio de 2,0 a 2,2 mg/L con retardo: escalón contra Bézier (saturación 0-100 %)', fontsize=12.5, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/adrc_bezier_retardo.pdf', bbox_inches='tight')
plt.show()
```

**Embrague que quita el sobrepico** (Quecan-Herrera, Garrido y Sira-Ramírez, CoDIT 2026)

Durante el transitorio, la parte de la orden que reacciona al cambio de consigna se multiplica por $F_s(t)$: $u = u_c + F_s(t)\,(\bar u - u_c)$, con $F_s = p_i + (1-p_i)(3\tau^2 - 2\tau^3)$, $\tau = t/t_2$, hasta $t_2 = 1/\omega_c$. El embrague va antes de la saturación y al observador le entra $b_0\,u_{sat}$. Con $p_i^\star = \omega_o/(\omega_o + 2\omega_c) = 5/7$ el polo del controlador cae sobre el cero del lazo cerrado. Escalón de 2,0 a 2,2 mg/L con la planta real, sin retardo y con $\theta$ = 50 s.

```python
# ============================================================
# Embrague p_i* = w_o/(w_o + 2 w_c), t2 = 1/w_c, escalando solo el cambio de la orden
#   u = u_c + Fs(t) (u_bar - u_c),  antes de la saturacion; al observador le entra b0 u_sat
# ============================================================
def _embrague():
    from scipy.linalg import expm
    wc_e, wo_e = 4 / Tau, 20 / Tau
    l1_e, l2_e = 2 * wo_e, wo_e**2; k1_e = wc_e * l1_e + l2_e
    pi_e, t2_e = l2_e / k1_e, 1 / wc_e
    dt_e = 0.5
    A_ = np.array([[0, 1], [0, 0.]]); B_ = np.array([[1], [0.]]); C_ = np.array([[1, 0.]]); L_ = np.array([[l1_e], [l2_e]])
    M_ = expm(np.block([[A_ - L_ @ C_, np.hstack([L_, B_])], [np.zeros((2, 4))]]) * dt_e)
    Ad_, Bd_ = M_[:2, :2], M_[:2, 2:]
    ag_, bg_ = np.exp(-dt_e / Tau), K_dc * (1 - np.exp(-dt_e / Tau))
    t_ = np.arange(-300, 3000, dt_e)
    def sim(emb, theta):
        d = int(round(theta / dt_e)); buf = [2.0 / K_dc] * (d + 1)
        y = 2.0; x = np.array([0.0, -2.0 / Tau]); uc = 2.0 / K_dc; Y, U = [], []
        for tk in t_:
            r_ = 2.2 if tk >= 0 else 2.0
            tau = tk / t2_e
            F = 1.0 if (not emb or tk < 0 or tau >= 1) else pi_e + (1 - pi_e) * (3 * tau**2 - 2 * tau**3)
            ub = -(wc_e * x[0] + x[1]) / bo
            u = min(max(uc + F * (ub - uc), 0.0), 100.0)
            Y.append(y); U.append(u); buf.append(u); ud = buf.pop(0)
            y = ag_ * y + bg_ * ud; x = Ad_ @ x + Bd_ @ np.array([Y[-1] - r_, bo * u])
        Y, U = np.array(Y), np.array(U); m = t_ >= 0; f = np.flatnonzero(np.abs(Y[m] - 2.2) > 0.02 * 0.2)
        return Y, U, 100 * max(Y.max() - 2.2, 0) / 0.2, t_[m][f[-1]] / 60, np.trapezoid(np.abs(Y[m] - 2.2), t_[m]), U.max()
    print('p_i* = %.4f   t2 = %.0f s' % (pi_e, t2_e))
    fig, axs = plt.subplots(2, 2, figsize=(12.5, 7), sharex=True, layout='constrained')
    for j, th in enumerate((0.0, 50.0)):
        for emb, st, lab in ((False, dict(color='C3', lw=1.4, ls='--'), 'sin embrague'),
                             (True, dict(color='k', lw=1.8), 'con embrague ($p_i^\\star$ = 5/7, $t_2 = 1/\\omega_c$)')):
            Y, U, Mp, tss, iae, um = sim(emb, th)
            print('theta %2.0f s  %-13s Mp %5.1f %%  ts %5.1f min  IAE %5.1f mg s/L  u max %5.1f %%'
                  % (th, 'con embrague' if emb else 'sin embrague', Mp, tss, iae, um))
            axs[0, j].plot(t_ / 60, Y, label=lab, **st); axs[1, j].plot(t_ / 60, U, **st)
        axs[0, j].plot([-5, 0, 0, 50], [2.0, 2.0, 2.2, 2.2], 'k:', lw=1)
        axs[0, j].set_title('%s · Salida, $\\theta$ = %.0f s' % ('ab'[j], th))
        axs[1, j].set_title('%s · Señal de control, $\\theta$ = %.0f s' % ('cd'[j], th))
        axs[1, j].set_xlabel('tiempo desde el cambio de consigna [min]')
        axs[0, j].set_ylim(1.98, 2.32); axs[1, j].set_ylim(35, 100)
    for ax in axs.flat:
        ax.grid(alpha=.3); ax.set_xlim(-5, 50)
    axs[0, 0].set_ylabel('OD [mg/L]'); axs[1, 0].set_ylabel('$u$ [%]'); axs[0, 0].legend(fontsize=9, frameon=False, loc='lower right')
    fig.suptitle('Escalón de 2,0 a 2,2 mg/L con y sin embrague (planta real, saturación 0-100 %)', fontsize=12.5, fontweight='bold')
    fig.savefig('Latex/Tesis/figuras/adrc_embrague.pdf', bbox_inches='tight'); plt.show()

_embrague()
```

# 4. El sistema diseñado contra la perturbación REAL de la planta
## 4.1 Observer ESO

Hasta aquí `ξ` ha sido un escalón de laboratorio. Esta sección usa la perturbación
medida en Tunja, exportada de `Datos/analisis/` con `exporta_perturbacion.py`
(que **no** modifica `analisis_dinamica.ipynb`: importa sus mismos módulos y
reproduce el camino 2 de su §14.2b).

**No se rediseña nada.** Entran tal cual el `sys_au_cl` de arriba —las mismas
`Aau_e`, `Bau_e`, `Cau_e`— y el prefiltro `F_ref` de la sección anterior. Para eso
se diseñaron.

### Lo único que hay que resolver: cómo entra `f`

La columna de `ξ` en `Bau_e` espera la perturbación **en % de `u`**, entrando por
`Bp_r`. La `f` del análisis no está en esas unidades ni mide lo mismo, y hacen
falta dos correcciones. Saltarse cualquiera de las dos hace que la simulación no
sirva.

### 4.1.1 · Unidades

El análisis de dinámica trabaja **por hora**; este cuaderno, **por segundo**:

| magnitud | en `Datos/analisis/` | aquí | conversión |
|---|---|---|---|
| `f` | mg/(L·h) | mg/(L·s) | ÷ 3600 |
| `b₀` | mg/L/(%·h) | mg/L/(%·s) | ÷ 3600 |
| tiempo | horas | segundos | × 3600 |
| muestreo de `f` | 1 min | — | se interpola |

Metida cruda en mg/(L·h), la perturbación queda unas **270 veces más pequeña** de
lo que es: el lazo no es que no la rechace, es que no hay nada que rechazar y la
gráfica sale plana.

### 4.1.2 · El polo de la planta, que si no se cuenta dos veces

`f` está definida por la forma que usa el ADRC,

$$\dot y = f + b_0\,u$$

así que **ya lleva dentro el término $-y/\tau$**. Y la planta que hay dentro de
`sys_au_e` es $K_{dc}/(\tau s+1)$, que también lo tiene. Hay que quitárselo antes
de inyectarla:

$$f=-\frac{y}{\tau}+b_0\,\xi
\qquad\Longrightarrow\qquad
\boxed{\;\xi=\frac{f+y/\tau}{b_0}\;}$$
Se usa $y\approx r$ porque el lazo regula. Sin esta corrección el resultado del OD
sale igual, pero `u` queda desplazada unos **40 puntos**, que es la diferencia
entre leer 36–116 % y leer 76–156 %.
### 4.1.3 Y al leer `ξ̂`, la vuelta
El observador trabaja sobre el integrador ideal, así que su `ξ̂` absorbe **también**
el polo de la planta. Para compararla con la aplicada hay que devolvérselo:
$$\hat\xi_{fis}=\frac{\hat\xi}{b_0}+\frac{y}{\tau\,b_0}$$
## 4.1.4 Dos decisiones del montaje
**La ventana es de 25 h**, no de 7 días. A `T_s` = 0,05 s —el del cuaderno, para
poder reusar `sys_au_cl` sin rediscretizar— una semana son 12 millones de muestras.
25 h son 1,8 millones, un ciclo diario completo, y corren en 20 s. Es además la
misma ventana y la misma consigna que usa el GPI en 2.2.4, para que la sección 5
pueda comparar los dos en la misma gráfica.
**La planta usa el `b₀` de diseño** (0,150 mg/L/(%·h)), no el medido (0,1338). Es
el caso optimista: un **12 % de desajuste** que aquí no se modela. Está dentro de
la incertidumbre conocida de `b₀`, y conviene declararlo en el documento.

```python
# ============================================================
# Lazo cerrado con la f medida en Tunja
#
#   REUSA el sistema que ya disenamos:  sys_au_cl  (las matrices de arriba, sin
#   tocar una sola) y  F_ref, el prefiltro simple de la seccion anterior.
#
#   Lo unico que hace falta resolver es como convertir la f medida a lo que la
#   columna de xi de Bau_e espera, que es "% de u". Son DOS correcciones:
#
#     1. UNIDADES.  f viene en mg/(L*h) y la columna espera % de u.  Metida
#        cruda queda ~270 veces mas pequena de lo que es: parece que el lazo
#        no rechaza nada cuando en realidad no hay perturbacion que rechazar.
#
#     2. EL POLO DE LA PLANTA.  f esta definida por  y' = f + b0*u,  asi que
#        YA lleva dentro el termino -y/Tau.  Y la planta de sys_au_cl tambien lo
#        tiene.  Si no se le quita, se cuenta dos veces y u sale desplazada
#        unos 40 puntos.
#
#            f = -y/Tau + b0 * xi        ->        xi = (f + y/Tau) / b0
#
#        Se usa y ~ r porque el lazo regula (el residuo queda por debajo del
#        0,2 % de sigma de f).
# ============================================================


DIR_AN = 'Datos/analisis/'
MOD    = 'M2'
HORAS  = 25                                     # un ciclo diario completo, el mismo que el GPI

NIV       = [2.0, 2.5, 1.5]                       # mg/L
HORAS_NIV = 4

par  = pd.read_csv(DIR_AN + 'f_parametros.csv', index_col=0)
b0_h = float(par.loc['b0', MOD])                # mg/L/(%*h), tal como viene del analisis
Fp   = pd.read_csv(DIR_AN + 'f_generada_%s_14d_1min.csv' % MOD)

t_f  = np.arange(0, HORAS * 3600, T_s)          # el T_s del cuaderno, no otro
f_h  = np.interp(t_f / 3600.0, Fp.h.to_numpy(), Fp.f.to_numpy())    # mg/(L*h)
f_s  = f_h / 3600.0                                                  # mg/(L*s)

r_op = np.full(len(t_f), 2.0)                   # consigna del operario [mg/L]
r_op       = np.array(NIV)[(t_f // (HORAS_NIV * 3600)).astype(int) % len(NIV)]
r_ff = np.asarray(lsim(F_ref, U=r_op, T=t_f)[0]).ravel()             # el prefiltro de arriba

#r_ff = np.asarray(lsim(C_PF, U=r_op, T=t_f)[0]).ravel() 
#r_ff = r_op
xi_f = (f_s +r_op / Tau  ) / bo                # -> % de u, con las dos correcciones

print('b0 del analisis = %.4f mg/L/(%%*h)   ·   bo de diseno = %.4f' % (b0_h, bo * 3600))
print('f   : media %+.2f, sigma %.2f mg/(L*h)' % (f_h.mean(), f_h.std()))
print('xi  : media %+.1f, sigma %.1f %% de u   <- lo que el actuador debe compensar'
      % (xi_f.mean(), xi_f.std()))
print('ventana de %d h -> %s muestras a T_s = %.2f s\n' % (HORAS, f'{len(t_f):,}', T_s))

[Y, T, X] = lsim(sys_au_cl, U=np.column_stack([r_ff, xi_f]), T=t_f, X0=np.zeros(n + n + m))

y_f   = (Cp_r @ X[:, :n].T).ravel()
eh_f  = X[:, 1]                                 # e_hat_y
xih_f = X[:, n + 1]                             # xi1_hat (la perturbacion,
                                                 # NO su derivada): el indice de la
                                                 # estimacion de f es SIEMPRE n+1, no
                                                 # n+m -- con m=1 coincidian, con m>1 no.
u_f   = -(K @ X[:, n:].T).ravel() / bo

# El estado xi_hat del observador ES la estimacion de la perturbacion total f,
# ya en mg/(L*s): basta pasarlo a mg/(L*h) para compararlo con la f medida.
# No hace falta pasar por % de u para nada.
f_est_h = xih_f * 3600.0                        # -> mg/(L*h), las unidades del analisis

# La misma cosa leida en puntos de soplador, por si hace falta el coste:
xi_est  = xih_f / bo  + y_f / (Tau * bo)        # -> % de u

# Trayectoria que impone el polo del controlador, tramo a tramo de consigna:
#     y(t) = r + (y(t0) - r) * exp(-omega_c * (t - t0))
w_c    = -pols[0]
nuevo  = np.r_[True, np.diff(r_op) != 0]        # True en cada cambio de consigna
tramo  = np.cumsum(nuevo) - 1
k0     = np.flatnonzero(nuevo)[tramo]           # muestra en que arranco su tramo
y_teo  = r_op + (y_f[k0] - r_op) * np.exp(-w_c * (t_f - t_f[k0]))

ok = t_f > 2 * 3600                             # fuera del transitorio de arranque
print('RESULTADO   (sys_au_cl y F_ref reusados, sin recrear ninguna matriz)')
print('   OD : IAE medio %.4f mg/L   ·   y entre %.3f y %.3f   ·   en banda +-0,2: %.1f %%'
      % (np.abs(y_f[ok] - r_op[ok]).mean(), y_f[ok].min(), y_f[ok].max(),
         100 * np.mean(np.abs(y_f[ok] - r_op[ok]) <= 0.2)))
print('   u  : media %.1f %%, sigma %.1f puntos, entre %.0f y %.0f %%'
      % (u_f[ok].mean(), u_f[ok].std(), u_f[ok].min(), u_f[ok].max()))
print('   fuera de 0-100 %% el %.1f %% del tiempo   <- el soplador no da para esto'
      % (100 * np.mean((u_f[ok] < 0) | (u_f[ok] > 100))))
print('   estimacion de f : max|f - f_est| = %.2f mg/(L*h)   (sigma de f = %.2f)'
      % (np.abs(f_h[ok] - f_est_h[ok]).max(), f_h.std()))
print('                     lo mismo en soplador: %.1f %% de u  (sigma de xi = %.1f)'
      % (np.abs(xi_f[ok] - xi_est[ok]).max(), xi_f.std()))
```

```python
hh = t_f / 3600.0                               # eje en horas

fig, axs = plt.subplots(4, 1, figsize=(13, 11), sharex=True, layout='constrained')

# a) el OD contra la consigna
ax = axs[0]
ax.plot(hh, y_f, color='C0', linewidth=1.1, label='$y$  (OD)')
ax.plot(hh, r_op, color='k', linewidth=1.3, linestyle='--', label='$r$  (consigna)')
ax.plot(hh, y_teo, color='C1', linewidth=1.8, linestyle=':',
        label=r'$r+(y_0-r)\,e^{-\omega_c t}$  (lo que impone el polo)')
ax.fill_between(hh, r_op - 0.2, r_op + 0.2, color='C2', alpha=.12, label='banda $\\pm$0,2 mg/L')
ax.set_title('a · Oxígeno disuelto contra la perturbación real de Tunja (%s, %d h)' % (MOD, HORAS))
ax.set_ylabel('OD [mg/L]'); ax.set_ylim(0 ,3)

# b) la perturbacion: aplicada contra estimada
ax = axs[1]
ax.plot(hh, f_h, color='k', linewidth=1.2, label='$f$ MEDIDA en la planta')
ax.plot(hh, f_est_h, color='C1', linewidth=1.0, linestyle='--',
        label='$\\hat f = 3600\\,\\hat\\xi$  estimada por el ESO')
ax.set_title('b · La perturbación total, en sus unidades: la que hay contra la que el ESO ve')
ax.set_ylabel('$f$ [mg/(L·h)]')
ax.text(.99, .06, 'máx. error de estimación: %.2f mg/(L·h)   ·   $\\sigma$ de $f$ = %.2f'
        % (np.abs(f_h[ok] - f_est_h[ok]).max(), f_h.std()),
        transform=ax.transAxes, ha='right', fontsize=9, color='#555555')

# segundo eje: la MISMA curva leida en puntos de soplador.  Delta_u = Delta_f/b0
# es lo que cuesta compensarla (el criterio de la seccion 10 del analisis), y es
# un escalado fijo, asi que los dos ejes quedan alineados exactamente.
ax_pts = ax.twinx()
ax_pts.set_ylim(np.array(ax.get_ylim()) / (bo * 3600.0))
ax_pts.set_ylabel('$f/b_0$  [% de $u$]', color='#555555')
ax_pts.tick_params(axis='y', colors='#555555')
ax_pts.grid(False)

# c) el soplador contra sus limites
ax = axs[2]
ax.plot(hh, u_f, color='C4', linewidth=1.0, label='$u = -K\\hat x / b_0$')
ax.axhline(100, color='C3', linestyle=':', linewidth=1.3, label='límite del soplador')
ax.axhline(0, color='C3', linestyle=':', linewidth=1.3)
ax.fill_between(hh, 100, np.maximum(u_f, 100), color='C3', alpha=.25)
ax.fill_between(hh, np.minimum(u_f, 0), 0, color='C3', alpha=.25)
ax.set_title('c · Señal de control: lo sombreado es lo que el actuador NO puede dar (%.1f %% del tiempo)'
             % (100 * np.mean((u_f[ok] < 0) | (u_f[ok] > 100))))
ax.set_ylabel('$u$ [%]')

# d) el error: el que ve el observador contra el que estima (estado X[:, n])
ax = axs[3]
e_obs_f = y_f - r_ff                            # e_y = y - F r: lo que el ESO tiene que estimar
ax.plot(hh, e_obs_f, color='C0', linewidth=1.8, label='$e_y = y - F\\,r$  (el que ve el observador)')
ax.plot(hh, eh_f, color='C1', linewidth=1.0, linestyle='--',
        label='$\\hat e_y$  estimado (estado del observador)')
ax.axhline(0, color='k', linewidth=.8)
lim = 1.15 * np.abs(np.r_[e_obs_f[ok], eh_f[ok]]).max()
ax.set_ylim(-lim, lim)
ax.set_title('d · El error: real contra estimado   ·   máx. $|e_y - \\hat e_y|$ = %.2e mg/L'
             % np.abs(e_obs_f[ok] - eh_f[ok]).max())
ax.set_ylabel('error [mg/L]'); ax.set_xlabel('tiempo [h]')

for ax in axs:
    ax.minorticks_on(); ax.legend(fontsize=9, frameon=False, loc='upper right')
    ax.grid(visible=True, which='major', alpha=0.35)
    ax.set_xlim(0, HORAS)
fig.suptitle('El sistema diseñado (sys_au_cl + $F_{ref}$) contra la perturbación medida en la PTAR',
             fontsize=13, fontweight='bold')
plt.show()
```

### 4.2 GPI contra la perturbación REAL de la planta

Los mismos bloques de arriba, sin tocar nada, ahora con la perturbación
medida en Tunja en vez del escalón. Se usa **la misma `f` y la misma conversión que la
sección 4** del ADRC, para que las dos arquitecturas queden medidas contra lo mismo:

$$\xi=\frac{f+y/\tau}{b_0}\qquad(y\approx r)$$

— unidades de mg/(L·h) a mg/(L·s), y restando el polo $-y/\tau$ que `f` ya lleva
dentro y que la planta real también tiene.

**La perturbación estimada, en unidades físicas.** Escribiendo la ley del GPI en la
forma del ADRC, $b_0u = u_0 - \hat f$ con $u_0 = k_2 e$, la estimación se lee de la
propia señal de control:

$$\hat f = k_2\,e - b_0\,u\quad[\mathrm{mg/(L\cdot s)}] \qquad\longrightarrow\qquad
3600\,\hat f\ \ [\mathrm{mg/(L\cdot h)}]$$
Es la lectura directamente comparable con la $\hat\xi$ del ESO de la sección 4.
**El escenario es el mismo que el del observador:** 25 h, consigna que salta cada
4 h entre 2,0 · 2,5 · 1,5 mg/L, y todo en tiempo discreto a $T_s$. Así la sección 5
puede poner las dos arquitecturas en la misma gráfica.

```python
# ============================================================
# GPI (n=1, m=2) contra la perturbacion medida en Tunja  ·  EN CONTINUO
#   Los mismos bloques G_yr, G_yd, G_ur, G_ud de arriba, sin tocar nada.
#   Escenario identico al del observador (seccion 4): 25 h y consigna que
#   salta cada 4 h entre 2,0 · 2,5 · 1,5 mg/L.
# ============================================================
DIR_AN = 'Datos/analisis/'
MOD    = 'M2'
HORAS  = 25                                       # un ciclo diario completo

NIV       = [2.0, 2.5, 1.5]                       # mg/L
HORAS_NIV = 4

Fp_g  = pd.read_csv(DIR_AN + 'f_generada_%s_14d_1min.csv' % MOD)
t_gr  = np.arange(0, HORAS * 3600, T_s)                                   # el T_s del cuaderno
f_h_g = np.interp(t_gr / 3600.0, Fp_g.h.to_numpy(), Fp_g.f.to_numpy())    # mg/(L*h)
f_s_g = f_h_g / 3600.0                                                    # mg/(L*s)

r_gr  = np.array(NIV)[(t_gr // (HORAS_NIV * 3600)).astype(int) % len(NIV)]   # consigna [mg/L]
xi_gr = (f_s_g + r_gr / Tau) / bo                 # -> % de u, igual que la seccion 4

y_gr = lsim(G_yr, r_gr, t_gr)[0] + lsim(G_yd, xi_gr, t_gr)[0]     # salida
u_gr = lsim(G_ur, r_gr, t_gr)[0] + lsim(G_ud, xi_gr, t_gr)[0]     # senal de control [%]

e_gr       = r_gr - y_gr
#f_est_h_gr = (k2_ * e_gr - bo * u_gr) * 3600.0    # f_est = k2*e - b0*u  ->  mg/(L*h)
f_est_h_gr = (- bo * u_gr) * 3600.0    # f_est = k2*e - b0*u  ->  mg/(L*h)
ok_g = t_gr > 2 * 3600
print('GPI CONTINUO contra la f real  (%s, %d h, %s muestras a T_s = %.2f s)'
      % (MOD, HORAS, f'{len(t_gr):,}', T_s))
print('   f   : media %+.2f, sigma %.2f mg/(L*h)' % (f_h_g[ok_g].mean(), f_h_g[ok_g].std()))
print('   OD  : IAE medio %.4f mg/L   ·   y entre %.3f y %.3f   ·   en banda +-0,2: %.1f %%'
      % (np.abs(e_gr[ok_g]).mean(), y_gr[ok_g].min(), y_gr[ok_g].max(),
         100 * np.mean(np.abs(e_gr[ok_g]) <= 0.2)))
print('   u   : media %.1f %%, entre %.0f y %.0f %%   ·   fuera de 0-100 el %.1f %% del tiempo'
      % (u_gr[ok_g].mean(), u_gr[ok_g].min(), u_gr[ok_g].max(),
         100 * np.mean((u_gr[ok_g] < 0) | (u_gr[ok_g] > 100))))
print('   f   : estimada media %+.2f  ·  max|f - f_est| = %.2f mg/(L*h)'
      % (f_est_h_gr[ok_g].mean(), np.abs(f_h_g[ok_g] - f_est_h_gr[ok_g]).max()))

# ============================================================
# Figuras
# ============================================================
hh_g = t_gr / 3600.0
fig, axs = plt.subplots(3, 1, figsize=(13, 11), sharex=True, layout='constrained')

ax = axs[0]
ax.plot(hh_g, y_gr, color='C0', linewidth=1.1, label='$y$  (OD)')
ax.plot(hh_g, r_gr, color='k', linewidth=1.3, linestyle='--', label='$r$  (consigna)')
ax.fill_between(hh_g, r_gr - 0.2, r_gr + 0.2, color='C2', alpha=.12, label='banda $\\pm$0,2 mg/L')
ax2 = ax.twinx()
ax2.plot(hh_g, e_gr, color='C3', linewidth=.9, alpha=.7)
ax2.set_ylabel('error $e=r-y$ [mg/L]', color='C3'); ax2.tick_params(axis='y', colors='C3')
ax2.grid(False)
ax.set_title('a · Oxígeno disuelto y error, GPI contra la perturbación real (%s, %d h)' % (MOD, HORAS))
ax.set_ylabel('OD [mg/L]')

ax = axs[1]
ax.plot(hh_g, f_h_g, color='k', linewidth=1.2, label='$f$ MEDIDA en la planta')
ax.plot(hh_g, f_est_h_gr, color='C1', linewidth=1.0, linestyle='--',
        label='$\\hat f = k_2e - b_0u$  (leída de la señal de control)')
ax.set_title('b · La perturbación total: medida contra la que el GPI compensa')
ax.set_ylabel('$f$ [mg/(L·h)]'); ax.set_ylim(-30, 5)
ax.text(.99, .06, 'máx. error de estimación: %.2f mg/(L·h)'
        % np.abs(f_h_g[ok_g] - f_est_h_gr[ok_g]).max(),
        transform=ax.transAxes, ha='right', fontsize=9, color='#555555')
ax_p = ax.twinx()
ax_p.set_ylim(np.array(ax.get_ylim()) / (bo * 3600.0))
ax_p.set_ylabel('$f/b_0$  [% de $u$]', color='#555555')
ax_p.tick_params(axis='y', colors='#555555'); ax_p.grid(False)

ax = axs[2]
ax.plot(hh_g, u_gr, color='C4', linewidth=1.0, label='$u$  (al soplador)')
ax.axhline(100, color='C3', linestyle=':', linewidth=1.3, label='límite del soplador')
ax.axhline(0, color='C3', linestyle=':', linewidth=1.3)
ax.fill_between(hh_g, 100, np.maximum(u_gr, 100), color='C3', alpha=.25)
ax.fill_between(hh_g, np.minimum(u_gr, 0), 0, color='C3', alpha=.25)
ax.set_title('c · Señal de control: lo sombreado es lo que el actuador NO puede dar (%.1f %% del tiempo)'
             % (100 * np.mean((u_gr[ok_g] < 0) | (u_gr[ok_g] > 100))))
ax.set_ylabel('$u$ [%]'); ax.set_ylim(-60, 200)



for ax, loc in zip(axs, ['lower right', 'lower left', 'lower left', 'lower right']):
    ax.minorticks_on(); ax.legend(fontsize=9, frameon=False, loc=loc)
    ax.grid(visible=True, which='major', alpha=0.35)
    ax.set_xlim(0, HORAS)

fig.suptitle('GPI $n=1$, $m=2$ CONTINUO contra la perturbación medida en la PTAR',
             fontsize=13, fontweight='bold')
plt.show()
```

# 5. Las dos arquitecturas, lado a lado

Última gráfica del cuaderno: el **ADRC con observador** (secciones 2.1 + 3, es decir
`sys_au_cl` con el prefiltro `F_ref`) y el **GPI sin observador** (sección 2.2,
los bloques `G_yr, G_yd, G_ur, G_ud`), contra **la misma perturbación medida en
Tunja y la misma consigna**: 25 h, saltos cada 4 h entre 2,0 · 2,5 · 1,5 mg/L.

No se rediseña ni se vuelve a discretizar nada: se reusan las señales `t_f`, `r_op`
y `xi_f` de la sección 4 y se pasan por los bloques del GPI.

**Qué se compara y qué no.** Los dos lazos ven la misma consigna del operario, pero
cada uno la trata como lo pide su propio diseño: el ADRC la filtra con `F_ref`
(Herbst & Madoński, Fig. 6.16 — sin prefiltro el escalón le mete un impulso al
observador), y el GPI la recibe cruda, porque es un lazo de un solo grado de
libertad. Por eso los saltos del GPI son más bruscos: es su diseño, no un
desajuste del experimento.

**Dónde se separan de verdad.** En la estimación de `f`. El ESO la estima con un
observador cuya velocidad se elige aparte (`pols_L`), mientras que en el GPI la
velocidad de estimación está amarrada a los polos del lazo cerrado. Es la
diferencia que se ve en el panel b.

```python
# ============================================================
# 5. ADRC con observador  contra  GPI, misma perturbacion y misma consigna
#
#   Del ESO (seccion 4, ya calculado):  y_f, u_f, f_est_h
#   Del GPI (seccion 2.2):              los bloques G_yr, G_yd, G_ur, G_ud
#   Entradas comunes:                   t_f, r_op, xi_f, f_h
# ============================================================

y_c = lsim(G_yr, r_op, t_f)[0] + lsim(G_yd, xi_f, t_f)[0]     # OD con el GPI
u_c = lsim(G_ur, r_op, t_f)[0] + lsim(G_ud, xi_f, t_f)[0]     # soplador con el GPI
#f_est_c = (k2_ * (r_op - y_c) - bo * u_c) * 3600.0            # f estimada por el GPI [mg/(L*h)]
f_est_c = ( - bo * u_c) * 3600.0            # f estimada por el GPI [mg/(L*h)]

print('MISMA perturbacion (%s, %d h) y MISMA consigna (%s mg/L cada %d h)\n'
      % (MOD, HORAS, NIV, HORAS_NIV))
print('  %-26s %12s %12s' % ('', 'ADRC-ESO', 'GPI m=2'))
for txt, a, b in (
    ('IAE del OD      [mg/L]', np.abs(y_f[ok] - r_op[ok]).mean(), np.abs(y_c[ok] - r_op[ok]).mean()),
    ('OD en banda +-0,2   [%]', 100 * np.mean(np.abs(y_f[ok] - r_op[ok]) <= 0.2),
                                100 * np.mean(np.abs(y_c[ok] - r_op[ok]) <= 0.2)),
    ('OD minimo       [mg/L]', y_f[ok].min(), y_c[ok].min()),
    ('OD maximo       [mg/L]', y_f[ok].max(), y_c[ok].max()),
    ('u media            [%]', u_f[ok].mean(), u_c[ok].mean()),
    ('u minima           [%]', u_f[ok].min(), u_c[ok].min()),
    ('u maxima           [%]', u_f[ok].max(), u_c[ok].max()),
    ('u fuera de 0-100   [%]', 100 * np.mean((u_f[ok] < 0) | (u_f[ok] > 100)),
                               100 * np.mean((u_c[ok] < 0) | (u_c[ok] > 100))),
    ('max|f - f_est| [mg/L/h]', np.abs(f_h[ok] - f_est_h[ok]).max(),
                                np.abs(f_h[ok] - f_est_c[ok]).max()),
    ('sigma del error de f   ', (f_h[ok] - f_est_h[ok]).std(), (f_h[ok] - f_est_c[ok]).std())):
    print('  %-26s %12.3f %12.3f' % (txt, a, b))

# ============================================================
# La grafica final
# ============================================================
hh = t_f / 3600.0
fig, axs = plt.subplots(4, 1, figsize=(13, 11.5), sharex=True, layout='constrained')

ax = axs[0]
ax.plot(hh, r_op, color='k', linewidth=1.3, linestyle='--', label='$r$  (consigna del operario)')
ax.fill_between(hh, r_op - 0.2, r_op + 0.2, color='C2', alpha=.12, label='banda $\\pm$0,2 mg/L')
ax.plot(hh, y_f, color='C0', linewidth=1.1, label='ADRC con observador  (IAE %.3f)'
        % np.abs(y_f[ok] - r_op[ok]).mean())
ax.plot(hh, y_c, color='C1', linewidth=1.1, label='GPI sin observador  (IAE %.3f)'
        % np.abs(y_c[ok] - r_op[ok]).mean())
ax.set_title('a · El oxígeno disuelto: las dos arquitecturas contra la misma consigna')
ax.set_ylabel('OD [mg/L]'); ax.set_ylim(0, 3.2)

ax = axs[1]
ax.plot(hh, f_h, color='k', linewidth=1.2, label='$f$ MEDIDA en la planta')
ax.plot(hh, f_est_h, color='C0', linewidth=1.0, linestyle='--',
        label='estimada por el ESO  (máx. error %.2f)' % np.abs(f_h[ok] - f_est_h[ok]).max())
ax.plot(hh, f_est_c, color='C1', linewidth=1.0, linestyle='--',
        label='estimada por el GPI  (máx. error %.2f)' % np.abs(f_h[ok] - f_est_c[ok]).max())
ax.set_title('b · La perturbación total: quién la sigue mejor')
ax.set_ylabel('$f$ [mg/(L·h)]'); ax.set_ylim(-30, 5)
ax_p = ax.twinx()
ax_p.set_ylim(np.array(ax.get_ylim()) / (bo * 3600.0))
ax_p.set_ylabel('$f/b_0$  [% de $u$]', color='#555555')
ax_p.tick_params(axis='y', colors='#555555'); ax_p.grid(False)

ax = axs[2]
ax.plot(hh, u_f, color='C0', linewidth=1.0, label='ADRC con observador  (%.1f %% fuera de rango)'
        % (100 * np.mean((u_f[ok] < 0) | (u_f[ok] > 100))))
ax.plot(hh, u_c, color='C1', linewidth=1.0, label='GPI sin observador  (%.1f %% fuera de rango)'
        % (100 * np.mean((u_c[ok] < 0) | (u_c[ok] > 100))))
ax.axhline(100, color='C3', linestyle=':', linewidth=1.3, label='límites del soplador')
ax.axhline(0, color='C3', linestyle=':', linewidth=1.3)
ax.set_title('c · Señal de control: lo que cada uno le pide al soplador')
ax.set_ylabel('$u$ [%]'); ax.set_ylim(-60, 200)

ax = axs[3]
ax.plot(hh, y_f - r_ff, color='C0', linewidth=1.8, label='ADRC: $e_y = y - F\\,r$  (el que ve el observador)')
ax.plot(hh, eh_f, color='k', linewidth=1.0, linestyle='--',
        label='ADRC: $\\hat e_y$  estimado (estado del observador)')
ax.plot(hh, y_c - r_op, color='C1', linewidth=1.0, alpha=.85,
        label='GPI: $e_y = y - r$  (sin observador, no hay $\\hat e_y$)')
ax.axhline(0, color='k', linewidth=.8)
lim = 1.15 * np.abs(np.r_[(y_f - r_ff)[ok], eh_f[ok], (y_c - r_op)[ok]]).max()
ax.set_ylim(-lim, lim)
ax.set_title('d · El error: el del ADRC contra su estimación, y el del GPI')
ax.set_ylabel('error [mg/L]'); ax.set_xlabel('tiempo [h]')

for ax, loc in zip(axs, ['lower right', 'lower left', 'upper left', 'lower right']):
    ax.minorticks_on(); ax.legend(fontsize=9, frameon=False, loc=loc)
    ax.grid(visible=True, which='major', alpha=0.35)
    ax.set_xlim(0, HORAS)

fig.suptitle('ADRC con observador  contra  GPI  ·  misma perturbación real y misma consigna',
             fontsize=13, fontweight='bold')
plt.show()
```

# 6. ADRC y GPI en el PLC ctrlX (Bosch Rexroth)

Simulación de los dos lazos en el software del PLC (`ADRC_Module.project`). El *trace* del ctrlX guarda como máximo 10 000 muestras
(≈ 500 s a $T_s$ = 50 ms), así que cada respuesta quedó en 10 archivos consecutivos con huecos cortos entre uno y otro:
`Datos/txtCtrl/01.txt … 10.txt` para el ADRC con observador y `G_01.txt … G_10.txt` para el GPI. Cada lazo se une en un solo
tramo a $T_s$, interpolando linealmente los huecos. Solo se registró la salida $y$.

Tiempo de asentamiento $t_s$: el instante desde el cual $|y-r| \le b\,r$ para siempre, con $b$ = 5 % y 2 %, medido en el reloj del PLC.

```python
# ============================================================
# ADRC y GPI en el PLC ctrlX (Bosch Rexroth): 10 trazas de y por lazo, unidas en un solo tramo
# ============================================================
import glob, os, re

R_PLC  = 2.0                                   # consigna de la simulacion en el PLC [mg/L]
BANDAS = (0.05, 0.02)
tau_c  = -1 / pols[0]                          # polo de diseno del controlador ADRC [s]
LAZOS  = {'ADRC': '[0-9][0-9].txt', 'GPI': 'GPI_*.txt'}

def lee_trazas(patron):
    arch = sorted(glob.glob(os.path.join('Datos', 'txtCtrl', patron)),
                  key=lambda a: int(re.findall(r'\d+', os.path.basename(a))[0]))
    d = pd.concat([pd.read_csv(a, sep=r'\s+', skiprows=3, header=None, names=['t', 'y'])
                     .assign(archivo=os.path.basename(a)[:-4]) for a in arch], ignore_index=True)
    d['t'] = d['t'] / 1000.0                   # ms -> s (reloj del PLC)
    return d

plc, res = {}, []
for lazo, patron in LAZOS.items():
    d = lee_trazas(patron)
    res.append(d.groupby('archivo').agg(t_ini=('t', 'first'), t_fin=('t', 'last'),
                                        y_ini=('y', 'first'), y_fin=('y', 'last'), n=('y', 'size'))
                .assign(lazo=lazo))
    # un solo tramo continuo a T_s: los huecos entre trazas se interpolan linealmente
    t_c = np.arange(d['t'].iloc[0], d['t'].iloc[-1], T_s)
    y_c = np.interp(t_c, d['t'], d['y'])
    t_s = {b: t_c[np.flatnonzero(np.abs(y_c - R_PLC) > b * R_PLC)[-1] + 1] for b in BANDAS}
    plc[lazo] = dict(t=t_c, y=y_c, t_s=t_s)
    print('%-4s  medidas %d  ·  tramo continuo %d (%.1f %% interpolado)  ·  '
          't_s 5 %% = %.0f s (%.1f min)  ·  t_s 2 %% = %.0f s (%.1f min)'
          % (lazo, len(d), len(t_c), 100 * (1 - len(d) / len(t_c)),
             t_s[0.05], t_s[0.05] / 60, t_s[0.02], t_s[0.02] / 60))
display(pd.concat(res).round(4))

# polo de diseno desde la primera muestra del ADRC
t0, y0  = plc['ADRC']['t'][0], plc['ADRC']['y'][0]
t_teo   = plc['GPI']['t'] if plc['GPI']['t'][-1] > plc['ADRC']['t'][-1] else plc['ADRC']['t']
y_teo   = R_PLC + (y0 - R_PLC) * np.exp(-(t_teo - t0) / tau_c)
print('polo de diseno: t_s 5 %% = %.0f s  ·  t_s 2 %% = %.0f s'
      % tuple(t0 + tau_c * np.log(abs(y0 - R_PLC) / (b * R_PLC)) for b in BANDAS))

estilo = {'ADRC': dict(color='C0', ls='-'), 'GPI': dict(color='C2', ls='-')}
marca  = {0.05: '--', 0.02: '-'}
fig, axs = plt.subplots(2, 1, figsize=(13, 9), layout='constrained')
for ax in axs:
    for lazo, p in plc.items():
        ax.plot(p['t'], p['y'], linewidth=2.2, label='$y$  %s' % lazo, **estilo[lazo])
        for b in BANDAS:
            ax.axvline(p['t_s'][b], color=estilo[lazo]['color'], linestyle=marca[b], linewidth=1.2,
                       label=r'$t_s$ %.0f %% %s = %.0f s  (%.1f min)'
                             % (100 * b, lazo, p['t_s'][b], p['t_s'][b] / 60))
    ax.plot(t_teo, y_teo, color='k', linestyle=':', linewidth=1.3,
            label=r'polo de diseño ADRC  $\tau_c$ = %.0f s' % tau_c)
    ax.axhline(R_PLC, color='k', linestyle='--', linewidth=1.2, label='$r$ = %.1f mg/L' % R_PLC)
    for b, c in zip(BANDAS, ('C1', 'C3')):
        ax.axhspan(R_PLC * (1 - b), R_PLC * (1 + b), color=c, alpha=.08)
    ax.set_xlabel('tiempo del PLC [s]'); ax.set_ylabel('OD [mg/L]')
    ax.minorticks_on(); ax.grid(visible=True, which='major', alpha=.35)
axs[0].set_title('a · Respuesta completa')
axs[0].legend(fontsize=8.5, frameon=False, loc='lower right', ncol=2)
axs[1].set_ylim(R_PLC * 0.9, R_PLC * 1.02)
axs[1].set_xlim(2000, max(p['t'][-1] for p in plc.values()) + 100)
axs[1].set_title('b · Acercamiento a las bandas del 5 % y 2 %')
fig.suptitle('ADRC y GPI en el PLC ctrlX (Bosch) · escalón de consigna a %.1f mg/L' % R_PLC,
             fontsize=13, fontweight='bold')
plt.show()
```

### 6.1 Trazas con la planta de diseño: Bézier y prefiltro (figura `plc_trazas.pdf` de la tesis)

Trazas del ctrlX con el bloque `PLANT` configurado como la planta de diseño ($\tau$ = 1200 s, $k_{dc}$ = 0,05) y la sintonía de diseño
($\omega_c = 4/\tau$, $\omega_o = 5\,\omega_c$, $b_0 = k_{dc}/\tau$, $T_s$ = 50 ms), en `PLC_ST/capturas_codesys/Pruebas`:
`BEZIER` (`BEZIER_REF` + `ADRC_ESO`) y `Filtro` (`PREFILTER_ADRC` + `ADRC_ESO`). Las capturas se unen por la marca de tiempo
del PLC y solo se dibujan datos medidos. El tiempo se cuenta en ciclos $\times\,T_s$, porque la tarea corre en promedio a 50,5 ms
mientras `PLANT` avanza 50 ms por ciclo. $t_s$ es el instante desde el cual $|y - r|$ queda dentro del 2 % del cambio de consigna.

```python
# ============================================================
# Figura de las trazas del ctrlX PLC con el bloque PLANT configurado como la planta de diseño
# (tau = 1200 s, k_dc = 0,05), wc = 4/tau, wo = 5 wc, b0 = k_dc/tau y Ts = 50 ms.
#   PLC_ST/capturas_codesys/Pruebas/BEZIER: BEZIER_REF + ADRC_ESO     columnas t, y, U, r, y*
#   PLC_ST/capturas_codesys/Pruebas/Filtro: PREFILTER_ADRC + ADRC_ESO columnas t, y, r, U
# Cada captura guarda 10 000 muestras. Las capturas se unen por la marca de tiempo del PLC y se
# quitan las muestras repetidas donde dos capturas se solapan. Solo se dibujan datos medidos.
# La tarea del PLC corre en promedio a 50,5 ms y no a 50 ms; como PLANT avanza Ts = 50 ms por
# ciclo, el tiempo se pasa de reloj del PLC a ciclos x Ts.
# ============================================================
import glob, os, re
import numpy as np
import matplotlib.pyplot as plt

PRU = os.path.join('PLC_ST', 'capturas_codesys', 'Pruebas')
TS = 0.05                     # periodo con que PLANT integra la planta [s]


def une_trazas(carpeta, cols):
    arch = sorted(glob.glob(os.path.join(PRU, carpeta, '*.txt')),
                  key=lambda a: int(re.findall(r'(\d+)\.txt', a)[0]))
    A = np.vstack([np.loadtxt(a, skiprows=3) for a in arch])
    A = A[np.argsort(A[:, 0], kind='stable')]
    A = A[np.unique(A[:, 0], return_index=True)[1]]           # solapes: muestras idénticas
    d = {k: A[:, j] for k, j in cols.items()}
    t = A[:, 0] / 1000.0                                        # ms -> s, reloj del PLC
    dt = np.diff(t); P = dt[dt < .3].mean()                     # periodo real de la tarea
    d['t'] = t * TS / P                                         # tiempo de la planta: ciclos x Ts
    print('%-7s %d capturas · %d muestras · periodo real %.2f ms · %.0f s · huecos > 0,3 s: %d, '
          'el mayor de %.0f s, %.0f s sin registrar en total' % (carpeta, len(arch), len(A), P * 1e3,
          d['t'][-1], (dt > .3).sum(), dt.max(), dt[dt > .3].sum()))
    return d


def indicadores(d, t0, t1, y0, y1):
    m = (d['t'] >= t0) & (d['t'] < t1)
    t, y, u = d['t'][m], d['y'][m], d['u'][m]
    z = (y - y0) / (y1 - y0); fuera = np.flatnonzero(np.abs(z - 1) > 0.02)
    dt = np.diff(t, append=t[-1] + 0.05)
    return dict(Mp=max((z.max() - 1) * 100, 0), ts=(t[fuera[-1] + 1] - t0) / 60, umax=u.max(),
                sat=dt[u >= 99.99].sum(), hueco=t[fuera[-1] + 1] - t[fuera[-1]])


BZ = une_trazas('BEZIER', dict(y=1, u=2, r=3, yr=4))
PF = une_trazas('Filtro', dict(y=1, r=2, u=3))

print('\n%-34s %-10s %9s %10s %9s %13s' % ('lazo', 'consigna', 'sobrepico', 't_s (2 %)', 'u máx.', 'u en 100 %'))
for nom, d in (('BEZIER_REF y ADRC_ESO', BZ), ('PREFILTER_ADRC y ADRC_ESO', PF)):
    c = np.flatnonzero(np.diff(d['r']) != 0) + 1                # cambios de consigna
    d['cambios'] = d['t'][c]
    bordes = list(d['t'][c]) + [d['t'][-1] + 1]
    for k, i in enumerate(c):
        y0, y1 = d['r'][i - 1], d['r'][i]
        r_ = indicadores(d, bordes[k], bordes[k + 1], y0, y1)
        print('%-34s %3.1f -> %3.1f %8.1f %% %7.1f min %7.1f %% %10.1f s   (hueco en el cruce: %.2f s)'
              % (nom if k == 0 else '', y0, y1, r_['Mp'], r_['ts'], r_['umax'], r_['sat'], r_['hueco']))

fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.6), layout='constrained', sharex='col')
for j, (tit, d) in enumerate((('BEZIER_REF y ADRC_ESO', BZ), ('PREFILTER_ADRC y ADRC_ESO', PF))):
    ax, bx = axs[0, j], axs[1, j]
    ax.plot(d['t'], d['r'], 'k:', lw=1.1, label='consigna $r$')
    if 'yr' in d: ax.plot(d['t'], d['yr'], color='C3', ls='--', lw=1.4, label='referencia $y^*$')
    ax.plot(d['t'], d['y'], 'k', lw=1.6, label='salida $y$')
    bx.plot(d['t'], d['u'], 'k', lw=1.4)
    bx.axhline(100, color='C3', ls=':', lw=1.2); bx.axhline(0, color='C3', ls=':', lw=1.2)
    ax.set_title('%s · Salida, %s' % ('ab'[j], tit)); bx.set_title('%s · Señal de control, %s' % ('cd'[j], tit))
    ax.set_ylim(-0.1, 2.45); bx.set_ylim(-5, 108); bx.set_xlabel('tiempo [s]')
    ax.legend(fontsize=9, frameon=False, loc='lower right')
    for a in (ax, bx):
        a.set_xlim(0, d['t'][-1]); a.minorticks_on(); a.grid(visible=True, which='major', alpha=0.35); eje_minutos(a)
axs[0, 0].set_ylabel('OD [mg/L]'); axs[1, 0].set_ylabel('$u$ [%]')
fig.suptitle('Trazas del PLC con la planta de diseño: consigna de 0 a 2 mg/L y de 2 a 2,2 mg/L',
             fontsize=13, fontweight='bold')
fig.savefig(os.path.join('Latex', 'Tesis', 'figuras', 'plc_trazas.pdf'), bbox_inches='tight')
plt.show()
```

# 7. Comparación PID ISA contra ADRC

**PID ISA ideal con filtro en la derivada**, sobre el mismo error
que el ADRC, $e = r - y = -e_y$:

$$C_{PID}(s) = K_p\left(1 + \frac{1}{T_i\,s} + \frac{T_d\,s}{1 + \dfrac{T_d}{N}\,s}\right)$$

**Sintonía: mismo esfuerzo inicial y misma robustez.** Primero se elige $K_p$ para que la señal de control
salte cerca de 40 % ante el escalón de 0,2 mg/L, de 40 a 80 %, el mismo pico del ADRC. En el instante del escalón la
acción proporcional y la derivativa suman $K_p(1 + N)\,\Delta r$, así que con $N = 2$ queda $K_p = 40/(3 \cdot 0{,}2) = 66{,}7$ %/(mg/L).
Luego, con la relación clásica $T_i = 4\,T_d$, se ajusta $T_d$ para que el PID tenga, sin retardo, la misma sensibilidad
máxima $M_s$ que el ADRC. El retardo $\theta = 50$ s solo se usa para evaluar.

**Evaluación.** Escalón de consigna de 2,0 a 2,2 mg/L, primero sin retardo (el caso de diseño) y luego con $\theta = 50$ s, sin ruido y con el ruido
del sensor medido en el capítulo 2 ($\sigma = 0{,}0011$ mg/L, blanco, a 50 ms), sumado a la medida.
El soplador se limita a 0-100 %: el ADRC realimenta $b_0\,u_{sat}$ al observador y el PID usa
el mismo anti-windup del GPI (Handbook 19.1.3), $v = \frac{P}{E_1}e - \frac{L - E_1}{E_1}u_{sat}$, con $E_1$ formado por dos polos del lazo cerrado, los más rápidos que se pueden tomar sin partir un par complejo.

```python
# ============================================================
# 7.1 PID ISA con el mismo esfuerzo inicial y la misma M_s que el ADRC (sin retardo)
#   C_PID(s) = Kp (1 + 1/(Ti s) + Td s/(1 + Td s/N)),  Ti = 4 Td,  N = 2
#   1) Kp: salto de u de 40 % ante el escalon de 0,2 mg/L:  Kp (1 + N) 0,2 = 40
#   2) Td (Ti = 4 Td) tal que M_s(PID) = M_s(ADRC) con theta = 0
#   El retardo theta = 50 s solo se usa para evaluar.
# ============================================================
from scipy.optimize import brentq
wc_c, wo_c = 4 / Tau, 20 / Tau
kp_c, l1_c, l2_c = wc_c, 2 * wo_c, wo_c**2
k2_c, k1_c, k0_c = kp_c + l1_c, kp_c * l1_c + l2_c, kp_c * l2_c
N_pid, TH = 2.0, 50.0
w_c = np.logspace(-6, 2, 30000); s_c = 1j * w_c
G_c = lambda th: K_dc / (Tau * s_c + 1) * np.exp(-th * s_c)
C_adrc = (1 / bo) * (k1_c * s_c + k0_c) / (s_c * (s_c + k2_c))          # (-e_y) -> u
C_pid = lambda Kp, Ti, Td: Kp * (1 + 1 / (Ti * s_c) + Td * s_c / (1 + Td * s_c / N_pid))

def robustez_c(C, th):
    L_ = C * G_c(th); S_ = 1 / (1 + L_); T_ = L_ * S_
    m = np.abs(L_); ph = np.unwrap(np.angle(L_))
    i = np.flatnonzero(np.diff(np.sign(m - 1)))[0]
    cr = np.flatnonzero(np.diff(np.sign(ph + np.pi)))
    aT = np.abs(T_); j = np.flatnonzero(aT >= 1 / np.sqrt(2))[-1]
    lw = np.log10(w_c); f = (1 / np.sqrt(2) - aT[j]) / (aT[j + 1] - aT[j])
    return dict(Ms=np.abs(S_).max(), Mt=aT.max(), PM=180 + np.degrees(ph[i]),
                GM=(1 / m[cr[0]] if len(cr) else np.inf), wbw=10**(lw[j] + f * (lw[j + 1] - lw[j])))

R_a0 = robustez_c(C_adrc, 0)
DU0, DR = 40.0, 0.2                                                    # salto inicial de u [%] y escalon [mg/L]
Kp_p = DU0 / ((1 + N_pid) * DR)
Td_p = brentq(lambda Td: robustez_c(C_pid(Kp_p, 4 * Td, Td), 0)['Ms'] - R_a0['Ms'], 10, 60, xtol=1e-6)
Ti_p = 4 * Td_p
print('PID ISA:  Kp = %.2f %%/(mg/L)   Ti = %.2f s   Td = %.2f s   N = %g\n' % (Kp_p, Ti_p, Td_p, N_pid))
rob = {}
for nom, C in (('ADRC', C_adrc), ('PID', C_pid(Kp_p, Ti_p, Td_p))):
    for th in (0.0, TH):
        rob[(nom, th)] = robustez_c(C, th)
        d = rob[(nom, th)]
        print('%-5s theta = %2.0f s:  Ms = %.3f   Mt = %.3f   MF = %5.1f grados   MG = %s   w_BW = %.1f rad/h'
              % (nom, th, d['Ms'], d['Mt'], d['PM'], ('inf' if np.isinf(d['GM']) else '%.2f' % d['GM']), d['wbw'] * 3600))

fig, axs = plt.subplots(1, 2, figsize=(12.5, 4.2), layout='constrained')
for nom, C, col in (('ADRC', C_adrc, 'k'), ('PID', C_pid(Kp_p, Ti_p, Td_p), 'C0')):
    for th, ls in ((0.0, '-'), (TH, '--')):
        L_ = C * G_c(th)
        axs[0].semilogx(w_c * 3600, 20 * np.log10(np.abs(1 / (1 + L_))), color=col, ls=ls,
                        label='%s, $\\theta$ = %.0f s' % (nom, th))
        axs[1].semilogx(w_c * 3600, 20 * np.log10(np.abs(L_ / (1 + L_))), color=col, ls=ls,
                        label='%s, $\\theta$ = %.0f s' % (nom, th))
axs[0].set_title('a · Sensibilidad $|S(j\\omega)|$'); axs[1].set_title('b · Sensibilidad complementaria $|T(j\\omega)|$')
for ax in axs:
    ax.set_xlim(1e-1, 1e3); ax.set_ylim(-40, 20); ax.axhline(0, color='0.6', lw=0.8)
    ax.axvline(R_a0['wbw'] * 3600, color='0.6', ls=':', lw=1)
    ax.set_xlabel('$\\omega$ [rad/h]'); ax.set_ylabel('magnitud [dB]')
    ax.grid(True, which='both', alpha=0.3); ax.legend(fontsize=8, frameon=False)
fig.savefig('Latex/Tesis/figuras/pid_adrc_sensibilidad.pdf', bbox_inches='tight')
plt.show()
```

```python
# ============================================================
# 7.1 Escalon de 2,0 a 2,2 mg/L sin retardo y con theta = 50 s, sin ruido y con el ruido del sensor
#   Simulacion muestra a muestra a 50 ms: planta exacta (ZOH), retardo como buffer,
#   soplador 0-100 %.  ADRC: observador con [e_y, b0 u_sat].
#   PID: el mismo anti-windup del GPI (Handbook 19.1.3): C = P/L -> v = (P e - (L - E1) u_sat)/E1,
#        con E1 = n polos del lazo cerrado con la planta nominal (los mas rapidos sin partir un par complejo); ZOH a dt_c.
# ============================================================
import math
from scipy.linalg import expm
SIG_RUIDO = 0.0011                      # mg/L a 50 ms (capitulo 2, M1)
dt_c = 0.05

import itertools
def pid_aw():
    """PID ISA (o PI si Td = 0) como C = P/L con L monico, y el anti-windup del Handbook 19.1.3:
    v = P/E1 e - (L - E1)/E1 u_sat, con E1 = n polos del lazo cerrado con la planta nominal, los mas rapidos sin partir un par complejo."""
    if Td_p > 0:
        a = N_pid / Td_p
        P = Kp_p / Ti_p * np.array([(1 + N_pid) * Ti_p, a * Ti_p + 1, a]); L = np.array([1.0, a, 0.0])
    else:
        P = np.array([Kp_p, Kp_p / Ti_p]); L = np.array([1.0, 0.0])
    n = len(L) - 1
    lc = np.polyadd(np.polymul(L, [Tau, 1.0]), K_dc * P)                  # L (Tau s + 1) + K_dc P
    p = np.roots(lc); E1 = None
    # n polos del lazo cerrado, los mas rapidos posibles sin partir un par complejo (E1 con coeficientes reales)
    for idx in sorted(itertools.combinations(range(len(p)), n), key=lambda q: max(p[list(q)].real)):
        cand = np.poly(p[list(idx)])
        if np.abs(cand.imag).max() < 1e-12: E1 = cand.real; break
    if E1 is None: E1 = np.poly([-np.abs(p.real).max()] * n)             # sin subconjunto real: polo mas rapido repetido
    c = E1[1:]; p_n = P[0]; resto = P[1:] - p_n * c                        # P/E1 = p_n + resto/E1
    A = np.zeros((n, n)); A[:, 0] = -c; A[:-1, 1:] = np.eye(n - 1)         # forma observable
    B = np.c_[resto, -(L[1:] - c)]                                         # columnas: e, u_sat
    D = c2d(ss(A, B, np.eye(1, n), np.zeros((1, 2))), dt_c)
    return np.asarray(D.A), np.asarray(D.B), p_n

def sim_c(ctrl, t_end, r_fun, eps, theta=TH, sigma=0.0, seed=1, umin=0.0, umax=100.0, y0=2.0):
    # y0 = 2,0: arranca en regimen en 2,0 mg/L;  y0 = 0: soplador apagado y OD en cero
    n_ = int(round(t_end / dt_c)); rng = np.random.default_rng(seed)
    nz = sigma * rng.standard_normal(n_) if sigma > 0 else np.zeros(n_)
    ap = math.exp(-dt_c / Tau); bp = K_dc * (1 - ap); d = int(round(theta / dt_c))
    uss = (y0 / K_dc - eps[0]) if y0 > 0 else 0.0; buf = [uss] * (d + 1); bi = 0
    x = y0; Y = np.empty(n_); U = np.empty(n_); XI = np.full(n_, np.nan)
    if ctrl == 'ADRC':
        A_ = np.array([[0, 1], [0, 0.]]); B_ = np.array([1, 0.]); C_ = np.array([1, 0.]); L_ = np.array([l1_c, l2_c])
        M = expm(np.block([[A_ - np.outer(L_, C_), np.c_[L_, B_]], [np.zeros((2, 4))]]) * dt_c)
        a11, a12, a21, a22 = M[0, 0], M[0, 1], M[1, 0], M[1, 1]; b11, b12, b21, b22 = M[0, 2], M[0, 3], M[1, 2], M[1, 3]
        h1, h2 = ((0.0, -y0 / Tau + bo * eps[0]) if y0 > 0 else (y0 - r_fun(0.0), 0.0))
    else:
        Ad_aw, Bd_aw, Dp_aw = pid_aw()
        xc = -np.linalg.solve(Ad_aw - np.eye(len(Ad_aw)), Bd_aw[:, 1]) * uss      # regimen: v = uss con e = 0
    for k in range(n_):
        y = x + nz[k]; r_ = r_fun(k * dt_c)
        if ctrl == 'ADRC':
            u_ = -(kp_c * h1 + h2) / bo; us = umin if u_ < umin else (umax if u_ > umax else u_)
            ey = y - r_; bu = bo * us; XI[k] = (h2 + y / Tau) / bo          # xi estimada, en % de u
            h1, h2 = a11 * h1 + a12 * h2 + b11 * ey + b12 * bu, a21 * h1 + a22 * h2 + b21 * ey + b22 * bu
        else:
            e = r_ - y; v = xc[0] + Dp_aw * e
            us = umin if v < umin else (umax if v > umax else v)
            xc = Ad_aw @ xc + Bd_aw @ np.array([e, us])
        Y[k] = x; U[k] = us
        buf[bi] = us; bi = (bi + 1) % (d + 1); x = max(0.0, ap * x + bp * (buf[bi] + eps[k]))   # el OD no baja de 0
    return Y, U, XI

T_e, t0_e = 3000.0, 200.0
t_e = np.arange(int(T_e / dt_c)) * dt_c; eps_e = np.zeros(len(t_e))
r1_e = 2.2; dr_e = r1_e - 2.0
r_e = lambda t: r1_e if t >= t0_e else 2.0
me = t_e >= t0_e; tq = t_e[me] - t0_e; mq = t_e >= T_e - 1000      # ventana para el ruido: sigma_u de (u con ruido - u sin ruido)
res_e, tab_e = {}, {}
for th in (0.0, TH):                                              # primero sin retardo (diseno), luego theta = 50 s
    for nom in ('ADRC', 'PID'):
        Y0, U0, _ = sim_c(nom, T_e, r_e, eps_e, theta=th)
        Yn, Un, _ = sim_c(nom, T_e, r_e, eps_e, theta=th, sigma=SIG_RUIDO)
        res_e[(nom, th)] = (Y0, U0, Yn, Un)
        e = Y0[me] - r1_e; fuera = np.flatnonzero(np.abs(e) > 0.02 * dr_e)
        tab_e[(nom, th)] = dict(IAE=np.sum(np.abs(e)) * dt_c, ITAE=np.sum(tq * np.abs(e)) * dt_c,
                                Mp=100 * (Y0.max() - r1_e) / dr_e, ts=(tq[fuera[-1]] if len(fuera) else 0.0),
                                umax=U0.max(), umin=U0.min(), tsat=np.mean((U0 <= 0) | (U0 >= 100)) * T_e,
                                TV=np.sum(np.abs(np.diff(U0))), sig_u=np.std((Un - U0)[mq]), TVn=np.sum(np.abs(np.diff(Un[me]))))
print('%-5s %5s %8s %9s %6s %7s %6s %6s %7s %7s %8s %9s' % ('', 'theta', 'IAE', 'ITAE', 'Mp%', 'ts[s]', 'umax', 'umin', 'tsat[s]', 'TV[%]', 'sig_u[%]', 'TVruido[%]'))
for (nom, th), d in tab_e.items():
    print('%-5s %5.0f %8.2f %9.0f %6.1f %7.0f %6.1f %6.1f %7.1f %7.0f %8.4f %9.0f' % (nom, th, d['IAE'], d['ITAE'], d['Mp'], d['ts'], d['umax'], d['umin'], d['tsat'], d['TV'], d['sig_u'], d['TVn']))

# Figura: sin retardo y con theta = 50 s, los dos con el ruido del sensor
fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.5), layout='constrained', sharex=True)
for j, th in enumerate((0.0, TH)):
    for nom, col, lw in (('ADRC', 'k', 1.3), ('PID', 'C0', 0.9)):
        Y0, U0, Yn, Un = res_e[(nom, th)]
        axs[0, j].plot(t_e - t0_e, Yn, color=col, lw=lw, label=nom)
        axs[1, j].plot(t_e - t0_e, Un, color=col, lw=0.6, alpha=0.8, label=nom)
    axs[0, j].plot(t_e - t0_e, np.where(t_e >= t0_e, r1_e, 2.0), color='0.5', ls='--', lw=1.2, label='$r$')
    axs[0, j].set_ylabel('OD [mg/L]')
    axs[1, j].axhline(100, color='0.4', ls=':', lw=1); axs[1, j].axhline(0, color='0.4', ls=':', lw=1)
    axs[1, j].set_ylim(-8, 110); axs[1, j].set_ylabel('$u$ [%]'); axs[1, j].set_xlabel('tiempo desde el escalón [s]')
    txt = 'sin retardo' if th == 0 else '$\\theta$ = %.0f s' % th
    axs[0, j].set_title('%s · Salida, %s' % ('ab'[j], txt)); axs[1, j].set_title('%s · Señal de control, %s' % ('cd'[j], txt))
for ax in axs.flat:
    ax.set_xlim(-200, 2000); ax.grid(True, alpha=0.3); ax.legend(fontsize=8, frameon=False); eje_minutos(ax)
fig.suptitle('PID ISA contra ADRC: escalón de 2,0 a 2,2 mg/L con el ruido del sensor', fontsize=13, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/pid_adrc_escalon.pdf', bbox_inches='tight')
plt.show()
```

### 7.2 Perturbación de entrada con el sistema en estacionario

Con el sistema en estacionario en 2,0 mg/L, a los 200 s entra una perturbación de entrada $\varepsilon = -15$ %, una caída
de la eficiencia del soplador, sin retardo y con $\theta = 50$ s y sin ruido. Se mide el IAE, la caída máxima del OD, el
tiempo hasta volver a $\pm 0{,}02$ mg/L (1 % de la consigna) y cuánto se pasa al recuperarse, en % de la caída.

```python
# ============================================================
# 7.2 Escalon de perturbacion de entrada eps = -15 % con el sistema en estacionario (2,0 mg/L), sin y con retardo
# ============================================================
T_p, t0_p, EPS = 4000.0, 200.0, -15.0
t_p = np.arange(int(T_p / dt_c)) * dt_c; eps_p = np.where(t_p >= t0_p, EPS, 0.0)
res_p, tab_p = {}, {}
for th in (0.0, TH):
    for nom in ('ADRC', 'PID'):
        Y0, U0, _ = sim_c(nom, T_p, lambda t: 2.0, eps_p, theta=th)
        Yn, Un, _ = sim_c(nom, T_p, lambda t: 2.0, eps_p, theta=th, sigma=SIG_RUIDO)   # para la figura
        res_p[(nom, th)] = (Y0, U0, Yn, Un)
        m = t_p >= t0_p; e = Y0[m] - 2.0; tq = t_p[m] - t0_p; fuera = np.flatnonzero(np.abs(e) > 0.02)
        tab_p[(nom, th)] = dict(IAE=np.sum(np.abs(e)) * dt_c, caida=-e.min(), t_caida=tq[np.argmin(e)],
                                trec=(tq[fuera[-1]] if len(fuera) else 0.0), umax=U0.max(), sobre=100 * max(0.0, e.max()) / -e.min())
print('%-5s %5s %8s %10s %9s %9s %7s %9s' % ('', 'theta', 'IAE', 'caida', 't_caida', 't_rec', 'umax', 'sobre[%]'))
for (nom, th), d in tab_p.items():
    print('%-5s %5.0f %8.2f %10.4f %9.0f %9.0f %7.1f %9.1f' % (nom, th, d['IAE'], d['caida'], d['t_caida'], d['trec'], d['umax'], d['sobre']))
fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.5), layout='constrained', sharex=True)
for j, th in enumerate((0.0, TH)):
    for nom, col, lw in (('ADRC', 'k', 1.6), ('PID', 'C0', 1.1)):
        _, _, Yn, Un = res_p[(nom, th)]                    # indices sin ruido; la figura, con el ruido del sensor
        axs[0, j].plot(t_p - t0_p, Yn, color=col, lw=lw, label=nom); axs[1, j].plot(t_p - t0_p, Un, color=col, lw=0.6, alpha=0.8, label=nom)
    axs[0, j].axhline(2.0, color='0.5', ls='--', lw=1.2, label='$r$'); axs[0, j].set_ylabel('OD [mg/L]')
    axs[1, j].axhline(100, color='0.4', ls=':', lw=1); axs[1, j].axhline(0, color='0.4', ls=':', lw=1)
    axs[1, j].set_ylim(-8, 110); axs[1, j].set_ylabel('$u$ [%]'); axs[1, j].set_xlabel('tiempo desde la perturbación [s]')
    txt = 'sin retardo' if th == 0 else '$\\theta$ = %.0f s' % th
    axs[0, j].set_title('%s · Salida, %s' % ('ab'[j], txt)); axs[1, j].set_title('%s · Señal de control, %s' % ('cd'[j], txt))
for ax in axs.flat:
    ax.set_xlim(-200, 3000); ax.grid(True, alpha=0.3); ax.legend(fontsize=8, frameon=False); eje_minutos(ax)
fig.suptitle('PID ISA contra ADRC: perturbación de entrada $\\varepsilon$ = %.0f %% con el sistema en estacionario y el ruido del sensor' % EPS, fontsize=13, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/pid_adrc_perturbacion_escalon.pdf', bbox_inches='tight')
plt.show()
```

### 7.2b Un PID que rechace como el ADRC

Para que el PID rechace la perturbación de entrada como el ADRC, sin perder la misma $M_s$ sin retardo, hay que subir $K_p$,
y con él el salto inicial. Se repite la sintonía con un salto de 80 % en vez de 40 % y se compara, con $\theta = 50$ s, en el
escalón de consigna y en la perturbación de entrada.

```python
# ============================================================
# 7.2b PID con salto inicial de 80 %: rechaza como el ADRC. Misma M_s sin retardo, Ti = 4 Td, mismo N
# ============================================================
Kp_r = 80.0 / ((1 + N_pid) * DR)
Td_r = brentq(lambda Td: robustez_c(C_pid(Kp_r, 4 * Td, Td), 0)['Ms'] - R_a0['Ms'], 5, 60, xtol=1e-6); Ti_r = 4 * Td_r
def sim_pid(Kp, Ti, Td, *a, **k):                    # sim_c con otras ganancias del PID
    global Kp_p, Ti_p, Td_p
    g = (Kp_p, Ti_p, Td_p); Kp_p, Ti_p, Td_p = Kp, Ti, Td
    try: return sim_c('PID', *a, **k)
    finally: Kp_p, Ti_p, Td_p = g
print('PID con salto de 80 %%:  Kp = %.2f   Ti = %.2f s   Td = %.2f s   N = %g' % (Kp_r, Ti_r, Td_r, N_pid))
for th in (0.0, TH):
    d = robustez_c(C_pid(Kp_r, Ti_r, Td_r), th)
    print('   theta = %2.0f s:  Ms = %.3f   MF = %5.1f grados   MG = %s' % (th, d['Ms'], d['PM'], 'inf' if np.isinf(d['GM']) else '%.2f' % d['GM']))
CTR = (('ADRC', 'ADRC', 'k', 1.6), ('PID', 'PID (salto de 40 %)', 'C0', 1.1), ('PIDr', 'PID con salto de 80 %', 'C3', 1.1))
def corre(nom, *a, **k):
    return sim_c(nom, *a, **k) if nom != 'PIDr' else sim_pid(Kp_r, Ti_r, Td_r, *a, **k)
res_r = {}
for nom, *_ in CTR:
    Ye, Ue, _ = corre(nom, T_e, r_e, eps_e, theta=TH); Yn, Un, _ = corre(nom, T_e, r_e, eps_e, theta=TH, sigma=SIG_RUIDO)
    Yp, Up, _ = corre(nom, T_p, lambda t: 2.0, eps_p, theta=TH)
    res_r[nom] = (Ye, Ue, Yp, Up)
    e = Ye[me] - r1_e; fuera = np.flatnonzero(np.abs(e) > 0.02 * dr_e); ep = Yp[t_p >= t0_p] - 2.0
    print('%-5s escalon: IAE %5.1f  Mp %5.1f %%  ts %5.0f s  umax %5.1f  tsat %4.1f s | perturbacion: IAE %5.1f  caida %.3f  sobrepaso %4.1f %% | sig_u %.3f %%'
          % (nom, np.sum(np.abs(e)) * dt_c, 100 * (Ye.max() - r1_e) / dr_e, tq[fuera[-1]], Ue.max(), np.sum((Ue <= 0) | (Ue >= 100)) * dt_c,
             np.sum(np.abs(ep)) * dt_c, -ep.min(), 100 * max(0.0, ep.max()) / -ep.min(), np.std((Un - Ue)[mq])))
fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.5), layout='constrained', sharex='col')
for nom, lab, col, lw in CTR:
    Ye, Ue, Yp, Up = res_r[nom]
    axs[0, 0].plot(t_e - t0_e, Ye, color=col, lw=lw, label=lab); axs[1, 0].plot(t_e - t0_e, Ue, color=col, lw=lw, label=lab)
    axs[0, 1].plot(t_p - t0_p, Yp, color=col, lw=lw, label=lab); axs[1, 1].plot(t_p - t0_p, Up, color=col, lw=lw, label=lab)
axs[0, 0].plot(t_e - t0_e, np.where(t_e >= t0_e, r1_e, 2.0), color='0.5', ls='--', lw=1.2, label='$r$')
axs[0, 1].axhline(2.0, color='0.5', ls='--', lw=1.2, label='$r$')
for ax in axs[1]:
    ax.axhline(100, color='0.4', ls=':', lw=1); ax.axhline(0, color='0.4', ls=':', lw=1); ax.set_ylim(-8, 110); ax.set_ylabel('$u$ [%]')
for ax in axs[0]: ax.set_ylabel('OD [mg/L]')
axs[0, 0].set_title('a · Salida, escalón de consigna'); axs[0, 1].set_title('b · Salida, perturbación $\\varepsilon$ = %.0f %%' % EPS)
axs[1, 0].set_title('c · Señal de control, escalón de consigna'); axs[1, 1].set_title('d · Señal de control, perturbación $\\varepsilon$ = %.0f %%' % EPS)
axs[1, 0].set_xlabel('tiempo desde el escalón [s]'); axs[1, 1].set_xlabel('tiempo desde la perturbación [s]')
for ax in axs[:, 0]: ax.set_xlim(-200, 2000)
for ax in axs[:, 1]: ax.set_xlim(-200, 3000)
for ax in axs.flat: ax.grid(True, alpha=0.3); ax.legend(fontsize=8, frameon=False)
fig.suptitle('PID con más rechazo: salto inicial de 80 %% contra 40 %%, $\\theta$ = %.0f s' % TH, fontsize=13, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/pid_rechazo_mejorado.pdf', bbox_inches='tight')
plt.show()
```

### 7.3 Perturbación medida

La perturbación total medida de M2 del capítulo 2, $\bar\xi[k] = (y[k+1]-y[k-1])/(2T_s) - b_0\,u[k-1]$
(`Datos/analisis/xi_medida_M2_1min.csv`, en mg/(L·h), sin filtro, 12 h desde las 09:30 del 1-may-2025), entra como
perturbación de entrada en % de $u$, $\varepsilon = (\bar\xi/3600 + y_{med}/\tau)/b_0$, con $y_{med}$ el OD medido. En ese
tramo el soplador de la planta no se satura al principio, se queda en su máximo de la hora 1,5 a la 6 y luego vuelve. Con $\theta = 50$ s y el soplador limitado a 0-100 %: arranque desde
0 mg/L (soplador apagado) hasta 2,0 mg/L, cambio a 2,2 mg/L a las 4 h y a 1,8 mg/L a las 8 h. Se corre sin
ruido y con el ruido del sensor, y se dibuja la perturbación que estima el observador del ADRC,
$3600\,(b_0\hat\varepsilon - y_{med}/\tau)$, en las mismas unidades que $\bar\xi$.

```python
# ============================================================
# 7.3 Perturbacion medida de M2 (capitulo 2): 12 h desde las 09:30 del 1-may-2025
#   xi = derivada centrada del OD - b0 u, a 1 min, sin filtro (cuaderno analisis_dinamica_limpio, celda 70)
#   theta = 50 s, soplador 0-100 %, sin ruido y con el ruido del sensor
#   consigna: arranque 0 -> 2,0, a las 4 h 2,2 y a las 8 h 1,8
# ============================================================
HORAS, H_CAMBIO = 12, 4
NIV_C = [2.0, 2.2, 1.8]
Fp_c = pd.read_csv('Datos/analisis/xi_medida_M2_1min.csv')                  # t_h, xi, od, u, sp
t_d = np.arange(int(HORAS * 3600 / dt_c)) * dt_c
xi_h = np.interp(t_d / 3600.0, Fp_c.t_h.to_numpy(), Fp_c.xi.to_numpy())     # mg/(L*h), medida
y_m = np.interp(t_d / 3600.0, Fp_c.t_h.to_numpy(), Fp_c.od.to_numpy())      # OD medido [mg/L]
eps_d = (xi_h / 3600.0 + y_m / Tau) / bo                                     # % de u
r_d = lambda t: NIV_C[min(int(t // (H_CAMBIO * 3600)), len(NIV_C) - 1)]
r_dv = np.array(NIV_C)[np.minimum(t_d // (H_CAMBIO * 3600), len(NIV_C) - 1).astype(int)]
CASOS = (('sin ruido', 0.0), ('con ruido', SIG_RUIDO))
res_d = {(nom, caso): sim_c(nom, HORAS * 3600, r_d, eps_d, sigma=sg, seed=2, y0=0.0)
         for nom in ('ADRC', 'PID') for caso, sg in CASOS}

cambios = [(0.0, 0.0, NIV_C[0])] + [(j * H_CAMBIO * 3600.0, NIV_C[j - 1], NIV_C[j]) for j in range(1, len(NIV_C))]
def indices_cambio(Y, t0, r0, r1, dur=2 * 3600):
    m = (t_d >= t0) & (t_d < t0 + dur); e = Y[m] - r1; sg = np.sign(r1 - r0)
    return dict(IAE=np.sum(np.abs(e)) * dt_c, Mp=100 * max(0.0, np.max(sg * e)) / abs(r1 - r0))
print('Por cambio de consigna (ventana de 2 h):')
print('%-18s | %9s %9s | %9s %9s | %7s %7s' % ('cambio', 'IAE sin A', 'IAE sin P', 'IAE con A', 'IAE con P', 'Mp A', 'Mp P'))
for t0, r0, r1 in cambios:
    f = {(nom, caso): indices_cambio(res_d[(nom, caso)][0], t0, r0, r1) for nom in ('ADRC', 'PID') for caso, _ in CASOS}
    print('%4.1f -> %3.1f (%2.0f h)   | %9.1f %9.1f | %9.1f %9.1f | %6.1f%% %6.1f%%'
          % (r0, r1, t0 / 3600, f[('ADRC', 'sin ruido')]['IAE'], f[('PID', 'sin ruido')]['IAE'],
             f[('ADRC', 'con ruido')]['IAE'], f[('PID', 'con ruido')]['IAE'],
             f[('ADRC', 'sin ruido')]['Mp'], f[('PID', 'sin ruido')]['Mp']))
reg = t_d >= 2 * 3600                                                          # despues del arranque
print('\nVentana de %d h (RMS y max|e| desde las 2 h):' % HORAS)
print('%-5s %-10s %12s %10s %8s %13s %10s %7s' % ('', '', 'IAE[mg h/L]', 'RMS[mg/L]', 'max|e|', 'sdu/v2 [%]', 'TV/h[%]', 'sat%'))
for (nom, caso), (Y, U, XI) in res_d.items():
    e = Y - r_dv
    print('%-5s %-10s %12.3f %10.4f %8.3f %13.4f %10.0f %7.2f'
          % (nom, caso, np.sum(np.abs(e)) * dt_c / 3600, np.sqrt(np.mean(e[reg]**2)), np.abs(e[reg]).max(),
             np.std(np.diff(U)) / np.sqrt(2), np.sum(np.abs(np.diff(U))) / HORAS, 100 * np.mean((U <= 0) | (U >= 100))))
for caso, _ in CASOS:                                                          # la perturbacion estimada, en mg/(L*h)
    xi_est = 3600.0 * (bo * res_d[('ADRC', caso)][2] - y_m / Tau)
    print('estimacion de xi (%s, desde las 2 h): error RMS %.2f mg/(L*h)' % (caso, np.sqrt(np.mean((xi_est - xi_h)[reg]**2))))

hh = t_d / 3600.0; sub = slice(None, None, 20)                                 # 1 s para dibujar
caso = 'con ruido'                                                              # la figura, solo con el ruido del sensor
fig, axs = plt.subplots(3, 1, figsize=(12.5, 8.4), layout='constrained', sharex=True)
xi_est = 3600.0 * (bo * res_d[('ADRC', caso)][2] - y_m / Tau)
for nom, col, lw, al in (('PID', 'C0', 0.6, 0.7), ('ADRC', 'k', 0.8, 0.9)):
    Y, U, _ = res_d[(nom, caso)]
    axs[0].plot(hh[sub], Y[sub], color=col, lw=lw, alpha=al, label=nom)
    axs[1].plot(hh[sub], U[sub], color=col, lw=lw, alpha=al, label=nom)
axs[0].plot(hh[sub], r_dv[sub], color='0.5', ls='--', lw=1.2, label='$r$')
axs[0].set_title('a · Salida'); axs[0].set_ylabel('OD [mg/L]')
axs[1].set_title('b · Señal de control'); axs[1].set_ylabel('$u$ [%]')
axs[1].axhline(100, color='0.4', ls=':', lw=1); axs[1].axhline(0, color='0.4', ls=':', lw=1)
axs[1].set_ylim(-8, 110)
axs[2].plot(hh[sub], xi_h[sub], color='0.5', lw=2.5, alpha=0.5, label='$\\bar{\\xi}$  perturbación total medida en M2 (1 min)')
axs[2].plot(hh[sub], xi_est[sub], color='k', lw=0.7, label='estimada por el ADRC, llevada a $\\bar{\\xi}$ (incluye $-\\dot{r}$)')
axs[2].set_ylim(-30, 0); axs[2].set_ylabel('$\\bar{\\xi}$ [mg/(L·h)]'); axs[2].set_title('c · Perturbación')
axs[2].set_xlabel('tiempo [h]')
for ax, loc in zip(axs, ('lower right', 'upper right', 'lower right')):
    ax.set_xticks(np.arange(0, HORAS + 1, 2)); ax.grid(True, alpha=0.3); ax.legend(fontsize=8, frameon=False, loc=loc)
fig.suptitle('PID ISA contra ADRC con la perturbación medida de M2, %d horas, $\\theta$ = 50 s, con ruido' % HORAS,
             fontsize=13, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/pid_adrc_perturbacion.pdf', bbox_inches='tight')
plt.show()
```

### 7.3b Referencia acondicionada con la perturbación medida

El mismo escenario de 7.3 (12 h, $\theta = 50$ s, ruido del sensor, consigna $0 \to 2{,}0 \to 2{,}2 \to 1{,}8$ mg/L), pero con el ADRC
recibiendo la referencia acondicionada: con la curva de Bézier del bloque `BEZIER_REF` ($T_F$ = 40 min de 0 a 2 y 20 min en los
otros cambios) y con el prefiltro del bloque `PREFILTER_ADRC`. Se compara el sobrepico y el esfuerzo en cada cambio, y el rechazo de la
perturbación cuando la referencia ya llegó a la consigna (figura `adrc_acond_perturbacion.pdf` de la tesis).

```python
# ============================================================
# 7.3b ADRC con la referencia acondicionada, con la perturbacion medida
#   Mismo escenario de 7.3: 12 h desde las 09:30 del 1-may-2025, theta = 50 s, soplador 0-100 %,
#   ruido del sensor; consigna 0 -> 2,0, a las 4 h 2,2 y a las 8 h 1,8.
#   Bezier como el bloque BEZIER_REF: segmento cubico desde el valor y la pendiente actuales de la
#     referencia hasta la consigna, con pendiente final cero; TF = 40 min de 0 a 2 y 20 min en los
#     otros cambios (los mismos de las trazas del PLC).
#   Prefiltro como el bloque PREFILTER_ADRC: k0/(k1 s + k0), k1/k0 = (wo + 2 wc)/(wc wo), Tustin a dt_c,
#     arranca en la medida (0 mg/L).
# ============================================================
CAMBIOS_R = [(0.0, NIV_C[0]), (H_CAMBIO * 3600.0, NIV_C[1]), (2 * H_CAMBIO * 3600.0, NIV_C[2])]
TF_R = [2400.0, 1200.0, 1200.0]

def ref_bezier(t, cambios, tfs, y_ini=0.0):
    """Referencia del bloque BEZIER_REF: en cada cambio replanifica desde (y*, dy*/dt) actuales."""
    ys = np.full(len(t), y_ini); yc, dyc = y_ini, 0.0
    for j, ((t0, r), tf) in enumerate(zip(cambios, tfs)):
        t1 = cambios[j + 1][0] if j + 1 < len(cambios) else np.inf
        m = (t >= t0) & (t < t1); tau = (t[m] - t0) / tf
        D = r - yc; c1, c2, c3 = dyc * tf, 3 * D - 2 * dyc * tf, -2 * D + dyc * tf
        ys[m] = np.where(tau < 1, yc + tau * (c1 + tau * (c2 + tau * c3)), r)
        tau1 = (t1 - t0) / tf if np.isfinite(t1) else 1.0                  # estado al llegar el siguiente cambio
        if tau1 < 1:
            yc, dyc = yc + tau1 * (c1 + tau1 * (c2 + tau1 * c3)), (c1 + tau1 * (2 * c2 + 3 * tau1 * c3)) / tf
        else:
            yc, dyc = r, 0.0
    return ys

def ref_prefiltro(r_esc, y_ini=0.0):
    """Referencia del bloque PREFILTER_ADRC (Tustin), arrancando en la medida."""
    Q = 2 * (wo_c + 2 * wc_c); a_f = (wc_c * wo_c * dt_c - Q) / (wc_c * wo_c * dt_c + Q)
    b_f = wc_c * wo_c * dt_c / (wc_c * wo_c * dt_c + Q)
    rf = np.empty(len(r_esc)); y1, r1 = y_ini, y_ini
    for k, r in enumerate(r_esc):
        y1 = -a_f * y1 + b_f * (r + r1); r1 = r; rf[k] = y1
    return rf

REF = {'Bézier': ref_bezier(t_d, CAMBIOS_R, TF_R), 'prefiltro': ref_prefiltro(r_dv)}
res_a = {nom: sim_c('ADRC', HORAS * 3600, (lambda t, R=R: R[min(int(round(t / dt_c)), len(R) - 1)]), eps_d,
                    sigma=SIG_RUIDO, seed=2, y0=0.0) for nom, R in REF.items()}
res_a['sin acondicionar'] = res_d[('ADRC', 'con ruido')]

print('Por cambio de consigna (ventana de 2 h), contra la consigna en escalones r:')
print('%-18s | %-16s %7s %8s %9s' % ('cambio', 'referencia', 'Mp[%]', 'u max[%]', 'IAE[mg s/L]'))
for t0, r0, r1 in cambios:
    m = (t_d >= t0) & (t_d < t0 + 2 * 3600); sg = np.sign(r1 - r0)
    for nom in ('sin acondicionar', 'Bézier', 'prefiltro'):
        Y, U, _ = res_a[nom]; e = Y[m] - r1
        print('%4.1f -> %3.1f (%2.0f h)   | %-16s %7.1f %8.1f %9.0f'
              % (r0, r1, t0 / 3600, nom, 100 * max(0.0, np.max(sg * e)) / abs(r1 - r0), U[m].max(), np.sum(np.abs(e)) * dt_c))
# rechazo: desde 1 h despues de cada cambio hasta el siguiente, donde las tres referencias ya llegaron a r
reg_r = np.zeros(len(t_d), bool)
for j, (t0, _) in enumerate(CAMBIOS_R):
    reg_r |= (t_d >= t0 + 3600) & (t_d < (CAMBIOS_R[j + 1][0] if j + 1 < len(CAMBIOS_R) else np.inf))
print('\nRechazo (1 h despues de cada cambio hasta el siguiente):')
for nom in ('sin acondicionar', 'Bézier', 'prefiltro'):
    Y, U, _ = res_a[nom]; e = (Y - r_dv)[reg_r]
    print('%-16s RMS %.4f mg/L   max|e| %.3f mg/L' % (nom, np.sqrt(np.mean(e**2)), np.abs(e).max()))

fig, axs = plt.subplots(3, 1, figsize=(12.5, 8.4), layout='constrained', sharex=True)
for nom, col, lw, al in (('Bézier', 'C0', 0.7, 0.8), ('prefiltro', 'k', 0.8, 0.9)):
    Y, U, XI = res_a[nom]
    axs[0].plot(hh[sub], Y[sub], color=col, lw=lw, alpha=al, label='ADRC con %s' % nom)
    axs[1].plot(hh[sub], U[sub], color=col, lw=lw, alpha=al, label='ADRC con %s' % nom)
axs[0].plot(hh[sub], REF['Bézier'][sub], color='C0', ls=':', lw=1.4, label='referencia de Bézier $y^*$')
axs[0].plot(hh[sub], REF['prefiltro'][sub], color='k', ls=':', lw=1.4, label='referencia del prefiltro $r_f$')
axs[0].plot(hh[sub], r_dv[sub], color='0.5', ls='--', lw=1.2, label='$r$')
axs[0].set_title('a · Salida'); axs[0].set_ylabel('OD [mg/L]')
axs[1].set_title('b · Señal de control'); axs[1].set_ylabel('$u$ [%]')
axs[1].axhline(100, color='0.4', ls=':', lw=1); axs[1].axhline(0, color='0.4', ls=':', lw=1); axs[1].set_ylim(-8, 110)
axs[2].plot(hh[sub], xi_h[sub], color='0.3', lw=1.2, label='$\\bar{\\xi}$  perturbación total medida en M2 (1 min)')
axs[2].set_ylim(-30, 0); axs[2].set_ylabel('$\\bar{\\xi}$ [mg/(L·h)]'); axs[2].set_title('c · Perturbación'); axs[2].set_xlabel('tiempo [h]')
for ax, loc in zip(axs, ('lower right', 'upper right', 'lower right')):
    ax.set_xticks(np.arange(0, HORAS + 1, 2)); ax.grid(True, alpha=0.3); ax.legend(fontsize=8, frameon=False, loc=loc)
fig.suptitle('ADRC con Bézier y con prefiltro, perturbación medida de M2, %d horas, $\\theta$ = 50 s, con ruido' % HORAS,
             fontsize=13, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/adrc_acond_perturbacion.pdf', bbox_inches='tight')
plt.show()
```

### 7.4 Comparación de desempeños

Con el mismo procedimiento de sintonía se busca un controlador con mejor desempeño que el ADRC, sin que la señal de control
llegue al 100 %:
- **PI** ($T_d = 0$): primero $K_p$ para que el salto inicial sea de 40 %; sin derivada, todo el salto es proporcional, $K_p = 40/0{,}2 = 200$ %/(mg/L).
  Luego $T_i$ para que tenga el mismo margen de fase sin retardo que el ADRC. Con la misma $M_s$ el PI queda casi inestable con $\theta = 50$ s.

```python
# ============================================================
# 7.4 Comparacion de desempenos, sin llegar al 100 %: PI con salto de 40 % y el MF del ADRC
# ============================================================
import itertools
C_pidN = lambda Kp, Ti, Td, N: Kp * (1 + 1 / (Ti * s_c) + (Td * s_c / (1 + Td * s_c / N) if Td > 0 else 0))
def sim_pidN(Kp, Ti, Td, N, *a, **k):                       # sim_pid con otro N
    global N_pid
    g = N_pid; N_pid = N
    try: return sim_pid(Kp, Ti, Td, *a, **k)
    finally: N_pid = g
def indices_m(par, th, sigma=0.0):
    Ye, Ue, _ = sim_pidN(*par, T_e, r_e, eps_e, theta=th, sigma=sigma) if par else sim_c('ADRC', T_e, r_e, eps_e, theta=th, sigma=sigma)
    Yp, Up, _ = sim_pidN(*par, T_p, lambda t: 2.0, eps_p, theta=th, sigma=sigma) if par else sim_c('ADRC', T_p, lambda t: 2.0, eps_p, theta=th, sigma=sigma)
    return Ye, Ue, Yp, Up
# PI: salto de 40 % y el MF del ADRC
Kp_pi = DU0 / DR
Ti_pi = brentq(lambda Ti: robustez_c(C_pid(Kp_pi, Ti, 0.0), 0)['PM'] - R_a0['PM'], 30, 3000, xtol=1e-6)
Ti_pi_ms = brentq(lambda Ti: robustez_c(C_pid(Kp_pi, Ti, 0.0), 0)['Ms'] - R_a0['Ms'], 30, 1000, xtol=1e-6)
print('PI con salto de 40 %% y el MF del ADRC:  Kp = %.0f   Ti = %.1f s' % (Kp_pi, Ti_pi))
print('   (con la misma Ms seria Ti = %.0f s, y con theta = 50 s: Ms = %.2f, MF = %.1f grados)'
      % (Ti_pi_ms, robustez_c(C_pid(Kp_pi, Ti_pi_ms, 0.0), TH)['Ms'], robustez_c(C_pid(Kp_pi, Ti_pi_ms, 0.0), TH)['PM']))
MEJ = (('ADRC', 'ADRC', None, 'k', 1.6), ('PID', 'PID (salto de 40 %)', (Kp_p, Ti_p, Td_p, N_pid), 'C0', 1.1),
       ('PI', 'PI', (Kp_pi, Ti_pi, 0.0, 1.0), 'C3', 1.3))
res_m, tab_m = {}, {}
print('%-6s %5s | %5s %5s %5s | %6s %7s %6s %5s %5s | %6s %6s | %7s' % ('', 'theta', 'Ms', 'MF', 'MG', 'IAEc', 'ITAEc', 'Mp', 'umax', 'tsat', 'IAEr', 'caida', 'sig_u'))
for nom, lab, par, col, lw in MEJ:
    Cm = C_adrc if par is None else C_pidN(*par)
    for th in (0.0, TH):
        Ye, Ue, Yp, Up = indices_m(par, th); Yen, Uen, Ypn, Upn = indices_m(par, th, sigma=SIG_RUIDO)
        res_m[(nom, th)] = (Yen, Uen, Ypn, Upn)
        d = robustez_c(Cm, th); e = Ye[me] - r1_e; ep = Yp[t_p >= t0_p] - 2.0
        tab_m[(nom, th)] = dict(Ms=d['Ms'], MF=d['PM'], MG=d['GM'], IAEc=np.sum(np.abs(e)) * dt_c, ITAEc=np.sum((t_e[me] - t0_e) * np.abs(e)) * dt_c, Mp=100 * (Ye.max() - r1_e) / dr_e,
                                umax=max(Ue.max(), Up.max()), tsat=np.sum((Ue <= 0) | (Ue >= 100)) * dt_c, IAEr=np.sum(np.abs(ep)) * dt_c,
                                caida=-ep.min(), sig_u=np.std((Uen - Ue)[mq]))
        t_ = tab_m[(nom, th)]
        print('%-6s %5.0f | %5.2f %5.1f %5s | %6.1f %7.0f %6.1f %5.1f %5.1f | %6.1f %6.3f | %7.3f' % (nom, th, t_['Ms'], t_['MF'], 'inf' if np.isinf(t_['MG']) else '%.2f' % t_['MG'],
              t_['IAEc'], t_['ITAEc'], t_['Mp'], t_['umax'], t_['tsat'], t_['IAEr'], t_['caida'], t_['sig_u']))
# figura en el tiempo, sin retardo y con theta = 50 s, con el ruido del sensor
fig, axs = plt.subplots(2, 4, figsize=(17, 7.5), layout='constrained', sharex='col')
for nom, lab, par, col, lw in MEJ:
    for j, th in enumerate((0.0, TH)):
        Yen, Uen, Ypn, Upn = res_m[(nom, th)]
        axs[0, j].plot(t_e - t0_e, Yen, color=col, lw=lw, label=lab); axs[1, j].plot(t_e - t0_e, Uen, color=col, lw=0.6, alpha=0.85, label=lab)
        axs[0, 2 + j].plot(t_p - t0_p, Ypn, color=col, lw=lw, label=lab); axs[1, 2 + j].plot(t_p - t0_p, Upn, color=col, lw=0.6, alpha=0.85, label=lab)
for j, th in enumerate((0.0, TH)):
    txt = 'sin retardo' if th == 0 else '$\\theta$ = %.0f s' % th
    axs[0, j].plot(t_e - t0_e, np.where(t_e >= t0_e, r1_e, 2.0), color='0.5', ls='--', lw=1.2, label='$r$')
    axs[0, 2 + j].axhline(2.0, color='0.5', ls='--', lw=1.2, label='$r$')
    axs[0, j].set_title('%s · Salida, escalón, %s' % ('ab'[j], txt)); axs[1, j].set_title('%s · Señal de control, escalón, %s' % ('ef'[j], txt))
    axs[0, 2 + j].set_title('%s · Salida, perturbación, %s' % ('cd'[j], txt)); axs[1, 2 + j].set_title('%s · Señal de control, perturbación, %s' % ('gh'[j], txt))
    axs[1, j].set_xlabel('tiempo desde el escalón [s]'); axs[1, 2 + j].set_xlabel('tiempo desde la perturbación [s]')
for ax in axs[1]: ax.axhline(100, color='0.4', ls=':', lw=1); ax.set_ylim(20, 105)
for ax in axs[:, :2].flat: ax.set_xlim(-200, 2000)
for ax in axs[:, 2:].flat: ax.set_xlim(-200, 3000)
axs[0, 0].set_ylabel('OD [mg/L]'); axs[1, 0].set_ylabel('$u$ [%]')
for ax in axs.flat: ax.grid(True, alpha=0.3); ax.legend(fontsize=7, frameon=False); eje_minutos(ax)
fig.suptitle('Comparación de desempeños', fontsize=13, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/pid_mejores.pdf', bbox_inches='tight')
plt.show()
# sensibilidad de los cuatro, sin y con retardo
fig, axs = plt.subplots(1, 2, figsize=(12.5, 4.2), layout='constrained', sharey=True)
for j, th in enumerate((0.0, TH)):
    for nom, lab, par, col, lw in MEJ:
        Cm = C_adrc if par is None else C_pidN(*par)
        axs[j].semilogx(w_c * 3600, 20 * np.log10(np.abs(1 / (1 + Cm * G_c(th)))), color=col, lw=lw, label=lab)
    axs[j].set_title('%s · Sensibilidad $|S(j\\omega)|$, %s' % ('ab'[j], 'sin retardo' if th == 0 else '$\\theta$ = %.0f s' % th))
    axs[j].axhline(0, color='0.6', lw=0.8); axs[j].set_xlim(1e-1, 1e3); axs[j].set_ylim(-40, 15)
    axs[j].set_xlabel('$\\omega$ [rad/h]'); axs[j].grid(True, which='both', alpha=0.3); axs[j].legend(fontsize=8, frameon=False)
axs[0].set_ylabel('magnitud [dB]')
fig.savefig('Latex/Tesis/figuras/pid_mejores_sensibilidad.pdf', bbox_inches='tight')
plt.show()
```

### Prefiltro $k_0/(k_1 s + k_0)$ con la planta (figura `adrc_prefiltro.pdf` de la tesis, Sección 5.2.2)

Escalón de consigna de 2,0 a 2,2 mg/L y perturbación de entrada $\varepsilon = -15~\%$ a los 40 min,
sin retardo y con $\theta = 50$ s, con el soplador limitado a 0-100 % y sin ruido, con y sin el
prefiltro $F(s) = k_0/(k_1 s + k_0)$. Usa el mismo simulador `sim_c` de la celda 7.1. Como el
prefiltro es de primer orden y de ganancia 1, la consigna filtrada tiene solución cerrada,
$r_f(t) = 2 + 0{,}2\,\bigl(1 - e^{-(t - t_0)\,k_0/k_1}\bigr)$.

Los índices se miden antes de la perturbación, entre el escalón y los 40 min: sobrepico $M_p$,
pico de $u$, $E_u = \int (u - u_\infty)^2\,dt$ en $\%^2\cdot$s, con $u_\infty = 2{,}2/k_{dc}$, y $t_s$ con
banda de 2 % del escalón. Después de la perturbación se mide el OD mínimo.

```python
# ============================================================
# Prefiltro k0/(k1 s + k0) con la planta: figura adrc_prefiltro.pdf (Seccion 5.2.2 de la tesis)
# ============================================================
T_pf = k1_c / k0_c                                     # constante de tiempo del prefiltro [s]
t0_pf = 300.0; T_end_pf = t0_pf + 80 * 60
t_pf = np.arange(int(round(T_end_pf / dt_c))) * dt_c
eps_pf = np.where(t_pf >= t0_pf + 40 * 60, -15.0, 0.0)
r_sin = lambda t: 2.2 if t >= t0_pf else 2.0
r_con = lambda t: 2.0 + 0.2 * (1 - math.exp(-(t - t0_pf) / T_pf)) if t >= t0_pf else 2.0
u_inf = 2.2 / K_dc                                     # orden que sostiene 2,2 mg/L [%]
ven = (t_pf >= t0_pf) & (t_pf < t0_pf + 40 * 60)       # del escalon a la perturbacion
res_pf, filas_pf = {}, []
for th in (0.0, TH):
    for caso, rf in (('sin prefiltro', r_sin), ('con prefiltro', r_con)):
        Y, U, _ = sim_c('ADRC', T_end_pf, rf, eps_pf, theta=th)
        res_pf[(caso, th)] = (Y, U)
        e = Y[ven] - 2.2; fuera = np.flatnonzero(np.abs(e) > 0.02 * 0.2)
        filas_pf.append(dict(theta=th, caso=caso, Mp=max(0.0, 100 * (Y[ven].max() - 2.2) / 0.2),
                             umax=U[ven].max(), Eu=np.sum((U[ven] - u_inf)**2) * dt_c,
                             ts=(t_pf[ven][fuera[-1]] - t0_pf) / 60 if len(fuera) else 0.0,
                             OD_min=Y[t_pf >= t0_pf + 40 * 60].min()))
print('Prefiltro k0/(k1 s + k0): k1/k0 = %.0f s (%.1f min)\n' % (T_pf, T_pf / 60))
print('%-6s %-14s %7s %8s %12s %9s %15s' % ('theta', 'caso', 'Mp [%]', 'umax [%]', 'Eu [%^2 s]', 'ts [min]', 'OD min [mg/L]'))
for f in filas_pf:
    print('%-6.0f %-14s %7.1f %8.1f %12.0f %9.1f %15.3f'
          % (f['theta'], f['caso'], f['Mp'], f['umax'], f['Eu'], f['ts'], f['OD_min']))

fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.0), layout='constrained', sharex=True)
tq = t_pf - t0_pf
for j, th in enumerate((0.0, TH)):
    for caso, col, ls, lw, lab in (('sin prefiltro', 'C3', '--', 1.3, 'sin prefiltro'),
                                   ('con prefiltro', 'k', '-', 1.6, 'con prefiltro $k_0/(k_1 s + k_0)$')):
        Y, U = res_pf[(caso, th)]
        axs[0, j].plot(tq, Y, color=col, ls=ls, lw=lw, label=lab)
        axs[1, j].plot(tq, U, color=col, ls=ls, lw=lw, label=lab)
    axs[0, j].plot(tq, np.where(tq >= 0, 2.2, 2.0), color='k', ls=':', lw=1.1)
    txt = '$\\theta$ = %.0f s' % th
    axs[0, j].set_title('%s · Salida, %s' % ('ab'[j], txt))
    axs[1, j].set_title('%s · Señal de control, %s' % ('cd'[j], txt))
    axs[0, j].set_ylim(1.98, 2.32); axs[1, j].set_ylim(35, 100)
    axs[1, j].set_xlabel('tiempo desde el cambio de consigna [s]')
axs[0, 0].set_ylabel('OD [mg/L]'); axs[1, 0].set_ylabel('$u$ [%]')
axs[0, 0].legend(fontsize=8.5, frameon=False, loc='lower right')
for ax in axs.flat:
    ax.set_xlim(-300, 80 * 60); ax.minorticks_on(); ax.grid(visible=True, which='major', alpha=0.35)
    eje_minutos(ax)
fig.suptitle('Escalón de 2,0 a 2,2 mg/L y $\\varepsilon$ = $-$15 % a los 40 min, con y sin prefiltro (saturación 0-100 %)',
             fontsize=13, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/adrc_prefiltro.pdf', bbox_inches='tight')
plt.show()
```
