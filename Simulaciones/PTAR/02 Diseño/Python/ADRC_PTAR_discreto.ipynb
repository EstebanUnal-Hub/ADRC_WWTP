### 1.1 Mass balance 
$$
\text{Acumulación}=\text{Entrada}-\text{Salida}+\text{Generación}-\text{Consumo}
$$

$$
\frac{dS}{dt}=\frac{1}{V}\sum_{i=1}^3 Q_{in,i}S_{in,i}-\frac{Q_{out}}{V}S+K_{L_a}\left(S^{*}-S\right)-r_z(S)
$$

$$
\underbrace{\frac{dS}{dt}}_{\frac{\mathrm{mg}}{\mathrm{L}\cdot\mathrm{h}}}
=
\underbrace{\frac{1}{V}\sum_{i=1}^3 Q_{in,i}S_{in,i}}_{\frac{1}{\mathrm{L}}
\left(\frac{\mathrm{L}}{\mathrm{h}}\right)
\left(\frac{\mathrm{mg}}{\mathrm{L}}\right)
=
\frac{\mathrm{mg}}{\mathrm{L}\cdot\mathrm{h}}}
-
\underbrace{\frac{Q_{out}}{V}S}_{\left(\frac{\mathrm{L}/\mathrm{h}}{\mathrm{L}}\right)
\left(\frac{\mathrm{mg}}{\mathrm{L}}\right)
=
\frac{\mathrm{mg}}{\mathrm{L}\cdot\mathrm{h}}}
+
\underbrace{K_{La}(S^{*}-S)}_{\left(\frac{1}{\mathrm{h}}\right)
\left(\frac{\mathrm{mg}}{\mathrm{L}}\right)
=
\frac{\mathrm{mg}}{\mathrm{L}\cdot\mathrm{h}}}
-
\underbrace{r_z(S)}_{\frac{\mathrm{mg}}{\mathrm{L}\cdot\mathrm{h}}}
$$
With $r_Z:$
\begin{equation}
    r_{Z} = -\frac{1 - Y_H}{Y_H}\,\rho_1 - \frac{4.57 - Y_A}{Y_A}\,\rho_3
\end{equation}

\begin{equation}
    \rho_1 = \mu_H \left(\frac{S_S}{K_S + S_S}\right)\left(\frac{S_O}{K_{O,H} + S_O}\right) X_{B,H}
\end{equation}

\begin{equation}
    \rho_3 = \mu_A \left(\frac{S_{NH}}{K_{NH} + S_{NH}}\right)\left(\frac{S_O}{K_{O,A} + S_O}\right) X_{B,A}
\end{equation}

If the oxygen transfer coefficient is considered proportional to the control input,

\begin{equation}
K_{La} \propto u
\end{equation}

then,

\begin{equation}
K_{La} = \alpha u
\end{equation}

where $\alpha$ is a proportionality constant.

Substituting into the dissolved oxygen mass balance yields

\begin{equation}
\dot S
=
\underbrace{
\frac{1}{V}\sum_{i=1}^3 Q_{in,i}S_{in,i}
-\frac{Q_{out}}{V}S
-r_z
}_{f(S,t)}
+
u\,
\underbrace{
\alpha\left(S^{*}-S\right)
}_{g(S,t)}.
\end{equation}

\begin{equation}
\dot S = f(S,t) + g(S,t)\,u
\end{equation}

where

\begin{equation}
g(S,t)=\alpha\left(S^{*}-S\right)
\end{equation}

and

\begin{equation}
f(S,t)
=
\frac{1}{V}\sum_{i=1}^3 Q_{in,i}S_{in,i}
-\frac{Q_{out}}{V}S
-r_z(S)
\end{equation}


All dynamics that differ from 
a pure integrator are grouped as a total disturbance $f$, 


\begin{equation}
    \dot{y} = f + b_0 \cdot K_{L_a}
\end{equation}

where the total disturbance $f$ is defined as:

\begin{equation}
f(S,t)
=
\frac{1}{V}\sum_{i=1}^3 Q_{in,i}S_{in,i}
-\frac{Q_{out}}{V}S
-r_z(S)+K_{L_a}\left(S^{*}-S\right)-b_0 \cdot K_{L_a}
\end{equation}

relationship between Oxygen transfer coefficient $K_{La}$ and $u$. Oxygen transfer coefficient is 
proportional to the control signal $u$ applied to the 
blowers:

\begin{equation}
    K_{L_a} \propto u
\end{equation}

\begin{equation}
     K_{L_a} = \alpha \cdot u
\end{equation}

with $k = bo \cdot \alpha$ 

$$\dot{y} = f + k \cdot u$$


The oxygen mass balance is affine in the transfer coefficient:

\begin{equation}
\dot S_O
=
\underbrace{
\frac{1}{V}\sum_{i=1}^{3} Q_{in,i}S_{O,in,i}
-\frac{Q_{out}}{V}S_O
- r_{O}(\mathbf{x})
}_{f(\mathbf{x},t)}
\;+\;
\underbrace{\left(S_O^{*}-S_O\right)}_{g(S_O)}\,K_{L_a}
\end{equation}

where $f$ collects convective transport and biological uptake, and
$g$ is the driving force for oxygen transfer, in $\mathrm{mg/L}$.

Over the operating range of the blowers,

\begin{equation}
K_{L_a} = k_a\,u,
\qquad k_a = \frac{K_{L_a}^{\max}}{100},
\qquad u \in [0,100]\,\%
\end{equation}

Substituting into the dissolved oxygen mass balance yields

\begin{equation}
\dot S_O
=
\underbrace{
\frac{1}{V}\sum_{i=1}^{3} Q_{in,i}S_{O,in,i}
-\frac{Q_{out}}{V}S_O
- r_{O}(\mathbf{x})
}_{f(\mathbf{x},t)}
\;+\;
\underbrace{k_a\left(S_O^{*}-S_O\right)}_{g(S_O)}\,u
\end{equation}

### 1.2 Simplified model
The ASM Transfer function is given by:
First order system
$$
G(s)=\frac{Y(s)}{U(s)}
=\frac{k_{dc}}{\tau s+1}
$$
In chain of integrators:
$$
s\cdot Y(s)=-\frac{Y(s)}{\tau}+\frac{U(s)\cdot k_{d_c}}{\tau}
$$

$$
\dot{y}(t)=-\frac{y(t)}{\tau}+\frac{u(t)\cdot k_{d_c}}{\tau}
$$

In disturbance form, grouping the plant's own pole into the total disturbance:
$$
\dot{y}(t)=\xi(t) + \frac{k_{d_c}}{\tau}\cdot u(t),
\qquad \xi(t):=-\frac{y(t)}{\tau}
$$

with
$$
b_0:=\frac{k_{d_c}}{\tau}
$$

$$
\dot{y}(t)=\xi(t) + b_0\cdot u(t)
$$

````python
# ============================================================
# Other functions
# ============================================================

def tf2sym(tf,symb=None):
    s = symbols(symb) if symb is not None else symbols('s')
    numTF = tf.num[0][0]   
    denTF = tf.den[0][0]
    num_symb = sum(numTF[i]*s**(len(numTF)-1-i)
                   for i in range(len(numTF)))
    den_symb = sum(denTF[i]*s**(len(denTF)-1-i)
                   for i in range(len(denTF)))
    return (num_symb / den_symb).simplify()

def sym2tf(expr,symb=None,Ts=None):

    s = symbols(symb) if symb is not None else symbols('s')
    num, den = fraction(expr)
    num_poly = Poly(num, s)
    den_poly = Poly(den, s)

    num_coeff = [float(i) for i  in num_poly.as_poly().all_coeffs()]
    den_coeff = [float(i) for i in den_poly.as_poly().all_coeffs()]
    if Ts!=None:
        return tf(num_coeff,den_coeff,Ts)
    else:
        return tf(num_coeff,den_coeff)

````

## 2. Control

### 2.1 Control Law with observer

#### 2.1.1 Continuos Time
**Now, we have to design to the ADRC**

We rewrite the second-order system as a second-order error system. Let $e_y = y - r$:

$$\dot{e}_y = k\cdot u + (\bar{\xi} -u^{*})$$

Defining $\xi = \bar{\xi} - u^{*}$:

$$\dot{e}_y = k\cdot u + \xi$$

Applying the same procedure as before, the estimation error dynamics become: We have to select to the states of the system.
$$\dot{x_1}=x_2+k\cdot u$$
$$\dot{x_2}=PMI(\xi,t)$$

**Extended Dynamics System**
$$\dot{X}_{au} = A_{au} \cdot X_{au} + B_{au} \cdot u + B_{\xi_{au}} \cdot \xi^{(m)}$$

with the state vectors defined as:

$$X_p = \begin{bmatrix} e_y  \end{bmatrix}, \qquad X_{\xi} = \begin{bmatrix} \xi \end{bmatrix}$$

and the augmented state:

$$X_{au} = \begin{bmatrix} X_p \\ X_{\xi}\end{bmatrix}$$
**Observer System**
$$\dot{\widehat{X}}_{au} = (A_{au} - LC_{au})\widehat{X}_{au} + B_{au} u + Le_y$$

with $e_y=C_{au}\cdot X_{au}$

**Estimation Error Dynamics**
$$(\dot{X}_{au} - \dot{\widehat{X}}_{au}) = (A_{au} - LC_{au})(X_{au} - \widehat{X}_{au}) + B_{{\xi}_{au}} {\xi}^{(m)}$$

$$\dot{\bar{e}} = (A_{au} - LC_{au})\bar{e} + B_{\xi_{au}} \xi^{(m)}$$





The control law is given by the following expression, where the observer provides the estimated plant and disturbance states:

\begin{equation}
    e^{(1)}=v 
\end{equation}
Ideal model:
\begin{equation}
   f_\delta^{(n)}=\beta \cdot u+\phi(f,\dot{f}_\delta,...,f^{(n-1)}_\delta) 
\end{equation}

Donde $\beta$, $f_\delta^{(n)}$ y $\phi$ corresponde a la dinámica no deseada.
Ahora bien de aquí se puede establecer la señal de control.

\begin{equation}
    u=\frac{1}{\beta}[-\phi-v(ef^{(n-1)},..,ef,f^{(n)*})]
\end{equation}



Now:


\begin{equation}
    u=\frac{1}{b_0}[-\hat{f}-v(ey^{(0)})]
\end{equation}



with $v$:
$$v = k_1 \cdot X_{au_1} + k_2 \cdot \widehat{X}_{au_2}$$


$$v=K\cdot \widehat{X}_{au}$$

#### 2.1.2 Discrte time

**System:**


\begin{equation}
    x[k+1] = A\,x[k] + b\bigl(u[k] + \xi[k]\bigr)
\end{equation}

**Disturbance system:**
\begin{align}
    x_{im}[k+1] &= A_{im}\,x_{im}[k] \\
    \xi[k]       &= C_{im}\,x_{im}
\end{align}

**Augmented system:**
\begin{equation}
    z[k+1] = A\,x[k] + b\,C_{im}\,x_{im}[k] + b\,u[k]
\end{equation}

In matrix form:
\begin{equation}
    \begin{bmatrix} x[k+1] \\ x_{im}[k+1] \end{bmatrix}
    =
    \underbrace{
    \begin{bmatrix} A & b\,C_{im} \\ 0 & A_{im} \end{bmatrix}
    }_{A_{ou}}
    \underbrace{
    \begin{bmatrix} x[k] \\ x_{im}[k] \end{bmatrix}
    }_{x_{au}}
    +
    \underbrace{
    \begin{bmatrix} b \\ 0 \end{bmatrix}
    }_{b_{au}}
    u
\end{equation}

\begin{equation}
    y = \underbrace{\begin{bmatrix} C & 0 \end{bmatrix}}_{C_{au}}
    \begin{bmatrix} x[k] \\ x_{im}[k] \end{bmatrix}
\end{equation}

**with Observer**

Augmented system:
\begin{equation}
    x_{au}[k+1] = A_{au}\,x_{u} + b_{au}\,u[k]
\end{equation}

Observer equation:
\begin{equation}
    \hat{x}_{au}[k+1] = A_{au}\,\hat{x}_{au}[k] 
    + b_{au}\,u[k] 
    + L_{au}\bigl(e[k] - \hat{e}[k]\bigr)
\end{equation}

Expanded form:
\begin{equation}
    \hat{x}_{au}[k+1] = 
    \bigl(A_{au} - L_{au}\,C_{au}\bigr)\,\hat{x}_{au} 
    + b_{au}\,u[k] 
    + L_{au}\,e[k]
\end{equation}

**Extended System**

$$
\begin{bmatrix}

\end{bmatrix}
=

\end{bmatrix}
+
\begin{bmatrix}

\end{bmatrix}

+
\begin{bmatrix}

\end{bmatrix}
\xi^{(m)}(k).
$$

The output equation is

$$
y(k)=
\begin{bmatrix}
C_{p} & 0_{(1,m)}
\end{bmatrix}
\begin{bmatrix}

\end{bmatrix}.
$$

In compact form,

$$
x_{au}(k+1)
=
A_{au}x_{au}(k)
+

+

$$

$$

$$

$$
x_{au}(k)=
\begin{bmatrix}

**Control Law**

\begin{equation}
    u=\frac{1}{k_{sys}}[\hat{\xi}[k]-v(ey^{(0)}[k])]
\end{equation}

\begin{equation}
    u = \frac{1}{b_{0}} \left[ -k_1\,\hat{x}_{ou_1}[k] + k_2\,\hat{x}_{ou}[k] \right]
\end{equation}

\begin{equation}

````python
n=Ap_z.shape[0]
m=Ad_z.shape[0]

Aau = np.block([[Ap_z,Bp_z@Cd_z],
                [np.zeros((m,n)),Ad_z]])
Bau = np.block([[Bp_z],
                [np.zeros((m,1))]])
Cau = np.block([Cp_z,np.zeros((1,m))])
Dau = np.block([np.zeros((1,1))])   
sys_au_z = ss(Aau,Bau,Cau,Dau,T_s)

display(Latex(r"State Space of Augmented System:"))
display(sys_au_z)
````

**Observer System**

The discrete-time observer is

$$
\widehat{x}_{au}(k+1)
=
\left(A_{au}-LC_{au}\right)\widehat{x}_{au}(k)
+
B_{au}u(k)
+
Le_y(k).
$$

where

$$
e_y(k)=C_{au}x_{au}(k).
$$

**Estimation Error Dynamics**

Defining the estimation error as

$$
\bar{e}(k)=x_{au}(k)-\widehat{x}_{au}(k),
$$

its dynamics are

$$
\bar{e}(k+1)
=
\left(A_{au}-LC_{au}\right)\bar{e}(k)
+
B_{\xi,au}\,\xi^{(m)}(k).
$$

**State Feedback Control Law**

The control law is

$$
u(k)
=
-k_1\widehat{x}_{au,1}(k)
-k_2\widehat{x}_{au,2}(k)
-\cdots
-k_{n+m}\widehat{x}_{au,n+m}(k),
$$

or, in compact form,

$$
u(k)=-K\widehat{x}_{au}(k).
$$

Since

$$
\widehat{x}_{au}(k)=
\begin{bmatrix}
\widehat{x}_p(k)\\
\widehat{\xi}(k)
\end{bmatrix},
$$

the control law can also be written as

$$
u(k)
=
-K_p\widehat{x}_p(k)
-1*\widehat{\xi}(k).
$$

**Closed-Loop Augmented System**

The closed-loop dynamics are

$$
\begin{bmatrix}
x_{au}(k+1)\\
\bar{e}(k+1)
\end{bmatrix}
=
\begin{bmatrix}
A_{au}-B_{au}K & B_{au}K\\
0 & A_{au}-LC_{au}
\end{bmatrix}
\begin{bmatrix}
x_{au}(k)\\
\bar{e}(k)
\end{bmatrix}\cdot u(k)
$$

````python
# ============================================================
# Control-Loop Design
# ============================================================


pols = np.array([
    -4/Tau,              # 4 veces mas rapido que el lazo abierto: tau_c = Tau/4 = 300 s
])
# Discrete-time poles: z = e^{s Ts}
poles_d = np.exp(pols * T_s)
K_1= acker(Ap_z,Bp_z,poles_d).reshape(1,n)
K = np.block([K_1,1]).reshape(1,n+m)
display(Latex(r"State Feedback Gain K:"))
display(K_1)
display(K)

# ============================================================
# Observer Design
# ============================================================
pols_L = np.array([
-0.0045,-0.0035, -0.0025
])
pols_L = np.array([
-5*4/Tau, -5*4/Tau     # 5 veces mas rapido que el controlador: 20/Tau = 0,0167 rad/s (60 rad/h)
])
#   72 y 86 1/h.  Antes estaba en -0.05,-0.06 (180 1/h): con esa el ESO
#   metia 16,4 puntos de soplador de RUIDO DE SONDA puro (ver el barrido
#   de la seccion 4).  A 72 1/h bajan a 7,1, y el IAE solo empeora 0,020
#   mg/L, que esta por debajo del ruido del propio sensor.
#   Coincide con el optimo 60-100 1/h del analisis de dinamica, 14.3.
# Convert to discrete time: z = e^{s Ts}

poles_L_d = np.exp(pols_L * T_s)
L = acker(Aau.T,Cau.T, poles_L_d.T).reshape(n+m,1)
display(Latex(r"Observer Gain L:"))
display(L)  

# ============================================================
# Validation of the design
# ============================================================

# Closed-loop poles
eig_cl = np.linalg.eigvals(Ap_z - Bp_z @ K_1)

# Observer poles
eig_obs = np.linalg.eigvals(Aau - L @ Cau)
print("Desired controller poles :", np.sort_complex(poles_d))

print("Obtained controller poles:", np.sort_complex(eig_cl))

print("Desired observer poles :", np.sort_complex(poles_L_d))
print("Obtained observer poles:", np.sort_complex(eig_obs))
````

**Resumen del diseño discreto** ($T_s$ = 50 ms, retenedor de orden cero)

$\mathbf{A}_d = e^{\mathbf{A}T_s}$, $\mathbf{B}_d = \int_0^{T_s} e^{\mathbf{A}t}\,dt\;\mathbf{B}$; polos $z = e^{sT_s}$, con $z_c = e^{-\omega_c T_s}$ y $z_o = e^{-\omega_o T_s}$ (doble). Para $n = m = 1$ el Ackermann discreto da

$$k_{p,d} = \frac{1-z_c}{T_s},\qquad \mathbf{L}_d = \begin{bmatrix} 2\,(1-z_o) \\ (1-z_o)^2/T_s \end{bmatrix},$$

y la controlabilidad de la planta y la observabilidad del sistema aumentado se mantienen: $\mathcal{C}_{p,d} = T_s$ y $\det\mathcal{O}_d = T_s \neq 0$.

````python
# ============================================================
# Resumen del diseno discreto (Ts = 50 ms, ZOH)
#   A_d = e^{A Ts}, B_d = int_0^Ts e^{A t} dt B;  polos z = e^{s Ts}
#   Ackermann discreto: k_p,d = (1 - z_c)/Ts,  L_d = [2(1 - z_o), (1 - z_o)^2/Ts]
# ============================================================
z_c, z_o = poles_d[0].real, poles_L_d[0].real
Ctrb_p = ctrb(Ap_z, Bp_z)                    # controlabilidad de la planta (cadena de integradores)
Obsv_d = obsv(Aau, Cau)                      # observabilidad del sistema aumentado
print('A_d =\n', Aau, '\nB_d =', Bau.ravel(), '   C =', Cau.ravel())
print('A_p,d = %s   B_p,d = %s' % (Ap_z.ravel(), Bp_z.ravel()))
print('controlabilidad: C_p,d = %s   rango %d' % (Ctrb_p.ravel(), np.linalg.matrix_rank(Ctrb_p)))
print('observabilidad : O_d = %s   det = %.4g   rango %d' % (Obsv_d.tolist(), np.linalg.det(Obsv_d), np.linalg.matrix_rank(Obsv_d)))
print('z_c = %.8f   z_o = %.8f (doble)' % (z_c, z_o))
print('k_p,d : acker %.6e   (1 - z_c)/Ts %.6e   continuo w_c %.6e' % (K_1[0, 0], (1 - z_c) / T_s, -pols[0]))
print('L_d   : acker %s   formula %s   Ts*L continuo %s'
      % (np.array2string(L.ravel(), precision=6), np.array2string(np.array([2 * (1 - z_o), (1 - z_o)**2 / T_s]), precision=6),
         np.array2string(T_s * np.array([-2 * pols_L[0], pols_L[0]**2]), precision=6)))
print('K_d =', K.ravel())
print('A_o = A_d - L_d C =\n', Ao, '\nB_o = [L_d  B_d] =\n', Bo)
num_g, den_g = np.ravel(Gz_real_tf.num[0][0]), np.ravel(Gz_real_tf.den[0][0])
print('planta real discreta: G(z) = %.6e / (z - %.8f)   [e^{-Ts/tau} = %.8f, K_dc (1 - e^{-Ts/tau}) = %.6e]'
      % (num_g[-1] / den_g[0], -den_g[1] / den_g[0], np.exp(-T_s / Tau), K_dc * (1 - np.exp(-T_s / Tau))))
````

**Discretización del controlador por distintos métodos** ($T_s$ = 50 ms)

El ADRC visto de $-e_y$ a $u$, $C(s) = \dfrac{1}{b_0}\dfrac{k_1 s + k_0}{s\,(s+k_2)}$, discretizado con:

- ZOH: $C(z) = (1 - z^{-1})\,\mathcal{Z}\{C(s)/s\}$
- Tustin (trapezoidal): $s \to \dfrac{2}{T_s}\dfrac{z-1}{z+1}$
- Euler hacia adelante: $s \to \dfrac{z-1}{T_s}$
- Euler hacia atrás: $s \to \dfrac{z-1}{T_s\,z}$

La respuesta en frecuencia de cada $C(z)$ se evalúa en $z = e^{j\omega T_s}$ hasta la frecuencia de Nyquist, $\pi/T_s$.

````python
# ============================================================
# Discretizacion del controlador por distintos metodos (Ts = 50 ms)
#   C(s) = (1/b0) (k1 s + k0)/(s (s + k2)),  de -e_y a u   (ganancias continuas)
#   ZOH, Tustin, Euler adelante, Euler atras
# ============================================================
import control as ct
wc_c, wo_c = 4 / Tau, 20 / Tau
k2_c, k1_c, k0_c = wc_c + 2 * wo_c, 2 * wc_c * wo_c + wo_c**2, wc_c * wo_c**2
C_s = ct.tf([k1_c / bo, k0_c / bo], [1, k2_c, 0])

metodos = [('ZOH', 'zoh', dict(color='C0', lw=2)), ('Tustin (trapezoidal)', 'bilinear', dict(color='C1', lw=1.6, ls='--')),
           ('Euler hacia adelante', 'euler', dict(color='C2', lw=1.6, ls='-.')), ('Euler hacia atrás', 'backward_diff', dict(color='C4', lw=1.6, ls=':'))]
C_z = {lab: ct.sample_system(C_s, T_s, method=m) for lab, m, _ in metodos}
estilos = dict((lab, st) for lab, _, st in metodos)

w_nyq = np.pi / T_s
w_b = np.logspace(-5, np.log10(w_nyq), 4000)                    # rad/s, hasta Nyquist
H_c = ct.frequency_response(C_s, w_b).frdata[0, 0]
fase_c = np.degrees(np.unwrap(np.angle(H_c)))

fig, axs = plt.subplots(2, 1, figsize=(10, 6.5), sharex=True, layout='constrained')
axs[0].semilogx(w_b * 3600, 20 * np.log10(np.abs(H_c)), color='0.6', lw=4, alpha=.6, label='continuo $C(s)$')
axs[1].semilogx(w_b * 3600, fase_c, color='0.6', lw=4, alpha=.6, label='continuo $C(s)$')
print('%-26s %14s %14s %16s' % ('', 'polos C(z)', '|dfase| en w_gc', 'dfase a Nyquist/10'))
w_gc = 0.0121
for lab, Cz in C_z.items():
    Hz = Cz(np.exp(1j * w_b * T_s))
    fz = np.degrees(np.unwrap(np.angle(Hz)))
    fz += 360 * np.round((fase_c[0] - fz[0]) / 360)           # misma rama que el continuo
    axs[0].semilogx(w_b * 3600, 20 * np.log10(np.abs(Hz)), label=lab, **estilos[lab])
    axs[1].semilogx(w_b * 3600, fz, **estilos[lab])
    i_gc = np.argmin(abs(w_b - w_gc)); i_n = np.argmin(abs(w_b - w_nyq / 10))
    print('%-26s %14s %13.4f° %15.2f°' % (lab, np.array2string(np.sort(np.real(ct.poles(Cz))), precision=5),
                                          abs(fz[i_gc] - fase_c[i_gc]), fz[i_n] - fase_c[i_n]))
for ax in axs:
    for x, txt in ((wc_c, '$\\omega_c$'), (wo_c, '$\\omega_o$'), (w_nyq, 'Nyquist')):
        ax.axvline(x * 3600, color='gray', ls=':', lw=1)
    ax.grid(visible=True, which='both', alpha=.3)
for x, txt in ((wc_c, '$\\omega_c$'), (wo_c, '$\\omega_o$'), (w_nyq, '$\\pi/T_s$')):
    axs[1].text(x * 3600 * 1.1, -170, txt, fontsize=9, color='#555555')
axs[0].set_ylabel('magnitud [dB]'); axs[0].set_ylim(-40, 110); axs[0].set_title('a · Magnitud')
axs[1].set_ylabel('fase [°]'); axs[1].set_title('b · Fase'); axs[1].set_ylim(-185, 5); axs[1].set_xlabel('frecuencia [rad/h]')
axs[0].legend(fontsize=8.5, frameon=False, loc='lower left')
fig.suptitle('Discretización del controlador con $T_s$ = 50 ms', fontsize=13, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/disc_metodos_bode.pdf', bbox_inches='tight')
plt.show()
````

**Lazo cerrado del ADRC discreto con $T_s$ = 50 ms y $T_s$ = 40 s**

El mismo escenario de la saturación en continuo (arranque desde 0 hasta 2,0 mg/L, cambio a 2,2 mg/L en $t$ = 4000 s y $\varepsilon = -15$ % en $t$ = 7000 s), con el soplador limitado entre 0 y 100 % y la orden aplicada realimentada al observador. El ADRC se diseña en discreto a cada $T_s$ ($k_{p,d} = (1-z_c)/T_s$, $\mathbf{L}_d = [2(1-z_o),\ (1-z_o)^2/T_s]$) y su orden se mantiene con ZOH; la planta real se integra cada 50 ms para ver lo que pasa entre muestras.

````python
# ============================================================
# Lazo cerrado del ADRC discreto con saturacion: Ts = 50 ms y Ts = 40 s
#   Planta real integrada exacta cada dt = 50 ms (ZOH);  el controlador se
#   ejecuta cada Ts y su orden se mantiene (ZOH) entre muestras.
#   Diseno discreto a cada Ts:  A_d = [[1, Ts], [0, 1]], B_d = [Ts, 0],
#   k_p,d = (1 - z_c)/Ts,  L_d = [2 (1 - z_o), (1 - z_o)^2 / Ts]
# ============================================================
dt_l = 0.05
t_l  = np.arange(0, 10000, dt_l)
r_l  = np.where(t_l >= 4000, 2.2, 2.0)
eps_l = np.where(t_l >= 7000, -15.0, 0.0)              # perturbacion [% de u]
a_g, b_g = np.exp(-dt_l / Tau), K_dc * (1 - np.exp(-dt_l / Tau))
wc_l, wo_l = 4 / Tau, 20 / Tau

def lazo_discreto(Ts_x):
    N = int(round(Ts_x / dt_l))
    zc, zo = np.exp(-wc_l * Ts_x), np.exp(-wo_l * Ts_x)
    kpd, l1d, l2d = (1 - zc) / Ts_x, 2 * (1 - zo), (1 - zo)**2 / Ts_x
    y = 0.0; x1 = x2 = 0.0; u = 0.0
    Y = np.empty(len(t_l)); U = np.empty(len(t_l)); E1 = np.empty(len(t_l)); XI = np.empty(len(t_l))
    for k in range(len(t_l)):
        if k % N == 0:                                   # instante de muestreo
            ey = y - r_l[k]
            u  = min(max(-(kpd * x1 + x2) / bo, 0.0), 100.0)
            e1_k, xi_k = x1, (x2 + y / Tau) / bo         # estimaciones en este instante
            x1, x2 = (1 - l1d) * x1 + Ts_x * x2 + Ts_x * bo * u + l1d * ey, -l2d * x1 + x2 + l2d * ey
        Y[k], U[k], E1[k], XI[k] = y, u, e1_k, xi_k
        y = a_g * y + b_g * (u + eps_l[k])
    return Y, U, E1, XI

casos_l = [(0.05, 'k', '$T_s$ = 50 ms'), (40.0, 'C0', '$T_s$ = 40 s')]
fig, axs = plt.subplots(3, 2, figsize=(12.5, 9), layout='constrained')
z0, z1 = 3950, 5400                                       # acercamiento al cambio de 2,0 a 2,2 mg/L
zm = (t_l >= z0) & (t_l <= z1)
mp_l = []                                                 # M_p de cada caso en el escalon de 2,0 a 2,2 mg/L
for Ts_x, col, lab in casos_l:
    Y, U, E1, XI = lazo_discreto(Ts_x)
    N_x = int(round(Ts_x / dt_l))
    Yk = np.repeat(Y[::N_x], N_x)[:len(Y)]                # y[k]: muestreada y retenida (lo que ve el controlador)
    Rk = np.repeat(r_l[::N_x], N_x)[:len(Y)]
    st = dict(color=col, lw=1.8 if Ts_x < 1 else 1.3, drawstyle='steps-post')
    axs[0, 0].plot(t_l, Yk, label='$y[k]$ (%s)' % lab, **st)
    axs[0, 1].plot(t_l, Yk - Rk, color=col, lw=3.5, alpha=0.3, drawstyle='steps-post', label='$e_y[k]$ real (%s)' % lab)
    axs[0, 1].plot(t_l, E1, color=col, lw=1.0, drawstyle='steps-post', label='$\\hat e_y[k]$ (%s)' % lab)
    axs[1, 0].plot(t_l, XI, label='$\\hat\\xi[k]$  perturbación total (%s)' % lab, **st)
    axs[1, 1].plot(t_l, U, label='$u_{sat}[k]$ (%s)' % lab, **st)
    axs[2, 0].plot(t_l[zm], Yk[zm], label='$y[k]$ (%s)' % lab, **st)
    axs[2, 1].plot(t_l[zm], U[zm], label='$u_{sat}[k]$ (%s)' % lab, **st)
    ms_ = (t_l >= 4000) & (t_l < 7000); i_pk = np.argmax(Yk[ms_])
    mp_l.append((t_l[ms_][i_pk], Yk[ms_][i_pk], 100 * (Yk[ms_][i_pk] - 2.2) / 0.2))
    print('%-12s y max tras 4000 s %.3f mg/L   u max %.1f %%   xi_hat final %.2f %%'
          % (lab.replace('$', ''), Y[t_l >= 4000].max(), U.max(), XI[-1]))
axs[0, 0].plot(t_l, r_l, color='0.5', ls='--', lw=1.2, label='$r$ (consigna)')
axs[2, 0].plot(t_l[zm], r_l[zm], color='0.5', ls='--', lw=1.2, label='$r$ (consigna)')
axs[1, 0].plot(t_l, eps_l, color='0.5', ls='--', lw=1.2, label='$\\varepsilon$  perturbación exógena')
# M_p de cada caso en el acercamiento: linea vertical gris, punto y valor con el color de su curva
ax_e = axs[2, 0]; y_lo, y_hi = 1.99, 2.265
for (t_pk, y_pk, Mp_x), (Ts_x, col, lab), y_txt in zip(mp_l, casos_l, [2.234, 2.247]):   # texto arriba, donde las curvas ya bajan
    ax_e.plot([t_pk, t_pk], [y_lo, y_pk], color='0.75', ls='--', lw=1)
    ax_e.plot(t_pk, y_pk, 'o', ms=3.5, color=col, zorder=5)
    ax_e.text(4650, y_txt, '$M_p$: %s %% (%s)' % (('%.1f' % Mp_x).replace('.', ','), lab),
              ha='left', va='bottom', fontsize=8.5, color=col)
    print('%-12s M_p %.1f %% en t = %.0f s' % (lab.replace('$', ''), Mp_x, t_pk))
ax_e.set_ylim(y_lo, y_hi)
for ax in (axs[1, 1], axs[2, 1]):
    ax.axhline(100, color='C3', ls=':', lw=1.2); ax.axhline(0, color='C3', ls=':', lw=1.2); ax.set_ylim(-8, 110)
axs[0, 0].set_title('a · Salida contra consigna'); axs[0, 0].set_ylabel('OD [mg/L]')
axs[0, 1].set_title('b · El error: real contra estimado'); axs[0, 1].set_ylabel('error [mg/L]')
axs[1, 0].set_title('c · Perturbación exógena y perturbación total'); axs[1, 0].set_ylabel('[% de $u$]')
axs[1, 1].set_title('d · Señal de control'); axs[1, 1].set_ylabel('$u$ [%]')
axs[2, 0].set_title('e · Cambio de consigna de 2,0 a 2,2 mg/L: salida'); axs[2, 0].set_ylabel('OD [mg/L]')
axs[2, 1].set_title('f · Cambio de consigna de 2,0 a 2,2 mg/L: señal de control'); axs[2, 1].set_ylabel('$u$ [%]')
for ax in axs.flat:
    ax.minorticks_on(); ax.grid(visible=True, which='major', alpha=0.35); ax.set_xlabel('tiempo [s]'); eje_minutos(ax)
    ax.legend(fontsize=8, frameon=False, loc='best')
fig.suptitle('Lazo cerrado del ADRC discreto con saturación: $T_s$ = 50 ms y $T_s$ = 40 s', fontsize=13, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/adrc_discreto_lazo_cerrado.pdf', bbox_inches='tight')
plt.show()
````

**GPI discreto por Tustin contra el ADRC discreto**

Cada función de transferencia del GPI con realimentación de la saturación se discretiza por Tustin, con el anti-windup escrito con el polinomio del observador, $E_1(s) = (s + \omega_o)^2$: $H_1(s) = \omega_o[(\omega_o + 2\omega_c)s + \omega_c\omega_o]/(s + \omega_o)^2$ (entra $-e_y$, luego $1/b_0$, sale $v_1$) y $H_2(s) = (\omega_c s - \omega_o^2)/(s + \omega_o)^2$ (entra $u_{sat}$, sale $v_2$); la planta real también se discretiza con Tustin, $G(z)$. En el cambio de 2,0 a 2,2 mg/L el soplador no se satura, así que $u = v$ y el controlador queda $C(z) = \frac{1}{b_0}\,\frac{H_1(z)}{1 + H_2(z)}$, que coincide con Tustin de $C(s)$. El lazo se resuelve con `tf`, `c2d`, `feedback` y `lsim`, y se compara con el ADRC discreto de la celda anterior (planta exacta), con $T_s$ = 50 ms y $T_s$ = 40 s.

````python
# ============================================================
# GPI discreto (Tustin por bloque) contra el ADRC discreto
#   Anti-windup con el polinomio del observador, E1(s) = (s + wo)^2:
#   H1(s) = wo [(wo + 2 wc) s + wc wo]/(s + wo)^2:  entra -e_y (luego 1/b0),  sale v1
#   H2(s) = (wc s - wo^2)/(s + wo)^2:               entra u_sat,              sale v2
#   v = v1/b0 - v2;  sin saturacion u = v  ->  C(z) = (1/b0) H1(z) / (1 + H2(z))
#   Como H1 = N1/D y H2 = N2/D tienen el mismo denominador, C(z) = (1/b0) N1/(D + N2):
#   el E1 se cancela en el papel y no hay que cancelar el polo doble numericamente
#   Planta: G(z) = Tustin{K_dc/(Tau s + 1)}, igual que el controlador
#   En el cambio de 2,0 a 2,2 mg/L (t = 4000 s) el soplador no se satura: el
#   lazo es lineal y se resuelve con tf(), c2d(), feedback() y lsim().
#   El ADRC es el de la celda anterior (lazo_discreto, planta exacta), a los mismos Ts.
# ============================================================
k2_g, k1_g, k0_g = wc_l + 2 * wo_l, wo_l * (wo_l + 2 * wc_l), wc_l * wo_l**2   # ganancias en funcion de wc y wo
E1_g = [1, 2 * wo_l, wo_l**2]                                                  # E1(s) = (s + wo)^2
H1_s = tf([wo_l * (wo_l + 2 * wc_l), wc_l * wo_l**2], E1_g)
H2_s = tf([wc_l, -wo_l**2], E1_g)
G_s  = tf([K_dc], [Tau, 1])
C_s  = tf([k1_g / bo, k0_g / bo], [1, k2_g, 0])           # GPI continuo, para comprobar
print('E1(s) = (s + wo)^2,  wc = %.4e  wo = %.4e  ->  k0 = %.4e   k1 = %.4e   k2 = %.4e   1/b0 = %.0f' % (wc_l, wo_l, k0_g, k1_g, k2_g, 1 / bo))

y0_g, u0_g = 2.0, 2.0 / K_dc                               # punto de operacion antes del cambio
z0, z1 = 3950, 4600
fig, axs = plt.subplots(1, 2, figsize=(12.5, 4.4), layout='constrained')
for Ts_x, col, lab in casos_l:
    H1_z = c2d(H1_s, Ts_x, 'tustin')
    H2_z = c2d(H2_s, Ts_x, 'tustin')
    G_z  = c2d(G_s, Ts_x, 'tustin')
    pad = lambda v: np.r_[np.zeros(3 - len(v)), v]
    D_  = np.ravel(H1_z.den[0][0]); N1_ = pad(np.ravel(H1_z.num[0][0])) / D_[0]
    N2_ = pad(np.ravel(H2_z.num[0][0])) / np.ravel(H2_z.den[0][0])[0]; D_ = D_ / D_[0]
    C_z  = tf(N1_ / bo, D_ + N2_, Ts_x)                    # (r - y) -> u
    C_ref = c2d(C_s, Ts_x, 'tustin')
    print('\nTs = %g s' % Ts_x)
    print('  H1(z) =', H1_z); print('  H2(z) =', H2_z); print('  G(z)  =', G_z)
    nc, dc = np.ravel(C_z.num[0][0]), np.ravel(C_z.den[0][0]); nr, dr_ = np.ravel(C_ref.num[0][0]), np.ravel(C_ref.den[0][0])
    print('  C(z) = (1/b0) H1/(1+H2) coincide con Tustin{C(s)}:',
          np.allclose(nc / dc[0], nr / dr_[0], rtol=1e-6) and np.allclose(dc / dc[0], dr_ / dr_[0], rtol=1e-9))
    T_yr = feedback(C_z * G_z, 1)                          # r -> y
    T_ur = feedback(C_z, G_z)                              # r -> u
    tk = np.arange(0, z1 + Ts_x, Ts_x)
    dr = np.where(tk >= 4000, 0.2, 0.0)
    y_g = y0_g + np.ravel(lsim(T_yr, dr, tk)[0])
    u_g = u0_g + np.ravel(lsim(T_ur, dr, tk)[0])
    # ADRC discreto de la celda anterior, en los instantes de muestreo
    N_x = int(round(Ts_x / dt_l))
    Y_a, U_a, _, _ = lazo_discreto(Ts_x)
    ta, ya, ua = t_l[::N_x], Y_a[::N_x], U_a[::N_x]
    ma, mg = (ta >= z0) & (ta <= z1), (tk >= z0) & (tk <= z1)
    sa = dict(color=col, lw=3.5, alpha=0.35, drawstyle='steps-post')     # ADRC: banda gruesa
    sg = dict(color=col, lw=1.3, ls='--', drawstyle='steps-post')        # GPI: discontinua encima
    axs[0].plot(ta[ma], ya[ma], label='ADRC (%s)' % lab, **sa)
    axs[0].plot(tk[mg], y_g[mg], label='GPI (%s)' % lab, **sg)
    axs[1].plot(ta[ma], ua[ma], label='ADRC (%s)' % lab, **sa)
    axs[1].plot(tk[mg], u_g[mg], label='GPI (%s)' % lab, **sg)
    yi = np.interp(tk[mg], ta, ya); ui = np.interp(tk[mg], ta, ua)
    print('  ADRC: y max %.4f mg/L  u max %.2f %%  |  GPI: y max %.4f mg/L  u max %.2f %%  |  max|dy| %.2e  max|du| %.2e'
          % (ya[ma].max(), ua[ma].max(), y_g[mg].max(), u_g[mg].max(),
             np.abs(y_g[mg] - yi).max(), np.abs(u_g[mg] - ui).max()))
tr = np.arange(z0, z1, 1.0)
axs[0].plot(tr, np.where(tr >= 4000, 2.2, 2.0), color='0.5', ls='--', lw=1.2, label='$r$ (consigna)')
axs[1].axhline(100, color='C3', ls=':', lw=1.2); axs[1].axhline(0, color='C3', ls=':', lw=1.2); axs[1].set_ylim(-8, 110)
axs[0].set_title('a · Salida $y[k]$'); axs[0].set_ylabel('OD [mg/L]')
axs[1].set_title('b · Señal de control $u[k]$'); axs[1].set_ylabel('$u$ [%]')
for ax in axs:
    ax.minorticks_on(); ax.grid(visible=True, which='major', alpha=0.35); ax.set_xlabel('tiempo [s]'); eje_minutos(ax)
    ax.legend(fontsize=8, frameon=False, loc='best')
fig.suptitle('GPI discreto (Tustin) contra ADRC discreto: cambio de 2,0 a 2,2 mg/L', fontsize=13, fontweight='bold')
fig.savefig('Latex/Tesis/figuras/gpi_discreto_vs_adrc.pdf', bbox_inches='tight')
plt.show()
````

**Coeficientes del GPI discreto con los dos $E_1$**

$H_1 = P/E_1$ y $H_2 = (L - E_1)/E_1$ por Tustin, con $E_1 = s^2 + k_2 s + k_1$ (con las ganancias) y con $E_1 = (s + \omega_o)^2$ (con el observador): la misma forma $H_1(z) = (\beta_0 + \beta_1 z^{-1} + \beta_2 z^{-2})/(1 + \alpha_1 z^{-1} + \alpha_2 z^{-2})$, $H_2(z) = (\gamma_0 + \gamma_1 z^{-1} + \gamma_2 z^{-2})/(1 + \alpha_1 z^{-1} + \alpha_2 z^{-2})$; solo cambian los valores. Se comparan las fórmulas con `c2d(..., 'tustin')`.

````python
# ============================================================
# Coeficientes del GPI discreto (Tustin) con los dos E1:
#   con las ganancias  E1 = s^2 + k2 s + k1  ->  H2 = -k1/E1, gamma = -rho (1, 2, 1), rho = k1 Ts^2/D0
#   con el observador  E1 = (s + wo)^2       ->  H2 = (wc s - wo^2)/E1
# ============================================================
def _coef_gpi():
    wc_, wo_ = 4 / Tau, 20 / Tau
    k2_, k1_, k0_ = wc_ + 2 * wo_, wo_ * (wo_ + 2 * wc_), wc_ * wo_**2
    def con_k(T):
        D0 = 4 + 2 * k2_ * T + k1_ * T**2
        return dict(D0=D0, a1=(2 * k1_ * T**2 - 8) / D0, a2=(4 - 2 * k2_ * T + k1_ * T**2) / D0,
                    b0=(k0_ * T**2 + 2 * k1_ * T) / D0, b1=2 * k0_ * T**2 / D0, b2=(k0_ * T**2 - 2 * k1_ * T) / D0,
                    g0=-k1_ * T**2 / D0, g1=-2 * k1_ * T**2 / D0, g2=-k1_ * T**2 / D0)
    def con_wo(T):
        D0 = (2 + wo_ * T)**2
        return dict(D0=D0, a1=-2 * (4 - wo_**2 * T**2) / D0, a2=(2 - wo_ * T)**2 / D0,
                    b0=wo_ * T * (wc_ * wo_ * T + 2 * (wo_ + 2 * wc_)) / D0, b1=2 * wc_ * wo_**2 * T**2 / D0,
                    b2=wo_ * T * (wc_ * wo_ * T - 2 * (wo_ + 2 * wc_)) / D0,
                    g0=T * (2 * wc_ - wo_**2 * T) / D0, g1=-2 * wo_**2 * T**2 / D0, g2=-T * (2 * wc_ + wo_**2 * T) / D0)
    casos = [('con las ganancias, E1 = s^2 + k2 s + k1', con_k, tf([k1_, k0_], [1, k2_, k1_]), tf([-k1_], [1, k2_, k1_])),
             ('con el observador, E1 = (s + wo)^2', con_wo, tf([k1_, k0_], [1, 2 * wo_, wo_**2]), tf([wc_, -wo_**2], [1, 2 * wo_, wo_**2]))]
    for nom, f, H1, H2 in casos:
        print(nom)
        print('   %-4s %16s %16s' % ('', 'Ts = 50 ms', 'Ts = 40 s'))
        c = {T: f(T) for T in (0.05, 40.0)}
        for k in ('D0', 'a1', 'a2', 'b0', 'b1', 'b2', 'g0', 'g1', 'g2'):
            print('   %-4s %16.8e %16.8e' % (k, c[0.05][k], c[40.0][k]))
        for T in (0.05, 40.0):
            ok = True
            for H, nm in ((H1, 'b'), (H2, 'g')):
                Hz = c2d(H, T, 'tustin'); n = np.ravel(Hz.num[0][0]); d = np.ravel(Hz.den[0][0]); n = np.r_[np.zeros(3 - len(n)), n]
                ok &= np.allclose(d / d[0], [1, c[T]['a1'], c[T]['a2']], rtol=1e-9) and \
                      np.allclose(n / d[0], [c[T][nm + '0'], c[T][nm + '1'], c[T][nm + '2']], rtol=1e-7, atol=1e-15)
            print('   Ts = %g s: formulas = c2d tustin: %s   polos de E1(z): %s   1 + a1 + a2 = %.3e'
                  % (T, ok, np.round(np.abs(np.roots([1, c[T]['a1'], c[T]['a2']])), 5), 1 + c[T]['a1'] + c[T]['a2']))
    for T in (0.05, 40.0):
        print('planta, Ts = %g s:  beta_g = %.6e   alpha_g = %.8f' % (T, K_dc * T / (2 * Tau + T), (T - 2 * Tau) / (2 * Tau + T)))

_coef_gpi()
````

**Prefiltro discreto con el ADRC y el GPI discretos** (figura `disc_prefiltro_retardo.pdf`, Tabla de índices de la sección 5.4.3)

El prefiltro $F(s) = k_0/(k_1 s + k_0)$ se discretiza con Tustin, $r_f[k] = -\alpha_f\,r_f[k-1] + \beta_f\,(r[k] + r[k-1])$, y el error es $e_y[k] = y[k] - r_f[k]$. Cambio de consigna de 2,0 a 2,2 mg/L con el sistema en estacionario (soplador en 40 %), planta exacta cada 50 ms con retardo $\theta$ = 0 y 50 s, soplador 0-100 %. ADRC discreto (el de la celda del lazo cerrado) y GPI discreto por Tustin con $E_1 = (s + \omega_o)^2$; el lazo algebraico del GPI ($\bar\gamma_0\,u_{sat}[k]$) se resuelve en la región lineal. $t_s$ con banda de 2 % del escalón sobre la salida de la planta.

````python
# ============================================================
# Prefiltro discreto (Tustin) con el ADRC discreto y el GPI discreto (Tustin, E1 = (s + wo)^2)
#   r_f[k] = -alpha_f r_f[k-1] + beta_f (r[k] + r[k-1]),   e_y[k] = y[k] - r_f[k]
#   Planta exacta cada dt = 50 ms con retardo theta; controlador cada Ts con ZOH; soplador 0-100 %.
#   GPI: v = v1/b0 - v2,  v2 = gamma0 u_sat[k] + ...  ->  lazo algebraico resuelto en la region lineal
# ============================================================
k2_p, k1_p, k0_p = wc_l + 2 * wo_l, wo_l * (wo_l + 2 * wc_l), wc_l * wo_l**2
dt_p = 0.05
t_p  = np.arange(0, 6000, dt_p)
r_p  = np.where(t_p >= 4000, 2.2, 2.0)
ag_p, bg_p = np.exp(-dt_p / Tau), K_dc * (1 - np.exp(-dt_p / Tau))
U0_p = 2.0 / K_dc                                          # soplador en regimen a 2,0 mg/L

def coef_pref(Ts_x):
    den = k0_p * Ts_x + 2 * k1_p
    return (k0_p * Ts_x - 2 * k1_p) / den, k0_p * Ts_x / den   # alpha_f, beta_f

def coef_gpi_wo(Ts_x):
    E1 = [1, 2 * wo_l, wo_l**2]
    H1 = c2d(tf([wo_l * (wo_l + 2 * wc_l), wc_l * wo_l**2], E1), Ts_x, 'tustin')
    H2 = c2d(tf([wc_l, -wo_l**2], E1), Ts_x, 'tustin')
    pad = lambda v: np.r_[np.zeros(3 - len(v)), v]
    D1, D2 = np.ravel(H1.den[0][0]), np.ravel(H2.den[0][0])
    return pad(np.ravel(H1.num[0][0])) / D1[0], pad(np.ravel(H2.num[0][0])) / D2[0], D1 / D1[0]

def lazo_prefiltro(ctrl, Ts_x, theta):
    N = int(round(Ts_x / dt_p)); d = int(round(theta / dt_p))
    af, bf = coef_pref(Ts_x)
    hist = np.full(len(t_p) + d + 1, U0_p)                 # orden aplicada, para el retardo
    y = 2.0; rf_1 = r_1 = 2.0; u = U0_p; yk = y
    if ctrl == 'ADRC':
        zc, zo = np.exp(-wc_l * Ts_x), np.exp(-wo_l * Ts_x)
        kpd, l1d, l2d = (1 - zc) / Ts_x, 2 * (1 - zo), (1 - zo)**2 / Ts_x
        x1, x2 = 0.0, -bo * U0_p                           # regimen: e_hat = 0, x2 = -b0 u
    else:
        be, ga, D = coef_gpi_wo(Ts_x); a1, a2 = D[1], D[2]
        x_h, v1_h, us_h, v2_h = [0.0, 0.0], [0.0, 0.0], [U0_p, U0_p], [-U0_p, -U0_p]   # H2(1) = -1
    Y = np.empty(len(t_p)); U = np.empty(len(t_p)); YK = np.empty(len(t_p)); RF = np.empty(len(t_p))
    for k in range(len(t_p)):
        if k % N == 0:                                     # instante de muestreo
            yk = y
            rf = -af * rf_1 + bf * (r_p[k] + r_1); rf_1, r_1 = rf, r_p[k]
            ey = y - rf
            if ctrl == 'ADRC':
                u = min(max(-(kpd * x1 + x2) / bo, 0.0), 100.0)
                x1, x2 = (1 - l1d) * x1 + Ts_x * x2 + Ts_x * bo * u + l1d * ey, -l2d * x1 + x2 + l2d * ey
            else:
                x = -ey
                v1 = be[0] * x + be[1] * x_h[0] + be[2] * x_h[1] - a1 * v1_h[0] - a2 * v1_h[1]
                R2 = ga[1] * us_h[0] + ga[2] * us_h[1] - a1 * v2_h[0] - a2 * v2_h[1]
                u = min(max((v1 / bo - R2) / (1 + ga[0]), 0.0), 100.0)
                x_h, v1_h, us_h, v2_h = [x, x_h[0]], [v1, v1_h[0]], [u, us_h[0]], [ga[0] * u + R2, v2_h[0]]
        hist[d + k] = u
        Y[k], U[k], YK[k], RF[k] = y, u, yk, rf_1
        y = ag_p * y + bg_p * hist[k]                      # la planta recibe la orden retrasada theta
    return Y, U, YK, RF

def _figura_prefiltro():
    fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.8), sharex=True, layout='constrained')
    mv = (t_p >= 3950) & (t_p <= 6000)
    filas = []
    for j, th in enumerate((0, 50)):
        for Ts_x, col, lab in casos_l:
            for ctrl in ('ADRC', 'GPI'):
                Y, U, YK, RF = lazo_prefiltro(ctrl, Ts_x, th)
                st = (dict(color=col, lw=3.5, alpha=0.35) if ctrl == 'ADRC' else dict(color=col, lw=1.3, ls='--'))
                axs[0, j].plot(t_p[mv], YK[mv], drawstyle='steps-post', label='%s (%s)' % (ctrl, lab), **st)
                axs[1, j].plot(t_p[mv], U[mv], drawstyle='steps-post', label='%s (%s)' % (ctrl, lab), **st)
                m = t_p >= 4000; f = np.flatnonzero(np.abs(Y[m] - 2.2) > 0.02 * 0.2)
                filas.append(dict(controlador=ctrl, Ts=lab.replace('$', ''), theta=th,
                                  ts_min=round((t_p[m][f[-1]] - 4000) / 60, 1), umax=round(U[m].max(), 1),
                                  Mp=round(max(100 * (Y[m].max() - 2.2) / 0.2, 0), 2)))
                if Ts_x < 1 and ctrl == 'ADRC':
                    axs[0, j].plot(t_p[mv], RF[mv], color='0.45', lw=1.2, label='$r_f$ (prefiltro)')
        axs[0, j].plot(t_p[mv], r_p[mv], color='0.5', ls='--', lw=1.2, label='$r$ (consigna)')
        axs[0, j].set_ylim(1.99, 2.215); axs[0, j].set_ylabel('OD [mg/L]')
        axs[1, j].axhline(100, color='C3', ls=':', lw=1.2); axs[1, j].axhline(0, color='C3', ls=':', lw=1.2)
        axs[1, j].set_ylim(-8, 110); axs[1, j].set_ylabel('$u$ [%]'); axs[1, j].set_xlabel('tiempo [s]')
    txt = ('sin retardo', '$\\theta$ = 50 s')
    for j in range(2):
        axs[0, j].set_title('%s · Salida $y[k]$, %s' % ('ab'[j], txt[j]))
        axs[1, j].set_title('%s · Señal de control $u[k]$, %s' % ('cd'[j], txt[j]))
    for ax in axs.flat:
        ax.minorticks_on(); ax.grid(visible=True, which='major', alpha=0.35); eje_minutos(ax)
        ax.legend(fontsize=8, frameon=False, loc='lower right' if ax in axs[0] else 'upper right')
    fig.suptitle('Prefiltro discreto con el ADRC discreto y el GPI discreto: cambio de 2,0 a 2,2 mg/L',
                 fontsize=13, fontweight='bold')
    fig.savefig('Latex/Tesis/figuras/disc_prefiltro_retardo.pdf', bbox_inches='tight')
    plt.show()
    display(pd.DataFrame(filas))

_figura_prefiltro()
````

**Planta con retardo: observador sincronizado** (Herbst & Madoński, §9.4, ec. 9.6)

Si se conoce el retardo $\theta$ de la planta, se sincroniza el observador con ella retrasando la orden aplicada que le entra, $n_d$ muestras:

$$\hat{\mathbf X}[k+1] = (\mathbf A_d - \mathbf L_d \mathbf C_d)\,\hat{\mathbf X}[k] + \mathbf B_d\, b_0\, u_{sat}[k - n_d] + \mathbf L_d\, e_y[k],\qquad n_d \approx \frac{\theta}{T_s}.$$

Las matrices y las ganancias no cambian. Escalón de 2,0 a 2,2 mg/L en $t$ = 0 y $\varepsilon = -15$ % a los 40 min, $T_s$ = 50 ms, soplador 0-100 %, $\theta$ = 20, 40, 50, 60 y 100 s: sin sincronizar ($n_d = 0$) y sincronizado ($n_d = \theta/T_s$). Al final, el desajuste: $\theta$ = 50 s real con $n_d T_s$ entre 0 y 100 s.

````python
# ============================================================
# ADRC discreto con planta con retardo: observador sincronizado
#   x_hat[k+1] = A_o x_hat[k] + B_d b0 u_sat[k - n_d] + L_d e_y[k],   n_d ~ theta / Ts
#   Herbst & Madonski, sec. 9.4, ec. (9.6)-(9.7).  Mismas ganancias del diseno discreto.
# ============================================================
Ts_dt = 0.05
zc_dt, zo_dt = np.exp(-wc_l * Ts_dt), np.exp(-wo_l * Ts_dt)
kpd_dt, l1d_dt, l2d_dt = (1 - zc_dt) / Ts_dt, 2 * (1 - zo_dt), (1 - zo_dt)**2 / Ts_dt
ag_dt, bg_dt = np.exp(-Ts_dt / Tau), K_dc * (1 - np.exp(-Ts_dt / Tau))
t_dt = np.arange(-300, 4800, Ts_dt)                       # [s] desde el escalon
r_dt = np.where(t_dt >= 0, 2.2, 2.0)
eps_dt = np.where(t_dt >= 2400, -15.0, 0.0)               # perturbacion a los 40 min [% de u]

def lazo_retardo(theta, nd):
    # arranca en regimen en 2,0 mg/L: u = 40 %, x_hat = [0, -y/Tau]
    d = int(round(theta / Ts_dt)); off = max(d, nd) + 1
    hist = np.full(len(t_dt) + off, 2.0 / K_dc)            # historial de u_sat
    y, x1, x2 = 2.0, 0.0, -2.0 / Tau
    Y = np.empty(len(t_dt)); U = np.empty(len(t_dt))
    for k in range(len(t_dt)):
        ey = y - r_dt[k]
        u = min(max(-(kpd_dt * x1 + x2) / bo, 0.0), 100.0); hist[off + k] = u
        ud = hist[off + k - nd]                            # orden que entra al observador
        x1, x2 = (1 - l1d_dt) * x1 + Ts_dt * x2 + Ts_dt * bo * ud + l1d_dt * ey, -l2d_dt * x1 + x2 + l2d_dt * ey
        Y[k], U[k] = y, u
        y = ag_dt * y + bg_dt * (hist[off + k - d] + eps_dt[k])   # la planta recibe u retrasada theta
    return Y, U

def ts_dt(Y):
    m = (t_dt >= 0) & (t_dt < 2400); f = np.flatnonzero(np.abs(Y[m] - 2.2) > 0.02 * 0.2)
    return t_dt[m][f[-1]] / 60

def _figura_retardo():
    thetas_dt = [(20, '20 s'), (40, '40 s'), (50, '50 s'), (60, '60 s'), (100, '100 s (extremo)')]
    col_dt = plt.cm.viridis(np.linspace(0.05, 0.85, len(thetas_dt)))
    Y0_dt, U0_dt = lazo_retardo(0, 0)
    fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.5), sharex=True, layout='constrained')
    tm = t_dt / 60
    for j, sinc in enumerate((False, True)):
        axs[0, j].plot(tm, Y0_dt, 'k', lw=2.4, label='sin retardo ($\\theta$ = 0)')
        axs[1, j].plot(tm, U0_dt, 'k', lw=2.4)
        for (th, txt), c in zip(thetas_dt, col_dt):
            nd = int(round(th / Ts_dt)) if sinc else 0
            Y, U = lazo_retardo(th, nd)
            lab = '$\\theta$ = %s' % txt + ('  ($n_d$ = %d)' % nd if sinc else '')
            axs[0, j].plot(tm, Y, color=c, lw=1.4, ls='--', label=lab)
            axs[1, j].plot(tm, U, color=c, lw=1.2, ls='--')
            m = (t_dt >= 0) & (t_dt < 2400)
            print('%-15s %-14s ymax %.3f   u en [%5.1f, %5.1f] %%   ts %5.1f min   OD min tras la perturbacion %.3f' % (
                txt, 'n_d = %d' % nd, Y[m].max(), U[m].min(), U[m].max(), ts_dt(Y), Y[t_dt >= 2400].min()))
        axs[0, j].plot(tm, r_dt, 'k:', lw=1)
        axs[0, j].set_ylim(1.95, 2.45); axs[0, j].set_ylabel('OD [mg/L]')
        axs[1, j].axhline(100, color='C3', ls=':', lw=1); axs[1, j].axhline(0, color='C3', ls=':', lw=1)
        axs[1, j].set_ylim(-5, 105); axs[1, j].set_ylabel('$u_{sat}$ [%]'); axs[1, j].set_xlabel('tiempo desde el escalón [min]')
    axs[0, 0].set_title('a · Observador sin sincronizar ($n_d$ = 0): salida')
    axs[0, 1].set_title('b · Observador sincronizado ($n_d = \\theta/T_s$): salida')
    axs[1, 0].set_title('c · Observador sin sincronizar: señal de control')
    axs[1, 1].set_title('d · Observador sincronizado: señal de control')
    for ax in axs.flat:
        ax.grid(alpha=.3); ax.minorticks_on(); ax.set_xlim(-5, 80)
    for ax in axs[0]:
        ax.legend(fontsize=7.8, frameon=False, loc='upper right')
    fig.suptitle('ADRC discreto ($T_s$ = 50 ms) con retardo en la planta: escalón de 2,0 a 2,2 mg/L y $\\varepsilon = -15$ % a los 40 min',
                 fontsize=12, fontweight='bold')
    fig.savefig('Latex/Tesis/figuras/adrc_discreto_retardo.pdf', bbox_inches='tight')
    plt.show()

    print('\nDesajuste: theta real = 50 s')
    for thn in (0, 25, 40, 50, 60, 75, 100):
        Y, U = lazo_retardo(50, int(round(thn / Ts_dt))); m = (t_dt >= 0) & (t_dt < 2400)
        print('n_d Ts = %3d s:  ymax %.3f   u en [%5.1f, %5.1f] %%   ts %5.1f min' % (thn, Y[m].max(), U[m].min(), U[m].max(), ts_dt(Y)))

_figura_retardo()
````

**Closed-Loop System for simulation**



The discrete-time closed-loop system is (with bo=1 and the Gain is into the model system)

$$
\begin{bmatrix}
x_{p}(k+1)\\
\widehat{x}_{au}(k+1)
\end{bmatrix}
=
\begin{bmatrix}
A_{p} & -B_{p}K\\
LC_{p} & A_{au}-LC_{au}-B_{au}K
\end{bmatrix}
\begin{bmatrix}
x_{p}(k)\\
\widehat{x}_{au}(k)
\end{bmatrix}
+
\begin{bmatrix}
0_{(n,1)} & B_{p}\\
-L & 0_{(n+m,1)}
\end{bmatrix}
\begin{bmatrix}
r(k)\\
\xi(k)
\end{bmatrix}.
$$



The discrete-time closed-loop system is (without bo into the model system)

**State Feedback Control Law**

The control law is
$$
u(k)
=
\frac{1}{b_0}
\left(
-k_1\widehat{x}_{au,1}(k)
-k_2\widehat{x}_{au,2}(k)
-\cdots
-k_{n+m}\widehat{x}_{au,n+m}(k)
\right)
$$

or, in compact form,

$$
u(k)=\frac{1}{b_0}
\left(-K\widehat{x}_{au}(k)\right)
$$

Since

$$
\widehat{x}_{au}(k)=
\begin{bmatrix}
\widehat{x}_p(k)\\
\widehat{\xi}(k)
\end{bmatrix},
$$
 with
 $$
 K=\begin{bmatrix}K_p &1\end{bmatrix}
 $$
the control law can also be written as

$$
u(k)
=\frac{1}{b_0}
\left(
-K_p\widehat{x}_p(k)
-1\cdot\widehat{\xi}(k)\right)
$$


**The closed-loop dynamics are**

$$
\begin{bmatrix}
x_{p}(k+1)\\
\widehat{x}_{au}(k+1)
\end{bmatrix}
=
\begin{bmatrix}
A_{p} & \frac{-B_{p}K}{b_0}\\
LC_{p} & A_{au}-LC_{au}-B_{au}K
\end{bmatrix}
\begin{bmatrix}
x_{p}(k)\\
\widehat{x}_{au}(k)
\end{bmatrix}
+
\begin{bmatrix}
0_{(n,1)} & B_{p}\\
-L & 0_{(n+m,1)}
\end{bmatrix}
\begin{bmatrix}
r(k)\\
\xi(k)
\end{bmatrix}.
$$

````python
## Lazo cerrado: ADRC basado en el error sobre integrador ideal
#
# El observador estima e_y = y - r.  Esa variable lleva la consigna dentro de su
# definicion, asi que el sistema del error tiene TRES entradas, no dos:
#
#       e_y' = e_y'(planta)  +  u0            <- conocida, se la pasas
#                            +  xi            <- desconocida, la estima el ESO
#                            -  r'            <- CONOCIDA, y aqui NO se le pasa
#
# Esta celda deja el lazo "desnudo", sin tratar la consigna: el observador no se
# entera de que r salta, deduce una xi falsa y la manda al actuador.  Es la
# propiedad que Herbst & Madonski documentan en la Fig. 6.13 ("giant control
# efforts") y que se mide abajo: u pico 419 %, t63 de 93 s con tau_c = 1063 s.
#
# NO es un error: el ADRC basado en el error es el de salida al que se le ha
# quitado el prefiltro (seccion 8.4, p. 141).  Las dos secciones siguientes se
# lo devuelven, cada una a su manera, SIN tocar ninguna de estas matrices.
#
# bo aparece en UN solo sitio: u = u0/bo, al cruzar al hardware.



Aau_cl = np.block([[Ap_r,      -Bp_r @ K / bo],
                  [L @ Cp_r,   Aau - L @ Cau - Bau @ K]])

Bau_cl = np.block([[np.zeros((n,1)),Bp_r],
                  [-L,np.zeros((n+m,1))]])

Cau_cl = np.block([Cp_r, np.zeros((1, n + m))])
Dau_cl = np.zeros((1, 2))

sys_au_cl = ss(Aau_cl, Bau_cl, Cau_cl, Dau_cl, T_s)

print('|eig| maximo del lazo cerrado : %.4f' % np.abs(np.linalg.eigvals(Aau_cl)).max())
print('L = %s   (la misma de la celda 37)' % np.array2string(L.ravel(), precision=6))
display(Latex(r"State Space of Extended System:"))
display(sys_au_cl)
````

````python
# ============================================================
# Simulacion del lazo cerrado
#   U = [r, xi, w]    r  -> consigna [mg/L]
#                     xi -> perturbacion de entrada [% de u]
#                     w  -> -dr/dt, la entrada conocida del sistema del error
# ============================================================
t  = np.arange(0, 10000, T_s)
r  = np.where(t >= 4000, 2.2, 2.0)          # consigna: arranca en 2.0, sube a 3.0 en t = 4000 s
xi = np.where(t >= 7000, -15.0, 0.0)        # perturbacion [% de u] en t = 7000 s
w  = np.r_[-np.diff(r), 0.0] / T_s          # -dr/dt: un pulso de una muestra en cada cambio
u  = np.column_stack([r, xi])

xp0    = np.zeros(n)                        # planta en reposo, y(0) = 0
xp0_es = np.array([0.0])             # e_hat_y(0) = y(0) - r(0): el error inicial MEDIDO
xd0_es = np.zeros(m)                        # de la perturbacion no sabemos nada
x0 = np.block([xp0, xp0_es, xd0_es])

[Y, T, X] = lsim(sys_au_cl, U=u, T=t, X0=x0)

# ------------------------------------------------------------
#   X[:, 0] = x_p       estado de la PLANTA
#   X[:, 1] = e_hat_y   estimacion del ERROR      [mg/L]   (Cp_z = 1)
#   X[:, 2] = xi_hat    estimacion de la PERTURBACION  [mg/L/s]
# ------------------------------------------------------------
y_out   = (Cp_r @ X[:, :n].T).ravel()
e_est   = X[:, n]                           # e_y ESTIMADO por el observador
xi_hat  = X[:, n + 1]           # xi1_hat: el indice SIEMPRE es n+1, no n+m (ver seccion 4)

e_real  = y_out - r                         # el error DE VERDAD

u0      = -K_1[0, 0] * e_est - xi_hat       # = -K x_hat   [mg/L/s]
u_ctrl  = u0 / bo                           # control fisico al soplador [%]

# xi_hat lleva dentro la autorregulacion de la planta (-y/Tau), que el
# integrador ideal mando a la perturbacion total. Lo que queda, en % de u:
xi_fisica = (xi_hat + y_out / Tau) / bo

# trayectoria que impone el polo del controlador, vectorizada:
#   en cada tramo de consigna,  y(t) = r + (y(t0) - r) e^{-omega_c (t - t0)}
w_c    = -pols[0]
k_pert = int(np.flatnonzero(xi != 0)[0])

nuevo = np.r_[True, np.diff(r) != 0]          # True en cada cambio de consigna
tramo = np.cumsum(nuevo) - 1                  # numero de tramo de cada muestra
k0    = np.flatnonzero(nuevo)[tramo]          # muestra en que arranco su tramo

y_teo = r + (y_out[k0] - r) * np.exp(-w_c * (t - t[k0]))
y_teo[k_pert:] = np.nan                       # a partir de xi manda el observador, no el polo

print('omega_c = %.6f rad/s  ->  tau_c = %.1f s      bo = %.4e\n' % (w_c, 1 / w_c, bo))
print('SEGUIMIENTO DE CONSIGNA  (%d escalones de consigna)' % nuevo.sum())
print('   max|y - trayectoria|   = %.2e mg/L' % np.nanmax(np.abs(y_out - y_teo)))
print('   t63 del primer escalon = %.1f s   (tau_c = %.1f s)'
      % (t[np.argmax(y_out >= 0.632 * r[0])], 1 / w_c))
print('\nestimacion del error : max|e_real - e_est| = %.2e mg/L' % np.abs(e_real - e_est).max())
print('perturbacion fisica  : estimada %.2f %%   ·   aplicada %.1f %%' % (xi_fisica[-1], xi[-1]))
print('control              : u pico %.1f %%, minimo %.1f %%' % (u_ctrl.max(), u_ctrl.min()))
````

#### 2.1.4 Discrete-time implementation

Design model (integrator chain + step disturbance), discretized with $T_s$:

$$
A=\begin{bmatrix}1&T_s\\0&1\end{bmatrix},\qquad
B=\begin{bmatrix}T_s\\0\end{bmatrix},\qquad
C=\begin{bmatrix}1&0\end{bmatrix}
$$

Observer, driven by the feedback error $e_y=y-r$:

$$
\widehat{x}[k+1]=(A-LC)\,\widehat{x}[k]+B\,u_0[k]+L\,e_y[k],
\qquad
A-LC=\begin{bmatrix}a_{11}-l_1c_1 & a_{12}-l_1c_2\\ a_{21}-l_2c_1 & a_{22}-l_2c_2\end{bmatrix}
$$

Row by row:

$$
\begin{aligned}
\widehat{x}_1[k+1]&=(a_{11}-l_1c_1)\,\widehat{x}_1[k]+(a_{12}-l_1c_2)\,\widehat{x}_2[k]+b_1u_0[k]+l_1e_y[k]\\[2pt]
\widehat{x}_2[k+1]&=(a_{21}-l_2c_1)\,\widehat{x}_1[k]+(a_{22}-l_2c_2)\,\widehat{x}_2[k]+b_2u_0[k]+l_2e_y[k]
\end{aligned}
$$

With $a_{11}=1,\ a_{12}=T_s,\ a_{21}=0,\ a_{22}=1,\ b_1=T_s,\ b_2=0,\ c_1=1,\ c_2=0$:

$$
\begin{aligned}
\widehat{x}_1[k+1]&=(1-l_1)\,\widehat{x}_1[k]+T_s\,\widehat{x}_2[k]+T_s\,u_0[k]+l_1\,e_y[k]\\[2pt]
\widehat{x}_2[k+1]&=-\,l_2\,\widehat{x}_1[k]+\widehat{x}_2[k]+l_2\,e_y[k]
\end{aligned}
$$

Control law with $K=\begin{bmatrix}k_p&1\end{bmatrix}$:

$$
u_0[k]=-\bigl(k_p\,\widehat{x}_1[k]+\widehat{x}_2[k]\bigr),
\qquad
u[k]=\frac{u_0[k]}{b_0}
$$

One scan of the PLC:

```
ey = y - r
u0 = -(kp*x1 + x2)
u  = u0 / b0                            // to the actuator, limited to 0-100 %
x1 = (1 - l1)*x1_old + Ts*x2_old + Ts*u0 + l1*ey
x2 = -l2*x1_old + x2_old + l2*ey
```

$$
\begin{array}{|l|l|}
\hline
\textbf{Module input} & \textbf{Role}\\
\hline
e_y[k]=y[k]-r[k] & \text{feedback error, the only signal from the process}\\
k_p & \text{control gain}\\
l_1,\ l_2 & \text{observer gains}\\
b_0 & \text{plant input gain}\\
T_s & \text{sampling time}\\
\hline
\end{array}
$$

$$
\begin{array}{|l|l|}
\hline
\textbf{Internal state (memory)} & \textbf{Role}\\
\hline
\widehat{x}_1[k] & \text{estimated error}\\
\widehat{x}_2[k] & \text{estimated total disturbance}\\
\hline
\end{array}
$$

$$
\begin{array}{|l|l|}
\hline
\textbf{Module output} & \textbf{Role}\\
\hline
u[k]=u_0[k]/b_0 & \text{control signal to the actuator, limited to } 0-100\ \%\\
\hline
\end{array}
$$

$u_0$ is the normalized control acting on the $1/s$ chain; $b_0$ appears only at the actuator.

#### 2.1.5 Relacion GPI with ESO:
$$\dot{\widehat{X}}_{au} = (A_{au} - LC_{au})\widehat{X}_{au} + B_{au} u + Le_y$$
The observer only sees the (1/s) term because the augmented system contains only the (1/s) dynamics. Therefore, (b_0) is not used by the observer.
$$u=-(K\cdot\widehat{X}_{au})$$


$$\dot{\widehat{X}}_{au} = (A_{au} - LC_{au}-B_{au}K)\widehat{X}_{au}  + Le_y$$
$$A_c = A_{au} - LC_{au}-B_{au}K$$
$$B_c = L_{(n+m,1)}$$
$$C_c = I_{(n+m,n+m)}$$
$$D_c = 0_{(n+m,1)}$$

Transfer function:

$$\frac{X_{au}(s)}{E_y(s)}=G_{au}(s)=C_c[sI-A_c]^{-1}B_c+D_c$$

$$G_{au}(s)=\begin{bmatrix} G_{au_1}(s) \\ G_{au_2}(s)\\ \vdots \\ G_{au_{(n+m)}}(s)\end{bmatrix}$$

$$U(s) = -\frac{1}{b_0} \cdot K\cdot G_{au}(s) \cdot E_y$$
$$U(s) = -\frac{1}{b_0} \cdot K_{(1,n+m)}\cdot G_{au}(s)_{(n+m,1)} \cdot E_y$$
$$U(s) = -\frac{1}{b_0} \cdot K_{(1,n+m)}\cdot G_{au}(s)_{(n+m,1)} \cdot E_y$$

#### 2.1.5.1 Building the controller from integrators (GPI / flat filter)

Error dynamics of the design model, with $u_0=b_0u$ and $\xi$ the total disturbance:

$$
\dot e_y=u_0+\xi
$$

Only $e_y$, $u_0$ and their integrals may be used: no derivatives, no observer. The gains are numbered so that the subscript matches the power of $s$ it ends up multiplying in the characteristic polynomial.

**A. Proportional only.**

$$
u_0=-k_0e_y
\qquad\Longrightarrow\qquad
e_y\,(s+k_0)=\xi
\qquad\Longrightarrow\qquad
e_y=\frac{\xi}{s+k_0}
$$

For a constant $\xi$ of value $\xi(\infty)$ the final value is $\frac{\xi(\infty)}{k_0}\neq0$: offset. The internal model of $\xi$ is missing.

**B. Proportional plus integral (minimum GPI).**

$$
u_0=-k_1e_y-k_0\!\int_0^t\! e_y(\tau)\,d\tau
\qquad\Longrightarrow\qquad
s\,e_y+k_1e_y+\frac{k_0}{s}e_y=\xi
$$

$$
e_y\big[s^{2}+k_1s+k_0\big]=s\,\xi
\qquad\Longrightarrow\qquad
\lim_{t\to\infty}e_y=0
$$

The $s$ on the right-hand side is the integrator of the controller: it is what kills a constant $\xi$. This already works, and it is the pure GPI, $C(s)=(k_1s+k_0)/s$.

**C. Replacing the measurement by a reconstruction.** Integrating the plant once,

$$
\int_0^t\!\dot e_y\,d\tau=\int_0^t\! u_0(\tau)\,d\tau+\int_0^t\!\xi(\tau)\,d\tau
$$
$$
\qquad\Longrightarrow\qquad
(e_y(t)- e_y(0))=\int_0^t\! u_0\,d\tau+\int_0^t\!\xi\,d\tau
\qquad\Longrightarrow\qquad
e_y(t)=\int_0^t\! u_0\,d\tau+\int_0^t\!\xi\,d\tau+ e_y(0)
$$

Dropping what is unknown, $e_y(0)$ and $\xi$, leaves the integral reconstructor of the output:

$$
\widehat e_y(t)=\int_0^t\! u_0(\tau)\,d\tau
$$

Feeding back that reconstruction instead of the measured error:

$$
u_0=-k_2\!\int_0^t\! u_0\,d\tau-k_1\!\int_0^t\! e_y\,d\tau
\qquad\Longrightarrow\qquad
u_0\Big(1+\frac{k_2}{s}\Big)=-\frac{k_1}{s}e_y
\qquad\Longrightarrow\qquad
u_0=-\frac{k_1}{s+k_2}\,e_y
$$

$$
e_y\big[s^{2}+k_2s+k_1\big]=(s+k_2)\,\xi
\qquad\Longrightarrow\qquad
\lim_{t\to\infty}e_y=\frac{k_2\,\xi(\infty)}{k_1}\neq0
$$

It fails. The $\int u_0$ moved the $1/s$ into the denominator, so $C(s)=k_1/(s+k_2)$ has no pole at the origin: the integrator that rejected $\xi$ in **B** was spent on the reconstructor.

**D. One more integral of the error: the flat filter.**

$$
u_0=-k_2\!\int_0^t\! u_0(\tau)\,d\tau-k_1\!\int_0^t\! e_y(\tau)\,d\tau-k_0\!\iint e_y(\tau)\,d\tau
$$

$$
u_0(s)=-\frac{k_2}{s}u_0(s)-\frac{k_1}{s}e_y(s)-\frac{k_0}{s^{2}}e_y(s)
\qquad\Longrightarrow\qquad
u_0\Big(1+\frac{k_2}{s}\Big)=-\frac{k_1s+k_0}{s^{2}}\,e_y
$$

$$
\boxed{\;u_0(s)=-\frac{k_1s+k_0}{s\,(s+k_2)}\,e_y(s)\;}
$$

Closing with $\dot e_y=u_0+\xi$:

$$
s\,e_y=-\frac{k_1s+k_0}{s\,(s+k_2)}e_y+\xi
\qquad\Longrightarrow\qquad
s\,e_y\,s(s+k_2)+(k_1s+k_0)\,e_y=s\,(s+k_2)\,\xi
$$

$$
e_y\big[s^{3}+k_2s^{2}+k_1s+k_0\big]=s\,(s+k_2)\,\xi
\qquad\Longrightarrow\qquad
\lim_{t\to\infty}e_y=0
$$

**Bookkeeping.** What has to be counted is not how many integrals are written, but how many survive in $C(s)$:

$$
\text{net integrators of }C(s)=\#\!\int\! e_y-\#\!\int\! u_0=2-1=1=m
$$

one per degree of the polynomial model of $\xi$. A ramp disturbance would need three integrals of the error and still one of $u_0$.

**Implementation, integrators only.** With $x=e_y/[s(s+k_2)]$:

$$
\dot x_1=x_2,\qquad \dot x_2=-k_2x_2+e_y,\qquad u_0=-k_0x_1-k_1x_2,\qquad u=\frac{u_0}{b_0}
$$

All the internal signals are $x$ and its derivatives: this is what makes it a *flat* filter. Design is pole placement on $s^{3}+k_2s^{2}+k_1s+k_0$, three coefficients for three roots.

#### 2.1.5.2 Error-based observer $\rightarrow$ $C(s)$, step by step ($n=1$, $m=1$, continuous)

Closing the observer with the control law gives a state-space system whose input is $e_y$ and whose output is $\widehat{x}_{au}$:

$$
C_c\big[sI-A_c\big]^{-1}B_c+D_c,
\qquad
A_c=A_{au}-LC_{au}-B_{au}K,\quad
B_c=L_{(n+m,1)},\quad
C_c=I_{(n+m,n+m)},\quad
D_c=0_{(n+m,1)}
$$

Design model (integrator chain + constant disturbance) and gains, with $k_p$ the control gain:

$$
A_{au}=\begin{bmatrix}0&1\\0&0\end{bmatrix},\quad
B_{au}=\begin{bmatrix}1\\0\end{bmatrix},\quad
C_{au}=\begin{bmatrix}1&0\end{bmatrix},\quad
L=\begin{bmatrix}l_1\\l_2\end{bmatrix},\quad
K=\begin{bmatrix}k_p&1\end{bmatrix}
$$

The second entry of $K$ is $1$: it is the term that cancels $\widehat{x}_2=\widehat f$ in $u_0=-K\widehat{x}_{au}$.

**Closed-loop observer matrix.**

$$
A_c=\underbrace{\begin{bmatrix}0&1\\0&0\end{bmatrix}}_{A_{au}}
-\underbrace{\begin{bmatrix}l_1\\l_2\end{bmatrix}\begin{bmatrix}1&0\end{bmatrix}}_{LC_{au}}
-\underbrace{\begin{bmatrix}1\\0\end{bmatrix}\begin{bmatrix}k_p&1\end{bmatrix}}_{B_{au}K}
=\begin{bmatrix}0&1\\0&0\end{bmatrix}
-\begin{bmatrix}l_1&0\\l_2&0\end{bmatrix}
-\begin{bmatrix}k_p&1\\0&0\end{bmatrix}
$$

$$
A_c=\begin{bmatrix}-(k_p+l_1)&0\\-l_2&0\end{bmatrix}
$$

The $1$ of $A_{au}$ and the $1$ of $B_{au}K$ cancel, so the $(1,2)$ entry is zero.

**Resolvent.** Subtracting $A_c$ changes the sign of every entry:

$$
sI-A_c=\begin{bmatrix}s+k_p+l_1&0\\l_2&s\end{bmatrix},
\qquad
\det(sI-A_c)=s\,(s+k_p+l_1)
$$

$$
\big[sI-A_c\big]^{-1}=\frac{1}{s\,(s+k_p+l_1)}
\begin{bmatrix}s&0\\-l_2&s+k_p+l_1\end{bmatrix}
$$

**States as a function of the error.** The input matrix is $B_c=L$, so the resolvent multiplies the gain vector row by row:

$$
\frac{\widehat{x}_{au}(s)}{e_y(s)}=\big[sI-A_c\big]^{-1}L
=\frac{1}{s\,(s+k_p+l_1)}
\begin{bmatrix}s&0\\-l_2&s+k_p+l_1\end{bmatrix}
\begin{bmatrix}l_1\\l_2\end{bmatrix}
$$

$$
=\frac{1}{s\,(s+k_p+l_1)}
\begin{bmatrix}s\cdot l_1+0\cdot l_2\\-l_2\cdot l_1+(s+k_p+l_1)\,l_2\end{bmatrix}
=\frac{1}{s\,(s+k_p+l_1)}
\begin{bmatrix}l_1s\\l_2(s+k_p)\end{bmatrix}
$$

In the second row $-l_1l_2$ cancels with the $+l_1l_2$ of the product, which is what leaves $l_2(s+k_p)$.

**Control law.** With $u=-\dfrac{1}{b_0}K\widehat{x}_{au}=-\dfrac{1}{b_0}\big(k_p\widehat{x}_1+\widehat{x}_2\big)$, the transfer function from $-e_y$ to $u$ is

$$
b_0\,C(s)=k_p\frac{l_1s}{s(s+k_p+l_1)}+\frac{l_2(s+k_p)}{s(s+k_p+l_1)}
=\frac{(k_pl_1+l_2)\,s+k_pl_2}{s\,(s+k_p+l_1)}
$$


**Same thing as a GPI (flat filter).** Writing the controller as $b_0C(s)=\dfrac{k_1s+k_0}{s\,(s+k_2)}$, the GPI gains are read directly from the observer gains:

$$
\boxed{\;k_2=k_p+l_1,\qquad k_1=k_pl_1+l_2,\qquad k_0=k_pl_2\;}
$$

$$
e_y\Big[s^{3}+k_2s^{2}+k_1s+k_0\Big]=s\,(s+k_2)\,\xi
$$

Both the characteristic polynomial and the zeros of $e_y/\xi$ agree, so this is the same controller written as a compensation network. Note the $s$ in the denominator: without it the controller has no integrator and a constant $\xi$ is not rejected. Going back, $k_p$ is any real root of the cubic and then $l_1=k_2-k_p$, $l_2=k_0/k_p$.

#### 2.1.5.3 The same development in discrete time ($n=1$, $m=1$)

Design model discretized with $T_s$, and the same gains as above ($k_p$ the control gain, $L$ the observer gains):

$$
A=\begin{bmatrix}1&T_s\\0&1\end{bmatrix},\quad
B=\begin{bmatrix}T_s\\0\end{bmatrix},\quad
C=\begin{bmatrix}1&0\end{bmatrix},\quad
L=\begin{bmatrix}l_1\\l_2\end{bmatrix},\quad
K=\begin{bmatrix}k_p&1\end{bmatrix}
$$

**Closed-loop observer matrix.**

$$
A_c=A-LC-BK
=\begin{bmatrix}1&T_s\\0&1\end{bmatrix}
-\begin{bmatrix}l_1&0\\l_2&0\end{bmatrix}
-\begin{bmatrix}k_pT_s&T_s\\0&0\end{bmatrix}
=\begin{bmatrix}1-l_1-k_pT_s&0\\-l_2&1\end{bmatrix}
$$

The $T_s$ of $A$ cancels with the $T_s$ of $BK$, exactly as the $1$ did in continuous time.

**Resolvent.**

$$
zI-A_c=\begin{bmatrix}z-1+l_1+k_pT_s&0\\l_2&z-1\end{bmatrix},
\qquad
\det(zI-A_c)=(z-1)\,(z-1+l_1+k_pT_s)
$$

$$
\big[zI-A_c\big]^{-1}=\frac{1}{(z-1)(z-1+l_1+k_pT_s)}
\begin{bmatrix}z-1&0\\-l_2&z-1+l_1+k_pT_s\end{bmatrix}
$$

**States as a function of the error**, multiplying by $B_c=L$ row by row:

$$
\frac{\widehat{x}_{au}(z)}{e_y(z)}=\big[zI-A_c\big]^{-1}L
=\frac{1}{(z-1)(z-1+l_1+k_pT_s)}
\begin{bmatrix}(z-1)\,l_1+0\cdot l_2\\-l_2l_1+(z-1+l_1+k_pT_s)\,l_2\end{bmatrix}
=\frac{1}{(z-1)(z-1+l_1+k_pT_s)}
\begin{bmatrix}l_1(z-1)\\l_2(z-1+k_pT_s)\end{bmatrix}
$$

**Controller.** With $u=-\dfrac{1}{b_0}\big(k_p\widehat{x}_1+\widehat{x}_2\big)$:

$$
b_0\,C(z)=\frac{(k_pl_1+l_2)\,(z-1)+k_pl_2T_s}{(z-1)\,(z-1+l_1+k_pT_s)}
$$

The pole at $z=1$ is the integrator (internal model of a constant $\xi$); the extra pole sits at $z=1-(l_1+k_pT_s)$ and is the reconstructor, the discrete counterpart of $-(k_p+l_1)$.

**Closed loop** against the design plant $G(z)=T_s/(z-1)$, with $\xi$ at the plant input:

$$
e_y\Big[(z-1)^{2}(z-1+l_1+k_pT_s)+T_s\big((k_pl_1+l_2)(z-1)+k_pl_2T_s\big)\Big]=T_s\,(z-1)\,(z-1+l_1+k_pT_s)\,\xi
$$

and the characteristic polynomial separates as in continuous time:

$$
(z-1+k_pT_s)\ \cdot\ \big(z^{2}+(l_1-2)z+1-l_1+T_sl_2\big)
$$

the first factor being the controller pole and the second $\det(zI-A+LC)$, the observer.

**Same thing as a discrete flat filter.** Replacing each integrator by an accumulator $T_s/(z-1)$:

$$
u_0=-k_2\,\Sigma u_0-k_1\,\Sigma e_y-k_0\,\Sigma\Sigma e_y
\qquad\Longrightarrow\qquad
b_0\,C(z)=\frac{k_1T_s(z-1)+k_0T_s^{2}}{(z-1)\,(z-1+k_2T_s)}
$$

$$

#### 2.1.6 Anti-Integral windup


**Controller implementation with saturation feedback** (*The Control Handbook*, 2nd ed., §19.1.3)

Any linear controller can be written as a ratio of polynomials, with $L(s)$ monic of degree $n$: For that, we use the faster poles.

$$
U(s) = C(s)\,E(s) = \frac{P(s)}{L(s)}\,E(s), \qquad L(s) = s^{n} + l_{n-1}s^{n-1} + \cdots + l_0
$$

It is implemented with saturation feedback, using a stable monic polynomial $E_1(s)$ of the same degree $n$:

$$
V(s) = \frac{P(s)}{E_1(s)}\,E(s) - \frac{L(s)-E_1(s)}{E_1(s)}\,U(s), \qquad U = \mathrm{sat}(V)
$$

**When the actuator does not saturate ($V = U$)** the original controller is recovered:

$$
\begin{aligned}
U &= \frac{P}{E_1}\,E - \frac{L-E_1}{E_1}\,U \\[4pt]
U\left[1 + \frac{L-E_1}{E_1}\right] &= \frac{P}{E_1}\,E \\[4pt]
U\left[\frac{L}{E_1}\right] &= \frac{P}{E_1}\,E \\[4pt]
U &= \frac{P}{L}\,E
\end{aligned}
$$

**Example** (book): plant $\dfrac{1}{s}$ and controller

$$
C(s) = \frac{6s+5}{s} \qquad\Longrightarrow\qquad P(s) = 6s+5, \quad L(s) = s, \quad n = 1
$$

Closed loop:

$$
G_{cl}(s) = \frac{\dfrac{6s+5}{s}\cdot\dfrac{1}{s}}{1+\dfrac{6s+5}{s}\cdot\dfrac{1}{s}} = \frac{6s+5}{s^{2}+6s+5}
$$

Poles: $(s+1)(s+5)$. $E_1$ takes the $n = 1$ fastest one:

$$
E_1(s) = s+5
$$

Then:

$$
\begin{aligned}
V &= \frac{6s+5}{s+5}\,E(s) - \frac{s-(s+5)}{s+5}\,U(s) \\[4pt]
V &= \frac{6s+5}{s+5}\,E(s) - \frac{-5}{s+5}\,U(s)
\end{aligned}
$$


````python
def antiWindup(C, Gp, symb=None):
    """C = P/L -> [P/E1, (L-E1)/E1].  Handbook 19.1.3.  symb: 's', 'z' or None (auto)."""
    if symb is None:
        symb = 'z' if getattr(C, 'dt', 0) else 's'
    P = np.asarray(C.num[0][0], float) / C.den[0][0][0]
    L = np.asarray(C.den[0][0], float) / C.den[0][0][0]     # monic
    n = len(L) - 1
    p = feedback(C * Gp, 1).poles()
    
    # E1: the n fastest closed-loop poles
    fast = p[np.argsort(np.abs(p))][:n] if symb == 'z' else p[np.argsort(p.real)][:n]
    E1 = np.poly(fast)
    arg = (T_s,) if symb == 'z' else ()
    return [tf(P, E1, *arg), tf(L - E1, E1, *arg)]

for nom, C, Gp in (('s', Ctrl_tf, Gs_real), ('z', Ctrl_z, tf(Gz_real))):
    C1, C2 = antiWindup(C, Gp, nom)
    display(Latex(r"$E_1(%s) = %s$" % (nom, C1.den)))
    display(Latex(r"$P/E_1$  (on the error):")); display(C1)
    display(Latex(r"$(L-E_1)/E_1$  (on the saturated $u$):")); display(C2)
````

Saturation feedback (Handbook §19.1.3):

$$

$$

The characteristic polynomial of the loop is

$$
1+C(s)G(s)=0
\qquad\Longrightarrow\qquad
\Delta(s)=s^{3}+k_2s^{2}+k_1s+k_0
$$

Taking $E_1$ as $\Delta$ without its constant term, divided by $s$ — a monic, stable polynomial of degree 2 built only from the gains, with no factorization:

$$
E_1(s)=s^{2}+k_2s+k_1
$$

$$
v_1=\frac{k_1s+k_0}{b_0\,(s^{2}+k_2s+k_1)},
\qquad
v_2=\frac{\big[s^{2}+k_2s\big]-\big[s^{2}+k_2s+k_1\big]}{s^{2}+k_2s+k_1}
=\frac{-k_1}{s^{2}+k_2s+k_1}
$$

Two properties come for free: $E_1$ is stable whenever $k_2,k_1>0$, and $v_2(0)=-k_1/k_1=-1$ exactly, so the integral action is recovered without any cancellation of nearly equal numbers. When the actuator is not saturated, $u=v$ and the loop returns $u=(P/L)\,e=C(s)\,e$.

**Discrete time.** The two blocks above are discretized with the substitution used throughout the notebook,

$$
s\to\frac{z-1}{T_s}
$$

$$
v_1(z)=\frac{\dfrac{k_1(z-1)}{T_s}+k_0}
{b_0\left[\left(\dfrac{z-1}{T_s}\right)^{2}+k_2\dfrac{z-1}{T_s}+k_1\right]}
=\frac{k_1T_s(z-1)+k_0T_s^{2}}{b_0\,E_1(z)}
$$

$$
v_2(z)=\frac{-k_1}{\left(\dfrac{z-1}{T_s}\right)^{2}+k_2\dfrac{z-1}{T_s}+k_1}
=\frac{-k_1T_s^{2}}{E_1(z)}
$$

$$
E_1(z)=(z-1)^{2}+k_2T_s(z-1)+k_1T_s^{2}
=z^{2}-(2-k_2T_s)\,z+\big(1-k_2T_s+k_1T_s^{2}\big)
$$

Both keep the same denominator, and each $k_i$ picks up one power of $T_s$ per integration. $v_2(1)=-k_1T_s^{2}/(k_1T_s^{2})=-1$, still exact, and the roots of $E_1(z)$ are $z=1+T_s\lambda_i$ with $\lambda_i$ those of $E_1(s)$, inside the unit circle for the usual $T_s$.

**Difference equations.** With $\varepsilon_1=-(2-k_2T_s)$ and $\varepsilon_0=1-k_2T_s+k_1T_s^{2}$:

$$
v_1[n]=-\varepsilon_1v_1[n-1]-\varepsilon_0v_1[n-2]

$$
v_2[n]=-\varepsilon_1v_2[n-1]-\varepsilon_0v_2[n-2]-k_1T_s^{2}\,u[n-2]
$$

$$
v[n]=v_1[n]-v_2[n],\qquad u[n]=\mathrm{sat}_{0}^{100}\big(v[n]\big)
$$

Four states, no divisions: it fits in any small controller. $E_1$ is a free choice, so if a faster or slower recovery from saturation is wanted, any monic stable $s^{2}+\alpha_1s+\alpha_0$ works and only changes $v_2=\big[(k_2-\alpha_1)s-\alpha_0\big]/E_1$.

**Recursive discrete-time difference equation ($n=1$, $m=1$, reconstructing $e_y$)**

Starting from the transfer function (output $u(z)$, input $e(z)$):

$$
u(z) = \frac{1}{b_0}\cdot\frac{k_1T_s(z-1)+k_0T_s^{2}}{(z-1)\,(z-1+k_2T_s)}\cdot e(z)
$$

Cross-multiply and expand both products:

$$
b_0\Big[z^{2}-(2-k_2T_s)\,z+(1-k_2T_s)\Big]\,u(z)
=\Big[k_1T_s\,z+(k_0T_s^{2}-k_1T_s)\Big]\,e(z)
$$

Divide every term by $z^{2}$ (i.e. multiply by $z^{-2}$) to put it in causal delay-operator form:

$$
b_0\Big[1-(2-k_2T_s)z^{-1}+(1-k_2T_s)z^{-2}\Big]\,u(z)
=\Big[k_1T_s\,z^{-1}+(k_0T_s^{2}-k_1T_s)\,z^{-2}\Big]\,e(z)
$$

Apply the inverse z-transform, using $z^{-1}X(z)\leftrightarrow x[n-1]$:

$$
b_0\Big(u[n]-(2-k_2T_s)\,u[n-1]+(1-k_2T_s)\,u[n-2]\Big)
=k_1T_s\,e[n-1]+(k_0T_s^{2}-k_1T_s)\,e[n-2]
$$

Solving for $u[n]$ gives the recursive control law, ready to implement on a PLC:

$$
\boxed{\,u[n] = (2-k_2T_s)\,u[n-1]-(1-k_2T_s)\,u[n-2]
+\frac{1}{b_0}\Big[k_1T_s\,e[n-1]+(k_0T_s^{2}-k_1T_s)\,e[n-2]\Big]\,}
$$

This form needs the last two control outputs ($u[n-1],\,u[n-2]$) and the last two error
samples ($e[n-1],\,e[n-2]$) in memory. There is **no** $e[n]$ term: the controller is
strictly proper because $e_y$ is not fed back directly, only its reconstruction, so the
current sample reaches $u$ one step later. With $k_2=0$ the equation collapses to the
incremental PI of the minimum GPI.

Equivalent form with accumulators, which is the flat filter written as it is programmed:

$$
\Sigma u[n]=\Sigma u[n-1]+T_s\,u_0[n-1],\qquad
\Sigma e[n]=\Sigma e[n-1]+T_s\,e_y[n-1],\qquad
\Sigma\Sigma e[n]=\Sigma\Sigma e[n-1]+T_s\,\Sigma e[n-1]
$$

$$
u_0[n]=-k_2\,\Sigma u[n]-k_1\,\Sigma e[n]-k_0\,\Sigma\Sigma e[n],
\qquad u[n]=\frac{u_0[n]}{b_0}
$$

### 2.2.3 Closed-loop simulation

Escalón de consigna de 2,0 a 2,5 mg/L y escalón de perturbación, contra la planta
real $K_{dc}/(\tau s+1)$.

**Sobre la "perturbación estimada":** esta forma del GPI no tiene observador, así
que no hay ningún estado $\hat\xi$ que leer. La estimación está **implícita en la
acción integral**. Con la ley de 2.2.2, que reconstruye $e_y$, el controlador es
estrictamente propio y ya no tiene parte proporcional: se parte en su rama
integradora y su rama de reconstrucción,

$$b_0\,C(z) = \underbrace{\frac{k_0T_s/k_2}{z-1}}_{\text{modelo interno de }f}
+\underbrace{\frac{T_s(k_1k_2-k_0)/k_2}{z-1+k_2T_s}}_{\text{reconstructor}}$$

Comparando con la forma del ADRC, $b_0u=u_0=-k_p\widehat{x}_1-\widehat f$, donde
$\widehat{x}_1$ es el error reconstruido (la rama de arriba, el error filtrado por
$1/(z-1+k_2T_s)$), la estimación se lee **directamente de la señal de control**, sin
simular ningún integrador suelto:

$$\hat f = -b_0\,u-k_p\,\widehat{x}_1
\qquad\xrightarrow{\;e_y\to0\;}\qquad
\hat f = -b_0\,u \qquad [\mathrm{mg/(L\cdot s)}]$$

El término que falta es proporcional al error reconstruido, así que desaparece en
estacionario y en los tramos sin cambio de consigna; es lo que se grafica abajo. (El
integrador suelto en lazo abierto no sirve aquí: con 1,8 millones de muestras
acumula su propio error numérico y la estimación se va.)

Lo acumulado por la integral contiene en estacionario dos cosas: el sesgo que sostiene la
consigna ($r/K_{dc}$) y la compensación de la perturbación. Despejando la segunda:

$$\hat\xi = \frac{r}{K_{dc}} + \frac{\hat f}{b_0} \qquad [\%\ \text{de }u]$$

Ésa es la lectura comparable con la $\hat\xi$ del ESO de la sección 2.1, y es lo que
se grafica abajo.

````python
# ============================================================
# Simulacion en lazo cerrado del GPI  (n=1, m=2)  ·  POR BLOQUES, EN DISCRETO
#
#                            xi (perturbacion, % de u)
#                                       |
#    r --->(+)---- e ---->[ C(z) ]--u-->(+)--->[ G(z) ]----+---> y
#          -^                                              |
#           +----------------------------------------------+
#
#   C(z) = (1/b0)[k2(z-1)^2 + k1 T_s(z-1) + k0 T_s^2]/(z-1)^2  -> Ctrl_z, celda 2.2.2
#   G(z) = ZOH{ K_dc/(Tau s+1) }                               -> Gz_real, la planta REAL
#
#   Dos entradas (r y xi) en un lazo lineal = SUPERPOSICION: cada una tiene su
#   propia funcion de transferencia y las respuestas se suman.  Las cuatro salen
#   de feedback(), sin escribir un solo cociente a mano:
#
#       y = G_yr r + G_yd xi              u = G_ur r + G_ud xi
#
#       G_yr = CG/(1+CG)                  G_ur =  C/(1+CG)
#       G_yd =  G/(1+CG)                  G_ud = -CG/(1+CG) = -G_yr
#
#   (G_ud = -G_yr no es casualidad: lo que el control tiene que quitar es
#    exactamente lo que la perturbacion mete por el mismo punto de suma.)
#
#   Y la perturbacion estimada sale de la identidad del ADRC, leyendo la propia
#   senal de control  --  sin simular el integrador suelto, que en lazo abierto
#   y con millones de muestras acumula su error numerico y se va:
#
#       b0 u = u0 - f_est,   u0 = k2 e      ->      f_est = k2 e - b0 u
# ============================================================

# lsim() siempre devuelve tambien el estado interno; con una funcion de
# transferencia eso lanza un aviso que aqui no aplica (solo se usa la salida)
import warnings; warnings.filterwarnings('ignore', message='return_x specified')

#k2_, k1_, k0_ = float(coeff[1]), float(coeff[2]), float(coeff[3])
k1_, k0_ = float(coeff[1]), float(coeff[2])
# --- los bloques ---
Gz_tf = tf(Gz_real)                        # la planta real, discreta, como TF

G_yr = feedback(Ctrl_z * Gz_tf, 1)         # r  -> y
G_yd = feedback(Gz_tf, Ctrl_z)             # xi -> y
G_ur = feedback(Ctrl_z, Gz_tf)             # r  -> u
G_ud = -G_yr                               # xi -> u

print('polos del lazo cerrado en z :', np.array2string(
      np.sort_complex(pole(G_yr)), precision=8))
print('   los mismos, llevados a s :', np.array2string(
      np.log(np.sort_complex(pole(G_yr))) / T_s, precision=6))

# --- simulacion ---
t_g  = np.arange(0, 20000, T_s)
r_g  = np.where(t_g >= 4000,  2.5,   2.0)      # consigna [mg/L]
xi_g = np.where(t_g >= 12000, -15.0, 0.0)      # perturbacion [% de u]

y_g = lsim(G_yr, r_g, t_g)[0] + lsim(G_yd, xi_g, t_g)[0]     # salida
u_g = lsim(G_ur, r_g, t_g)[0] + lsim(G_ud, xi_g, t_g)[0]     # senal de control [%]

e_g      = r_g - y_g                                          # error
#f_est_g  = k2_ * e_g - bo * u_g                               # f estimada [mg/(L*s)]
f_est_g  = - bo * u_g      
xi_est_g = r_g / K_dc + f_est_g / bo                          # la misma, en % de u

k_ref = int(11000 / T_s); k_fin = -1
print('\nantes de la perturbacion (t=11000 s): y=%.4f  u=%.2f %%  xi_est=%+.2f %%  (aplicada %+.1f)'
      % (y_g[k_ref], u_g[k_ref], xi_est_g[k_ref], xi_g[k_ref]))
print('al final                 (t=20000 s): y=%.4f  u=%.2f %%  xi_est=%+.2f %%  (aplicada %+.1f)'
      % (y_g[k_fin], u_g[k_fin], xi_est_g[k_fin], xi_g[k_fin]))
print('error final |y - r| = %.2e mg/L' % abs(y_g[k_fin] - r_g[k_fin]))

# ============================================================
# Figuras
# ============================================================
fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.5), layout='constrained')
ax = axs[0, 0]
ax.plot(t_g, y_g, color='C0', linewidth=2, label='$y$  (OD)')
ax.plot(t_g, r_g, color='k', linewidth=1.4, linestyle='--', label='$r$  (consigna)')
ax.set_title('a · Salida contra consigna'); ax.set_ylabel('OD [mg/L]')

ax = axs[0, 1]
ax.plot(t_g, e_g, color='C0', linewidth=2, label='$e = r - y$')
ax.axhline(0, color='k', linewidth=1, linestyle=':')
ax.set_title('b · El error'); ax.set_ylabel('error [mg/L]')
ax.text(.97, .9, 'error final: %.1e mg/L' % abs(e_g[-1]), transform=ax.transAxes,
        ha='right', fontsize=9, color='#555555')

ax = axs[1, 0]
ax.plot(t_g, xi_est_g, color='C2', linewidth=2,
        label=r'$\hat\xi = r/K_{dc} + \hat f/b_0$  (leida de $u$)')
ax.plot(t_g, xi_g, color='k', linewidth=1.5, linestyle='--', label='$\\xi$ APLICADA')
ax.set_title('c · La perturbación: estimada contra aplicada'); ax.set_ylabel('[% de $u$]')

ax = axs[1, 1]
ax.plot(t_g, u_g, color='C4', linewidth=2, label='$u$  (al soplador)')
ax.axhline(100, color='C3', linestyle=':', linewidth=1.2, label='límite del soplador')
ax.axhline(0, color='C3', linestyle=':', linewidth=1.2)
ax.set_title('d · Señal de control'); ax.set_ylabel('$u$ [%]')

for ax in axs.flat:
    ax.minorticks_on(); ax.legend(fontsize=8.5, frameon=False, loc='best')
    ax.grid(visible=True, which='major', alpha=.35); ax.set_xlabel('tiempo [s]')

fig.suptitle('GPI $n=1$, $m=2$ · lazo cerrado DISCRETO contra la planta real',
             fontsize=13, fontweight='bold')
plt.show()
````

## 3.With Prefiltro LD:

### 3.1 Closed-Loop System for simulation — con prefiltro de consigna



El ADRC basado en el error es el de salida **al que se le ha quitado el prefiltro**
(Herbst & Madoński, §8.4, p. 141). Por eso la consigna entra cruda al observador y
su derivada acaba dentro de la perturbación total: con un escalón, `ṙ` es un impulso
y el ESO responde con lo que el libro llama *"giant control efforts"* (Fig. 6.13).

La solución del propio libro es devolverle el prefiltro (Fig. A.6.1):

$$u(z)=C_{FB}(z)\cdot\bigl[\,C_{PF}(z)\,r(z)-y(z)\,\bigr]$$

**Las matrices del lazo no cambian.** Se filtra `r` antes de entrar, y ya.

### El prefiltro, Tabla A.6.1 (primer orden)

$$C_{PF}(z)=\frac{\dfrac{\gamma_0}{\beta_0}+\dfrac{\gamma_1}{\beta_0}z^{-1}+\dfrac{\gamma_2}{\beta_0}z^{-2}}
{1+\dfrac{\beta_1}{\beta_0}z^{-1}}$$
### Los coeficientes, Tabla A.6.2, columna *General terms*
Se usa esa columna y no la de *bandwidth parameterization* porque ésta supone
**polo doble** de observador, y tus `pols_L` son dos polos distintos:
$$\beta_0=\frac{k_1l_1+l_2}{b_0}\qquad
\beta_1=\frac{T_sk_1l_2-k_1l_1-l_2}{b_0}$$

$$\gamma_0=\frac{k_1}{b_0}\qquad
\gamma_1=\frac{k_1}{b_0}(T_sl_2+l_1-2)\qquad
\gamma_2=\frac{k_1}{b_0}(1-l_1)$$
con `k1 = K_1[0,0]` y `l1, l2 = L`, los mismos de la celda de diseño. Nota que
`C_PF` sólo usa **cocientes** de estos coeficientes, así que `b0` se cancela y el
prefiltro no depende de él.

````python
# ============================================================
# Prefiltro de consigna C_PF(z)   ·   Herbst & Madonski, Tablas A.6.1 y A.6.2
#   Las matrices del lazo (Aau_e, Bau_e, Cau_e) NO se tocan.
# ============================================================

k1 = K_1[0, 0]                 # ganancia del controlador, de la celda de diseno
l1 = L[0, 0]                   # ganancias del observador
l2 = L[1, 0]

# Tabla A.6.2, columna "General terms".  b0 se cancela en C_PF, asi que se omite.
beta_0  = k1 * l1 + l2
beta_1  = T_s * k1 * l2 - k1 * l1 - l2
gamma_0 = k1
gamma_1 = k1 * (T_s * l2 + l1 - 2)
gamma_2 = k1 * (1 - l1)

print('k1 = %.6e     l1 = %.6e     l2 = %.6e' % (k1, l1, l2))
print('beta_0  = %+.6e   beta_1  = %+.6e' % (beta_0, beta_1))
print('gamma_0 = %+.6e   gamma_1 = %+.6e   gamma_2 = %+.6e' % (gamma_0, gamma_1, gamma_2))

# Tabla A.6.1, primer orden
C_PF = tf([gamma_0 / beta_0, gamma_1 / beta_0, gamma_2 / beta_0],
          [1.0,              beta_1  / beta_0, 0.0], T_s)

display(Latex(r"Reference signal prefilter $C_{PF}(z)$:"))
display(C_PF)
print('ganancia DC = %.6f   (tiene que ser 1: el prefiltro no cambia el regimen)'
      % float(dcgain(C_PF)))
````

````python
# ============================================================
# Simulacion del lazo cerrado CON prefiltro
#   Mismo sys_au_cl de arriba.  Lo unico que cambia: r pasa por C_PF antes.
# ============================================================
t  = np.arange(0, 10000, T_s)
r  = np.where(t >= 4000, 3.0, 2.0)          # consigna que teclea el operario [mg/L]
xi = np.where(t >= 7000, -15.0, 0.0)        # perturbacion [% de u] en t = 7000 s

r_pf = np.asarray(lsim(C_PF, U=r, T=t)[0]).ravel()      # <-- el prefiltro

u  = np.column_stack([r_pf, xi])            # al lazo entra la consigna FILTRADA

x0 = np.zeros(n + n + m)                    # sin inicializar nada

[Y, T, X] = lsim(sys_au_cl, U=u, T=t, X0=x0)

y_out   = (Cp_r @ X[:, :n].T).ravel()
xhat_au = X[:, n:].T
e_est   = xhat_au[0, :]
xi_hat  = xhat_au[1, :]
e_real  = y_out - r                         # error contra la consigna DEL OPERARIO
e_obs   = y_out - r_pf                      # error que el observador SI ve (consigna filtrada)
u_ctrl  = -(K @ xhat_au).ravel() / bo

# xi_hat esta en mg/(L*s) y lleva dentro -y/Tau, porque el modelo de diseno es el
# integrador ideal y la autorregulacion de la planta se fue a la perturbacion total.
xi_fisica = (xi_hat + y_out / Tau) / bo     # -> % de u, comparable con la xi aplicada

# trayectoria que impone el polo del controlador
w_c    = -pols[0]
k_pert = int(np.flatnonzero(xi != 0)[0])
nuevo  = np.r_[True, np.diff(r) != 0]
tramo  = np.cumsum(nuevo) - 1
k0     = np.flatnonzero(nuevo)[tramo]
y_teo  = r + (y_out[k0] - r) * np.exp(-w_c * (t - t[k0]))
y_teo[k_pert:] = np.nan

print('omega_c = %.6f rad/s  ->  tau_c = %.1f s\n' % (w_c, 1 / w_c))
print('CON prefiltro:')
print('   t63 del primer escalon = %.1f s   (tau_c = %.1f s)'
      % (t[np.argmax(y_out >= 0.632 * r[0])], 1 / w_c))
print('   u pico = %.1f %%   minimo = %.1f %%' % (u_ctrl.max(), u_ctrl.min()))
# C_PF tiene ganancia 1 en continua pero ~36 en alta frecuencia, asi que la consigna
# filtrada pega un pico de UNA muestra en cada escalon. La calidad del observador se
# mide fuera de esas muestras.
lejos = np.ones(len(t), bool)
lejos[:20] = False
for kk in np.flatnonzero(np.diff(r) != 0) + 1:
    lejos[max(0, kk - 5):kk + 20] = False
print('   pico de r_pf en el escalon = %.1f mg/L  (ganancia de C_PF en alta frecuencia)'
      % r_pf.max())
print('   max|e_obs - e_est| fuera de las transiciones = %.2e mg/L'
      % np.abs(e_obs - e_est)[lejos].max())
print('   perturbacion: estimada %.2f %%   ·   aplicada %.1f %%' % (xi_fisica[-1], xi[-1]))

# lo mismo sin prefiltro, para comparar
[_, _, X0_] = lsim(sys_au_cl, U=np.column_stack([r, xi]), T=t, X0=x0)
u0_ = -(K @ X0_[:, n:].T).ravel() / bo
y0_ = (Cp_r @ X0_[:, :n].T).ravel()
print('\nSIN prefiltro (para comparar):')
print('   t63 del primer escalon = %.1f s' % t[np.argmax(y0_ >= 0.632 * r[0])])
print('   u pico = %.1f %%' % u0_.max())
````

````python
fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.5), layout='constrained')

# a) salida contra consigna y contra el polo diseñado
ax = axs[0, 0]
ax.plot(T, y_out, color='C0', linewidth=2.4, label='$y$  (OD medido)')
ax.plot(T, r, color='k', linewidth=1.4, linestyle='--', label='$r$  (consigna del operario)')
ax.plot(T, y_teo, color='C1', linewidth=1.8, linestyle=':',
        label=r'$r+(y_0-r)\,e^{-\omega_c t}$')
ax.axvline(T[k_pert], color='C3', linewidth=1, alpha=.5)
ax.set_title('a · Salida, consigna cruda y consigna filtrada')
ax.set_ylabel('OD [mg/L]'); ax.set_ylim(0, 3.2)

# b) el error: real contra estimado
ax = axs[0, 1]
ax.plot(T, e_obs, color='C0', linewidth=3, label='$e_y$ que ve el observador $= y - C_{PF}r$')
ax.plot(T, e_est, color='C1', linewidth=1.6, linestyle='--', label='$\\hat e_y$  ESTIMADO')
ax.plot(T, e_real, color='C7', linewidth=1.1, alpha=.8, label='$y - r$  (contra la consigna cruda)')
ax.set_title('b · El error: el que ve el observador contra el que estima')
ax.set_ylabel('error [mg/L]')

# c) la perturbación
ax = axs[1, 0]
ax.plot(T, xi_fisica, color='C2', linewidth=2,
        label='$(\\hat\\xi + y/\\tau)/b_0$  = perturbación física estimada')
ax.plot(T, xi, color='k', linewidth=1.6, linestyle='--', label='$\\xi$ APLICADA')
ax.set_title('c · La perturbación física: estimada contra aplicada')
ax.set_ylabel('[% de $u$]')

# d) señal de control, con y sin prefiltro
ax = axs[1, 1]
ax.plot(T, u0_, color='C3', linewidth=1.2, alpha=.65, label='sin prefiltro (pico %.0f %%)' % u0_.max())
ax.plot(T, u_ctrl, color='C4', linewidth=2, label='con prefiltro (pico %.0f %%)' % u_ctrl.max())
ax.axhline(100, color='C3', linestyle=':', linewidth=1.2, label='límite del soplador')
ax.axhline(0, color='C3', linestyle=':', linewidth=1.2)
ax.set_title('d · Señal de control: lo que compra el prefiltro')
ax.set_ylabel('$u$ [%]'); ax.set_ylim(-20, 200)

for ax in axs.flat:
    ax.minorticks_on(); ax.legend(fontsize=8.5, frameon=False, loc='best')
    ax.grid(visible=True, which='major', alpha=0.35)
    ax.set_xlabel('tiempo [s]')

fig.suptitle('Lazo cerrado con prefiltro de consigna $C_{PF}(z)$  ·  Tablas A.6.1 y A.6.2',
             fontsize=13, fontweight='bold')
plt.show()
````

### 3.2 Closed-Loop System for simulation — con prefiltro simple de elección propia

El `C_PF(z)` de la Tabla A.6.1 es el prefiltro **exacto**: reproduce clavada la
respuesta del ADRC de salida. El precio es que tiene tres coeficientes en el
numerador, dos en el denominador, y una ganancia en alta frecuencia de ~36 — que
hace que la consigna filtrada pegue un pico de 72 mg/L durante una muestra.

El libro ofrece la alternativa en la Fig. 6.16: un prefiltro **de elección propia**
en el camino de la consigna.

> *"a custom prefilter `F(s)` could be optionally attached in the reference signal
> path… to help with the transient tracking performance"* (§6.4, receta)

> *"Using customized prefilters for error-based ADRC enables **true 2DOF design
> possibilities**, whereas the output-based ADRC design is fixed at a critically
> damped response"* (Fig. 6.17)

Para `N = 1` el más simple es de primer orden, con **un solo parámetro**:

$$F(s)=\frac{1}{T_{ref}\,s+1}$$

Igual que antes, **las matrices del lazo no cambian**: solo se filtra `r` antes de
entrar.

### Cómo se elige `T_ref`

`T_ref` es literalmente *"cuánto quiero que tarde la consigna en llegar"*. Y se
traduce directo a esfuerzo de control, porque lo que la consigna le pide al
actuador es proporcional a $\dot r = \Delta r/T_{ref}$. La celda de abajo lo barre.

El punto interesante está en `T_ref = 600 s`: da **el mismo pico de control que el
`C_PF` exacto** (75,1 %) siendo casi el doble de rápido, y sin el pico de 72 mg/L.
Pasado ese punto, subir `T_ref` ya no baja `u` — porque ese 75,1 % lo fija el
escalón de perturbación de `t` = 7000 s, no la consigna.


````python
# ============================================================
# Prefiltro simple de eleccion propia   ·   Herbst & Madonski, Fig. 6.16
#   F(s) = 1/(T_ref s + 1).  Un parametro.
#   Las matrices del lazo (Aau_e, Bau_e, Cau_e) siguen sin tocarse.
# ============================================================

T_ref = 1065.0                               # [s] cuanto quiero que tarde la consigna

F_ref = c2d(tf([1.0], [T_ref, 1.0]), T_s)
display(Latex(r"Prefiltro de eleccion propia $F(z)$:"))
display(F_ref)
print('ganancia DC = %.6f' % float(dcgain(F_ref)))

t  = np.arange(0, 10000, T_s)
r  = np.where(t >= 4000, 3.0, 2.0)          # consigna que teclea el operario [mg/L]
xi = np.where(t >= 7000, -15.0, 0.0)        # perturbacion [% de u] en t = 7000 s

# --- barrido de T_ref, para elegir con numeros ---
def _corre(rr):
    _, _, Xq = lsim(sys_au_cl, U=np.column_stack([rr, xi]), T=t, X0=np.zeros(n + n + m))
    yq = (Cp_r @ Xq[:, :n].T).ravel()
    uq = -(K @ Xq[:, n:].T).ravel() / bo
    return yq, uq

print('\n  filtro                    u pico     t63 de y    pico de r filtrada')
yq, uq = _corre(r)
print('  sin filtro              %8.1f %% %10.1f s %14.2f mg/L'
      % (uq.max(), t[np.argmax(yq >= 0.632 * r[0])], r.max()))
r_cpf = np.asarray(lsim(C_PF, U=r, T=t)[0]).ravel()
yq, uq = _corre(r_cpf)
print('  C_PF del libro          %8.1f %% %10.1f s %14.2f mg/L'
      % (uq.max(), t[np.argmax(yq >= 0.632 * r[0])], r_cpf.max()))
for Tq in [150.0, 300.0, 600.0, 1063.0, 2000.0]:
    rq = np.asarray(lsim(c2d(tf([1.0], [Tq, 1.0]), T_s), U=r, T=t)[0]).ravel()
    yq, uq = _corre(rq)
    marca = '   <- T_ref elegido' if Tq == T_ref else ''
    print('  F(s)=1/(%6.0f s+1)     %8.1f %% %10.1f s %14.2f mg/L%s'
          % (Tq, uq.max(), t[np.argmax(yq >= 0.632 * r[0])], rq.max(), marca))

# ============================================================
# Simulacion con el T_ref elegido
# ============================================================
r_f  = np.asarray(lsim(F_ref, U=r, T=t)[0]).ravel()      # <-- el prefiltro
x0   = np.zeros(n + n + m)

[Y, T, X] = lsim(sys_au_cl, U=np.column_stack([r_f, xi]), T=t, X0=x0)

y_out   = (Cp_r @ X[:, :n].T).ravel()
xhat_au = X[:, n:].T
e_est   = xhat_au[0, :]
xi_hat  = xhat_au[1, :]
e_real  = y_out - r                         # error contra la consigna DEL OPERARIO
e_obs   = y_out - r_f                       # error que el observador SI ve
u_ctrl  = -(K @ xhat_au).ravel() / bo

xi_fisica = (xi_hat + y_out / Tau) / bo     # -> % de u, comparable con la xi aplicada

w_c    = -pols[0]
k_pert = int(np.flatnonzero(xi != 0)[0])
nuevo  = np.r_[True, np.diff(r) != 0]
tramo  = np.cumsum(nuevo) - 1
k0     = np.flatnonzero(nuevo)[tramo]
y_teo  = r + (y_out[k0] - r) * np.exp(-w_c * (t - t[k0]))
y_teo[k_pert:] = np.nan

print('\nCON F(s) = 1/(%.0f s + 1):' % T_ref)
print('   t63 del primer escalon = %.1f s   (tau_c = %.1f s)'
      % (t[np.argmax(y_out >= 0.632 * r[0])], 1 / w_c))
print('   u pico = %.1f %%   minimo = %.1f %%' % (u_ctrl.max(), u_ctrl.min()))
print('   max|e_obs - e_est| = %.2e mg/L   (sin picos: F no amplifica en alta frecuencia)'
      % np.abs(e_obs - e_est)[20:].max())
print('   perturbacion: estimada %.2f %%   ·   aplicada %.1f %%' % (xi_fisica[-1], xi[-1]))
````

````python
# para comparar en la figura: las tres versiones
r_sin  = r
r_cpf_ = np.asarray(lsim(C_PF, U=r, T=t)[0]).ravel()
u_sin  = _corre(r_sin)[1]
u_cpf  = _corre(r_cpf_)[1]

fig, axs = plt.subplots(2, 2, figsize=(12.5, 7.5), layout='constrained')

# a) salida contra consigna cruda y filtrada
ax = axs[0, 0]
ax.plot(T, y_out, color='C0', linewidth=2.4, label='$y$  (OD medido)')
ax.plot(T, r, color='k', linewidth=1.4, linestyle='--', label='$r$  (consigna del operario)')
ax.plot(T, r_f, color='C2', linewidth=1.6, label='$F(z)\\,r$  (lo que ve el lazo)')
ax.axvline(T[k_pert], color='C3', linewidth=1, alpha=.5)
ax.set_title('a · Salida, consigna cruda y consigna filtrada')
ax.set_ylabel('OD [mg/L]'); ax.set_ylim(0, 3.4)

# b) el error: el que ve el observador contra el que estima
ax = axs[0, 1]
ax.plot(T, e_obs, color='C0', linewidth=3, label='$e_y$ que ve el observador $= y - F r$')
ax.plot(T, e_est, color='C1', linewidth=1.6, linestyle='--', label='$\\hat e_y$  ESTIMADO')
ax.plot(T, e_real, color='C7', linewidth=1.1, alpha=.8, label='$y - r$  (contra la cruda)')
ax.set_title('b · El error: el que ve el observador contra el que estima')
ax.set_ylabel('error [mg/L]')

# c) la perturbación física
ax = axs[1, 0]
ax.plot(T, xi_fisica, color='C2', linewidth=2,
        label='$(\\hat\\xi + y/\\tau)/b_0$  estimada')
ax.plot(T, xi, color='k', linewidth=1.6, linestyle='--', label='$\\xi$ APLICADA')
ax.set_title('c · La perturbación física: estimada contra aplicada')
ax.set_ylabel('[% de $u$]')

# d) las tres señales de control
ax = axs[1, 1]
ax.plot(T, u_sin, color='C3', linewidth=1.1, alpha=.6, label='sin filtro (%.0f %%)' % u_sin.max())
ax.plot(T, u_cpf, color='C1', linewidth=1.4, alpha=.9, label='$C_{PF}$ del libro (%.0f %%)' % u_cpf.max())
ax.plot(T, u_ctrl, color='C4', linewidth=2.2, label='$F(s)=1/(%.0f s+1)$  (%.0f %%)' % (T_ref, u_ctrl.max()))
ax.axhline(100, color='C3', linestyle=':', linewidth=1.2, label='límite del soplador')
ax.axhline(0, color='C3', linestyle=':', linewidth=1.2)
ax.set_title('d · Señal de control: los tres casos')
ax.set_ylabel('$u$ [%]'); ax.set_ylim(-20, 200)

for ax in axs.flat:
    ax.minorticks_on(); ax.legend(fontsize=8.5, frameon=False, loc='best')
    ax.grid(visible=True, which='major', alpha=0.35)
    ax.set_xlabel('tiempo [s]')

fig.suptitle('Lazo cerrado con prefiltro simple $F(s)=1/(T_{ref}s+1)$  ·  Fig. 6.16',
             fontsize=13, fontweight='bold')
plt.show()
````

# 4. El sistema diseñado contra la perturbación REAL de la planta
## 4.1 Observer ESO

Hasta aquí `ξ` ha sido un escalón de laboratorio. Esta sección usa la perturbación
medida en Tunja, exportada de `Datos/analisis/` con `exporta_perturbacion.py`
(que **no** modifica `analisis_dinamica.ipynb`: importa sus mismos módulos y
reproduce el camino 2 de su §14.2b).

**No se rediseña nada.** Entran tal cual el `sys_au_cl` de arriba —las mismas
`Aau_e`, `Bau_e`, `Cau_e`— y el prefiltro `F_ref` de la sección anterior. Para eso
se diseñaron.

### Lo único que hay que resolver: cómo entra `f`

La columna de `ξ` en `Bau_e` espera la perturbación **en % de `u`**, entrando por
`Bp_r`. La `f` del análisis no está en esas unidades ni mide lo mismo, y hacen
falta dos correcciones. Saltarse cualquiera de las dos hace que la simulación no
sirva.

### 4.1.1 · Unidades

El análisis de dinámica trabaja **por hora**; este cuaderno, **por segundo**:

| magnitud | en `Datos/analisis/` | aquí | conversión |
|---|---|---|---|
| `f` | mg/(L·h) | mg/(L·s) | ÷ 3600 |
| `b₀` | mg/L/(%·h) | mg/L/(%·s) | ÷ 3600 |
| tiempo | horas | segundos | × 3600 |
| muestreo de `f` | 1 min | — | se interpola |

Metida cruda en mg/(L·h), la perturbación queda unas **270 veces más pequeña** de
lo que es: el lazo no es que no la rechace, es que no hay nada que rechazar y la
gráfica sale plana.

### 4.1.2 · El polo de la planta, que si no se cuenta dos veces

`f` está definida por la forma que usa el ADRC,

$$\dot y = f + b_0\,u$$

así que **ya lleva dentro el término $-y/\tau$**. Y la planta que hay dentro de
`sys_au_e` es $K_{dc}/(\tau s+1)$, que también lo tiene. Hay que quitárselo antes
de inyectarla:

$$f=-\frac{y}{\tau}+b_0\,\xi
\qquad\Longrightarrow\qquad
\boxed{\;\xi=\frac{f+y/\tau}{b_0}\;}$$
Se usa $y\approx r$ porque el lazo regula. Sin esta corrección el resultado del OD
sale igual, pero `u` queda desplazada unos **40 puntos**, que es la diferencia
entre leer 36–116 % y leer 76–156 %.
### 4.1.3 Y al leer `ξ̂`, la vuelta
El observador trabaja sobre el integrador ideal, así que su `ξ̂` absorbe **también**
el polo de la planta. Para compararla con la aplicada hay que devolvérselo:
$$\hat\xi_{fis}=\frac{\hat\xi}{b_0}+\frac{y}{\tau\,b_0}$$
## 4.1.4 Dos decisiones del montaje
**La ventana es de 25 h**, no de 7 días. A `T_s` = 0,05 s —el del cuaderno, para
poder reusar `sys_au_cl` sin rediscretizar— una semana son 12 millones de muestras.
25 h son 1,8 millones, un ciclo diario completo, y corren en 20 s. Es además la
misma ventana y la misma consigna que usa el GPI en 2.2.4, para que la sección 5
pueda comparar los dos en la misma gráfica.
**La planta usa el `b₀` de diseño** (0,150 mg/L/(%·h)), no el medido (0,1338). Es
el caso optimista: un **12 % de desajuste** que aquí no se modela. Está dentro de
la incertidumbre conocida de `b₀`, y conviene declararlo en el documento.

````python
# ============================================================
# Lazo cerrado con la f medida en Tunja
#
#   REUSA el sistema que ya disenamos:  sys_au_cl  (las matrices de arriba, sin
#   tocar una sola) y  F_ref, el prefiltro simple de la seccion anterior.
#
#   Lo unico que hace falta resolver es como convertir la f medida a lo que la
#   columna de xi de Bau_e espera, que es "% de u". Son DOS correcciones:
#
#     1. UNIDADES.  f viene en mg/(L*h) y la columna espera % de u.  Metida
#        cruda queda ~270 veces mas pequena de lo que es: parece que el lazo
#        no rechaza nada cuando en realidad no hay perturbacion que rechazar.
#
#     2. EL POLO DE LA PLANTA.  f esta definida por  y' = f + b0*u,  asi que
#        YA lleva dentro el termino -y/Tau.  Y la planta de sys_au_cl tambien lo
#        tiene.  Si no se le quita, se cuenta dos veces y u sale desplazada
#        unos 40 puntos.
#
#            f = -y/Tau + b0 * xi        ->        xi = (f + y/Tau) / b0
#
#        Se usa y ~ r porque el lazo regula (el residuo queda por debajo del
#        0,2 % de sigma de f).
# ============================================================


DIR_AN = 'Datos/analisis/'
MOD    = 'M2'
HORAS  = 25                                     # un ciclo diario completo, el mismo que el GPI

NIV       = [2.0, 2.5, 1.5]                       # mg/L
HORAS_NIV = 4

par  = pd.read_csv(DIR_AN + 'f_parametros.csv', index_col=0)
b0_h = float(par.loc['b0', MOD])                # mg/L/(%*h), tal como viene del analisis
Fp   = pd.read_csv(DIR_AN + 'f_generada_%s_14d_1min.csv' % MOD)

t_f  = np.arange(0, HORAS * 3600, T_s)          # el T_s del cuaderno, no otro
f_h  = np.interp(t_f / 3600.0, Fp.h.to_numpy(), Fp.f.to_numpy())    # mg/(L*h)
f_s  = f_h / 3600.0                                                  # mg/(L*s)

r_op = np.full(len(t_f), 2.0)                   # consigna del operario [mg/L]
r_op       = np.array(NIV)[(t_f // (HORAS_NIV * 3600)).astype(int) % len(NIV)]
r_ff = np.asarray(lsim(F_ref, U=r_op, T=t_f)[0]).ravel()             # el prefiltro de arriba

#r_ff = np.asarray(lsim(C_PF, U=r_op, T=t_f)[0]).ravel() 
#r_ff = r_op
xi_f = (f_s +r_op / Tau  ) / bo                # -> % de u, con las dos correcciones

print('b0 del analisis = %.4f mg/L/(%%*h)   ·   bo de diseno = %.4f' % (b0_h, bo * 3600))
print('f   : media %+.2f, sigma %.2f mg/(L*h)' % (f_h.mean(), f_h.std()))
print('xi  : media %+.1f, sigma %.1f %% de u   <- lo que el actuador debe compensar'
      % (xi_f.mean(), xi_f.std()))
print('ventana de %d h -> %s muestras a T_s = %.2f s\n' % (HORAS, f'{len(t_f):,}', T_s))

[Y, T, X] = lsim(sys_au_cl, U=np.column_stack([r_ff, xi_f]), T=t_f, X0=np.zeros(n + n + m))

y_f   = (Cp_r @ X[:, :n].T).ravel()
eh_f  = X[:, 1]                                 # e_hat_y
xih_f = X[:, n + 1]                             # xi1_hat (la perturbacion,
                                                 # NO su derivada): el indice de la
                                                 # estimacion de f es SIEMPRE n+1, no
                                                 # n+m -- con m=1 coincidian, con m>1 no.
u_f   = -(K @ X[:, n:].T).ravel() / bo

# El estado xi_hat del observador ES la estimacion de la perturbacion total f,
# ya en mg/(L*s): basta pasarlo a mg/(L*h) para compararlo con la f medida.
# No hace falta pasar por % de u para nada.
f_est_h = xih_f * 3600.0                        # -> mg/(L*h), las unidades del analisis

# La misma cosa leida en puntos de soplador, por si hace falta el coste:
xi_est  = xih_f / bo  + y_f / (Tau * bo)        # -> % de u

# Trayectoria que impone el polo del controlador, tramo a tramo de consigna:
#     y(t) = r + (y(t0) - r) * exp(-omega_c * (t - t0))
w_c    = -pols[0]
nuevo  = np.r_[True, np.diff(r_op) != 0]        # True en cada cambio de consigna
tramo  = np.cumsum(nuevo) - 1
k0     = np.flatnonzero(nuevo)[tramo]           # muestra en que arranco su tramo
y_teo  = r_op + (y_f[k0] - r_op) * np.exp(-w_c * (t_f - t_f[k0]))

ok = t_f > 2 * 3600                             # fuera del transitorio de arranque
print('RESULTADO   (sys_au_cl y F_ref reusados, sin recrear ninguna matriz)')
print('   OD : IAE medio %.4f mg/L   ·   y entre %.3f y %.3f   ·   en banda +-0,2: %.1f %%'
      % (np.abs(y_f[ok] - r_op[ok]).mean(), y_f[ok].min(), y_f[ok].max(),
         100 * np.mean(np.abs(y_f[ok] - r_op[ok]) <= 0.2)))
print('   u  : media %.1f %%, sigma %.1f puntos, entre %.0f y %.0f %%'
      % (u_f[ok].mean(), u_f[ok].std(), u_f[ok].min(), u_f[ok].max()))
print('   fuera de 0-100 %% el %.1f %% del tiempo   <- el soplador no da para esto'
      % (100 * np.mean((u_f[ok] < 0) | (u_f[ok] > 100))))
print('   estimacion de f : max|f - f_est| = %.2f mg/(L*h)   (sigma de f = %.2f)'
      % (np.abs(f_h[ok] - f_est_h[ok]).max(), f_h.std()))
print('                     lo mismo en soplador: %.1f %% de u  (sigma de xi = %.1f)'
      % (np.abs(xi_f[ok] - xi_est[ok]).max(), xi_f.std()))
````

````python
hh = t_f / 3600.0                               # eje en horas

fig, axs = plt.subplots(4, 1, figsize=(13, 11), sharex=True, layout='constrained')

# a) el OD contra la consigna
ax = axs[0]
ax.plot(hh, y_f, color='C0', linewidth=1.1, label='$y$  (OD)')
ax.plot(hh, r_op, color='k', linewidth=1.3, linestyle='--', label='$r$  (consigna)')
ax.plot(hh, y_teo, color='C1', linewidth=1.8, linestyle=':',
        label=r'$r+(y_0-r)\,e^{-\omega_c t}$  (lo que impone el polo)')
ax.fill_between(hh, r_op - 0.2, r_op + 0.2, color='C2', alpha=.12, label='banda $\\pm$0,2 mg/L')
ax.set_title('a · Oxígeno disuelto contra la perturbación real de Tunja (%s, %d h)' % (MOD, HORAS))
ax.set_ylabel('OD [mg/L]'); ax.set_ylim(0 ,3)

# b) la perturbacion: aplicada contra estimada
ax = axs[1]
ax.plot(hh, f_h, color='k', linewidth=1.2, label='$f$ MEDIDA en la planta')
ax.plot(hh, f_est_h, color='C1', linewidth=1.0, linestyle='--',
        label='$\\hat f = 3600\\,\\hat\\xi$  estimada por el ESO')
ax.set_title('b · La perturbación total, en sus unidades: la que hay contra la que el ESO ve')
ax.set_ylabel('$f$ [mg/(L·h)]')
ax.text(.99, .06, 'máx. error de estimación: %.2f mg/(L·h)   ·   $\\sigma$ de $f$ = %.2f'
        % (np.abs(f_h[ok] - f_est_h[ok]).max(), f_h.std()),
        transform=ax.transAxes, ha='right', fontsize=9, color='#555555')

# segundo eje: la MISMA curva leida en puntos de soplador.  Delta_u = Delta_f/b0
# es lo que cuesta compensarla (el criterio de la seccion 10 del analisis), y es
# un escalado fijo, asi que los dos ejes quedan alineados exactamente.
ax_pts = ax.twinx()
ax_pts.set_ylim(np.array(ax.get_ylim()) / (bo * 3600.0))
ax_pts.set_ylabel('$f/b_0$  [% de $u$]', color='#555555')
ax_pts.tick_params(axis='y', colors='#555555')
ax_pts.grid(False)

# c) el soplador contra sus limites
ax = axs[2]
ax.plot(hh, u_f, color='C4', linewidth=1.0, label='$u = -K\\hat x / b_0$')
ax.axhline(100, color='C3', linestyle=':', linewidth=1.3, label='límite del soplador')
ax.axhline(0, color='C3', linestyle=':', linewidth=1.3)
ax.fill_between(hh, 100, np.maximum(u_f, 100), color='C3', alpha=.25)
ax.fill_between(hh, np.minimum(u_f, 0), 0, color='C3', alpha=.25)
ax.set_title('c · Señal de control: lo sombreado es lo que el actuador NO puede dar (%.1f %% del tiempo)'
             % (100 * np.mean((u_f[ok] < 0) | (u_f[ok] > 100))))
ax.set_ylabel('$u$ [%]')

# d) el error: el que ve el observador contra el que estima (estado X[:, n])
ax = axs[3]
e_obs_f = y_f - r_ff                            # e_y = y - F r: lo que el ESO tiene que estimar
ax.plot(hh, e_obs_f, color='C0', linewidth=1.8, label='$e_y = y - F\\,r$  (el que ve el observador)')
ax.plot(hh, eh_f, color='C1', linewidth=1.0, linestyle='--',
        label='$\\hat e_y$  estimado (estado del observador)')
ax.axhline(0, color='k', linewidth=.8)
lim = 1.15 * np.abs(np.r_[e_obs_f[ok], eh_f[ok]]).max()
ax.set_ylim(-lim, lim)
ax.set_title('d · El error: real contra estimado   ·   máx. $|e_y - \\hat e_y|$ = %.2e mg/L'
             % np.abs(e_obs_f[ok] - eh_f[ok]).max())
ax.set_ylabel('error [mg/L]'); ax.set_xlabel('tiempo [h]')

for ax in axs:
    ax.minorticks_on(); ax.legend(fontsize=9, frameon=False, loc='upper right')
    ax.grid(visible=True, which='major', alpha=0.35)
    ax.set_xlim(0, HORAS)
fig.suptitle('El sistema diseñado (sys_au_cl + $F_{ref}$) contra la perturbación medida en la PTAR',
             fontsize=13, fontweight='bold')
plt.show()
````

### 4.2 GPI contra la perturbación REAL de la planta

Los mismos bloques de arriba, sin tocar nada, ahora con la perturbación
medida en Tunja en vez del escalón. Se usa **la misma `f` y la misma conversión que la
sección 4** del ADRC, para que las dos arquitecturas queden medidas contra lo mismo:

$$\xi=\frac{f+y/\tau}{b_0}\qquad(y\approx r)$$

— unidades de mg/(L·h) a mg/(L·s), y restando el polo $-y/\tau$ que `f` ya lleva
dentro y que la planta real también tiene.

**La perturbación estimada, en unidades físicas.** Escribiendo la ley del GPI en la
forma del ADRC, $b_0u = u_0 - \hat f$ con $u_0 = k_2 e$, la estimación se lee de la
propia señal de control:

$$\hat f = k_2\,e - b_0\,u\quad[\mathrm{mg/(L\cdot s)}] \qquad\longrightarrow\qquad
3600\,\hat f\ \ [\mathrm{mg/(L\cdot h)}]$$
Es la lectura directamente comparable con la $\hat\xi$ del ESO de la sección 4.
**El escenario es el mismo que el del observador:** 25 h, consigna que salta cada
4 h entre 2,0 · 2,5 · 1,5 mg/L, y todo en tiempo discreto a $T_s$. Así la sección 5
puede poner las dos arquitecturas en la misma gráfica.

# 6. ADRC y GPI en el PLC ctrlX (Bosch Rexroth)

Simulación de los dos lazos en el software del PLC (`ADRC_Module.project`). El *trace* del ctrlX guarda como máximo 10 000 muestras
(≈ 500 s a $T_s$ = 50 ms), así que cada respuesta quedó en 10 archivos consecutivos con huecos cortos entre uno y otro:
`Datos/txtCtrl/01.txt … 10.txt` para el ADRC con observador y `G_01.txt … G_10.txt` para el GPI. Cada lazo se une en un solo
tramo a $T_s$, interpolando linealmente los huecos. Solo se registró la salida $y$.

Tiempo de asentamiento $t_s$: el instante desde el cual $|y-r| \le b\,r$ para siempre, con $b$ = 5 % y 2 %, medido en el reloj del PLC.

````python
# ============================================================
# ADRC y GPI en el PLC ctrlX (Bosch Rexroth): 10 trazas de y por lazo, unidas en un solo tramo
# ============================================================
import glob, os, re

R_PLC  = 2.0                                   # consigna de la simulacion en el PLC [mg/L]
BANDAS = (0.05, 0.02)
tau_c  = -1 / pols[0]                          # polo de diseno del controlador ADRC [s]
LAZOS  = {'ADRC': '[0-9][0-9].txt', 'GPI': 'GPI_*.txt'}

def lee_trazas(patron):
    arch = sorted(glob.glob(os.path.join('Datos', 'txtCtrl', patron)),
                  key=lambda a: int(re.findall(r'\d+', os.path.basename(a))[0]))
    d = pd.concat([pd.read_csv(a, sep=r'\s+', skiprows=3, header=None, names=['t', 'y'])
                     .assign(archivo=os.path.basename(a)[:-4]) for a in arch], ignore_index=True)
    d['t'] = d['t'] / 1000.0                   # ms -> s (reloj del PLC)
    return d

plc, res = {}, []
for lazo, patron in LAZOS.items():
    d = lee_trazas(patron)
    res.append(d.groupby('archivo').agg(t_ini=('t', 'first'), t_fin=('t', 'last'),
                                        y_ini=('y', 'first'), y_fin=('y', 'last'), n=('y', 'size'))
                .assign(lazo=lazo))
    # un solo tramo continuo a T_s: los huecos entre trazas se interpolan linealmente
    t_c = np.arange(d['t'].iloc[0], d['t'].iloc[-1], T_s)
    y_c = np.interp(t_c, d['t'], d['y'])
    t_s = {b: t_c[np.flatnonzero(np.abs(y_c - R_PLC) > b * R_PLC)[-1] + 1] for b in BANDAS}
    plc[lazo] = dict(t=t_c, y=y_c, t_s=t_s)
    print('%-4s  medidas %d  ·  tramo continuo %d (%.1f %% interpolado)  ·  '
          't_s 5 %% = %.0f s (%.1f min)  ·  t_s 2 %% = %.0f s (%.1f min)'
          % (lazo, len(d), len(t_c), 100 * (1 - len(d) / len(t_c)),
             t_s[0.05], t_s[0.05] / 60, t_s[0.02], t_s[0.02] / 60))
display(pd.concat(res).round(4))

# polo de diseno desde la primera muestra del ADRC
t0, y0  = plc['ADRC']['t'][0], plc['ADRC']['y'][0]
t_teo   = plc['GPI']['t'] if plc['GPI']['t'][-1] > plc['ADRC']['t'][-1] else plc['ADRC']['t']
y_teo   = R_PLC + (y0 - R_PLC) * np.exp(-(t_teo - t0) / tau_c)
print('polo de diseno: t_s 5 %% = %.0f s  ·  t_s 2 %% = %.0f s'
      % tuple(t0 + tau_c * np.log(abs(y0 - R_PLC) / (b * R_PLC)) for b in BANDAS))

estilo = {'ADRC': dict(color='C0', ls='-'), 'GPI': dict(color='C2', ls='-')}
marca  = {0.05: '--', 0.02: '-'}
fig, axs = plt.subplots(2, 1, figsize=(13, 9), layout='constrained')
for ax in axs:
    for lazo, p in plc.items():
        ax.plot(p['t'], p['y'], linewidth=2.2, label='$y$  %s' % lazo, **estilo[lazo])
        for b in BANDAS:
            ax.axvline(p['t_s'][b], color=estilo[lazo]['color'], linestyle=marca[b], linewidth=1.2,
                       label=r'$t_s$ %.0f %% %s = %.0f s  (%.1f min)'
                             % (100 * b, lazo, p['t_s'][b], p['t_s'][b] / 60))
    ax.plot(t_teo, y_teo, color='k', linestyle=':', linewidth=1.3,
            label=r'polo de diseño ADRC  $\tau_c$ = %.0f s' % tau_c)
    ax.axhline(R_PLC, color='k', linestyle='--', linewidth=1.2, label='$r$ = %.1f mg/L' % R_PLC)
    for b, c in zip(BANDAS, ('C1', 'C3')):
        ax.axhspan(R_PLC * (1 - b), R_PLC * (1 + b), color=c, alpha=.08)
    ax.set_xlabel('tiempo del PLC [s]'); ax.set_ylabel('OD [mg/L]')
    ax.minorticks_on(); ax.grid(visible=True, which='major', alpha=.35)
axs[0].set_title('a · Respuesta completa')
axs[0].legend(fontsize=8.5, frameon=False, loc='lower right', ncol=2)
axs[1].set_ylim(R_PLC * 0.9, R_PLC * 1.02)
axs[1].set_xlim(2000, max(p['t'][-1] for p in plc.values()) + 100)
axs[1].set_title('b · Acercamiento a las bandas del 5 % y 2 %')
fig.suptitle('ADRC y GPI en el PLC ctrlX (Bosch) · escalón de consigna a %.1f mg/L' % R_PLC,
             fontsize=13, fontweight='bold')
plt.show()
````

# 7. Precisión numérica: REAL frente a LREAL

El mismo escalón de 2,0 a 2,2 mg/L con el GPI en forma directa I y con el ADRC con observador, calculado con números de
32 bits (`REAL`) y de 64 bits (`LREAL`), con $T_s$ = 50 ms y con $T_s$ = 1 s. Sustenta la recomendación de tipo de dato de la
discretización del GPI (capítulo 5 de la tesis): con 50 ms, $1 + \alpha_1 + \alpha_2 \approx 10^{-6}$ y un `REAL` ya no lo
representa bien; con 1 s el resultado es el mismo con los dos tipos. El observador del ADRC no es sensible al tipo de dato.

````python
# ============================================================
# Precision numerica: REAL (32 bits) frente a LREAL (64 bits)
# Escalon de 2,0 a 2,2 mg/L con la planta nominal, sin saturacion ni retardo, arrancando en estacionario.
#   GPI  : forma directa I con los coeficientes de Tustin, E1 = s^2 + k2 s + k1
#   ADRC : observador de dos estados (ZOH) con K_d y L_d por ancho de banda
# Celda independiente: no usa variables de las celdas anteriores.
# ============================================================
import numpy as np

def _real_vs_lreal(horas_antes=1, horas_despues=2):
    tau, kdc = 1200.0, 0.05                              # planta nominal: [s] y [mg/(L·%)]
    b0 = kdc / tau; wc = 4 / tau; wo = 5 * wc
    k2, k1, k0 = wc + 2 * wo, 2 * wc * wo + wo**2, wc * wo**2

    def planta(T, Ts):                                   # planta exacta (ZOH): y[k+1] = a y[k] + b u[k]
        return T(np.exp(-Ts / tau)), T(kdc * (1 - np.exp(-Ts / tau)))

    def gpi(T, Ts):
        D0 = 4 + 2 * k2 * Ts + k1 * Ts**2
        a1, a2 = T((2 * k1 * Ts**2 - 8) / D0), T((4 - 2 * k2 * Ts + k1 * Ts**2) / D0)
        be0, be1, be2 = T((k0 * Ts**2 + 2 * k1 * Ts) / D0), T(2 * k0 * Ts**2 / D0), T((k0 * Ts**2 - 2 * k1 * Ts) / D0)
        g0, g1, g2 = T(-k1 * Ts**2 / D0), T(-2 * k1 * Ts**2 / D0), T(-k1 * Ts**2 / D0)
        a, b = planta(T, Ts); b0_T = T(b0); uno = T(1)
        N = int((horas_antes + horas_despues) * 3600 / Ts); kc = int(horas_antes * 3600 / Ts)
        y = T(2.0); x1 = x2 = T(0); v11 = v12 = T(0)     # estacionario en 2,0 mg/L con u = 40 %
        u1 = u2 = T(40.0); v21 = v22 = T(-40.0)
        Y = np.empty(N); U = np.empty(N)
        for k in range(N):
            x = (T(2.0) if k < kc else T(2.2)) - y       # x = -e_y
            v1 = be0 * x + be1 * x1 + be2 * x2 - a1 * v11 - a2 * v12
            resto = g1 * u1 + g2 * u2 - a1 * v21 - a2 * v22
            u = (v1 / b0_T - resto) / (uno + g0)         # v = v1/b0 - v2, con v2 = g0 u + resto y u = v
            v2 = g0 * u + resto
            Y[k], U[k] = y, u
            x2, x1, v12, v11, v22, v21, u2, u1 = x1, x, v11, v1, v21, v2, u1, u
            y = a * y + b * u
        suma = (1 + float(a1) + float(a2), 4 * k1 * Ts**2 / D0)   # 1 + a1 + a2: guardado en T, y exacto
        return Y, U, kc, suma

    def adrc(T, Ts):
        zc, zo = np.exp(-wc * Ts), np.exp(-wo * Ts)
        kp, l1, l2 = T((1 - zc) / Ts), T(2 * (1 - zo)), T((1 - zo)**2 / Ts)
        a, b = planta(T, Ts); b0_T = T(b0); Ts_T = T(Ts)
        N = int((horas_antes + horas_despues) * 3600 / Ts); kc = int(horas_antes * 3600 / Ts)
        y = T(2.0); xh1 = T(0); xh2 = T(-40.0 * b0)      # estacionario: u0 = -(kp xh1 + xh2) = 40 b0
        Y = np.empty(N); U = np.empty(N)
        for k in range(N):
            e = y - (T(2.0) if k < kc else T(2.2))
            u0 = -(kp * xh1 + xh2); u = u0 / b0_T
            Y[k], U[k] = y, u
            err = e - xh1
            xh1, xh2 = xh1 + Ts_T * xh2 + Ts_T * u0 + l1 * err, xh2 + l2 * err
            y = a * y + b * u
        return Y, U, kc, None

    print('%-6s %-8s %-6s %12s %12s %10s %12s' % ('lazo', 'Ts', 'tipo', 'OD antes', 'pico OD', 'pico u', 'OD a las 2 h'))
    for nombre, f in (('GPI', gpi), ('ADRC', adrc)):
        for Ts in (0.05, 1.0):
            for tipo, T in (('LREAL', np.float64), ('REAL', np.float32)):
                Y, U, kc, suma = f(T, Ts)
                print('%-6s %-8s %-6s %12.4f %12.4f %10.1f %12.4f' % (nombre, '%g s' % Ts, tipo, Y[kc - 1], Y[kc:].max(), U[kc:].max(), Y[-1])
                      + ('   1+a1+a2 = %.3e (exacto %.3e)' % suma if suma else ''))
    print('wo*Ts: %.5f con 50 ms y %.4f con 1 s; resolucion de un REAL cerca de 2: %.1e' % (wo * 0.05, wo * 1.0, np.spacing(np.float32(1.998))))

_real_vs_lreal()
````
